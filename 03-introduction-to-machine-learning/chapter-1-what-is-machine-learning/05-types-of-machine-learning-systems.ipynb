# **Các loại hệ thống Học máy (Types of Machine Learning Systems)**

- **Học máy (Machine Learning)** cho phép các hệ thống học các **mẫu (Patterns)** từ dữ liệu mà không cần được lập trình tường minh cho từng quy tắc riêng lẻ. Nhưng chính xác thì các hệ thống này học như thế nào?

- Hóa ra, không chỉ có một cách duy nhất. Các thuật toán **Machine Learning** thường được phân loại dựa trên *loại* nhiệm vụ học tập mà chúng thực hiện và *đặc điểm* của dữ liệu mà chúng học từ đó.

- Ba nhóm phổ biến nhất mà bạn sẽ gặp là:

  1. **Học có giám sát (Supervised Learning)**
  2. **Học không giám sát (Unsupervised Learning)**
  3. **Học tăng cường (Reinforcement Learning)**

## **1. Học có giám sát (Supervised Learning): Học với dữ liệu có nhãn**

- Hãy tưởng tượng bạn đang dạy một đứa trẻ nhận biết các loại trái cây khác nhau. Bạn cho trẻ xem một quả táo và nói “táo”, cho trẻ xem một quả chuối và nói “chuối”, cứ như vậy tiếp tục.

- Bạn cung cấp cả ví dụ (hình ảnh trái cây) và câu trả lời chính xác (tên loại trái cây). Đây chính là bản chất của **Supervised Learning**.

- Trong **Supervised Learning**, thuật toán **Machine Learning** được huấn luyện trên một **tập dữ liệu (Dataset)**, trong đó mỗi điểm dữ liệu bao gồm các **đặc trưng đầu vào (Input Features)** *và* đầu ra chính xác tương ứng, hay còn gọi là **nhãn (Label)**.

- Mục tiêu là để thuật toán học một **hàm ánh xạ (Mapping Function)** có khả năng dự đoán nhãn đầu ra cho những đặc trưng đầu vào mới mà nó chưa từng gặp.

- Phương pháp này được gọi là “có giám sát” vì các nhãn chính xác định hướng quá trình học, tương tự như cách giáo viên hướng dẫn học sinh.

### **1.1. Hai loại bài toán chính trong Supervised Learning**

- **Phân loại (Classification):**
  - Mục tiêu là dự đoán một **danh mục hoặc nhãn lớp rời rạc (Discrete Category/Class Label)**.

  - Hãy nghĩ đến những câu hỏi như: **“Email này có phải thư rác không?”, “Khối u này là ác tính hay lành tính?” hoặc “Trong bức ảnh này là loại động vật nào (mèo, chó hay chim)?”**.

  - Kết quả đầu ra là một danh mục cụ thể.

- **Hồi quy (Regression):**
  - Mục tiêu là dự đoán một **giá trị số liên tục (Continuous Numerical Value)**.

  - Hãy nghĩ đến những câu hỏi như: **“Nhiệt độ ngày mai sẽ là bao nhiêu?”, “Ngôi nhà này sẽ được bán với giá bao nhiêu?” hoặc “Tuần tới sẽ có bao nhiêu khách hàng đến cửa hàng?”**.

  - Kết quả đầu ra là một giá trị số trên một thang đo.

> *Quy trình cơ bản của **Supervised Learning**: Thuật toán học từ dữ liệu có nhãn để tạo ra một mô hình có khả năng dự đoán nhãn cho dữ liệu mới chưa được gán nhãn.*

- Hiện nay, phần lớn các ứng dụng **Machine Learning** thực tế sử dụng **Supervised Learning**, bởi vì dữ liệu có nhãn thường giúp đưa ra các dự đoán chính xác hơn cho những nhiệm vụ cụ thể.

- **Chương 3 và Chương 4** sẽ tìm hiểu sâu hơn về các kỹ thuật **Classification** và **Regression**.

## **2. Học không giám sát (Unsupervised Learning): Tìm kiếm cấu trúc trong dữ liệu không có nhãn**

- Bây giờ, hãy tưởng tượng bạn đưa cho đứa trẻ đó một chiếc hộp lớn chứa nhiều mảnh ghép Lego lẫn lộn mà không kèm theo bất kỳ hướng dẫn hay nhãn nào. Đứa trẻ có thể bắt đầu phân loại chúng dựa trên màu sắc, hình dạng hoặc kích thước, từ đó phát hiện ra những nhóm tự nhiên giữa các mảnh ghép.

- Điều này tương tự với **Unsupervised Learning**.

- Trong **Unsupervised Learning**, thuật toán được cung cấp dữ liệu đầu vào *mà không có* các nhãn đầu ra tương ứng.

- Mục tiêu là để thuật toán tự khám phá dữ liệu và tìm ra cấu trúc, mẫu hình hoặc mối quan hệ có ý nghĩa. Phương pháp này được gọi là “không giám sát” vì không có giáo viên hoặc câu trả lời chính xác định hướng quá trình học.

### **2.1. Các loại nhiệm vụ phổ biến trong Unsupervised Learning**

- **Phân cụm (Clustering):**
  - Nhóm các điểm dữ liệu tương tự nhau dựa trên những đặc trưng của chúng.

  - Ví dụ: nhóm những khách hàng có thói quen mua sắm tương tự để thực hiện tiếp thị có mục tiêu, hoặc nhóm các bài báo cùng nói về một chủ đề.

- **Giảm chiều dữ liệu (Dimensionality Reduction):**
  - Đơn giản hóa dữ liệu bằng cách giảm số lượng **đặc trưng (chiều dữ liệu — Dimensions)** trong khi vẫn giữ lại những thông tin quan trọng.

  - Phương pháp này có thể hữu ích cho việc **trực quan hóa dữ liệu (Data Visualization)** hoặc cải thiện hiệu suất của các thuật toán **Machine Learning** khác.

- **Học luật kết hợp (Association Rule Learning)**
  - Khám phá các quy tắc mô tả mối quan hệ giữa những mặt hàng trong các tập dữ liệu lớn.
  
  - Một ví dụ kinh điển là phát hiện rằng những khách hàng mua tã thường cũng mua bia.

> *Quy trình cơ bản của **Unsupervised Learning**: Thuật toán phân tích dữ liệu không có nhãn để tìm ra các mẫu hình hoặc nhóm ẩn.*

- **Unsupervised Learning** thường được sử dụng trong **phân tích dữ liệu khám phá (Exploratory Data Analysis — EDA)** và có thể tiết lộ những thông tin mà bạn chưa từng nghĩ đến.

- **Chương 5** sẽ giới thiệu một thuật toán **Clustering**.

## **3. Học tăng cường (Reinforcement Learning): Học thông qua thử và sai**

- Hãy nghĩ về cách bạn huấn luyện một chú chó.

  - Bạn ra lệnh “ngồi xuống”. Nếu chú chó thực hiện, bạn thưởng cho nó một món ăn ngon (**phần thưởng — Reward**). Nếu nó không thực hiện, nó sẽ không nhận được phần thưởng, hoặc có thể nhận được một sự điều chỉnh nhẹ (**hình phạt — Penalty**).

  - Theo thời gian, chú chó học được những hành động nào mang lại phần thưởng.

- **Reinforcement Learning** hoạt động dựa trên một nguyên lý tương tự.

- Trong **Reinforcement Learning**, một **tác nhân (Agent)** — tức thuật toán học — tương tác với một **môi trường (Environment)**.

- Tác nhân quan sát trạng thái hiện tại của môi trường và lựa chọn một hành động. Dựa trên hành động đó, môi trường chuyển sang trạng thái mới và cung cấp phản hồi cho tác nhân dưới dạng **Reward** (mang tính tích cực) hoặc **Penalty** (mang tính tiêu cực).

- Mục tiêu của tác nhân là học một chiến lược, được gọi là **chính sách (Policy)**, nhằm tối đa hóa tổng phần thưởng tích lũy theo thời gian.

### **3.1. Các thành phần chính trong Reinforcement Learning**

- **Agent:** Đối tượng học hoặc đưa ra quyết định.
- **Environment:** Môi trường mà tác nhân tương tác.
- **Trạng thái (State):** Tình huống hoặc cấu hình hiện tại của môi trường.
- **Hành động (Action):** Lựa chọn mà tác nhân thực hiện trong một trạng thái nhất định.
- **Reward/Penalty:** Phản hồi từ môi trường, cho biết kết quả của một hành động.

> *Vòng lặp tương tác trong **Reinforcement Learning**: Agent thực hiện hành động, Environment phản hồi bằng trạng thái mới và thông tin phản hồi, sau đó Agent học cách cải thiện hành động của mình.*

- **Reinforcement Learning** đặc biệt phù hợp với những bài toán cần đưa ra quyết định theo chuỗi, trong đó tác động của một hành động có thể không thể hiện rõ ngay lập tức.

- Ví dụ bao gồm:

  - Dạy robot đi bộ hoặc thực hiện các nhiệm vụ.
  - Phát triển trí tuệ nhân tạo chơi trò chơi (Game-Playing AI), chẳng hạn như AlphaGo.
  - Tối ưu hóa hệ thống điều khiển đèn giao thông.
  - Cá nhân hóa hệ thống đề xuất theo thời gian.

- Mặc dù rất mạnh mẽ, **Reinforcement Learning** thường phức tạp hơn trong quá trình triển khai so với **Supervised Learning** hoặc **Unsupervised Learning**. Vì vậy, phương pháp này thường được trình bày trong các khóa học nâng cao hơn.

## **4. Các biến thể khác (Other Variations)**

- Mặc dù ba loại trên là những nhóm chính, đôi khi bạn cũng có thể nghe nói đến các phương pháp khác:

  - **Học bán giám sát (Semi-Supervised Learning):**
    - Sử dụng kết hợp một lượng nhỏ dữ liệu có nhãn và một lượng lớn dữ liệu không có nhãn.

    - Phương pháp này hữu ích khi việc thu thập nhãn tốn kém hoặc mất nhiều thời gian.

  - **Học tự giám sát (Self-Supervised Learning):** Một dạng **Unsupervised Learning**, trong đó các nhãn được tạo tự động từ chính dữ liệu đầu vào.

    - Ví dụ: dự đoán từ tiếp theo trong một câu dựa trên những từ xuất hiện trước đó.

- Khi tiếp tục khóa học, bạn sẽ thấy các ví dụ chủ yếu đến từ **Supervised Learning** và **Unsupervised Learning**. Đây là hai nền tảng quan trọng của nhiều ứng dụng phổ biến.
