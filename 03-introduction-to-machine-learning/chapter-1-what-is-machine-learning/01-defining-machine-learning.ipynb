# **Defining Machine Learning**

- **Học máy (Machine Learning)** là lĩnh vực tập trung vào việc tạo ra các hệ thống có khả năng học từ data. Thay vì yêu cần lập trình viên viết lại các chỉ dẫn tường minh, từng bước cho mọi tình huống có thể xảy ra, một **hệ thống học máy (machine learning system)** xác định các **mẫu (patterns)** trong dữ liệu và xây dựng logic riêng dựa trên những **patterns** đó.

- Ví dụ: Cách bạn viết một chương trình truyền thống để lọc email rác (spam) hay không. Bạn sẽ cần tạo ra một danh sách quy tắc dài: *NẾu email chứa một số từ nhất định ('viagra', 'free', 'offer',...), hãy đánh dấu email đó là spam. Nếu email có định dạng bất thường, hãy đánh dấu là spam*. Danh sách quy tắc này có thể trở nên rất dài và khó duy trì. Những kẻ gửi thư rác liên tục thay đổi thủ đoạn, vì vậy bạn sẽ thường xuyên cập nhật các quy tắc.

- **Machine Learning** sử dụng một số cách tiếp cận khác. Thay vì tự viết các quy tắc, bạn cung cập cho hệ thống một số lượng lớn email đã được gặp nhãn là "thư rác (spam)" hoặc "không phải thư rác (ham)". Sử dụng một **thuật toán (algorithm)** cụ thể, hệ thống sẽ phân tích các ví dụ này và *học những patterns** thường dùng để phân biệt spam với email hợp lệ*. Hệ thống có thể học được rằng một số từ nhất định, tên miền của người gửi hoặc tổ hợp các **đặc trưng (features)** có khả năng cho thấy email là spam rất cao, mà bạn không cần phải lập trình tương minh những quy tắc cụ thể.

 > *Tom Mitchell (1997): "Một chương trình máy tính được cho là học từ kinh nghiệm E đối với một nhóm nhiệm vụ T và thước đo hiệu suất P nào đó, nếu hiệu suất của chương trình khi thực hiện các nhiệm vụ thuộc T, được đo bằng P, được cải thiện nhờ kinh nghiệm E".*

## **Hãy phân tích định nghĩa của Mitchell:**
  - **Nhiệm vụ (Task, T):** Hệ thống được kỳ vọng sẽ thực hiện công việc gì? Ví dụ bao gồm:
    - Phân loại email thành **spam** hoặc **ham**.
    - Dự đoán giá nhà dựa trên các **features** như diện tịch, vị trí, ...
    - Nhận dạng các chữ số viết tay.
  - **Kinh nghiệm (Experience, E):** Hệ thống học từ dữ liệu nào? Thông thường, đó là một **tập dữ liệu (dataset)** gồm các **samples**. Đối với việc lọc **spam**, kinh ngiệm chính là tập hợp các email đã được gắn nhãn. Đối với việc dự đoán giá nhà, đó là dữ liệu bạn nhà trong quá khứ, bao gồm các **features** của ngôi nhà với mức gia tương ứng.

  - **Thước đo hiệu suất (Performance Measure, P):** Làm sao để đánh giá hệ thông hoạt động tốt hay không? Đối với việc lọc **spam**, **P** có thể là độ chênh lệch trung bình giữa giá được dự đoán và giá thực tế.

- Mục tiêu là khi hệ thống có thêm kinh nghiệm (nhiều dữ liệu hơn), hiệu suất thực hiện nhiệm vụ của hệ thống sẽ được cải thiện.
