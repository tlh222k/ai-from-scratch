# **What is Python Programming?**

- **Lập trình (Programming)** cho phép chúng ta truyền đạt các chỉ dẫn cho máy tính hiểu. Về bản chất máy tính chỉ hiểu các chuỗi số $0$ hoặc $1$, thường được gọi là **mã máy (Machine Code)**. Vì viết trực tiếp machine code vô cùng khó khăn và tốn thời gian. Vậy nên, **Ngôn ngữ lập trình (Prgoramming Languages)** được tạo ra để đóng vai trò trong gian, cung cấp các viết chỉ dẫn để dễ đọc hơn đối với con người. Sau đó các chỉ dẫn này được chuyển đổi thành dạng mà máy tính có thể thực thi.

- Python là một ngôn ngữ lập trình như vậy. Python được biết đến như một ngôn ngữ **bậc cao (High-Level)**, **thông dịch - (Interpreted)** và **đa dụng (General-Purpose)**


## **1. Ngôn ngữ bậc cao (High-Level)**

- Điều này có nghĩa là cú pháp Python là **Syntax** được thiết kế gần với ngôn ngữ con người hơn, đồng thời che giấu nhiều chi tiết phức tạp bên dưới của phần cứng máy tính.

- Hãy hình dung điều này giống với việc chỉ đường khi lái xe: Thay vì mô tả từng lần xoay vô lăng và nhấn bàn đạp, bạn chỉ cần đưa ra chỉ dẫn như: "Rẽ trái ở giao lộ tiếp theo". Vì vậy, mã Python nhìn chung là dễ đọc, dễ viết và dễ bảo trì hơn sơ với các ngôn ngữ bậc thấp như C, Assembly, ...

## **2. Ngôn ngữ thông dịch (Interpreted)**

- Mã Python thường được thực thi từng dòng bởi một chương trình gọi là **trình thông dịch (Interpreter)**. Khi bạn chạy một **tập lệnh Python (Python Script)**, **Interpreter** sẽ đọc mã của bạn và gần như ngay lập tức thực hiện các hành động được chỉ định.

- Điều này khác với các **ngôn ngữ biên dịch (Compiled Langauges)**, trong đó toàn bộ mã trức tiên được chuyển thành **Machine code** rồi mới có thể chạy. Đặc tính **Interpreter** giúp đẩy nhanh quá trình phát triển, đặc biệt là khi kiểm thử và gỡ lỗi.

    ```text
    Mã Python  ──>  Trình thông dịch  ──>  Hành động của máy tính
    (print(...))     (Interpreter)       (hiển thị thông báo)
    ```

## **3. Ngôn ngữ đa dụng (General-Purpose)**

- Python không được thiết kế chỉ để thực hiện một nhiệm vụ cụ thể. Đây là ngôn ngữ linh hoạt, được sử dụng trong nhiều lĩnh vực, bao gồm:
  - Phát triển web (Web Development), chẳng hạn như các hệ thống phía máy chủ (Backend Systems) của website.
  - Phân tích dữ liệu (Data Analysis).
  - Trí tuệ nhân tạo (Artificial Intelligence — AI).
  - Tính toán khoa học (Scientific Computing).
  - Tự động hóa tác vụ (Task Automation), còn gọi là viết tập lệnh (Scripting).
  - Và nhiều lĩnh vực khác.

# **Nguồn gốc và triết lý thiết kế của Python**

- Python được tạo ra vào cuối thập niên 1980 bởi Guido van Rossum. Một mục tiêu quan trọng trong thiết kế là khả năng dễ đọc của mã (Code Readability).

- Cú pháp Python được cố ý thiết kế gọn gàng và nhấn mạnh việc sử dụng **khoảng trắng đầu dòng (Indentation)** để cấu trúc các khối mã. Điều này góp phần đáng kể vào tính rõ ràng của ngôn ngữ. Thông thường, mã Python được viết tốt gần như có thể đọc giống những câu tiếng Anh thông thường.

- Ví dụ, một chỉ dẫn đơn giản để hiển thị thông báo:

````python
print("Hello, Python learner!")
````

- Chỉ một dòng này đã thể hiện rõ mục đích của nó. Để đạt được kết quả tương tự bằng các ngôn ngữ bậc thấp, có thể cần nhiều bước thiết lập hơn đáng kể và mã cũng kém trực quan hơn.

# **Các đặc điểm cốt lõi của Python (Core Characteristics of Python)**

- Một số đặc điểm góp phần tạo nên sự phổ biến của Python, đặc biệt với người mới bắt đầu:
  1. **Dễ đọc và đơn giản (Readability and Simplicity)**: Như đã đề cập, cú pháp rõ ràng giúp việc học các kiến thức nền tảng về lập trình trở nên dễ dàng hơn, mà không bị sa lầy vào những quy tắc phức tạp.

  2. **Thư viện chuẩn phong phú (Large Standard Library)**:
      - Python đi kèm một bộ sưu tập lớn các mô-đun mã được viết sẵn, gọi là **thư viện chuẩn (Standard Library)**. Các mô-đun này xử lý những tác vụ phổ biến như làm việc với văn bản, số, tệp, mạng và nhiều thứ khác.

      - Điều này có nghĩa là bạn thường không cần tự viết mã từ đầu cho những thao tác cơ bản.

  3. **Cộng đồng và hệ sinh thái sôi động (Vibrant Community and Ecosystem)**:
      - Python có một cộng đồng toàn cầu đông đảo và tích cực. Nhờ đó, có rất nhiều tài liệu học tập, diễn đàn hỗ trợ và **một kho lớn các gói của bên thứ ba (Third-Party Packages)**, còn gọi là **thư viện (Libraries)**, có thể tìm thấy thông qua các công cụ như `pip`.

      - Những gói này mở rộng khả năng của Python sang gần như mọi lĩnh vực có thể hình dung.
  4. **Định kiểu động (Dynamic Typing)**:
      -  Trong Python, nhìn chung bạn không cần khai báo rõ ràng **kiểu dữ liệu (Data Type)** mà một biến sẽ lưu giữ — chẳng hạn đó là số hay văn bản. Python tự xác định kiểu dữ liệu này khi chương trình chạy.
      
      - Điều này mang lại sự linh hoạt, nhưng việc hiểu các kiểu dữ liệu (sẽ được trình bày trong chương tiếp theo) vẫn rất quan trọng.

# **Vì sao nên chọn Python làm ngôn ngữ đầu tiên?**

- Sự kết hợp giữa tính đơn giản, sức mạnh và tính linh hoạt khiến Python trở thành lựa chọn tuyệt vời cho ngôn ngữ lập trình đầu tiên. Python cung cấp nền tảng vững chắc có thể áp dụng trong nhiều lĩnh vực phát triển phần mềm và khoa học dữ liệu (Data Science).
