# **Perceptron đa tầng (Multi-Layer Perceptrons – MLPs): Bổ sung chiều sâu (Adding Depth)**

**Perceptron một tầng (Single-layer Perceptron)**, mặc dù có vai trò nền tảng, nhưng vẫn tồn tại một hạn chế đáng kể: nó chỉ có thể học các **mẫu hình phân tách tuyến tính (Linearly Separable Patterns)**. Những bài toán như **hàm XOR (Exclusive OR)** kinh điển nằm ngoài khả năng của nó, bởi vì các lớp (đầu ra $0$ và $1$) không thể được phân tách bằng một đường thẳng duy nhất trong không gian đầu vào. Để xử lý những mối quan hệ phi tuyến tính như vậy, chúng ta cần đưa thêm độ phức tạp vào cấu trúc mạng. Đây chính là lúc **MLPs** phát huy vai trò.

Ý tưởng cốt lõi đằng sau **MLPs** rất đơn giản nhưng hiệu quả: chèn một hoặc nhiều tầng neuron vào giữa tầng đầu vào và tầng đầu ra. Những tầng trung gian này được gọi là **tầng ẩn (Hidden Layers)**. Chúng được gọi là “ẩn” vì đầu ra của chúng không được quan sát trực tiếp như kết quả cuối cùng của mạng; thay vào đó, chúng đóng vai trò là đầu vào cho các tầng tiếp theo.

## **1. Kiến trúc của MLP (Architecture of an MLP)**

Một **MLP** bao gồm:

1. **Tầng đầu vào (Input Layer):** Tầng này tiếp nhận dữ liệu ban đầu (**Features**). Số lượng neuron trong tầng này tương ứng với số lượng **Features** trong dữ liệu đầu vào. Tầng này không thực hiện phép tính; nó chỉ truyền dữ liệu đến tầng ẩn đầu tiên.

2. **Các tầng ẩn (Hidden Layers):** Một hoặc nhiều tầng nằm giữa tầng đầu vào và tầng đầu ra. Mỗi tầng ẩn chứa một số neuron (còn được gọi là **đơn vị ẩn (Hidden Units)**). Những neuron này thực hiện các phép tính: chúng lấy tổng có trọng số của các đầu ra từ tầng trước đó, cộng thêm một **độ lệch (Bias)**, sau đó áp dụng một **hàm kích hoạt (Activation Function)**. Sự hiện diện của các **Hidden Layers** cho phép mạng học được những mẫu hình và cách biểu diễn phức tạp từ dữ liệu.

3. **Tầng đầu ra (Output Layer):** Đây là tầng cuối cùng tạo ra dự đoán của mạng. Số lượng neuron và **Activation Function** được sử dụng trong tầng này phụ thuộc vào nhiệm vụ cụ thể (ví dụ: một neuron với **Activation Function** dạng **Sigmoid** cho bài toán **phân loại nhị phân (Binary Classification)**, nhiều neuron với **Softmax** cho bài toán **phân loại đa lớp (Multi-class Classification)**, hoặc một neuron với **Activation Function** tuyến tính cho bài toán **hồi quy (Regression)**).

Thông tin trong một **MLP** thường truyền theo một hướng: từ tầng đầu vào, đi qua các **Hidden Layers** rồi đến tầng đầu ra. Loại kiến trúc này được gọi là **mạng truyền thẳng (Feedforward Network)**. Trong quá trình truyền thẳng (**Forward Pass**, giai đoạn dự đoán), không có vòng lặp hoặc kết nối nào gửi thông tin ngược trở lại bên trong cấu trúc các tầng. Nhìn chung, mỗi neuron trong một tầng được kết nối với *tất cả* neuron ở tầng tiếp theo, tạo thành cấu trúc được gọi là **các tầng kết nối đầy đủ (Fully Connected Layers)** hoặc **các tầng dày đặc (Dense Layers)**.

## **2. Khắc phục những hạn chế của tính tuyến tính (Overcoming Linearity Limitations)**

Các **Hidden Layers** giải quyết vấn đề mà **Single-layer Perceptron** gặp phải như thế nào? Điểm mấu chốt nằm ở sự kết hợp giữa nhiều tầng và các **Activation Functions** *phi tuyến tính*.

Hãy hình dung mỗi neuron trong tầng ẩn đầu tiên học một ranh giới tuyến tính đơn giản, tương tự như một **Perceptron** đơn lẻ. Tuy nhiên, đầu ra của những neuron này, sau khi được biến đổi bởi một **Activation Function** phi tuyến tính, sẽ trở thành đầu vào cho tầng tiếp theo. Khi đó, các neuron trong những tầng tiếp theo có thể kết hợp các phép biến đổi phi tuyến tính này.

Về bản chất, các **Hidden Layers** học cách biến đổi dữ liệu đầu vào thành một không gian biểu diễn mới. Trong không gian đã được biến đổi này, bài toán ban đầu vốn không thể phân tách tuyến tính có thể trở thành bài toán có thể phân tách tuyến tính. Đối với bài toán **Exclusive OR**, một **Hidden Layer** có thể học được các cách biểu diễn mà trong đó những điểm thuộc các lớp khác nhau có thể được neuron ở **Output Layer** phân tách bằng một đường thẳng. Việc bổ sung chiều sâu (nhiều tầng hơn) cho phép mạng học được những hàm ngày càng phức tạp và các **đặc trưng phân cấp (Hierarchical Features)** từ dữ liệu.

Để một **MLP** có thể học các hàm phi tuyến tính, những **Activation Functions** được sử dụng trong các neuron ẩn *bắt buộc* phải là hàm phi tuyến tính (ví dụ: **Sigmoid**, **Tanh**, **ReLU**). Nếu chỉ sử dụng các **Activation Functions** tuyến tính trong toàn bộ mạng, thì toàn bộ **MLP**, bất kể có bao nhiêu tầng, về mặt toán học sẽ thu gọn thành một mô hình tuyến tính một tầng tương đương, đưa chúng ta trở lại những hạn chế của **Perceptron**.

-----
Tóm lại, **MLPs** khắc phục những hạn chế của các mô hình một tầng bằng cách đưa vào các **Hidden Layers**. Những tầng này, khi kết hợp với các **Activation Functions** phi tuyến tính, cho phép **MLPs** học được các ánh xạ phi tuyến tính phức tạp giữa đầu vào và đầu ra, khiến chúng trở thành công cụ mạnh mẽ cho nhiều loại bài toán **máy học (Machine Learning)**. Chúng là nền tảng của nhiều kiến trúc **học sâu (Deep Learning)** tiên tiến hơn.
