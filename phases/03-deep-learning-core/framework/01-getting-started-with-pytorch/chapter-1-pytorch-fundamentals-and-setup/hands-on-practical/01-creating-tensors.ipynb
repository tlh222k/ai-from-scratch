# **1. Tạo Tensor (Creating Tensors)**

- Hãy thực hành tọa **Tensor** bằng nhiều phương pháp đã được giới thiệu.


## **1.1. Tạo Tensor từ danh sách Python (From Python Lists)**

- Tạo một **Tensor** có kích thước: `shape(2x3)` từ một danh sách Python lồng nhau.

```python
# Tạo Tensor từ Python List
data = [[1, 2, 3],
        [4, 5, 6]]

tensor_from_list = torch.tensor(data)
print(f"tensor_from_list:\n {tensor_from_list}")
print(f"Shape: {tensor_from_list.shape}")
print(f"Data type: {tensor_from_list.dtype}") # Mặc định là int64
```

## **1.2. Chỉ định kiểu dữ liệu (Specifying Data Type)**

- Tạo **Tensor** tương tự như trên nhưng chỉ định rõ kiểu dữ liệu là số thực dấu phẩy động 32 bit.

```python
# Tạo Tensor với dtype: float32
tensor_float32 = torch.tensor(data, dtype=torch.float32)

print(f"tensor_float32:\n {tensor_float32}")
print(f"Shape: {tensor_float32.shape}")
print(f"Data type: {tensor_float32.dtype}")
```

## **1.3. Sử dụng các hàm tạo Tensor (Using Factory Functions)**

- Tạo **Tensor** với kích thước và giá trị khởi tạo cụ thể.

```python
from numpy import random
# Tạo Tensor 3x4 chứa các số 0
zero_tensor = torch.zeros(3,4)
print(f"Zero Tensor (3x4): {zero_tensor}")
print(f"Shape: {zero_tensor.shape}")
print(f"Data type: {zero_tensor.dtype}")
print(50*"-")

# Tạo Tensor 3x4 chứa các số 1 với kiển int32
ones_tensor_int32 = torch.ones(3,4, dtype=torch.int32)
print(f"One Tensor (3x4): {ones_tensor_int32}")
print(f"Shape: {ones_tensor_int32.shape}")
print(f"Data type: {ones_tensor_int32.dtype}")
print(50*"-")

# Tạo Tensor một chiều biễu diễn một dãy số
range_tensor = torch.arange(0, 5, 1)
print(f"Range Tensor: {range_tensor}")
print(f"Shape: {range_tensor.shape}")
print(f"Data type: {range_tensor.dtype}")
print(50*"-")

# Tạo Tensor 2x3 với các giá trị ngẫu nhiên (phân phối đều từ 0 -> 1)
random = torch.rand(2,3)
print(f"Random Tensor (2x3): {random}")
print(f"Shape: {random.shape}")
print(f"Data type: {random.dtype}")
```
