## **Các phép toán cơ bản trên Tensor (Basic Tensor Operations)**

```python
a = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
b = torch.ones(2, 2) # Mặc định là float32

print("Tensor 'a':")
print(a)
print("Tensor 'b':")
print(b)

# Phép cộng từng phần tử
sum_tensor = a + b
# Cú pháp thay thế: sum_tensor = torch.add(a, b)
print("\nElement-wise Sum (a + b):")
print(sum_tensor)

# Phép nhân từng phần tử
prod_tensor = a * b
# Cú pháp thay thế: prod_tensor = torch.mul(a, b)
print("\nElement-wise Product (a * b):")
print(prod_tensor)

# Phép nhân với một số vô hướng
scalar_mult = a * 3
print("\nScalar Multiplication (a * 3):")
print(scalar_mult)

# Phép cộng tại chỗ (thay đổi trực tiếp 'a')
print(f"\n'a' before in-place add: ID {id(a)}")
a.add_(b) # Chú ý dấu gạch dưới ở cuối đối với các phép toán tại chỗ
print("'a' after in-place add (a.add_(b)):")
print(a)
print(f"'a' after in-place add: ID {id(a)}") # ID vẫn giữ nguyên

# Phép nhân ma trận
# Đảm bảo các kích thước tương thích với phép nhân ma trận
# Tạo các tensor có kích thước tương thích: x (2x3), y (3x2)
x = torch.rand(2, 3)
y = torch.rand(3, 2)
matmul_result = torch.matmul(x, y)
# Cú pháp thay thế: matmul_result = x @ y
print("\nMatrix Multiplication (x @ y):")
print(f"Shape of x: {x.shape}, Shape of y: {y.shape}")
print(f"Result shape: {matmul_result.shape}")
print(matmul_result)
```

- Hãy chú ý sự khác biệt giữa các **phép toán từng phần tử (element-wise operations)**, chẳng hạn $+$, $*$, và **phép nhân ma trận (matrix multiplication)** thông qua **`torch.matmul`** hoặc toán tử **@**.

- Ngoài ra, hãy lưu ý rằng các **phép toán tại chỗ (in-place operations)**, chẳng hạn **`add_`**, sẽ thay đổi trực tiếp Tensor hiện có thay vì tạo ra một đối tượng mới.
