# **Tạo Tensor (Creating Tensors)**

- **Tensor** là cấu trúc dữ liệu nền tảng trong PyTorch, dùng để biểu diễn **các mảng đa chiều (multidimensional arrays)** được sử dụng trong **các mô hình học sâu (deep learning models)**. PyTorch cung cấp nhiều phương pháp thực tế để tạo Tensor trong mã nguồn của bạn. Thư viện này có một tập hợp hàm linh hoạt để tạo Tensor, dù bạn sử dụng dữ liệu có sẵn hay cần tạo Tensor với hình dạng và giá trị khởi tạo cụ thể.


## **1. Tạo Tensor từ dữ liệu có sẵn (Creating Tensors from Existing Data)**

- Cách trực tiếp nhất để tạo Tensor thường là sử dụng các cấu trúc dữ liệu Python có sẵn, chẳng hạn như **danh sách (list) hoặc mảng NumPy**. Hàm chính được sử dụng cho mục đích này là **`torch.tensor()`**. Hàm này tự động suy luận kiểu dữ liệu (**`dtype`**) từ dữ liệu đầu vào, nhưng bạn cũng có thể chỉ định kiểu dữ liệu một cách tường minh.

- Điều quan trọng cần lưu ý là **`torch.tensor()` luôn sao chép dữ liệu đầu vào**.

- Hãy cùng xem cách hoạt động. Trước tiên, hãy bảo đảm rằng bạn đã nhập các thư viện PyTorch và NumPy:




```python
import torch
import numpy as np

# Tạo Tensor từ một danh sách Python
data_list = [[1, 2], [3, 4]]
tensor_from_list = torch.tensor(data_list)

print("Tensor from list:")
print(tensor_from_list)
print(f"Data type: {tensor_from_list.dtype}")
print(f"Shape: {tensor_from_list.shape}")
```

- Hãy chú ý rằng PyTorch đã tự suy luận kiểu dữ liệu là **`torch.int64`** (số nguyên 64 bit) dựa trên các số nguyên trong dữ liệu đầu vào.

- Bạn cũng có thể đạt được kết quả tương tự bằng cách bắt đầu với một mảng NumPy:

```python
# Tạo Tensor từ một mảng NumPy
data_numpy = np.array([[5.0, 6.0], [7.0, 8.0]])
tensor_from_numpy = torch.tensor(data_numpy)

print("\nTensor from NumPy array:")
print(tensor_from_numpy)
print(f"Data type: {tensor_from_numpy.dtype}")
print(f"Shape: {tensor_from_numpy.shape}")
```

- Trong trường hợp này, kiểu dữ liệu được suy luận là **`torch.float64`** vì mảng NumPy chứa các số thực dấu phẩy động (floating-point numbers). Trong phần sau của chương này, chúng ta sẽ tìm hiểu mối quan hệ giữa Tensor của PyTorch và mảng NumPy, bao gồm những phương pháp chuyển đổi hiệu quả hơn.



## **2. Tạo Tensor với hình dạng và giá trị cụ thể (Creating Tensors with Specific Shapes and Values)**

- Trong nhiều trường hợp, bạn cần tạo Tensor có kích thước nhất định ngay cả khi chưa có dữ liệu, chẳng hạn như khởi tạo Tensor bằng các giá trị $0$, giá trị $1$ hoặc giá trị ngẫu nhiên. PyTorch cung cấp các hàm chuyên dụng cho những trường hợp phổ biến này.

- Hình dạng (shape) thường được truyền vào dưới dạng một bộ (tuple) hoặc một chuỗi các số nguyên.



### **2.1. Tensor chứa các giá trị $0$, giá trị $1$ hoặc dữ liệu chưa được khởi tạo (Tensors of Zeros, Ones, or Uninitialized Data):**

- **`torch.zeros(*size, ...)`**: Tạo một Tensor chứa toàn các giá trị $0$.
- **`torch.ones(*size, ...)`**: Tạo một Tensor chứa toàn các giá trị $1$.
- **`torch.empty(*size, ...)`**: Tạo một Tensor có kích thước được chỉ định nhưng các phần tử chưa được khởi tạo. Giá trị của chúng sẽ phụ thuộc vào dữ liệu đang tồn tại tại vị trí bộ nhớ đó vào thời điểm cấp phát. Sử dụng **`torch.empty`** có thể nhanh hơn một chút nếu bạn dự định điền dữ liệu vào Tensor ngay sau đó. Tuy nhiên, hãy thận trọng vì các giá trị ban đầu không được xác định trước.

```python
# Xác định hình dạng mong muốn
shape = (2, 3) # 2 hàng, 3 cột

# Tạo Tensor với các giá trị cụ thể
zeros_tensor = torch.zeros(shape)
ones_tensor = torch.ones(shape)
empty_tensor = torch.empty(shape) # Các giá trị tùy ý

print(f"\nZeros Tensor (shape {shape}):")
print(zeros_tensor)

print(f"\nOnes Tensor (shape {shape}):")
print(ones_tensor)

print(f"\nEmpty Tensor (shape {shape}):")
print(empty_tensor)
```

- Theo mặc định, các hàm này tạo Tensor có **`dtype=torch.float32`**. Bạn có thể chỉ định kiểu dữ liệu khác bằng tham số **`dtype`**:



```python
# Tạo Tensor chứa các giá trị 1 với kiểu số nguyên
ones_int_tensor = torch.ones(shape, dtype=torch.int32)
print(f"\nOnes Tensor (dtype=torch.int32):")
print(ones_int_tensor)
```

### **2.2. Tensor chứa các giá trị ngẫu nhiên (Tensors with Random Values)**

- Việc khởi tạo các tham số mô hình thường bắt đầu bằng những giá trị ngẫu nhiên. PyTorch cung cấp các hàm sau để thực hiện việc này:

  - **`torch.rand(*size, ...)`**: Tạo Tensor có các phần tử được lấy mẫu theo phân phối đều (uniform distribution) trong khoảng **`[0, 1)`**.
  - **`torch.randn(*size, ...)`**: Tạo Tensor có các phần tử được lấy mẫu từ **phân phối chuẩn tắc (standard normal distribution)**, với giá trị trung bình bằng $0$ và phương sai bằng $1$.



```python
# Tạo Tensor chứa các giá trị ngẫu nhiên
rand_tensor = torch.rand(shape) # Phân phối đều [0, 1)
randn_tensor = torch.randn(shape) # Phân phối chuẩn tắc

print(f"\nRandom Tensor (Uniform [0, 1), shape {shape}):")
print(rand_tensor)

print(f"\nRandom Tensor (Standard Normal, shape {shape}):")
print(randn_tensor)
```

## **3. Tạo Tensor dựa trên Tensor khác (Creating Tensors Based on Other Tensors)**

- Đôi khi, bạn cần tạo một Tensor mới có cùng các thuộc tính với một Tensor hiện có, chẳng hạn như hình dạng (**`shape`**) và kiểu dữ liệu (**`dtype`**). Điều này hữu ích để bảo đảm tính tương thích giữa các phép toán. PyTorch cung cấp các biến thể hàm có hậu tố **`_like`** cho mục đích này:

  - **`torch.zeros_like(input_tensor, ...)`**
  - **`torch.ones_like(input_tensor, ...)`**
  - **`torch.rand_like(input_tensor, ...)`**
  - **`torch.randn_like(input_tensor, ...)`**

- Các hàm này nhận một Tensor hiện có làm đầu vào và trả về một Tensor mới chứa các giá trị được chỉ định — chẳng hạn như số $0$, số $1$ hoặc giá trị ngẫu nhiên — nhưng vẫn giữ nguyên hình dạng và kiểu dữ liệu (**`dtype`**) của Tensor đầu vào, trừ khi bạn chỉ định ghi đè các thuộc tính này.


```python
# Sử dụng một Tensor hiện có làm mẫu
base_tensor = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
print(f"\nBase Tensor (shape {base_tensor.shape}, dtype {base_tensor.dtype}):")
print(base_tensor)

# Tạo các Tensor có thuộc tính tương ứng với Tensor cơ sở
zeros_like_base = torch.zeros_like(base_tensor)
rand_like_base = torch.rand_like(base_tensor)

print("\nZeros tensor like base:")
print(zeros_like_base)
print(f"Shape: {zeros_like_base.shape}, dtype: {zeros_like_base.dtype}")

print("\nRandom tensor like base:")
print(rand_like_base)
print(f"Shape: {rand_like_base.shape}, dtype: {rand_like_base.dtype}")
```

---

- Như bạn có thể thấy, các Tensor mới **`zeros_like_base`** và **`rand_like_base`** tự động kế thừa thuộc tính **`shape`** (`torch.Size([2, 2])`) và **`dtype`** (`torch.float32`) từ **`base_tensor`**.

- Những phương pháp này cung cấp một bộ công cụ toàn diện để tạo ra các Tensor cần thiết, tạo nền tảng cho các phép tính được thực hiện trong mô hình học sâu. Trong các phần tiếp theo, chúng ta sẽ tìm hiểu cách thao tác với Tensor và thực hiện những phép toán thiết yếu.
