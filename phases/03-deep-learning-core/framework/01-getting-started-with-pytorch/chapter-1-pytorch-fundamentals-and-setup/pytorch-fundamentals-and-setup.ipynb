# **Phần 1. PyTorch là gì? (What is PyTorch?)**

- **PyTorch** là một thư viện và **Khung phát triển (Framework)** mã nguồn mở dành cho **Học máy (Machine Learning)**, dựa trên ngôn ngữ Python và được phát triển chủ yếu bởi Meta AI. Thư viện này đã trở nên phổ biến trong cả lĩnh vực nghiên cứu và công nghiệp nhờ khả năng xây dựng và huấn luyện các mô hình **Học sâu (Deep Learning)**. Đây là một bộ công cụ mạnh mẽ, được thiết kế chuyên biệt để đáp ứng nhu cầu của các ứng dụng **Machine Learning** hiện đại.

## **1.1. Tính toán Tensor (Tensor Computing)**

- Tương tự như **Mảng NumPy (NumPy Array)**, **PyTorch** cung cấp các **Mảng đa chiều (Multidimensional Array)** gọi là **Tensor**. Tuy nhiên, **Tensor** của PyTorch có một ưu điểm lớn:
  - Chúng có thể được xử lý dễ dàng trên **Bộ xử lý đồ họa (Graphics Processing Unit - GPU)** hoặc những bộ tăng tốc phần cứng chuyên dụng khác.

- Khả năng này giúp tăng tốc đáng kể các phép tính số học cần thiết để huấn luyện những **Mạng nơ-ron (Neural Network)** lớn, thường nhanh hơn nhiều bậc so với việc chỉ sử dụng **Bộ xử lý trung tâm (Central Processing Unit - CPU)**.

- Nếu bạn từng làm việc với NumPy, bạn sẽ thấy **Giao diện lập trình ứng dụng (Application Programming Interface - API)** dùng để thao tác với **Tensor** khá quen thuộc, giúp quá trình chuyển sang sử dụng **PyTorch** trở nên dễ dàng hơn.

## **1.2. Tự động tính vi phân (Automatic Differentiation)**

- Việc huấn luyện **Neural Network** đòi hỏi phải điều chỉnh các **Tham số (Parameter)** của mô hình dựa trên **Gradient** của **Hàm mất mát (Loss Function)**. Việc tính toán các **Gradient** này bằng tay khá phức tạp và dễ xảy ra sai sót, đặc biệt đối với các kiến trúc mạng sâu.

- PyTorch tích hợp một công cụ tự động tính vi phân mạnh mẽ có tên là **`Autograd`**. Khi các phép toán được thực hiện trên Tensor, **`Autograd`** sẽ xây dựng một **Đồ thị tính toán (Computation Graph)** một cách linh hoạt trong quá trình thực thi. Đồ thị này ghi lại chuỗi các phép toán đã thực hiện, cho phép PyTorch tự động tính toán **Gradient** bằng **Quy tắc dây chuyền (Chain Rule)** khi được yêu cầu, thường thông qua lời gọi **`.backward()`**.

- Tính linh hoạt này mang lại nhiều lợi thế trong việc thiết kế mô hình so với các **Framework** yêu cầu phải định nghĩa **Đồ thị tính toán tĩnh (Static Computation Graph)** ngay từ đầu.

<div style="text-align: center;">

</div>

## **1.3. Tại sao nên chọn PyTorch? (Why Choose PyTorch?)**

- Có nhiều yếu tố góp phần giúp PyTorch được sử dụng rộng rãi:

  - **Giao diện thân thiện với Python (Pythonic Interface)**: PyTorch tích hợp chặt chẽ với hệ sinh thái khoa học dữ liệu của Python. **API** của thư viện được thiết kế để mang lại cảm giác trực quan và tự nhiên cho các lập trình viên Python, giúp việc học và sử dụng tương đối dễ dàng. Việc **Gỡ lỗi (Debugging)** thường tương tự như gỡ lỗi mã Python thông thường.

  - **Tính linh hoạt (Flexibility)**: **Đồ thị tính toán động (Dynamic Computation Graph)**, được xác định trong quá trình chạy chương trình, cho phép cấu trúc của mạng thay đổi trong khi thực thi. Điều này đặc biệt hữu ích đối với một số loại mô hình, chẳng hạn như **Mạng nơ-ron hồi quy (Recurrent Neural Network - RNN)**, trong đó độ dài của chuỗi dữ liệu có thể thay đổi, hoặc khi cần triển khai các luồng điều khiển phức tạp bên trong mô hình.

  - **Hệ sinh thái (Ecosystem)** phong phú: PyTorch được hỗ trợ bởi một tập hợp đa dạng các thư viện và công cụ. Những thư viện như **`torchvision`**, **`torchaudio`** và **`torchtext`** cung cấp:
    - **Tập dữ liệu dựng sẵn (Pre-built Datasets)**
    - **Kiến trúc mô hình (Model Architectures)**
    - **Phép biến đổi dữ liệu (Data Transformations)** dành cho những lĩnh vực cụ thể, lần lượt là **Thị giác máy tính (Computer Vision)**, **Âm thanh (Audio)** và **Xử lý ngôn ngữ tự nhiên (Natural Language Processing - NLP)**.
    - Việc tích hợp với các công cụ như **TensorBoard** để trực quan hóa dữ liệu cũng giúp cải thiện quy trình phát triển.

  - **Nghiên cứu (Research)** và **Triển khai thực tế (Production)**: Ban đầu, PyTorch được cộng đồng nghiên cứu ưa chuộng nhờ tính linh hoạt. Tuy nhiên, thư viện này đã phát triển đáng kể và hiện có các công cụ như **TorchServe** và **TorchScript**, khiến nó trở thành một lựa chọn khả thi để triển khai mô hình vào môi trường vận hành thực tế.

### **2.1.1. Conda**

- **Conda** thường được ưu tiên trong các lĩnh vực **Khoa học dữ liệu (Data Science)** và **Machine Learning**.

- **Conda** quản lý cả gói phần mềm lẫn môi trường, giúp đơn giản hóa việc xử lý các **Phần phụ thuộc (Dependencies)** phức tạp, bao gồm cả những thư viện không thuộc Python. Nếu bạn đã sử dụng Anaconda, đây thường là phương pháp thuận tiện nhất.

### **2.1.2. Pip**

- **Pip** là trình cài đặt gói tiêu chuẩn của Python.
- **Pip** hoạt động tốt, đặc biệt nếu bạn muốn quản lý môi trường Python bằng các công cụ như **venv**. Nếu cần hỗ trợ **GPU**, bạn cần bảo đảm trình điều khiển NVIDIA đã được cài đặt sẵn trên hệ thống (xem mục 2.3).

### **2.2.1. Sử dụng Conda (Using Conda)**

- Trước tiên, tạo và kích hoạt một môi trường Conda mới. Thay `pytorch_env` bằng tên bạn muốn sử dụng và chọn phiên bản Python phù hợp:

```bash
# Tạo môi trường (có thể chọn phiên bản Python khác, ví dụ 3.10, 3.12)
conda create -n pytorch_env python=3.11

# Kích hoạt môi trường
conda activate pytorch_env
```

- Tiếp theo, truy cập trang web PyTorch và chọn hệ điều hành, Pip, phiên bản Python và phiên bản CUDA mong muốn (hoặc CPU), vì PyTorch được cài đặt thông qua Pip ngay cả khi chạy bên trong môi trường Conda.
- Sao chép lệnh được tạo:

```bash
# Lệnh Pip ví dụ cho CUDA 11.8, chạy bên trong môi trường Conda đã kích hoạt
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118

# Lệnh Pip ví dụ chỉ dành cho CPU, chạy bên trong môi trường Conda đã kích hoạt
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu
```

> Số hiệu CUDA (như `cu118`) chỉ là ví dụ; hãy thay bằng phiên bản mà trang pytorch.org đề xuất tại thời điểm bạn cài đặt.

### **2.2.2. Sử dụng Pip (Using Pip)**

- Trước tiên, tạo và kích hoạt một môi trường ảo mới bằng **venv**:

```bash
# Tạo môi trường (trong thư mục dự án)
python -m venv pytorch_env

# Kích hoạt môi trường
# Linux/macOS:
source pytorch_env/bin/activate
# Windows (Command Prompt):
.\pytorch_env\Scripts\activate
# Windows (PowerShell):
.\pytorch_env\Scripts\Activate.ps1
```

- Tiếp theo, truy cập trang web PyTorch và chọn hệ điều hành, Pip, phiên bản Python và phiên bản CUDA mong muốn (hoặc CPU). Sao chép lệnh được tạo. Lệnh này thường yêu cầu chỉ định URL chỉ mục gói (index URL) và có thể bao gồm cả phiên bản CUDA:

```bash
# Lệnh Pip ví dụ cho CUDA 11.8
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118

# Lệnh Pip ví dụ chỉ dành cho CPU
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu
```

- Chạy lệnh do trang web cung cấp trong môi trường ảo đã được kích hoạt. Hãy bảo đảm rằng Pip đã được cập nhật lên phiên bản mới nhất:

```bash
pip install --upgrade pip
```

## **2.3. Hỗ trợ GPU với CUDA (GPU Support with CUDA)**

- Các phép tính **Deep Learning**, đặc biệt là **Phép nhân ma trận (Matrix Multiplication)**, có thể được thực hiện nhanh hơn đáng kể trên **GPU** của NVIDIA bằng cách sử dụng nền tảng tính toán song song CUDA.

  - **Yêu cầu (Requirement):** Bạn cần có **GPU** NVIDIA tương thích với CUDA.

  - **Trình điều khiển và thư viện CUDA (Drivers & Libraries):**
    - Hệ thống của bạn phải có **Trình điều khiển (Driver)** đồ họa NVIDIA đủ mới để hỗ trợ phiên bản CUDA mà bản PyTorch bạn chọn được xây dựng dựa trên đó, chẳng hạn như `11.8`, `12.1`…
    - Các bản PyTorch dựng sẵn (cài bằng **Pip** hoặc **Conda**) thường đã đi kèm các thư viện CUDA runtime cần thiết, nên thông thường bạn không cần cài riêng **Bộ công cụ CUDA (CUDA Toolkit)** — trừ khi bạn tự biên dịch mã CUDA tùy chỉnh.
    - Hãy tham khảo tài liệu của NVIDIA và hướng dẫn cài đặt PyTorch để biết thông tin cụ thể.

  - **Lệnh cài đặt (Installation Command):** Khi sử dụng công cụ tạo lệnh cài đặt trên trang web PyTorch, việc chọn phiên bản CUDA sẽ điều chỉnh lệnh để cài đặt các tệp nhị phân tương thích với **GPU**.

  - Nếu bạn không có **GPU** tương thích hoặc chưa cần tăng tốc bằng **GPU**, chỉ cần chọn tùy chọn **CPU** trong quá trình cài đặt. Bạn vẫn có thể thực hiện mọi thao tác trên **CPU**, mặc dù tốc độ có thể chậm hơn khi xử lý các mô hình lớn.

## **2.4. Xác minh cài đặt (Verifying the Installation)**

- Sau khi cài đặt hoàn tất, bạn có thể xác minh bằng cách mở trình thông dịch Python trong môi trường ảo đã kích hoạt (hoặc chạy ô code bên dưới) với đoạn mã sau.

- Nếu các lệnh này chạy mà không phát sinh lỗi và hiển thị đúng phiên bản cùng trạng thái CUDA (`True` nếu bạn đã cài đặt phiên bản **GPU** và phần cứng cũng như **Driver** đã được thiết lập phù hợp, ngược lại là `False`), thì quá trình cài đặt của bạn đã thành công.

````python
import torch

# Kiểm tra phiên bản PyTorch
print(f"PyTorch Version: {torch.__version__}")

# Kiểm tra CUDA (hỗ trợ GPU) có khả dụng hay không
cuda_available = torch.cuda.is_available()
print(f"CUDA Available: {cuda_available}")

if cuda_available:
    # Lấy số lượng GPU khả dụng
    print(f"Number of GPUs: {torch.cuda.device_count()}")
    # Lấy tên của GPU hiện tại
    print(f"Current GPU Name: {torch.cuda.get_device_name(torch.cuda.current_device())}")
else:
    print("PyTorch is using CPU.")

# Tạo một Tensor đơn giản
x = torch.rand(2, 3)
print("Successfully created a tensor:")
print(x)
````

# **Phần 4. Tạo Tensor (Creating Tensors)**

- **Tensor** là cấu trúc dữ liệu nền tảng trong PyTorch, dùng để biểu diễn các **Multidimensional Array** được sử dụng trong các mô hình **Deep Learning**. PyTorch cung cấp nhiều phương pháp thực tế để tạo Tensor trong mã nguồn của bạn. Thư viện này có một tập hợp hàm linh hoạt để tạo Tensor, dù bạn sử dụng dữ liệu có sẵn hay cần tạo Tensor với **Hình dạng (Shape)** và giá trị khởi tạo cụ thể.

## **4.1. Tạo Tensor từ dữ liệu có sẵn (Creating Tensors from Existing Data)**

- Cách trực tiếp nhất để tạo Tensor thường là sử dụng các cấu trúc dữ liệu Python có sẵn, chẳng hạn như **List** hoặc **NumPy Array**. Hàm chính được sử dụng cho mục đích này là **`torch.tensor()`**. Hàm này tự động suy luận **Kiểu dữ liệu (Data Type)** (**`dtype`**) từ dữ liệu đầu vào, nhưng bạn cũng có thể chỉ định kiểu dữ liệu một cách tường minh.

- Điều quan trọng cần lưu ý là **`torch.tensor()` luôn sao chép dữ liệu đầu vào**.

- Hãy cùng xem cách hoạt động. Trước tiên, hãy bảo đảm rằng bạn đã nhập các thư viện PyTorch và NumPy:

<div style="text-align: center;">
    <br>
    <em>Hình 18. torch.tensor() sao chép dữ liệu đầu vào và tự suy luận Data Type: số nguyên → int64, số thực từ NumPy → float64.</em>
</div>

````python
import torch
import numpy as np

# Tạo Tensor từ một danh sách Python
data_list = [[1, 2], [3, 4]]
tensor_from_list = torch.tensor(data_list)

print("Tensor from list:")
print(tensor_from_list)
print(f"Data type: {tensor_from_list.dtype}")
print(f"Shape: {tensor_from_list.shape}")
````

- Hãy chú ý rằng PyTorch đã tự suy luận **Data Type** là **`torch.int64`** (số nguyên 64 bit) dựa trên các số nguyên trong dữ liệu đầu vào.

- Bạn cũng có thể đạt được kết quả tương tự bằng cách bắt đầu với một **NumPy Array**:

````python
# Tạo Tensor từ một mảng NumPy
data_numpy = np.array([[5.0, 6.0], [7.0, 8.0]])
tensor_from_numpy = torch.tensor(data_numpy)

print("Tensor from NumPy array:")
print(tensor_from_numpy)
print(f"Data type: {tensor_from_numpy.dtype}")
print(f"Shape: {tensor_from_numpy.shape}")
````

- Trong trường hợp này, **Data Type** được suy luận là **`torch.float64`** vì mảng NumPy chứa các số thực dấu phẩy động (floating-point numbers). Trong phần sau của chương này (Phần 6), chúng ta sẽ tìm hiểu mối quan hệ giữa Tensor của PyTorch và mảng NumPy, bao gồm những phương pháp chuyển đổi hiệu quả hơn.

## **4.2. Tạo Tensor với hình dạng và giá trị cụ thể (Creating Tensors with Specific Shapes and Values)**

- Trong nhiều trường hợp, bạn cần tạo Tensor có kích thước nhất định ngay cả khi chưa có dữ liệu, chẳng hạn như khởi tạo Tensor bằng các giá trị 0, giá trị 1 hoặc giá trị ngẫu nhiên. PyTorch cung cấp các hàm chuyên dụng cho những trường hợp phổ biến này.

- **Shape** thường được truyền vào dưới dạng một **Bộ (Tuple)** hoặc một chuỗi các số nguyên.

### **4.2.1. Tensor chứa toàn số 0, toàn số 1 hoặc dữ liệu chưa được khởi tạo (Tensors of Zeros, Ones, or Uninitialized Data)**

- **`torch.zeros(*size, ...)`**: Tạo một Tensor chứa toàn các giá trị 0.
- **`torch.ones(*size, ...)`**: Tạo một Tensor chứa toàn các giá trị 1.
- **`torch.empty(*size, ...)`**: Tạo một Tensor có kích thước được chỉ định nhưng các phần tử chưa được khởi tạo. Giá trị của chúng sẽ phụ thuộc vào dữ liệu đang tồn tại tại vị trí bộ nhớ đó vào thời điểm cấp phát. Sử dụng **`torch.empty`** có thể nhanh hơn một chút nếu bạn dự định điền dữ liệu vào Tensor ngay sau đó. Tuy nhiên, hãy thận trọng vì các giá trị ban đầu không được xác định trước.

<div style="text-align: center;">
    <br>
    <em>Hình 19. Ba hàm tạo Tensor theo hình dạng: zeros (toàn 0), ones (toàn 1) và empty (giá trị bất kỳ, chưa khởi tạo).</em>
</div>

````python
# Xác định hình dạng mong muốn
shape = (2, 3)  # 2 hàng, 3 cột

# Tạo Tensor với các giá trị cụ thể
zeros_tensor = torch.zeros(shape)
ones_tensor = torch.ones(shape)
empty_tensor = torch.empty(shape)  # Các giá trị tùy ý

print(f"Zeros Tensor (shape {shape}):")
print(zeros_tensor)

print(f"\nOnes Tensor (shape {shape}):")
print(ones_tensor)

print(f"\nEmpty Tensor (shape {shape}):")
print(empty_tensor)
````

- Theo mặc định, các hàm này tạo Tensor có **`dtype=torch.float32`**. Bạn có thể chỉ định **Data Type** khác bằng tham số **`dtype`**:

````python
# Tạo Tensor chứa các giá trị 1 với kiểu số nguyên
ones_int_tensor = torch.ones(shape, dtype=torch.int32)
print("Ones Tensor (dtype=torch.int32):")
print(ones_int_tensor)
````

<div style="text-align: center;">
    <br>
    <em>Hình 20. Các Data Type thường gặp: float32 là mặc định của zeros/ones/rand; int64 và float64 xuất hiện khi suy luận từ dữ liệu có sẵn.</em>
</div>

### **4.2.2. Tensor chứa các giá trị ngẫu nhiên (Tensors with Random Values)**

- Việc khởi tạo các **Parameter** mô hình thường bắt đầu bằng những giá trị ngẫu nhiên. PyTorch cung cấp các hàm sau để thực hiện việc này:

  - **`torch.rand(*size, ...)`**: Tạo Tensor có các phần tử được lấy mẫu theo **Phân phối đều (Uniform Distribution)** trong khoảng **`[0, 1)`**.
  - **`torch.randn(*size, ...)`**: Tạo Tensor có các phần tử được lấy mẫu từ **Phân phối chuẩn tắc (Standard Normal Distribution)**, với **Giá trị trung bình (Mean)** bằng 0 và **Phương sai (Variance)** bằng 1.

````python
# Cố định "hạt giống" để kết quả ngẫu nhiên có thể tái lập
torch.manual_seed(42)

# Tạo Tensor chứa các giá trị ngẫu nhiên
rand_tensor = torch.rand(shape)    # Phân phối đều [0, 1)
randn_tensor = torch.randn(shape)  # Phân phối chuẩn tắc

print(f"Random Tensor (Uniform [0, 1), shape {shape}):")
print(rand_tensor)

print(f"\nRandom Tensor (Standard Normal, shape {shape}):")
print(randn_tensor)
````

<div style="text-align: center;">
    <br>

</div>

## **4.3. Tạo Tensor dựa trên Tensor khác (Creating Tensors Based on Other Tensors)**

- Đôi khi, bạn cần tạo một Tensor mới có cùng các thuộc tính với một Tensor hiện có, chẳng hạn như **Shape** (**`shape`**) và **Data Type** (**`dtype`**). Điều này hữu ích để bảo đảm tính tương thích giữa các phép toán. PyTorch cung cấp các biến thể hàm có hậu tố **`_like`** cho mục đích này:

  - **`torch.zeros_like(input_tensor, ...)`**
  - **`torch.ones_like(input_tensor, ...)`**
  - **`torch.rand_like(input_tensor, ...)`**
  - **`torch.randn_like(input_tensor, ...)`**

- Các hàm này nhận một Tensor hiện có làm đầu vào và trả về một Tensor mới chứa các giá trị được chỉ định — chẳng hạn như số 0, số 1 hoặc giá trị ngẫu nhiên — nhưng vẫn giữ nguyên **Shape** và **Data Type** của Tensor đầu vào, trừ khi bạn chỉ định ghi đè các thuộc tính này.

<div style="text-align: center;">
    <br>
    <em>Hình 22. Các hàm hậu tố _like tạo Tensor mới với giá trị khác nhưng giữ nguyên Shape và Data Type của Tensor mẫu.</em>
</div>

````python
# Sử dụng một Tensor hiện có làm mẫu
base_tensor = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
print(f"Base Tensor (shape {base_tensor.shape}, dtype {base_tensor.dtype}):")
print(base_tensor)

# Tạo các Tensor có thuộc tính tương ứng với Tensor cơ sở
zeros_like_base = torch.zeros_like(base_tensor)
rand_like_base = torch.rand_like(base_tensor)

print("\nZeros tensor like base:")
print(zeros_like_base)
print(f"Shape: {zeros_like_base.shape}, dtype: {zeros_like_base.dtype}")

print("\nRandom tensor like base:")
print(rand_like_base)
print(f"Shape: {rand_like_base.shape}, dtype: {rand_like_base.dtype}")
````

---

- Như bạn có thể thấy, các Tensor mới **`zeros_like_base`** và **`rand_like_base`** tự động kế thừa thuộc tính **`shape`** (`torch.Size([2, 2])`) và **`dtype`** (`torch.float32`) từ **`base_tensor`**.

- Những phương pháp này cung cấp một bộ công cụ toàn diện để tạo ra các Tensor cần thiết, tạo nền tảng cho các phép tính được thực hiện trong mô hình **Deep Learning**. Trong các phần tiếp theo, chúng ta sẽ tìm hiểu cách thao tác với Tensor và thực hiện những phép toán thiết yếu.

Những phép toán này tạo thành nền tảng cho các phép tính bên trong **Neural Network**.

## **5.1. Các phép toán số học trên từng phần tử (Element-wise Arithmetic Operations)**

- Các phép toán phổ biến nhất bao gồm áp dụng những phép tính số học tiêu chuẩn một cách độc lập lên từng phần tử của các Tensor tham gia phép toán. Thông thường, các Tensor cần có **Hình dạng tương thích (Compatible Shapes)** để thực hiện những phép toán này. Chúng ta sẽ tìm hiểu thêm về tính tương thích của hình dạng khi đề cập đến **Phép quảng bá (Broadcasting)** trong chương tiếp theo.

- Bạn có thể sử dụng các toán tử số học tiêu chuẩn của Python hoặc những hàm tương ứng trong PyTorch:

  - **Phép cộng (Addition):** `+` hoặc `torch.add()`
  - **Phép trừ (Subtraction):** `-` hoặc `torch.sub()`
  - **Phép nhân (Multiplication):** `*` hoặc `torch.mul()`
  - **Phép chia (Division):** `/` hoặc `torch.div()`
  - **Phép lũy thừa (Exponentiation):** `**` hoặc `torch.pow()`

<div style="text-align: center;">
    <br>
    <em>Hình 24. Minh họa Element-wise Operation với phép cộng: các phần tử cùng vị trí của a và b được cộng độc lập với nhau.</em>
</div>

````python
# Tạo hai Tensor
a = torch.tensor([[1., 2.], [3., 4.]])
b = torch.tensor([[5., 6.], [7., 8.]])

# Phép cộng
sum_tensor = a + b
print("Addition (a + b):\n", sum_tensor)
print("Addition (torch.add(a, b)):\n", torch.add(a, b))

# Phép trừ
diff_tensor = a - b
print("\nSubtraction (a - b):\n", diff_tensor)

# Phép nhân trên từng phần tử
mul_tensor = a * b
print("\nElement-wise Multiplication (a * b):\n", mul_tensor)
print("Element-wise Multiplication (torch.mul(a, b)):\n", torch.mul(a, b))

# Phép chia
div_tensor = a / b
print("\nDivision (a / b):\n", div_tensor)

# Phép lũy thừa
pow_tensor = a ** 2
print("\nExponentiation (a ** 2):\n", pow_tensor)
print("Exponentiation (torch.pow(a, 2)):\n", torch.pow(a, 2))
````

- Các phép toán này tạo ra những Tensor mới chứa kết quả. Các Tensor ban đầu `a` và `b` vẫn không thay đổi.

## **5.2. Các phép toán tại chỗ (In-place Operations)**

- PyTorch cũng cung cấp các phiên bản **Phép toán tại chỗ (In-place Operation)** cho nhiều phép toán. Những phép toán này sửa đổi trực tiếp Tensor hiện có mà không tạo ra một đối tượng mới, nhờ đó có thể tiết kiệm bộ nhớ. Các hàm tại chỗ thường có thể được nhận biết thông qua dấu gạch dưới ở cuối tên hàm `_`, chẳng hạn như `add_` và `mul_`.

<div style="text-align: center;">
    <br>
    <em>Hình 25. Phép toán thông thường tạo vùng nhớ mới; In-place Operation sửa trực tiếp trên vùng nhớ cũ.</em>
</div>

````python
a = torch.tensor([[1., 2.], [3., 4.]])
b = torch.tensor([[5., 6.], [7., 8.]])

print("Original tensor 'a':\n", a)

# Thực hiện phép cộng tại chỗ
a.add_(b)  # a được sửa đổi trực tiếp
print("\nTensor 'a' after a.add_(b):\n", a)

# Một phép toán tại chỗ khác
a.mul_(2)  # Nhân 'a' với 2 tại chỗ
print("\nTensor 'a' after a.mul_(2):\n", a)
````

- Để thấy rõ sự khác biệt, hãy so sánh địa chỉ bộ nhớ (**`data_ptr()`**) của Tensor trước và sau phép toán. Lưu ý: câu lệnh `c = c + 1` **không** sửa vùng nhớ cũ; nó tạo ra một Tensor mới rồi gán lại tên `c` cho Tensor đó.

````python
c = torch.tensor([1., 2.])
ptr_goc = c.data_ptr()

c.add_(1)  # Tại chỗ: vẫn dùng vùng nhớ cũ
print("Sau c.add_(1)  -> cùng vùng nhớ:", c.data_ptr() == ptr_goc, "| c =", c)

c = c + 1  # Thông thường: tạo Tensor mới
print("Sau c = c + 1  -> cùng vùng nhớ:", c.data_ptr() == ptr_goc, "| c =", c)
````

- Mặc dù các **In-place Operation** có thể giúp sử dụng bộ nhớ hiệu quả hơn, bạn cần thận trọng khi sử dụng chúng. Việc sửa đổi Tensor tại chỗ có thể gây ra vấn đề trong quá trình tính toán **Gradient** bằng **`Autograd`** (được giới thiệu trong Chương 3), nếu giá trị ban đầu vẫn cần thiết ở một vị trí khác trong **Computation Graph**.

- Khi mới học, việc sử dụng các phép toán thông thường trả về Tensor mới thường là lựa chọn an toàn hơn.

## **5.3. Các phép toán với giá trị vô hướng (Scalar Operations)**

- Bạn có thể thực hiện các phép toán số học giữa một Tensor và một số đơn lẻ, được gọi là **Scalar**. PyTorch tự động mở rộng **Scalar** để phù hợp với **Shape** của Tensor khi thực hiện phép toán trên từng phần tử.

<div style="text-align: center;">
    <br>
    <em>Hình 26. Scalar 10 được tự động mở rộng thành Tensor cùng Shape rồi cộng theo từng phần tử.</em>

````python
t = torch.tensor([[1, 2, 3], [4, 5, 6]])
scalar = 10

# Cộng với giá trị vô hướng
print("t + scalar:\n", t + scalar)

# Nhân với giá trị vô hướng
print("\nt * scalar:\n", t * scalar)

# Trừ giá trị vô hướng
print("\nt - scalar:\n", t - scalar)
````

## **5.4. Các hàm toán học khác (Other Mathematical Functions)**

- PyTorch cung cấp một thư viện phong phú gồm các **Hàm toán học (Mathematical Function)** hoạt động trên từng phần tử của Tensor, tương tự như các **Hàm phổ quát (Universal Function - ufunc)** của NumPy.

<div style="text-align: center;">
    <br>
    <em>Hình 27. Đồ thị thực tế của sqrt, exp, log và abs, được tính bằng chính các hàm của PyTorch.</em>
</div>

````python
t = torch.tensor([[1., 4.], [9., 16.]])

# Căn bậc hai
print("Square Root (torch.sqrt(t)):\n", torch.sqrt(t))

# Hàm mũ
print("\nExponential (torch.exp(t)):\n", torch.exp(t))  # e^x

# Logarit tự nhiên
# Lưu ý: Bảo đảm các giá trị dương khi tính log
t_pos = torch.abs(t) + 1e-6  # Thêm một epsilon nhỏ để tăng tính ổn định nếu có giá trị 0
print("\nNatural Log (torch.log(t_pos)):\n", torch.log(t_pos))

# Giá trị tuyệt đối
t_neg = torch.tensor([[-1., 2.], [-3., 4.]])
print("\nAbsolute Value (torch.abs(t_neg)):\n", torch.abs(t_neg))
````

## **5.5. Các phép toán rút gọn (Reduction Operations)**

- **Phép toán rút gọn (Reduction Operation)** làm giảm số lượng phần tử trong một Tensor, thường nhằm tổng hợp thông tin. Những ví dụ phổ biến bao gồm tính tổng, **Mean**, giá trị nhỏ nhất và giá trị lớn nhất.

````python
t = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print("Original Tensor:\n", t)

# Tính tổng tất cả các phần tử
total_sum = torch.sum(t)
print("\nSum of all elements (torch.sum(t)):", total_sum)

# Tính giá trị trung bình của tất cả các phần tử
# Lưu ý: torch.mean yêu cầu Tensor kiểu số thực
mean_val = torch.mean(t.float())
print("Mean of all elements (torch.mean(t.float())):", mean_val)

# Giá trị lớn nhất
max_val = torch.max(t)
print("Max value in tensor (torch.max(t)):", max_val)

# Giá trị nhỏ nhất
min_val = torch.min(t)
print("Min value in tensor (torch.min(t)):", min_val)
````

- Bạn cũng có thể thực hiện các phép toán rút gọn dọc theo một chiều cụ thể bằng cách sử dụng đối số `dim`. Thao tác này loại bỏ chiều được chỉ định, trả về một Tensor có số chiều ít hơn một chiều so với Tensor ban đầu.

<div style="text-align: center;">
    <br>
    <em>Hình 28. Reduction Operation theo dim=0 (gộp các hàng thành kết quả theo cột), dim=1 (gộp các cột thành kết quả theo hàng) và trên toàn bộ Tensor.</em>
</div>

````python
t = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print("Original Tensor:\n", t)

# Tính tổng dọc theo chiều 0 (gộp các hàng, cho kết quả theo từng cột)
sum_dim0 = torch.sum(t, dim=0)
print("\nSum along dim=0 (one value per column):\n", sum_dim0)

# Tính tổng dọc theo chiều 1 (gộp các cột, cho kết quả theo từng hàng)
sum_dim1 = torch.sum(t, dim=1)
print("\nSum along dim=1 (one value per row):\n", sum_dim1)

# Tính giá trị trung bình dọc theo chiều 1
mean_dim1 = torch.mean(t.float(), dim=1)
print("\nMean along dim=1 (one value per row):\n", mean_dim1)
````

- Tính tổng Tensor `[[1, 2, 3], [4, 5, 6]]` dọc theo `dim=1` sẽ cho kết quả: `[1+2+3, 4+5+6] = [6, 15]`.

- Việc hiểu rõ cách hoạt động của `dim` rất quan trọng đối với nhiều phép toán trong **Deep Learning**, chẳng hạn như tính **Loss Function** cho từng phần tử trong một **Batch** hoặc áp dụng **Chuẩn hóa theo lô (Batch Normalization)**.

## **6.1. Cấu trúc tương tự và khả năng khác nhau (Similar Structures and Different Capabilities)**

### **6.1.1. Cấu trúc tương tự (Similar Structures)**

Về bản chất, cả PyTorch Tensor và mảng NumPy đều biểu diễn **Dữ liệu số dạng đặc (Dense Numerical Data)** trong nhiều chiều. Nhiều phép toán bạn thực hiện trên mảng NumPy đều có phép toán tương ứng trực tiếp trên PyTorch Tensor, thường sử dụng cách đặt tên tương tự:

- **Tạo dữ liệu (Creation):** Cả hai thư viện đều cung cấp các hàm để tạo mảng hoặc Tensor chứa toàn số 0, toàn số 1, các số ngẫu nhiên hoặc được tạo từ những **List** Python có sẵn.

<div style="text-align: center;">
    <br>
    <em>Hình 32. Bảng so sánh: hai thư viện giống nhau ở các thao tác cơ bản, nhưng chỉ PyTorch có GPU và Automatic Differentiation.</em>
</div>

````python
import numpy as np
import torch

# Tạo mảng/tensor toàn số 0, toàn số 1, từ danh sách Python
np_zeros = np.zeros((2, 3))
torch_zeros = torch.zeros((2, 3))

np_ones = np.ones((2, 3))
torch_ones = torch.ones((2, 3))

np_list = np.array([[1, 2], [3, 4]])
torch_list = torch.tensor([[1, 2], [3, 4]])

print(f"np_zeros: {np_zeros}")
print(f"torch_zeros: {torch_zeros}")
print(50*"-")

print(f"np_ones: {np_ones}")
print(f"torch_ones: {torch_ones}")
print(50*"-")

print(f"np_list: {np_list}")
print(f"torch_list: {torch_list}")
````

- **Các phép toán toán học (Mathematical Operations):** Các phép cộng, trừ, nhân, chia, lũy thừa theo từng phần tử và các hàm lượng giác hoạt động tương tự nhau.

````python
# Phép toán theo từng phần tử (căn bậc hai và phép cộng với một vô hướng)
np_a = np.array([1.0, 4.0, 9.0])
torch_a = torch.tensor([1.0, 4.0, 9.0])

np_result = np.sqrt(np_a) + 2
torch_result = torch.sqrt(torch_a) + 2

print(f"np_result: {np_result}")
print(f"torch_result: {torch_result}")
````

- **Indexing** và **Slicing**: Việc truy cập, sửa đổi các phần tử hoặc những mảng con/Tensor con sử dụng cú pháp tương đương.

````python
np_matrix = np.array([[1, 2, 3], [4, 5, 6]])
torch_matrix = torch.tensor([[1, 2, 3], [4, 5, 6]])

# Cắt lát (lấy tất cả hàng, từ cột 1 trở đi)
np_slice = np_matrix[:, 1:]
torch_slice = torch_matrix[:, 1:]

print(f"np_slice: {np_slice}")
print(f"torch_slice: {torch_slice}")
````

- **Thao tác với hình dạng (Shape Manipulation):** Việc **Reshaping**, **Chuyển vị (Transpose)** và nối các mảng/Tensor tuân theo những nguyên tắc tương tự.

````python
np_orig = np.arange(6)
torch_orig = torch.arange(6)

# Định hình lại (Reshape) và chuyển vị (Transpose)
np_reshaped = np_orig.reshape(2, 3)
torch_reshaped = torch_orig.reshape(2, 3)

np_transposed = np_reshaped.T
torch_transposed = torch_reshaped.T

print(f"np_reshaped: {np_reshaped}")
print(f"torch_reshaped: {torch_reshaped}")
print(50*"-")

print(f"np_transposed: {np_transposed}")
print(f"torch_transposed: {torch_transposed}")
````

### **6.1.2. Khả năng khác nhau (Different Capabilities)**

Tuy nhiên, vẫn có những khác biệt cơ bản, chủ yếu xuất phát từ việc PyTorch được thiết kế để phục vụ **Deep Learning**:

- **GPU Acceleration**:

  - PyTorch Tensor có thể được chuyển sang và xử lý trên **GPU**. Điều này cho phép thực hiện các phép tính song song quy mô lớn, giúp tăng tốc đáng kể những **Matrix Multiplication** và các phép toán khác thường gặp trong **Deep Learning**.

  - Trong khi đó, mảng NumPy chủ yếu được thiết kế để thực hiện các phép tính trên **CPU**.

<div style="text-align: center;">
    <br>
    <em>Hình 33. Nhắc lại: NumPy Array chỉ tính trên CPU, còn Tensor PyTorch có thể chuyển sang GPU.</em>
</div>

````python
# Kiểm tra xem GPU (CUDA) có sẵn hay không
if torch.cuda.is_available():
    # Tạo một Tensor trực tiếp trên GPU
    gpu_tensor = torch.tensor([[1.0, 2.0], [3.0, 4.0]], device='cuda')

    # Hoặc chuyển một Tensor từ CPU sang GPU
    cpu_tensor = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
    gpu_tensor_transferred = cpu_tensor.to('cuda')

    # Thực hiện phép nhân ma trận trên GPU
    result = torch.matmul(gpu_tensor, gpu_tensor)
    print(f"Kết quả tính toán trên GPU:\n{result}")
else:
    print("CUDA không khả dụng trên thiết bị này.")
````

- **Automatic Differentiation**:

  - PyTorch Tensor có hỗ trợ tích hợp cho tự động tính vi phân thông qua hệ thống **`Autograd`** (được trình bày trong Chương 3).
  - Cơ chế này tự động theo dõi các phép toán được thực hiện trên những Tensor cần tính **Gradient**, sau đó tính toán các **Gradient** này trong quá trình **Backpropagation**. Đây là thành phần thiết yếu để huấn luyện **Neural Network**.
  - Mảng NumPy không có khả năng này.

<div style="text-align: center;">
    <br>
    <em>Hình 34. Ví dụ số: với x = [2, 3] và y = x² + 3x, Autograd cho x.grad = 2x + 3 = [7, 9].</em>
</div>

````python
# Khởi tạo Tensor và yêu cầu PyTorch theo dõi gradient bằng requires_grad=True
x = torch.tensor([2.0, 3.0], requires_grad=True)

# Thực hiện một phép toán (ví dụ: y = x^2 + 3x)
y = x**2 + 3 * x

# Giả sử đây là hàm mất mát (loss) cần tối ưu, ta tính tổng để tạo thành một giá trị vô hướng (scalar)
loss = y.sum()

# Thực hiện lan truyền ngược (backpropagation) để tính đạo hàm của loss theo x
loss.backward()

# In ra gradient của x (dloss/dx = 2x + 3)
# Tại x = [2.0, 3.0], dloss/dx lần lượt là [2(2)+3, 2(3)+3] = [7.0, 9.0]
print(f"Giá trị tensor x: {x}")
print(f"Gradient của x (x.grad): {x.grad}")
````

## **6.2. Chuyển đổi giữa mảng NumPy và Tensor (Converting Between NumPy Arrays and Tensors)**

- PyTorch giúp việc chuyển đổi qua lại giữa hai cấu trúc dữ liệu này trở nên đơn giản.

### **6.2.1. Chuyển mảng NumPy thành Tensor PyTorch (NumPy Array to PyTorch Tensor)**

- Bạn có thể tạo trực tiếp một PyTorch Tensor từ **NumPy Array** bằng cách sử dụng hàm **`torch.from_numpy()`**.

````python
# Tạo một mảng NumPy
numpy_array = np.array([[1, 2], [3, 4]], dtype=np.float32)
print(f"NumPy array:\n{numpy_array}")
print(f"NumPy array type: {numpy_array.dtype}")

# Chuyển mảng NumPy thành Tensor PyTorch
pytorch_tensor = torch.from_numpy(numpy_array)
print(f"\nPyTorch Tensor:\n{pytorch_tensor}")
print(f"PyTorch Tensor type: {pytorch_tensor.dtype}")
````

**Lưu ý quan trọng về** **Chia sẻ bộ nhớ (Memory Sharing)**:

- Khi sử dụng **`torch.from_numpy()`**, PyTorch Tensor được tạo ra và mảng NumPy ban đầu sẽ **chia sẻ cùng một vùng bộ nhớ nền tảng** trên **CPU**.
- Điều này có nghĩa là việc sửa đổi một đối tượng sẽ ảnh hưởng đến đối tượng còn lại. Cách hoạt động này có hiệu quả cao vì không cần sao chép dữ liệu, nhưng bạn cần lưu ý đặc điểm này khi sử dụng.

<div style="text-align: center;">
    <br>

- Như bạn có thể thấy, các thay đổi được phản ánh trên cả hai đối tượng vì chúng cùng tham chiếu đến một vùng dữ liệu trong bộ nhớ.

### **6.2.2. Chuyển Tensor PyTorch thành mảng NumPy (PyTorch Tensor to NumPy Array)**

- Ngược lại, bạn có thể chuyển một PyTorch Tensor đang nằm trên **CPU** trở lại thành mảng NumPy bằng phương thức **`.numpy()`**.

````python
# Tạo một Tensor PyTorch trên CPU
cpu_tensor = torch.tensor([[10.0, 20.0], [30.0, 40.0]])
print(f"Original PyTorch Tensor (CPU):\n{cpu_tensor}")

# Chuyển Tensor thành mảng NumPy
numpy_array_converted = cpu_tensor.numpy()
print(f"\nConverted NumPy array:\n{numpy_array_converted}")
print(f"NumPy array type: {numpy_array_converted.dtype}")
````

- Một lần nữa, mảng NumPy được tạo ra và Tensor trên **CPU** ban đầu sẽ chia sẻ cùng một vùng bộ nhớ nền tảng. Vì vậy, việc sửa đổi một đối tượng sẽ ảnh hưởng đến đối tượng còn lại.

````python
# Sửa đổi Tensor
cpu_tensor[0, 1] = 25.0
print(f"Modified PyTorch Tensor:\n{cpu_tensor}")
print(f"NumPy array after modifying Tensor:\n{numpy_array_converted}")

# Sửa đổi mảng NumPy
numpy_array_converted[1, 0] = 35.0
print(f"\nModified NumPy array:\n{numpy_array_converted}")
print(f"Tensor after modifying NumPy array:\n{cpu_tensor}")
````

**Tensor trên GPU (GPU Tensors):**

- Phương thức **`.numpy()`** chỉ hoạt động với các Tensor được lưu trữ trên **CPU**.
- Nếu Tensor nằm trên **GPU**, trước tiên bạn phải chuyển nó về **CPU** bằng phương thức **`.cpu()`**, sau đó mới chuyển thành mảng NumPy.
- Việc gọi trực tiếp **`.numpy()`** trên một Tensor nằm trên **GPU** sẽ gây ra lỗi.

<div style="text-align: center;">
    <br>
    <em>Hình 36. Tensor trên GPU phải đi qua .cpu() trước khi gọi .numpy(); gọi .numpy() trực tiếp sẽ gây lỗi.</em>
</div>

````python
# Ví dụ chỉ chạy khi GPU khả dụng
if torch.cuda.is_available():
    gpu_tensor = torch.tensor([[1.0, 2.0], [3.0, 4.0]], device='cuda')
    print(f"Tensor on GPU:\n{gpu_tensor}")

    # Dòng lệnh sau sẽ gây ra lỗi:
    # numpy_from_gpu = gpu_tensor.numpy()

    # Cách đúng: Chuyển Tensor về CPU trước
    cpu_tensor_from_gpu = gpu_tensor.cpu()
    numpy_from_gpu = cpu_tensor_from_gpu.numpy()
    print(f"\nConverted NumPy array (from GPU Tensor):\n{numpy_from_gpu}")

    # Lưu ý: numpy_from_gpu chia sẻ bộ nhớ với cpu_tensor_from_gpu,
    # nhưng KHÔNG chia sẻ bộ nhớ với gpu_tensor ban đầu.
else:
    print("CUDA not available, skipping GPU to NumPy example.")
````
