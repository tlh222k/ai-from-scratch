# **Mối quan hệ với NumPy (Relationship with NumPy)**

- Nếu bạn đã có kinh nghiệm về **tính toán khoa học bằng Python (scientific computing in Python)**, có lẽ bạn đã quen thuộc với **NumPy** và đối tượng **`ndarray`** của thư viện này. NumPy cung cấp cấu trúc mảng N chiều mạnh mẽ, vốn đã trở thành tiêu chuẩn cho các phép toán số học trong Python. PyTorch nhận thấy mức độ phổ biến này và cung cấp khả năng tương thích rất tốt với các mảng NumPy.
- Trên thực tế, **Tensor của PyTorch (PyTorch Tensor)** có nhiều điểm tương đồng với mảng NumPy: cả hai đều là các cấu trúc biểu diễn những lưới số đa chiều.

- Mối quan hệ chặt chẽ này giúp việc chuyển đổi qua lại giữa hai cấu trúc dữ liệu trở nên đơn giản. Nhờ đó, bạn có thể tận dụng mã nguồn hoặc thư viện NumPy hiện có trong khi làm việc với hệ sinh thái PyTorch.

## **1. Cấu trúc tương tự và khả năng khác nhau (Similar Structures and Different Capabilities)**

### **1.1. Cấu trúc tương tự (Similar Structures)**

Về bản chất, cả PyTorch Tensor và mảng NumPy đều biểu diễn **dữ liệu số dạng đặc (dense numerical data)** trong nhiều chiều. Nhiều phép toán bạn thực hiện trên mảng NumPy đều có phép toán tương ứng trực tiếp trên PyTorch Tensor, thường sử dụng cách đặt tên tương tự:

- **Tạo dữ liệu (Creation):** Cả hai thư viện đều cung cấp các hàm để tạo mảng hoặc Tensor chứa toàn số $0$, toàn số $1$, các số ngẫu nhiên hoặc được tạo từ những danh sách Python có sẵn.

```python
import numpy as np
import torch

# Tạo mảng/tensor toàn số 0, toàn số 1, từ danh sách Python
np_zeros = np.zeros((2, 3))
torch_zeros = torch.zeros((2, 3))

np_ones = np.ones((2, 3))
torch_ones = torch.ones((2, 3))

np_list = np.array([[1, 2], [3, 4]])
torch_list = torch.tensor([[1, 2], [3, 4]])

print(f"np_zeros: {np_zeros}")
print(f"torch_zeros: {torch_zeros}")
print(50*"-")

print(f"np_ones: {np_ones}")
print(f"torch_ones: {torch_ones}")
print(50*"-")

print(f"np_list: {np_list}")
print(f"torch_list: {torch_list}")
```

- **Các phép toán toán học (Mathematical Operations):** Các phép cộng, trừ, nhân, chia, lũy thừa theo từng phần tử và các hàm lượng giác hoạt động tương tự nhau.

```python
# Phép toán theo từng phần tử (element-wise addition, exponentiation, trigonometric)
np_a = np.array([1.0, 4.0, 9.0])
torch_a = torch.tensor([1.0, 4.0, 9.0])

np_result = np.sqrt(np_a) + 2
torch_result = torch.sqrt(torch_a) + 2

print(f"np_result: {np_result}")
print(f"torch_result: {torch_result}")
```

- **Lập chỉ mục và cắt lát (Indexing and Slicing):** Việc truy cập, sửa đổi các phần tử hoặc những mảng con/Tensor con sử dụng cú pháp tương đương.

```python
np_matrix = np.array([[1, 2, 3], [4, 5, 6]])
torch_matrix = torch.tensor([[1, 2, 3], [4, 5, 6]])

# Cắt lát (lấy tất cả hàng, từ cột 1 trở đi)
np_slice = np_matrix[:, 1:]
torch_slice = torch_matrix[:, 1:]

print(f"np_slice: {np_slice}")
print(f"torch_slice: {torch_slice}")
```

- **Thao tác với hình dạng (Shape Manipulation):** Việc định hình lại, chuyển vị và nối các mảng/Tensor tuân theo những nguyên tắc tương tự.

```python
np_orig = np.arange(6)
torch_orig = torch.arange(6)

# Định hình lại (Reshape) và chuyển vị (Transpose)
np_reshaped = np_orig.reshape(2, 3)
torch_reshaped = torch_orig.reshape(2, 3)

np_transposed = np_reshaped.T
torch_transposed = torch_reshaped.T

print(f"np_reshaped: {np_reshaped}")
print(f"torch_reshaped: {torch_reshaped}")
print(50*"-")

print(f"np_transposed: {np_transposed}")
print(f"torch_transposed: {torch_transposed}")
```

### **1.2. Khả năng khác nhau (Different Capabilities)**

Tuy nhiên, vẫn có những khác biệt cơ bản, chủ yếu xuất phát từ việc PyTorch được thiết kế để phục vụ **học sâu (deep learning)**:

- **Tăng tốc bằng GPU (GPU Acceleration):**

  - PyTorch Tensor có thể được chuyển sang và xử lý trên **bộ xử lý đồ họa (Graphics Processing Unit — GPU)**. Điều này cho phép thực hiện các phép tính song song quy mô lớn, giúp tăng tốc đáng kể những phép nhân ma trận và các phép toán khác thường gặp trong Deep Learning.
  
  - Trong khi đó, mảng NumPy chủ yếu được thiết kế để thực hiện các phép tính trên CPU.

```python
import torch

# Kiểm tra xem GPU (CUDA) có sẵn hay không
if torch.cuda.is_available():
    # Tạo một Tensor trực tiếp trên GPU
    gpu_tensor = torch.tensor([[1.0, 2.0], [3.0, 4.0]], device='cuda')

    # Hoặc chuyển một Tensor từ CPU sang GPU
    cpu_tensor = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
    gpu_tensor_transferred = cpu_tensor.to('cuda')

    # Thực hiện phép tính nhân ma trận trên GPU
    result = torch.matmul(gpu_tensor, gpu_tensor)
    print(f"Kết quả tính toán trên GPU:\n{result}")
else:
    print("CUDA không khả dụng trên thiết bị này.")
```

- **Tự động tính đạo hàm (Automatic Differentiation):**

  - PyTorch Tensor có hỗ trợ tích hợp cho **tự động tính đạo hàm (automatic differentiation)** thông qua hệ thống **`Autograd`** (được trình bày trong Chương 3).
  - Cơ chế này tự động theo dõi các phép toán được thực hiện trên những Tensor cần tính gradient, sau đó tính toán các gradient này trong quá trình **lan truyền ngược (backpropagation)**. Đây là thành phần thiết yếu để huấn luyện Neural Networks.
  - Mảng NumPy không có khả năng này.

```python
import torch

# Khởi tạo Tensor và yêu cầu PyTorch theo dõi gradient bằng requires_grad=True
x = torch.tensor([2.0, 3.0], requires_grad=True)

# Thực hiện một phép toán (ví dụ: y = x^2 + 3x)
y = x**2 + 3 * x

# Giả sử đây là hàm mất mát (loss) cần tối ưu, ta tính tổng để tạo thành một giá trị vô hướng (scalar)
loss = y.sum()

# Thực hiện lan truyền ngược (backpropagation) để tính đạo hàm của loss theo x
loss.backward()

# In ra gradient của x (dy/dx = 2x + 3)
# Tại x = [2.0, 3.0], dy/dx lần lượt là [2(2)+3, 2(3)+3] = [7.0, 9.0]
print(f"Giá trị tensor x: {x}")
print(f"Gradient của x (dx): {x.grad}")
```

## **2. Chuyển đổi giữa mảng NumPy và Tensor (Converting Between NumPy Arrays and Tensors)**

- PyTorch giúp việc chuyển đổi qua lại giữa hai cấu trúc dữ liệu này trở nên đơn giản.

### **2.1. Chuyển mảng NumPy thành Tensor PyTorch (NumPy Array to PyTorch Tensor)**

- Bạn có thể tạo trực tiếp một PyTorch Tensor từ mảng NumPy bằng cách sử dụng hàm **`torch.from_numpy()`**.

```python
import numpy as np
import torch

# Tạo một mảng NumPy
numpy_array = np.array([[1, 2], [3, 4]], dtype=np.float32)
print(f"NumPy array:\n{numpy_array}")
print(f"NumPy array type: {numpy_array.dtype}")

# Chuyển mảng NumPy thành Tensor PyTorch
pytorch_tensor = torch.from_numpy(numpy_array)
print(f"\nPyTorch Tensor:\n{pytorch_tensor}")
print(f"PyTorch Tensor type: {pytorch_tensor.dtype}")
```

**Lưu ý quan trọng về việc chia sẻ bộ nhớ (Memory Sharing):**

- Khi sử dụng **`torch.from_numpy()`**, PyTorch Tensor được tạo ra và mảng NumPy ban đầu sẽ **chia sẻ cùng một vùng bộ nhớ nền tảng trên CPU**.
- Điều này có nghĩa là việc sửa đổi một đối tượng sẽ ảnh hưởng đến đối tượng còn lại. Cách hoạt động này có hiệu quả cao vì không cần sao chép dữ liệu, nhưng bạn cần lưu ý đặc điểm này khi sử dụng.

```python
# Sửa đổi mảng NumPy
numpy_array[0, 0] = 99
print(f"\nModified NumPy array:\n{numpy_array}")
print(f"PyTorch Tensor after modifying NumPy array:\n{pytorch_tensor}")

# Sửa đổi Tensor PyTorch
pytorch_tensor[1, 1] = -1
print(f"\nModified PyTorch Tensor:\n{pytorch_tensor}")
print(f"NumPy array after modifying PyTorch Tensor:\n{numpy_array}")
```

- Như bạn có thể thấy, các thay đổi được phản ánh trên cả hai đối tượng vì chúng cùng tham chiếu đến một vùng dữ liệu trong bộ nhớ.

### **2.2. Chuyển Tensor PyTorch thành mảng NumPy (PyTorch Tensor to NumPy Array)**

- Ngược lại, bạn có thể chuyển một PyTorch Tensor đang nằm trên CPU trở lại thành mảng NumPy bằng phương thức **`.numpy()`**.

```python
# Tạo một Tensor PyTorch trên CPU
cpu_tensor = torch.tensor([[10.0, 20.0], [30.0, 40.0]])
print(f"Original PyTorch Tensor (CPU):\n{cpu_tensor}")

# Chuyển Tensor thành mảng NumPy
numpy_array_converted = cpu_tensor.numpy()
print(f"\nConverted NumPy array:\n{numpy_array_converted}")
print(f"NumPy array type: {numpy_array_converted.dtype}")
```

- Một lần nữa, mảng NumPy được tạo ra và Tensor CPU ban đầu sẽ chia sẻ cùng một vùng bộ nhớ nền tảng. Vì vậy, việc sửa đổi một đối tượng sẽ ảnh hưởng đến đối tượng còn lại.

```python
# Sửa đổi Tensor
cpu_tensor[0, 1] = 25.0
print(f"\nModified PyTorch Tensor:\n{cpu_tensor}")
print(f"NumPy array after modifying Tensor:\n{numpy_array_converted}")

# Sửa đổi mảng NumPy
numpy_array_converted[1, 0] = 35.0
print(f"\nModified NumPy array:\n{numpy_array_converted}")
print(f"Tensor after modifying NumPy array:\n{cpu_tensor}")
```

**Tensor trên GPU (GPU Tensors):**

- Phương thức **`.numpy()`** chỉ hoạt động với các Tensor được lưu trữ trên CPU.
- Nếu Tensor nằm trên GPU, trước tiên bạn phải chuyển nó về CPU bằng phương thức **`.cpu()`**, sau đó mới chuyển thành mảng NumPy.
- Việc gọi trực tiếp **`.numpy()`** trên một Tensor nằm trên GPU sẽ gây ra lỗi.

```python
# Ví dụ giả định rằng GPU khả dụng
if torch.cuda.is_available():
    gpu_tensor = torch.tensor([[1.0, 2.0], [3.0, 4.0]], device='cuda')
    print(f"\nTensor on GPU:\n{gpu_tensor}")

    # Dòng lệnh sau sẽ gây ra lỗi:
    # numpy_from_gpu = gpu_tensor.numpy()

    # Cách đúng: Chuyển Tensor về CPU trước
    cpu_tensor_from_gpu = gpu_tensor.cpu()
    numpy_from_gpu = cpu_tensor_from_gpu.numpy()
    print(f"\nConverted NumPy array (from GPU Tensor):\n{numpy_from_gpu}")

    # Lưu ý: numpy_from_gpu chia sẻ bộ nhớ với cpu_tensor_from_gpu,
    # nhưng KHÔNG chia sẻ bộ nhớ với gpu_tensor ban đầu.
else:
    print("\nCUDA not available, skipping GPU to NumPy example.")
```
