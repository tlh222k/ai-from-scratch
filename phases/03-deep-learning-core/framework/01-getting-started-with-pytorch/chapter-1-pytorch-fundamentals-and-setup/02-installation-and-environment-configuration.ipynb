# **Cài đặt và cấu hình môi trường (Installation and Environment Configuration)**

- Trước khi bắt đầu xây dựng **các mô hình học sâu (deep learning models)**, bạn cần thiết lập môi trường phát triển bằng cách cài đặt PyTorch. Quá trình này bao gồm cài đặt thư viện cốt lõi và cấu hình không gian làm việc.

# **1. Lựa chọn phương pháp cài đặt: Conda và Pip (Choosing Your Installation Method: Conda vs. Pip)**

- PyTorch có thể được cài đặt bằng hai **trình quản lý gói (package manager)** chính:
  - **Conda**: một phần của các bản phân phối **Anaconda** hoặc **Miniconda**
  - **Pip**: trình cài đặt gói mặc định của Python.

## **1.1 Conda**

- **Conda**: Thường được ưu tiên trong các lĩnh vực **khoa học dữ liệu (data science)** và **học máy (machine learning)**.

- **Conda** quản lý cả gói phần mềm lẫn môi trường **(packages and environments)**, giúp đơn giản hóa việc xử lý các phần phụ thuộc phức tạp, bao gồm cả những thư viện không thuộc Python như bộ công cụ CUDA cần thiết để tăng tốc bằng GPU. Nếu bạn đã sử dụng Anaconda, đây thường là phương pháp thuận tiện nhất.

## **1.2. Pip**

- **Pip**: Là trình cài đặt gói tiêu chuẩn của Python.
- **Pip** hoạt động tốt, đặc biệt nếu bạn muốn quản lý môi trường Python bằng các công cụ như **venv**. Nếu cần hỗ trợ GPU, bạn có thể phải tự xử lý việc cài đặt trình điều khiển NVIDIA và bộ công cụ CUDA riêng biệt.

---
- Để bảo đảm khả năng tái lập kết quả và tránh xung đột với các dự án khác, bạn nên cài đặt PyTorch trong một **môi trường ảo (virtual environment) riêng biệt**, bất kể lựa chọn **Conda** hay **Pip**.

# **2. Các bước cài đặt (Installation Steps)**

- Lệnh cài đặt chính xác phụ thuộc vào hệ điều hành của bạn (Linux, macOS hoặc Windows), trình quản lý gói (Conda hoặc Pip) và việc bạn muốn sử dụng CPU hay GPU có hỗ trợ CUDA.

- Thực hành tốt nhất là luôn tham khảo trang web chính thức của [PyTorch](https://pytorch.org/)  để sử dụng công cụ tạo lệnh cài đặt được cập nhật mới nhất. Hãy chọn các tùy chọn phù hợp với nhu cầu của bạn, trang web sẽ cung cấp chính xác lệnh cần chạy.


### **2.1. Sử dụng Conda (Using Conda)**

- Trước tiên, tạo và kích hoạt một môi trường Conda mới. Thay pytorch_env bằng tên bạn muốn sử dụng và chọn phiên bản Python phù hợp:


```python
# Tạo môi trường
conda create -n pytorch_env python=3.9 # Hoặc 3.8, 3.10, v.v.

# Kích hoạt môi trường
conda activate pytorch_env
```

- Tiếp theo, truy cập trang web PyTorch và chọn hệ điều hành, Pip, phiên bản Python và phiên bản CUDA mong muốn (hoặc CPU), vì PyTorch được cài đặt thông qua Pip ngay cả khi chạy bên trong môi trường Conda.
- Sao chép lệnh được tạo.

```python
# Lệnh Pip ví dụ cho CUDA 11.8, chạy bên trong môi trường Conda đã kích hoạt
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118

# Lệnh Pip ví dụ chỉ dành cho CPU, chạy bên trong môi trường Conda đã kích hoạt
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu
```

### **2.2. Sử dụng Pip (Using Pip)**

- Trước tiên, tạo và kích hoạt một môi trường ảo mới bằng **venv**:

```python
# Tạo môi trường (trong thư mục dự án)
python -m venv pytorch_env

# Kích hoạt môi trường
# Linux/macOS:
source pytorch_env/bin/activate
# Windows (Command Prompt):
.\pytorch_env\Scripts\activate
# Windows (PowerShell):
.\pytorch_env\Scripts\Activate.ps1
```

- Tiếp theo, truy cập trang web PyTorch và chọn hệ điều hành, Pip, phiên bản Python và phiên bản CUDA mong muốn (hoặc CPU). Sao chép lệnh được tạo. Lệnh này thường yêu cầu chỉ định URL chỉ mục gói (index URL) và có thể bao gồm cả phiên bản CUDA.

```python
# Lệnh Pip ví dụ cho CUDA 11.8
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118

# Lệnh Pip ví dụ chỉ dành cho CPU
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu
```

- Chạy lệnh do trang web cung cấp trong môi trường ảo đã được kích hoạt. Hãy bảo đảm rằng pip đã được cập nhật lên phiên bản mới nhất:

```python
pip install --upgrade pip
```

## **3. GPU Support (CUDA)**

- Các phép tính học sâu, đặc biệt là **phép nhân ma trận (matrix multiplication)**, có thể được thực hiện nhanh hơn đáng kể trên GPU của NVIDIA bằng cách sử dụng nền tảng tính toán song song CUDA.

  - **Yêu cầu (Requirement):** Bạn cần có GPU NVIDIA tương thích với CUDA.

  - **Trình điều khiển và bộ công cụ (Drivers & Toolkit):**
    - Trước khi cài đặt phiên bản PyTorch có hỗ trợ CUDA, hệ thống của bạn phải có trình điều khiển đồ họa NVIDIA phù hợp và **Bộ công cụ CUDA (CUDA Toolkit)**.
    - Phiên bản **CUDA Toolkit** cần tương thích với phiên bản mà PyTorch được xây dựng dựa trên đó, chẳng hạn như `11.7`, `11.8` hoặc `12.1`. **Conda** có thể tự xử lý việc cài đặt bộ công cụ này, trong khi với **Pip**, bạn thường cần cài đặt nó trên toàn hệ thống trước.
    
    - Hãy tham khảo tài liệu của NVIDIA và hướng dẫn cài đặt PyTorch để biết thông tin cụ thể.

  - **Lệnh cài đặt (Installation Command):** Khi sử dụng công cụ tạo lệnh cài đặt trên trang web PyTorch, việc chọn phiên bản CUDA sẽ điều chỉnh lệnh để cài đặt các tệp nhị phân tương thích với GPU.

  - Nếu bạn không có GPU tương thích hoặc chưa cần tăng tốc bằng GPU, chỉ cần chọn tùy chọn **CPU** trong quá trình cài đặt. Bạn vẫn có thể thực hiện mọi thao tác trên CPU, mặc dù tốc độ có thể chậm hơn khi xử lý các mô hình lớn.

## **4. Xác minh cài đặt (Verifying the Installation)**

- Sau khi cài đặt hoàn tất, bạn có thể xác minh bằng cách mở trình thông dịch Python trong môi trường ảo đã kích hoạt và chạy đoạn mã sau:

- Nếu các lệnh này chạy mà không phát sinh lỗi và hiển thị đúng phiên bản cùng trạng thái CUDA (`True` nếu bạn đã cài đặt phiên bản GPU và phần cứng cũng như trình điều khiển đã được thiết lập phù hợp và ngược lại là `False`), thì quá trình cài đặt của bạn đã thành công.



```python
import torch

# Kiểm tra phiên bản PyTorch
print(f"PyTorch Version: {torch.__version__}")

# Kiểm tra CUDA (hỗ trợ GPU) có khả dụng hay không
cuda_available = torch.cuda.is_available()
print(f"CUDA Available: {cuda_available}")

if cuda_available:
    # Lấy số lượng GPU khả dụng
    print(f"Number of GPUs: {torch.cuda.device_count()}")
    # Lấy tên của GPU hiện tại
    print(f"Current GPU Name: {torch.cuda.get_device_name(torch.cuda.current_device())}")
else:
    print("PyTorch is using CPU.")

# Tạo một Tensor đơn giản
x = torch.rand(2, 3)
print("Successfully created a tensor:")
print(x)
```

## **5. Lưu ý về cấu hình môi trường (Environment Configuration Notes)**

- **Tính nhất quán (Consistency):** Luôn kích hoạt môi trường riêng **`pytorch_env`** hoặc tên khác mà bạn đã đặt, trước khi làm việc với các dự án PyTorch hoặc chạy tập lệnh.

- **Các thư viện phổ biến (Common Libraries):** Bạn thường sẽ sử dụng những thư viện khác cùng với PyTorch. Hãy cài đặt chúng vào cùng môi trường bằng lệnh **`conda install <package>`** hoặc **`pip install <package>`**:

  - **`numpy`**: Dùng cho các phép toán số học và tương tác với những thư viện khác.
  - **`matplotlib`** hoặc **`seaborn`**: Dùng để vẽ biểu đồ và trực quan hóa dữ liệu.
  - **`scikit-learn`**: Dùng cho các tác vụ và tiện ích học máy truyền thống.
  - **`jupyterlab`** hoặc **`notebook`**: Dùng cho quá trình phát triển tương tác.

- **Môi trường phát triển tích hợp (Integrated Development Environments — IDEs):** Cấu hình môi trường phát triển tích hợp (IDE) như VS Code hoặc PyCharm để sử dụng trình thông dịch Python nằm bên trong môi trường ảo của bạn. Điều này bảo đảm các tính năng hoàn thành mã (code completion) và gỡ lỗi (debugging) hoạt động chính xác với PyTorch.
