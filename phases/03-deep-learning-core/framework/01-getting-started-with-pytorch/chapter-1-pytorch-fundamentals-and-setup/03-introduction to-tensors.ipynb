# **Giới thiệu về Tensor (Introduction to Tensors)**

- **Tensor** là cấu trúc dữ liệu cốt lõi khi làm việc với PyTorch. Nếu bạn đã có kinh nghiệm sử dụng NumPy, bạn sẽ thấy Tensor của PyTorch khá quen thuộc. Tensor là **các mảng đa chiều (multidimensional arrays)**, rất giống với **`ndarray`** của NumPy.

- Hãy xem Tensor là sự khái quát hóa của những đối tượng toán học quen thuộc:

  - **Vô hướng (scalar):** Một số đơn lẻ, chẳng hạn như `7`, là một **Tensor không chiều (0-dimensional tensor)**, hay **Tensor hạng 0 (rank-0 tensor)**.

  - **Vectơ (vector):** Một danh sách hoặc mảng một chiều chứa các số, chẳng hạn như **`[1, 2, 3]`**, là **một Tensor một chiều (1-dimensional tensor)**, hay **Tensor hạng 1 (rank-1 tensor)**.

  - **Ma trận (matrix):** Một bảng số hai chiều, chẳng hạn như **`[[1, 2], [3, 4]]`**, là một **Tensor hai chiều (2-dimensional tensor)**, hay **Tensor hạng 2 (rank-2 tensor)**.

  - Mở rộng khái niệm này, ta có thể tạo **Tensor ba chiều (3-dimensional tensor)**, chẳng hạn như:
    - Một khối số thường được dùng để biểu diễn **hình ảnh RGB** với các chiều **chiều cao × chiều rộng × số kênh (height × width × channels)**.
    - **Tensor bốn chiều (4-dimensional tensor)**, thường được dùng để biểu diễn một **lô hình ảnh (batch of images)** với các chiều **kích thước lô × chiều cao × chiều rộng × số kênh (batch_size × height × width × channels)**.
    - Và các Tensor có số chiều lớn hơn nữa.

- Trong lĩnh vực học sâu (deep learning), Tensor được sử dụng để biểu diễn hầu như mọi thứ:

  - **Dữ liệu đầu vào (Input data):** Một lô hình ảnh, các chuỗi văn bản hoặc những bảng đặc trưng.
  - **Tham số mô hình (Model parameters):** Các trọng số (weights) và độ lệch (biases) của các lớp trong mạng nơ-ron.
  - **Giá trị kích hoạt trung gian (Intermediate activations):** Đầu ra của các lớp bên trong mạng.
  - **Gradient:** Các giá trị được tính toán trong **quá trình lan truyền ngược (backpropagation)**, dùng để cập nhật tham số mô hình.

- Điều gì khiến Tensor của PyTorch đặc biệt phù hợp với học sâu hơn so với các danh sách Python thông thường hoặc thậm chí các mảng NumPy?

### **1. Tăng tốc bằng GPU (GPU Acceleration)**

- Tensor có thể dễ dàng được chuyển đến và xử lý trên **các bộ xử lý đồ họa (Graphics Processing Units — GPUs)** hoặc những bộ tăng tốc phần cứng khác.

- Điều này giúp tăng tốc đáng kể các phép tính có cường độ tính toán cao thường gặp trong học sâu.


### **2. Tự động tính vi phân (Automatic Differentiation)**

- Tensor của PyTorch được tích hợp khả năng tự động tính vi phân thông qua hệ thống **`Autograd`** — nội dung này sẽ được trình bày trong **Chương 3 (Chapter 3)**. Cơ chế này tự động tính toán các gradient, vốn là thành phần nền tảng để huấn luyện mạng nơ-ron thông qua **quá trình lan truyền ngược (backpropagation)**.
