# **Các phép toán Tensor cơ bản (Basic Tensor Operations)**

- Thực hiện các phép toán cơ bản trên Tensor của PyTorch là một kỹ năng nền tảng. Tensor hỗ trợ nhiều phép toán **toán học và logic (mathematical and logical operations)**, trong đó có nhiều phép toán hoạt động tương tự như các phép toán tương ứng trong NumPy, bằng cách thực hiện phép toán trên từng phần tử (**element-wise**). Những phép toán này tạo thành nền tảng cho các phép tính bên trong **Mạng nơ-ron (Neural Networks)**.

## **1. Các phép toán số học trên từng phần tử (Element-wise Arithmetic Operations)**

- Các phép toán phổ biến nhất bao gồm áp dụng những phép tính số học tiêu chuẩn một cách độc lập lên từng phần tử của các Tensor tham gia phép toán. Thông thường, các Tensor cần có hình dạng tương thích (**compatible shapes**) để thực hiện những phép toán này. Chúng ta sẽ tìm hiểu thêm về tính tương thích của hình dạng khi đề cập đến **Phép quảng bá (Broadcasting)** trong chương tiếp theo.

- Bạn có thể sử dụng các toán tử số học tiêu chuẩn của Python hoặc những hàm tương ứng trong PyTorch:

  - **Phép cộng (Addition):** `+` hoặc `torch.add()`
  - **Phép trừ (Subtraction):** `-` hoặc `torch.sub()`
  - **Phép nhân (Multiplication):** `*` hoặc `torch.mul()`
  - **Phép chia (Division):** `/` hoặc `torch.div()`
  - **Phép lũy thừa (Exponentiation):** `**` hoặc `torch.pow()`

```python
import torch

# Tạo hai Tensor
a = torch.tensor([[1., 2.], [3., 4.]])
b = torch.tensor([[5., 6.], [7., 8.]])

# Phép cộng
sum_tensor = a + b
print("Addition (a + b):\n", sum_tensor)
print("Addition (torch.add(a, b)):\n", torch.add(a, b))

# Phép trừ
diff_tensor = a - b
print("\nSubtraction (a - b):\n", diff_tensor)

# Phép nhân trên từng phần tử
mul_tensor = a * b
print("\nElement-wise Multiplication (a * b):\n", mul_tensor)
print("Element-wise Multiplication (torch.mul(a, b)):\n", torch.mul(a, b))

# Phép chia
div_tensor = a / b
print("\nDivision (a / b):\n", div_tensor)

# Phép lũy thừa
pow_tensor = a ** 2
print("\nExponentiation (a ** 2):\n", pow_tensor)
print("Exponentiation (torch.pow(a, 2)):\n", torch.pow(a, 2))
```

- Các phép toán này tạo ra những Tensor mới chứa kết quả. Các Tensor ban đầu `a` và `b` vẫn không thay đổi.

## **2. Các phép toán tại chỗ (In-place Operations)**

- PyTorch cũng cung cấp các phiên bản **In-place Operations** cho nhiều phép toán. Những phép toán này sửa đổi trực tiếp Tensor hiện có mà không tạo ra một đối tượng mới, nhờ đó có thể tiết kiệm bộ nhớ. Các hàm tại chỗ thường có thể được nhận biết thông qua dấu gạch dưới ở cuối tên hàm `_`, chẳng hạn như `add_` và `mul_`.

```python
import torch

a = torch.tensor([[1., 2.], [3., 4.]])
b = torch.tensor([[5., 6.], [7., 8.]])

print("Original tensor 'a':\n", a)

# Thực hiện phép cộng tại chỗ
a.add_(b) # a được sửa đổi trực tiếp
print("\nTensor 'a' after a.add_(b):\n", a)

# Nếu bỏ dấu chú thích, dòng lệnh này sẽ gây ra lỗi,
# vì kết quả của a + b là một Tensor mới,
# không phù hợp để gán trực tiếp trở lại vùng bộ nhớ của 'a'
# a = a + b # Phép cộng thông thường tạo ra một Tensor mới

# Một phép toán tại chỗ khác
a.mul_(2) # Nhân 'a' với 2 tại chỗ
print("\nTensor 'a' after a.mul_(2):\n", a)
```

- Mặc dù các **In-place Operations** có thể giúp sử dụng bộ nhớ hiệu quả hơn, bạn cần thận trọng khi sử dụng chúng. Việc sửa đổi Tensor tại chỗ có thể gây ra vấn đề trong quá trình tính toán **Gradient (Gradient Calculations)** bằng **Autograd** (được giới thiệu trong Chương 3), nếu giá trị ban đầu vẫn cần thiết ở một vị trí khác trong **Đồ thị tính toán (Computation Graph)**.

- Khi mới học, việc sử dụng các phép toán thông thường trả về Tensor mới thường là lựa chọn an toàn hơn.

## **3. Các phép toán với giá trị vô hướng (Scalar Operations)**

- Bạn có thể thực hiện các phép toán số học giữa một Tensor và một số đơn lẻ, được gọi là **Giá trị vô hướng (Scalar)**. PyTorch tự động mở rộng **Scalar** để phù hợp với hình dạng của Tensor khi thực hiện phép toán trên từng phần tử.

```python
import torch

t = torch.tensor([[1, 2, 3], [4, 5, 6]])
scalar = 10

# Cộng với giá trị vô hướng
print("t + scalar:\n", t + scalar)

# Nhân với giá trị vô hướng
print("\nt * scalar:\n", t * scalar)

# Trừ giá trị vô hướng
print("\nt - scalar:\n", t - scalar)
```

## **4. Các hàm toán học khác (Other Mathematical Functions)**

- PyTorch cung cấp một thư viện phong phú gồm các **Hàm toán học (Mathematical Functions)** hoạt động trên từng phần tử của Tensor, tương tự như các **NumPy universal functions (ufuncs)**.

```python
import torch

t = torch.tensor([[1., 4.], [9., 16.]])

# Căn bậc hai
print("Square Root (torch.sqrt(t)):\n", torch.sqrt(t))

# Hàm mũ
print("\nExponential (torch.exp(t)):\n", torch.exp(t)) # e^x

# Logarit tự nhiên
# Lưu ý: Bảo đảm các giá trị dương khi tính log
t_pos = torch.abs(t) + 1e-6 # Thêm một epsilon nhỏ để tăng tính ổn định nếu có giá trị 0
print("\nNatural Log (torch.log(t_pos)):\n", torch.log(t_pos))

# Giá trị tuyệt đối
t_neg = torch.tensor([[-1., 2.], [-3., 4.]])
print("\nAbsolute Value (torch.abs(t_neg)):\n", torch.abs(t_neg))
```

- Nhiều hàm khác cũng có sẵn trong mô-đun `torch`, chẳng hạn như `torch.sin()`, `torch.cos()`, `torch.tanh()` và `torch.sigmoid()`.

## **5. Các phép toán rút gọn (Reduction Operations)**

- **Reduction Operations** làm giảm số lượng phần tử trong một Tensor, thường nhằm tổng hợp thông tin. Những ví dụ phổ biến bao gồm tính tổng, giá trị trung bình, giá trị nhỏ nhất và giá trị lớn nhất.

```python
import torch

t = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print("Original Tensor:\n", t)

# Tính tổng tất cả các phần tử
total_sum = torch.sum(t)
print("\nSum of all elements (torch.sum(t)):", total_sum)

# Tính giá trị trung bình của tất cả các phần tử
# Lưu ý: Cần Tensor kiểu số thực để tính giá trị trung bình
mean_val = torch.mean(t.float())
print("Mean of all elements (torch.mean(t.float())):", mean_val)

# Giá trị lớn nhất
max_val = torch.max(t)
print("Max value in tensor (torch.max(t)):", max_val)

# Giá trị nhỏ nhất
min_val = torch.min(t)
print("Min value in tensor (torch.min(t)):", min_val)
```

- Bạn cũng có thể thực hiện các phép toán rút gọn dọc theo một chiều cụ thể bằng cách sử dụng đối số `dim`. Thao tác này loại bỏ chiều được chỉ định, trả về một Tensor có số chiều ít hơn một chiều so với Tensor ban đầu.

```python
import torch

t = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print("Original Tensor:\n", t)

# Tính tổng dọc theo chiều 0 (cộng các hàng)
sum_dim0 = torch.sum(t, dim=0)
print("\nSum along dim=0 (columns):\n", sum_dim0)

# Tính tổng dọc theo chiều 1 (cộng các cột)
sum_dim1 = torch.sum(t, dim=1)
print("\nSum along dim=1 (rows):\n", sum_dim1)

# Tính giá trị trung bình dọc theo chiều 1
mean_dim1 = torch.mean(t.float(), dim=1)
print("\nMean along dim=1 (rows):\n", mean_dim1)
```

- Tính tổng Tensor `[[1, 2, 3], [4, 5, 6]]` dọc theo `dim=1` sẽ cho kết quả: `[1+2+3, 4+5+6] = [6, 15]`.

- Việc hiểu rõ cách hoạt động của `dim` rất quan trọng đối với nhiều phép toán trong học sâu, chẳng hạn như tính **Hàm mất mát (Loss Function)** cho từng phần tử trong một **Batch Item** hoặc **áp dụng Chuẩn hóa theo lô (Batch Normalization)**.

## **6. Các phép toán so sánh (Comparison Operations)**

- Bạn có thể so sánh các Tensor trên từng phần tử bằng cách sử dụng những toán tử so sánh tiêu chuẩn: `>`, `<`, `>=`, `<=`, `==` và `!=`.
- Kết quả trả về là một Tensor chứa các giá trị **Boolean (giá trị logic)**, có kiểu dữ liệu `torch.bool`.

```python
import torch

a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[1, 5], [0, 4]])

print("Tensor 'a':\n", a)
print("Tensor 'b':\n", b)

# Kiểm tra tính bằng nhau
print("\na == b:\n", a == b)

# Kiểm tra lớn hơn
print("\na > b:\n", a > b)

# Kiểm tra nhỏ hơn hoặc bằng
print("\na <= b:\n", a <= b)
```

- Tensor Boolean rất hữu ích cho các thao tác **Tạo mặt nạ (Masking Operations)**, mà bạn sẽ tìm hiểu trong các phần nâng cao hơn về thao tác với Tensor.

## **7. Các phép toán logic (Logical Operations)**

- Các phép toán logic như `torch.logical_and()`, `torch.logical_or()` và `torch.logical_not()` hoạt động trên từng phần tử của Tensor Boolean hoặc những Tensor có thể được đánh giá trong ngữ cảnh Boolean. Trong trường hợp này, giá trị `0` được xem là `False`, còn giá trị khác `0` được xem là `True`.

```python
import torch

bool_a = torch.tensor([[True, False], [True, True]])
bool_b = torch.tensor([[False, True], [True, False]])

print("Boolean Tensor 'bool_a':\n", bool_a)
print("Boolean Tensor 'bool_b':\n", bool_b)

# Phép AND logic
print("\ntorch.logical_and(bool_a, bool_b):\n", torch.logical_and(bool_a, bool_b))

# Phép OR logic
print("\ntorch.logical_or(bool_a, bool_b):\n", torch.logical_or(bool_a, bool_b))

# Phép NOT logic
print("\ntorch.logical_not(bool_a):\n", torch.logical_not(bool_a))
```

- Trong chương tiếp theo, chúng ta sẽ tìm hiểu các phương pháp thao tác với Tensor nâng cao hơn, bao gồm **Lập chỉ mục (Indexing)**, **Cắt lát (Slicing)**, **Định hình lại Tensor (Reshaping)** và **Broadcasting**.
