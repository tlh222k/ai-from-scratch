# **Dữ liệu: Nhiên liệu cho Học máy (Data: The Fuel for Machine Learning)**

- Các thuật toán **Học máy (Machine Learning)** học hỏi từ kinh nghiệm, tương tự như cách con người học hỏi. Nhưng điều gì được xem là “kinh nghiệm” đối với một thuật toán? Câu trả lời chính là **dữ liệu (data)**.

- Nếu xem một mô hình **Machine Learning** như một động cơ, thì dữ liệu chính là nhiên liệu của nó. Không có dữ liệu, động cơ không thể hoạt động vì nó không có gì để học hỏi.

- Trong lĩnh vực **Machine Learning**, dữ liệu thường đề cập đến một tập hợp các quan sát hoặc ví dụ. Đối với nhiều bài toán nhập môn, dữ liệu thường được tổ chức theo định dạng có cấu trúc, chẳng hạn như:
  - Bảng hoặc bảng tính.
    - Trong đó, mỗi hàng đại diện cho một quan sát riêng lẻ (còn được gọi là **mẫu (sample)**, **thực thể (instance)** hoặc **điểm dữ liệu (data point)**).
    - Còn mỗi cột đại diện cho một đặc điểm cụ thể hoặc **đặc trưng (feature)** của quan sát đó.

## **1. Thực thể (Instance)**

- **Thực thể (Instance / Sample / Data Point / Record / Observation):** Mỗi hàng trong bảng đại diện cho một email cụ thể. Đây là một ví dụ riêng lẻ mà mô hình có thể học hỏi từ đó.

## **2. Đặc trưng (Features)**

- **Đặc trưng (Features):** Đây là những đặc điểm hoặc thuộc tính có thể đo lường được của một thực thể, được sử dụng làm đầu vào cho mô hình.
- Trong ví dụ này: **Contains 'Viagra'**, **All Caps Subject** và **Known Sender** là các **features**. Chúng là những biến đầu vào $x$ mà mô hình sử dụng để đưa ra dự đoán. Tùy thuộc vào bài toán, bạn có thể có một vài **features** hoặc thậm chí hàng nghìn **features**.

## **3. Nhãn (Label)**

- **Nhãn (Label / Target / Output / Response Variable):** Đây là kết quả hoặc giá trị mà chúng ta muốn mô hình **Machine Learning** dự đoán.

- Trong ví dụ về email rác, cột **Is Spam?** chính là **Label**. Trong giai đoạn **học có giám sát (Supervised Learning)**, **Label** đại diện cho *“đáp án đúng” của từng thực thể*. Mô hình học mối quan hệ giữa các **features $x$** và **Label $y$**. Lưu ý rằng không phải tất cả các loại **Machine Learning** đều sử dụng **Label**; chúng ta sẽ tìm hiểu về **học không giám sát (Unsupervised Learning)** trong phần sau.

> Chất lượng và số lượng dữ liệu có ý nghĩa vô cùng quan trọng đối với sự thành công của một dự án **Machine Learning**.

## **4. Số lượng (Quantity)**

- Nhìn chung, càng có nhiều dữ liệu thì mô hình càng có khả năng học được những quy luật đáng tin cậy hơn.

- Một mô hình chỉ được huấn luyện bằng mười email có thể không khái quát hóa tốt khi gặp những email mới mà nó chưa từng thấy, so với một mô hình được huấn luyện bằng hàng triệu email.

## **5. Chất lượng (Quality)**

- Dữ liệu phải phù hợp và chính xác.

- Nếu các **features** được thu thập không có mối liên hệ nào với việc một email có phải là thư rác hay không, mô hình sẽ không thể học một cách hiệu quả.

- Tương tự, nếu nhãn thường xuyên bị gán sai (ví dụ: đánh dấu email hợp lệ là thư rác), mô hình sẽ học những quy luật không chính xác.

- Điều này dẫn đến một câu nói phổ biến trong lĩnh vực này: **“Đầu vào rác, đầu ra rác” (Garbage In, Garbage Out)**. Dữ liệu kém chất lượng chắc chắn sẽ dẫn đến một mô hình kém chất lượng.

- Dữ liệu có thể tồn tại dưới nhiều dạng khác nhau, chẳng hạn như:

  - **Dữ liệu số (Numerical):** Ví dụ: nhiệt độ và chiều cao.
  - **Dữ liệu phân loại (Categorical):** Ví dụ: màu sắc và tên miền của người gửi email.
  - **Dữ liệu văn bản (Text):** Dữ liệu được biểu diễn dưới dạng văn bản.

- Việc chuẩn bị dữ liệu thô thành định dạng phù hợp với các thuật toán là một phần quan trọng trong quy trình **Machine Learning**. Nội dung này sẽ được trình bày trong **Chương 6 - Chuẩn bị dữ liệu của bạn (Preparing Your Data)**.

- Ở thời điểm hiện tại, điều quan trọng nhất cần ghi nhớ là dữ liệu tạo thành nền tảng để xây dựng các mô hình **Machine Learning**.
