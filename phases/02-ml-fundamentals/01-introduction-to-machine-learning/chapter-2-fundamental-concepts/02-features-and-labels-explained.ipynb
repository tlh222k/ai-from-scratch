# **Giải thích về Đặc trưng và Nhãn (Features and Labels Explained)**

Dữ liệu là thành phần thiết yếu của **Máy học (Machine Learning)**. Việc hiểu rõ những gì cấu thành nên dữ liệu là nền tảng để triển khai các ứng dụng Machine Learning hiệu quả. Một **tập dữ liệu (Dataset)** trong Machine Learning là tập hợp các mẫu (examples), trong đó mỗi mẫu được mô tả bằng các đặc điểm hoặc thuộc tính của nó. Những đặc điểm này được chia thành hai thành phần cơ bản: **Đặc trưng (Features)** và **Nhãn (Labels)**.



## **1. Đặc trưng: Đầu vào hoặc Biến dự đoán (Features: The Inputs or Predictors)**

Hãy xem **Features** là những thuộc tính hoặc đặc điểm riêng lẻ, có thể đo lường được của hiện tượng mà bạn đang quan sát. Đây chính là các *đầu vào (inputs)* được đưa vào mô hình Machine Learning để đưa ra dự đoán hoặc tìm kiếm quy luật. Mỗi **Feature** đại diện cho một phần thông tin về một điểm dữ liệu hoặc một mẫu cụ thể.

Dưới đây là một số tình huống minh họa:

  - **Dự đoán giá nhà (Predicting House Prices):** Nếu bạn muốn dự đoán giá của một ngôi nhà, các **Features** có thể bao gồm diện tích (tính bằng feet vuông), số phòng ngủ, số phòng tắm, tuổi của ngôi nhà và khoảng cách từ ngôi nhà đến trung tâm thành phố. Mỗi yếu tố đều là một Feature.

  - **Nhận diện email rác (Identifying Spam Emails):** Để phân loại một email là thư rác hay không phải thư rác, các **Features** có thể bao gồm địa chỉ email của người gửi, sự xuất hiện của một số từ khóa nhất định (như "offer", "free", "winner"), số lượng chữ cái viết hoa được sử dụng hoặc việc email có chứa tệp đính kèm hay không.

  - **Nhận dạng chữ số viết tay (Recognizing Handwritten Digits):** Đối với một hệ thống nhận dạng chữ số viết tay từ hình ảnh, các **Features** có thể là giá trị của từng pixel riêng lẻ trong lưới ảnh.

**Features** thường được biểu diễn dưới dạng các cột trong bảng tính hoặc bảng cơ sở dữ liệu, trong đó mỗi hàng tương ứng với một mẫu riêng biệt (chẳng hạn như một ngôi nhà, một email hoặc một hình ảnh cụ thể). Chúng có thể là dữ liệu số (như diện tích hoặc giá trị pixel) hoặc dữ liệu phân loại (như tên miền của địa chỉ email người gửi hoặc sự hiện diện/vắng mặt của một từ khóa).

Các thuật ngữ khác thường được sử dụng thay thế cho **Features** bao gồm:

  - **Predictors** (Biến dự đoán)
  - **Inputs** (Đầu vào)
  - **Attributes** (Thuộc tính)
  - **Independent Variables** (Biến độc lập)



## **2. Nhãn: Đầu ra hoặc Mục tiêu (Labels: The Output or Target)**

**Nhãn (Label)** (hoặc **Target**) là đối tượng cụ thể mà bạn đang cố gắng dự đoán bằng mô hình Machine Learning. Đây chính là "đáp án" hoặc kết quả mà bạn muốn mô hình học cách liên kết với các **Features** đầu vào.

**Labels** chủ yếu gắn liền với **Học có giám sát (Supervised Learning)**, một loại hình Machine Learning mà chúng ta sẽ tìm hiểu kỹ hơn ở phần sau. Trong phương pháp này, tập dữ liệu bao gồm đầu ra chính xác tương ứng với từng mẫu đầu vào. Mô hình học bằng cách so sánh các dự đoán của nó với những **Labels** đã biết này.

Hãy xem xét lại các ví dụ trên:

  - **Dự đoán giá nhà (Predicting House Prices):** **Label** sẽ là giá bán thực tế của ngôi nhà (ví dụ: 350.000 USD). Đây là bài toán **Hồi quy (Regression)** vì Label là một giá trị số liên tục.

  - **Nhận diện email rác (Identifying Spam Emails):** **Label** sẽ là kết quả phân loại email, có thể được biểu diễn bằng "spam" hoặc "not spam" (hoặc đôi khi bằng giá trị số: 1 là spam, 0 là không phải spam). Đây là bài toán **Phân loại (Classification)** vì Label thuộc một tập hợp các danh mục rời rạc.

  - **Nhận dạng chữ số viết tay (Recognizing Handwritten Digits):** **Label** của mỗi hình ảnh sẽ là chữ số thực tế mà hình ảnh đó biểu thị (0, 1, 2, ..., 9). Đây cũng là một bài toán Classification.

Trong các tập dữ liệu được sử dụng cho **Supervised Learning**, Label thường được biểu diễn bằng một cột riêng biệt, tách biệt với các cột Features.

Các thuật ngữ đồng nghĩa thường gặp với **Label** bao gồm:

  - **Target Variable** (Biến mục tiêu)
  - **Output** (Đầu ra)
  - **Response** (Biến phản hồi)
  - **Dependent Variable** (Biến phụ thuộc)
  - **Class** (Lớp – đặc biệt trong các bài toán Classification)



## **3. Features và Labels khi kết hợp (Features and Labels Together)**

Mục tiêu cơ bản của **Supervised Learning** là sử dụng **Features** để dự đoán **Label**. Mô hình học mối quan hệ hoặc quy luật tiềm ẩn kết nối các Features đầu vào với Label đầu ra dựa trên những mẫu được cung cấp trong **Dữ liệu huấn luyện (Training Data)**.

Dưới đây là cách biểu diễn đơn giản về Features và Labels trong một tập dữ liệu nhỏ dùng để dự đoán giá nhà:

| ID mẫu (Example ID) | Diện tích (feet vuông) (Feature 1) | Số phòng ngủ (Feature 2) | Tuổi nhà (năm) (Feature 3) | Giá ($) (Label) |
| ---------- | ------------------------ | -------------------- | ----------------------- | ----------------- |
| 1 | 1500 | 3 | 10 | 350000 |
| 2 | 2100 | 4 | 5 | 480000 |
| 3 | 1200 | 2 | 25 | 290000 |
| 4 | 1800 | 3 | 8 | 410000 |

> *Bảng biểu diễn dữ liệu đơn giản, trong đó các mẫu (hàng) được mô tả bằng các Features (các cột đầu vào) và Label tương ứng (cột đầu ra).*

Trong bảng này:

  - Mỗi hàng là một mẫu (một ngôi nhà cụ thể).
  - Các cột "Size (sq ft)", "Bedrooms" và "Age (years)" là các **Features**.
  - Cột "Price ( \$ )" là **Label**.

Mô hình Machine Learning sẽ được huấn luyện bằng dữ liệu này (hoặc một lượng dữ liệu lớn hơn có cấu trúc tương tự) để học cách diện tích, số lượng phòng ngủ và tuổi của ngôi nhà có mối quan hệ như thế nào với giá bán cuối cùng. Sau khi được huấn luyện, bạn có thể cung cấp cho mô hình các Features của một ngôi nhà *mới* (ví dụ: diện tích 1.600 feet vuông, 3 phòng ngủ, tuổi nhà 12 năm), và mô hình sẽ dự đoán Label (mức giá có khả năng xảy ra).

Cần lưu ý rằng không phải tất cả các tác vụ Machine Learning đều sử dụng Labels. Trong **Học không giám sát (Unsupervised Learning)**, mục tiêu thường là tìm ra cấu trúc hoặc quy luật bên trong dữ liệu chỉ dựa trên Features, mà không cần có các đáp án chính xác được xác định trước. Chúng ta sẽ tìm hiểu chi tiết hơn về các loại hình học khác nhau ở phần sau.



----
Việc hiểu rõ sự khác biệt giữa **Features** và **Labels** là nền tảng quan trọng. Điều này giúp bạn xác định đúng bài toán, chuẩn bị dữ liệu phù hợp và lựa chọn các thuật toán Machine Learning thích hợp. Khi được cung cấp một tập dữ liệu, một trong những bước đầu tiên thường là xác định những cột nào đại diện cho các Features đầu vào và cột nào (nếu có) đại diện cho Label mục tiêu mà bạn muốn dự đoán.
