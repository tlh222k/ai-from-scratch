# **Tập huấn luyện, Tập xác thực và Tập kiểm thử (Training, Validation, and Test Sets)**

Hãy tưởng tượng bạn đang dạy một học sinh một môn học mới. Bạn cung cấp cho học sinh tài liệu học tập và các bài tập thực hành. Làm thế nào để biết học sinh thực sự hiểu và nắm vững các khái niệm, hay chỉ đơn thuần ghi nhớ đáp án của những bài tập cụ thể mà bạn đã đưa ra? Bạn sẽ không kiểm tra học sinh bằng chính xác những bài tập đã dùng để luyện tập, đúng không? Thay vào đó, bạn sẽ đưa ra một bài kiểm tra mới với những câu hỏi khác nhau nhưng vẫn bao quát các chủ đề tương tự.

Các mô hình **Machine Learning** cũng phải đối mặt với một thách thức tương tự. Nếu chúng ta huấn luyện mô hình bằng toàn bộ dữ liệu hiện có rồi kiểm tra mô hình trên chính tập dữ liệu đó, mô hình có thể đạt hiệu suất hoàn hảo chỉ vì đã "ghi nhớ" các đáp án. Điều này không cho chúng ta biết mô hình sẽ hoạt động tốt như thế nào trên những dữ liệu mới mà nó chưa từng nhìn thấy trong tương lai, trong khi đây thường chính là mục tiêu cốt lõi của việc xây dựng mô hình. Khả năng hoạt động tốt trên dữ liệu mới này được gọi là **Khả năng khái quát hóa (Generalization)**.

Để đánh giá đúng khả năng khái quát hóa và xây dựng các mô hình đáng tin cậy, chúng ta chia tập dữ liệu thành những tập con riêng biệt. Phương pháp phổ biến nhất sử dụng ba tập dữ liệu khác nhau:



## **1. Tập huấn luyện (Training Set)**

Đây là phần dữ liệu lớn nhất, thường chiếm từ 60% đến 80% tổng số dữ liệu. Thuật toán Machine Learning sử dụng tập dữ liệu này để thực sự học các quy luật và mối quan hệ giữa các **Features** đầu vào và **Labels** đầu ra (trong **Supervised Learning**).

Hãy xem đây là "sách giáo khoa" và "bài tập về nhà" của mô hình. Mô hình liên tục điều chỉnh các tham số nội bộ bằng cách quan sát những mẫu trong **Training Set**, nhằm cố gắng giảm thiểu các lỗi mà nó mắc phải trên tập dữ liệu này.



## **2. Tập xác thực (Validation Set) (hoặc Tập phát triển – Development Set)**

Đây là một phần dữ liệu nhỏ hơn, độc lập, có thể chiếm từ 10% đến 20% tổng số dữ liệu. Mô hình *không* trực tiếp được huấn luyện trên tập dữ liệu này. Thay vào đó, **Validation Set** phục vụ hai mục đích chính:





### **2.1. Điều chỉnh siêu tham số (Hyperparameter Tuning)**

Như chúng ta đã tìm hiểu trước đó, **siêu tham số (Hyperparameters)** là những thiết lập do chúng ta lựa chọn *trước khi* quá trình huấn luyện bắt đầu (khác với các tham số - **parameters**, được mô hình tự học *trong quá trình* huấn luyện). Ví dụ về siêu tham số có thể là việc lựa chọn độ phức tạp của mô hình hoặc thiết lập **tốc độ học (Learning Rate)** cho thuật toán huấn luyện.

   - Chúng ta thường thử nhiều tổ hợp **Hyperparameters** khác nhau, huấn luyện các mô hình riêng biệt bằng **Training Set** cho từng tổ hợp, sau đó đánh giá hiệu suất của mỗi mô hình đã huấn luyện trên **Validation Set**. Các thiết lập **Hyperparameters** mang lại hiệu suất tốt nhất trên **Validation Set** thường được lựa chọn cho mô hình cuối cùng.

### **2.2. Lựa chọn mô hình và kiểm tra hiện tượng quá khớp (Model Selection & Overfitting Check)**

Chúng ta cũng có thể sử dụng **Validation Set** để so sánh các loại mô hình khác nhau (ví dụ: nên sử dụng **Hồi quy tuyến tính (Linear Regression)** hay **KNN (K-Nearest Neighbors)** cho bài toán này?).

   - Ngoài ra, bằng cách theo dõi hiệu suất của mô hình trên **Validation Set** trong quá trình huấn luyện, chúng ta có thể phát hiện liệu mô hình có đang bắt đầu đạt hiệu suất rất tốt trên dữ liệu huấn luyện nhưng lại hoạt động kém trên dữ liệu xác thực hay không. Sự khác biệt này thường cho thấy hiện tượng **Quá khớp (Overfitting)**, trong đó mô hình đã ghi nhớ dữ liệu huấn luyện quá mức, bao gồm cả những nhiễu trong dữ liệu, khiến khả năng khái quát hóa của mô hình bị suy giảm. **Validation Set** đóng vai trò như một hệ thống cảnh báo sớm.

Hãy xem **Validation Set** như những bài "kiểm tra ngắn" (quizzes) hoặc "bài thi thử" (practice exams) được sử dụng trong quá trình học tập. Nó giúp định hướng chiến lược học tập (điều chỉnh Hyperparameters) và kiểm tra mức độ hiểu biết thực sự (khả năng khái quát hóa) mà không cần sử dụng các câu hỏi của bài thi cuối cùng.



## **3. Tập kiểm thử (Test Set)**

Đây là tập con cuối cùng, thường cũng chiếm từ 10% đến 20% tổng số dữ liệu. Tập dữ liệu này được giữ riêng và chỉ được sử dụng *một lần duy nhất* sau khi toàn bộ quá trình huấn luyện và điều chỉnh Hyperparameters đã hoàn tất. Mô hình chưa từng nhìn thấy dữ liệu này trong suốt quá trình phát triển.

> "Mục đích duy nhất của **Test Set** là cung cấp một đánh giá cuối cùng, không thiên lệch về hiệu suất của mô hình đã được lựa chọn. Nó mô phỏng cách mô hình được kỳ vọng sẽ hoạt động trên những dữ liệu hoàn toàn mới."


Điều vô cùng quan trọng là không được sử dụng **Test Set** cho bất kỳ quyết định huấn luyện hoặc điều chỉnh nào. Nếu bạn đánh giá mô hình trên **Test Set**, sau đó quay lại điều chỉnh mô hình hoặc Hyperparameters dựa trên kết quả đó rồi tiếp tục đánh giá lại, bạn đã thực sự "làm nhiễm bẩn" **Test Set**. Khi đó, tập dữ liệu này không còn cung cấp một ước lượng khách quan nữa vì các quyết định của bạn đã bị ảnh hưởng bởi chính dữ liệu cụ thể trong tập kiểm thử.

Hãy hình dung **Test Set** giống như một kỳ thi chính thức cuối cùng. Bạn chỉ làm bài một lần và điểm số đó phản ánh kiến thức của bạn. Bạn không được phép liên tục làm lại bài thi sau khi đã xem các câu hỏi.

## **5. Các phương pháp thực hành phổ biến (Common Practices)**

Dữ liệu thường được **xáo trộn ngẫu nhiên (Random Shuffling)** trước khi phân chia nhằm bảo đảm mỗi tập con có tính đại diện cho phân phối tổng thể của dữ liệu.

Tỷ lệ phân chia chính xác (ví dụ: 70/15/15, 80/10/10) có thể thay đổi tùy thuộc vào quy mô tập dữ liệu và bài toán cụ thể. Đối với những tập dữ liệu rất lớn, các tỷ lệ nhỏ hơn có thể đã đủ để phục vụ việc xác thực và kiểm thử.

Đôi khi, đặc biệt trong các quy trình làm việc đơn giản hơn hoặc khi tập dữ liệu có quy mô nhỏ, người ta có thể chỉ chia dữ liệu thành Training Set và Test Set (ví dụ: 80/20). Trong trường hợp này, có thể sử dụng các kỹ thuật như **Kiểm định chéo (Cross-validation)** trên Training Set để thực hiện việc điều chỉnh Hyperparameters và lựa chọn mô hình (chúng ta có thể sẽ tìm hiểu về Cross-validation trong các phần sau).

Tuy nhiên, phương pháp phân chia thành ba tập (train/validate/test) vẫn là nền tảng và là thực tiễn được chấp nhận rộng rãi, đặc biệt khi phát triển các mô hình phức tạp hơn.


---
Tóm lại, việc chia dữ liệu thành Training Set, Validation Set và Test Set là một kỹ thuật nền tảng trong Machine Learning. Phương pháp này cho phép chúng ta huấn luyện mô hình, điều chỉnh mô hình một cách hiệu quả và thu được một ước lượng đáng tin cậy về khả năng hoạt động của mô hình trên dữ liệu mới. Đây là yếu tố thiết yếu để xây dựng những ứng dụng đáng tin cậy và hữu ích.

Phương pháp này cũng trực tiếp hỗ trợ việc chẩn đoán và ngăn ngừa những vấn đề phổ biến như Overfitting, một chủ đề mà chúng ta sẽ tìm hiểu ngay sau đây.

```python
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# 1. Tạo tập dữ liệu giả lập
X, y = make_classification(n_samples=1000, n_features=20, random_state=42)

# 2. Chia dữ liệu thành: 70% Train, 15% Validation, 15% Test
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42
)
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full, test_size=0.176, random_state=42
)

# 3. Huấn luyện mô hình với siêu tham số (Hyperparameter: max_depth = 3)
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)

# 4. Đánh giá trên Validation Set để kiểm tra hiệu suất
val_preds = model.predict(X_val)
print("Validation Accuracy:", accuracy_score(y_val, val_preds))

# 5. Đánh giá lần cuối trên Test Set
test_preds = model.predict(X_test)
print("Test Accuracy:", accuracy_score(y_test, test_preds))
```

**Phân tách tập dữ liệu (Dataset Split)**: Toàn bộ dữ liệu ban đầu được chia ngẫu nhiên thành ba tập riêng biệt để phục vụ các mục đích khác nhau trong vòng đời phát triển mô hình:

- **Training Set (60% - 80%)**: Dùng để huấn luyện các tham số bên trong của mô hình. Mô hình học cách nhận diện các mối quan hệ giữa đặc trưng đầu vào và nhãn thông qua tập này.

- **Validation Set (10% - 20%)**:  Không dùng để huấn luyện trực tiếp. Tập này đóng vai trò như các bài kiểm tra định kỳ (quizzes) để:

  - **Tinh chỉnh siêu tham số (Hyperparameter Tuning)**: Thử nghiệm nhiều cấu hình khác nhau (như tốc độ học hoặc độ phức tạp của mô hình) và chọn ra cấu hình mang lại kết quả tốt nhất.
  - **Kiểm tra hiện tượng quá khớp (Overfitting)**: Theo dõi hiệu suất trên tập này để phát hiện khi nào mô hình bắt đầu học thuộc lòng dữ liệu huấn luyện thay vì học quy luật tổng quát.

- **Test Set (10% - 20%)**: Được cất giữ hoàn toàn độc lập và chỉ sử dụng đúng một lần duy nhất ở bước cuối cùng. Mục đích là cung cấp thước đo khách quan và không chệch lệch về khả năng tổng quát hóa (generalization) của mô hình trên dữ liệu thực tế hoàn toàn chưa từng thấy.
