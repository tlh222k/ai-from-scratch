# **Tổng quan về quy trình Học máy (The Machine Learning Workflow Overview)**

- Xây dựng một hệ thống **Học máy (Machine Learning)** không phải là điều kỳ diệu, mà là một quy trình có cấu trúc rõ ràng. Mặc dù các chi tiết cụ thể có thể khác nhau tùy thuộc vào bài toán và dữ liệu, hầu hết các dự án Machine Learning đều tuân theo một chuỗi bước tổng quát.

- Hãy hình dung quy trình này như một bản đồ lộ trình, hướng dẫn bạn từ ý tưởng ban đầu đến một mô hình hoạt động được. Hiểu rõ quy trình sẽ giúp bạn nắm được bối cảnh của các khái niệm và kỹ thuật được trình bày trong những phần tiếp theo.

- Dưới đây là tổng quan về quy trình Machine Learning điển hình:



## **1. Xác định bài toán và nhìn nhận tổng thể (Frame the Problem and Look at the Big Picture)**

> Giải pháp cần đạt được mục tiêu kinh doanh nào? Hiệu suất của mô hình nên được đo lường như thế nào? Đây là loại bài toán gì (ví dụ: học có giám sát, học không giám sát, phân loại hay hồi quy)?

- Trả lời những câu hỏi này ngay từ đầu sẽ giúp xác định phạm vi và tiêu chí đánh giá thành công của dự án.

- Ví dụ: nếu bạn muốn dự đoán giá nhà, đây là một bài toán **hồi quy có giám sát (Supervised Regression)**. Bạn có thể đo lường mức độ thành công bằng cách xem các giá trị dự đoán gần với giá bán thực tế đến đâu.

## **2. Thu thập dữ liệu (Get the Data)**

- Các mô hình **Machine Learning** học từ dữ liệu, vì vậy việc thu thập dữ liệu phù hợp là yếu tố nền tảng.

- Quá trình này có thể bao gồm truy vấn cơ sở dữ liệu, sử dụng các bộ dữ liệu công khai, thu thập dữ liệu từ trang web (một cách có đạo đức và tuân thủ pháp luật) hoặc sử dụng API.

- Chất lượng và số lượng dữ liệu có ảnh hưởng đáng kể đến hiệu suất của mô hình.



## **3. Khám phá và chuẩn bị dữ liệu (Explore and Prepare the Data)**

- **Dữ liệu thô (Data Raw)** thường chưa được sắp xếp gọn gàng và có thể chứa nhiều vấn đề. Đây thường là bước tốn nhiều thời gian nhất, bao gồm:

- **Phân tích dữ liệu khám phá (Exploratory Data Analysis – EDA):** Tìm hiểu dữ liệu thông qua trực quan hóa và các thống kê tóm tắt. Quan sát các mẫu, mối tương quan, giá trị ngoại lệ và phân phối dữ liệu.

- **Làm sạch dữ liệu (Data Cleaning):** Xử lý các giá trị bị thiếu, sửa lỗi và giải quyết những điểm không nhất quán trong dữ liệu.

- **Tạo và lựa chọn đặc trưng (Feature Engineering/Selection):** Tạo ra các đặc trưng mới từ những đặc trưng hiện có hoặc lựa chọn các đặc trưng phù hợp nhất cho mô hình.

- **Biến đổi dữ liệu (Data Transformation):** Chuẩn hóa thang đo của các đặc trưng số hoặc mã hóa các đặc trưng phân loại sang định dạng phù hợp với thuật toán.

- Các kỹ thuật cụ thể sẽ được trình bày trong Chương 6



## **4. Lựa chọn và huấn luyện mô hình (Select and Train a Model)**

- Dựa trên loại bài toán và kết quả khám phá dữ liệu, bạn sẽ lựa chọn một hoặc nhiều mô hình tiềm năng. Ví dụ:

- **Hồi quy tuyến tính (Linear Regression):** Dùng để dự đoán các giá trị số.
- **K láng giềng gần nhất (K-Nearest Neighbors – KNN):** Dùng cho bài toán phân loại.
- **K-Means:** Dùng để phân cụm dữ liệu.

- Sau đó, bạn tiến hành **huấn luyện mô hình (Model Training)** bằng cách cung cấp dữ liệu đã được chuẩn bị cho mô hình. Tập dữ liệu này được gọi là **tập huấn luyện (Training Set)**.

- Trong quá trình huấn luyện, thuật toán học các mẫu hoặc mối quan hệ tồn tại bên trong dữ liệu.

- Các nội dung chi tiết về những mô hình như Linear Regression, KNN và K-Means sẽ được trình bày trong Chương 3, Chương 4 và Chương 5.



## **5. Đánh giá mô hình (Evaluate the Model)**

- Sau khi mô hình được huấn luyện, bạn cần đánh giá mức độ hoạt động hiệu quả của nó.

- Việc đánh giá được thực hiện bằng dữ liệu mà mô hình chưa từng nhìn thấy trước đó, được gọi là **tập kiểm tra (Test Set)**.

- Bạn sử dụng các **chỉ số đánh giá (Evaluation Metrics)** phù hợp với từng loại bài toán để đo lường hiệu suất, chẳng hạn như:

- **Độ chính xác (Accuracy):** Dùng cho bài toán phân loại.
- **Sai số bình phương trung bình (Mean Squared Error – MSE):** Dùng cho bài toán hồi quy.

- Bước này giúp xác định liệu mô hình đã đủ tốt hay cần tiếp tục điều chỉnh và cải thiện.

- Chương 2 đã giới thiệu các chỉ số cơ bản, và chúng ta sẽ tiếp tục xem xét việc đánh giá mô hình trong những chương sau.


## **6. Tinh chỉnh và lặp lại (Fine-Tune and Iterate)**

- Dựa trên kết quả đánh giá, bạn có thể cần điều chỉnh mô hình. Chẳng hạn, bạn có thể thay đổi các thiết lập được gọi là **siêu tham số (Hyperparameters)**, đã được giới thiệu sơ lược trong Chương 2

- Bạn thậm chí có thể cần quay lại những bước trước đó. Có thể bạn cần thêm dữ liệu, xây dựng các đặc trưng tốt hơn hoặc lựa chọn một mô hình hoàn toàn khác.

- **Machine Learning** thường là một quy trình lặp đi lặp lại, bao gồm nhiều chu kỳ huấn luyện, đánh giá và tinh chỉnh mô hình.



## **7. Trình bày giải pháp và triển khai (Present Solution and Deploy)**

*(Không bắt buộc trong khóa học này)*

- Khi đã hài lòng với hiệu suất của mô hình, bạn sẽ trình bày những kết quả thu được và có thể triển khai mô hình vào **môi trường vận hành thực tế (Production Environment)**.

- Tại đó, mô hình có thể đưa ra dự đoán dựa trên dữ liệu mới được thu thập trong thực tế.



---
- Quy trình này không phải lúc nào cũng diễn ra theo một trình tự tuyến tính nghiêm ngặt. Khi hiểu thêm về dữ liệu hoặc nhận thấy cách mô hình hoạt động, bạn có thể cần quay lại các bước trước đó.

- Việc hiểu những bước tổng quát này sẽ cung cấp một khuôn khổ để chúng ta tìm hiểu các thuật toán và kỹ thuật cụ thể trong những chương tiếp theo.

- Chúng ta sẽ quay lại quy trình này trong Chương 7, khi xây dựng một mô hình đơn giản từ đầu đến cuối.
