# **Khái niệm tối ưu hóa trong học máy (The Concept of Optimization in ML)**

Hãy cùng tìm hiểu quá trình **huấn luyện (training)** một mô hình học máy thực sự có ý nghĩa gì. Các mô hình **học máy (Machine Learning)** thường có thể được biểu diễn dưới dạng các **hàm số (mathematical functions)**.

Chẳng hạn, một mô hình **hồi quy tuyến tính (linear regression)** đơn giản cố gắng dự đoán giá trị đầu ra $y$ dựa trên đầu vào $x$ bằng một hàm có dạng $f(x)=mx+b$. Nhưng làm thế nào để tìm được những giá trị *tốt nhất* cho các **tham số (parameters)** $m$ (độ dốc) và $b$ (hệ số chặn) dựa trên dữ liệu huấn luyện? Đây chính là lúc **tối ưu hóa (optimization)** phát huy vai trò.

Trong bối cảnh **Machine Learning**, **Optimization** là quá trình tìm ra tập hợp các **Parameters** của một mô hình sao cho mô hình đạt được hiệu suất tốt nhất có thể trên dữ liệu mà nó được huấn luyện. Nhưng thế nào là “tốt nhất”? Chúng ta cần một cách để đo lường mức độ hoạt động hiệu quả (hoặc thường gặp hơn là mức độ hoạt động kém hiệu quả) của mô hình. Thước đo này thường được gọi là **hàm chi phí (cost function)** hoặc **hàm mất mát (loss function)**.

Một **Cost Function**, thường được ký hiệu là $J$, nhận các giá trị dự đoán của mô hình và các giá trị mục tiêu thực tế từ dữ liệu huấn luyện, sau đó tính toán một con số duy nhất để định lượng tổng thể sai số hoặc “chi phí”. Ví dụ, một **Cost Function** phổ biến trong các bài toán hồi quy là **Sai số bình phương trung bình (Mean Squared Error – MSE)**, được tính bằng cách lấy trung bình bình phương độ chênh lệch giữa các giá trị dự đoán và giá trị thực tế.

Nếu mô hình có các **Parameters** được biểu diễn bằng một vector $\theta$ (theta), thì **Cost Function** sẽ phụ thuộc vào các tham số này: $J(\theta)$. Khi đó, mục tiêu của **Optimization** là tìm ra tập hợp tham số cụ thể $\theta^*$ giúp tối thiểu hóa hàm chi phí:

$$
\theta^*=\arg\min_{\theta}J(\theta)
$$

Phương trình này được hiểu là: “$\theta^*$ là giá trị của các đối số (các giá trị tham số) $\theta$ giúp tối thiểu hóa hàm chi phí $J(\theta)$”.

Hãy hình dung **Cost Function** như một bề mặt, trong đó vị trí trên bề mặt được xác định bởi các giá trị của tham số mô hình, còn độ cao của bề mặt biểu thị chi phí. Mục tiêu của chúng ta là tìm ra điểm thấp nhất trên bề mặt này.

Ví dụ, hãy tưởng tượng một trường hợp rất đơn giản, trong đó chúng ta chỉ có một tham số, gọi là $w$. **Cost Function** $J(w)$ có thể có dạng một đường cong. **Optimization** khi đó có nghĩa là tìm giá trị của $w$ tại đáy đường cong này.

Đối với những mô hình có nhiều tham số (lên đến hàng triệu tham số trong trường hợp **học sâu (Deep Learning)**), “bề mặt” này tồn tại trong một không gian nhiều chiều, khiến việc trực quan hóa trở nên bất khả thi. Chúng ta không thể chỉ nhìn vào một đồ thị để tìm điểm cực tiểu. Thay vào đó, chúng ta cần một phương pháp có hệ thống, dựa trên thuật toán, để di chuyển trên bề mặt phức tạp này và tìm ra điểm thấp nhất, hoặc ít nhất là một điểm có giá trị rất thấp.

Đây chính là lúc **giải tích (Calculus)** trở nên không thể thiếu. Như đã đề cập sơ lược trong phần trước, **đạo hàm (derivatives)** đo lường tốc độ thay đổi hoặc độ dốc của một hàm số. Bằng cách tính đạo hàm (hoặc phiên bản tương đương trong trường hợp nhiều biến là **gradient**) của **Cost Function** theo các tham số của mô hình, chúng ta có thể xác định hướng mà chi phí tăng nhanh nhất. Để tối thiểu hóa chi phí, chúng ta chỉ cần di chuyển theo hướng *ngược lại*. Quá trình lặp đi lặp lại việc xác định hướng và thực hiện từng bước di chuyển này là nền tảng của nhiều thuật toán tối ưu hóa trong **Machine Learning**, nổi bật nhất là **Gradient Descent**.

Về bản chất, **Optimization** là cầu nối giữa việc có một cấu trúc mô hình và việc sở hữu một mô hình thực sự hoạt động hiệu quả trên dữ liệu. Đây chính là động cơ thúc đẩy quá trình học, còn **Calculus** cung cấp những công cụ nền tảng để xây dựng và tìm hiểu động cơ này.
