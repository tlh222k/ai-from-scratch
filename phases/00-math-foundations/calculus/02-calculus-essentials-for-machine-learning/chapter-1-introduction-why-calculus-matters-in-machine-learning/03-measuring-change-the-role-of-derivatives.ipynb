# **Đo lường sự thay đổi: Vai trò của đạo hàm (Measuring Change: The Role of Derivatives)**

Các mô hình **học máy (Machine Learning)** thường được biểu diễn dưới dạng các hàm số. Mục tiêu thường là tối ưu hóa những hàm số này, tức là tìm ra các đầu vào (tham số) mang lại đầu ra tốt nhất có thể (ví dụ: sai số thấp nhất). Vậy làm thế nào để tìm ra những tham số “tốt nhất” này một cách có hệ thống? Việc thử các giá trị ngẫu nhiên đơn thuần sẽ không hiệu quả, đặc biệt khi mô hình có đến hàng triệu tham số. Chúng ta cần một phương pháp giúp xác định *hướng* cần điều chỉnh các tham số.

Hãy tưởng tượng bạn đang đứng trên một sườn đồi trong làn sương mù dày đặc và muốn đi xuống chân đồi (điểm thấp nhất). Bạn không thể nhìn thấy toàn bộ khu vực xung quanh, nhưng có thể cảm nhận được độ dốc của mặt đất ngay tại vị trí mình đang đứng. Nếu mặt đất dốc xuống về phía bên trái, bạn sẽ bước sang trái. Nếu mặt đất dốc xuống về phía trước, bạn sẽ bước về phía trước. “Độ dốc” này cho bạn biết hướng đi xuống nhanh nhất.

Trong toán học, công cụ dùng để đo độ dốc tức thời hoặc tốc độ thay đổi tức thời của một hàm số tại một điểm cụ thể được gọi là **đạo hàm (derivative)**.





### **1. Đạo hàm: Đo lường sự thay đổi tức thời (The Derivative: Measuring Instantaneous Change)**

Hãy xét một hàm số $f(x)$ biểu diễn chi phí của mô hình (mức độ mô hình hoạt động kém hiệu quả) dựa trên một tham số duy nhất $x$. Chúng ta muốn tìm giá trị của $x$ sao cho $f(x)$ đạt giá trị nhỏ nhất. Đạo hàm của $f(x)$ theo $x$, thường được viết là $f'(x)$ hoặc $\frac{df}{dx}$, cho biết giá trị đầu ra $f(x)$ thay đổi bao nhiêu khi đầu vào $x$ thay đổi một lượng rất nhỏ.

Xét về mặt hình học, đạo hàm $f'(a)$ cho biết **độ dốc (slope)** của đường tiếp tuyến với đồ thị $y=f(x)$ tại điểm có hoành độ $x=a$.

- **Đạo hàm dương ($f'(a)>0$)**: Hàm số đang tăng tại $x=a$. Nếu chúng ta tăng nhẹ giá trị $x$, thì $f(x)$ sẽ tăng. Đường tiếp tuyến có độ dốc hướng lên.
- **Đạo hàm âm ($f'(a)<0$)**: Hàm số đang giảm tại $x=a$. Nếu chúng ta tăng nhẹ giá trị $x$, thì $f(x)$ sẽ giảm. Đường tiếp tuyến có độ dốc hướng xuống.
- **Đạo hàm bằng 0 ($f'(a)=0$)**: Hàm số tạm thời có độ dốc bằng 0 tại $x=a$. Đây có thể là điểm cực tiểu, điểm cực đại hoặc điểm yên ngựa (một điểm mà tại đó hàm số có độ dốc bằng 0 nhưng không phải điểm cực tiểu hay cực đại). Đường tiếp tuyến nằm ngang.

Ví dụ, xét hàm số $f(x)=x^2-2x+3$. Hãy cùng hình dung độ dốc của hàm số tại các điểm khác nhau:


> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/derivative_tangent_slope.html) ·


> *Đồ thị biểu diễn hàm số $f(x)=x^2-2x+3$. Tại $x=-1$, đường tiếp tuyến (nét đứt màu đỏ) có độ dốc hướng xuống (đạo hàm âm), cho thấy hàm số đang giảm. Tại $x=1$, đường tiếp tuyến (nét đứt màu vàng) nằm ngang (đạo hàm bằng 0), cho thấy đây có thể là điểm cực tiểu. Tại $x=3$, đường tiếp tuyến (nét đứt màu xanh lá) có độ dốc hướng lên (đạo hàm dương), cho thấy hàm số đang tăng.*

### **2. Tại sao đạo hàm quan trọng đối với tối ưu hóa? (Why Derivatives Matter for Optimization)**

Khái niệm này được áp dụng trực tiếp vào việc tối ưu hóa các mô hình **Machine Learning**. “Hàm chi phí” chính là hàm số mà chúng ta muốn tối thiểu hóa. Các tham số của mô hình đóng vai trò là đầu vào của hàm này.

Nếu tính đạo hàm của hàm chi phí theo một tham số cụ thể:

1. **Đạo hàm âm** cho biết việc tăng tham số sẽ làm *giảm* chi phí. Vì vậy, để tối thiểu hóa chi phí, chúng ta nên *tăng* giá trị của tham số đó.
2. **Đạo hàm dương** cho biết việc tăng tham số sẽ làm *tăng* chi phí. Để tối thiểu hóa chi phí, chúng ta nên *giảm* giá trị của tham số đó.
3. **Đạo hàm bằng 0** cho thấy có thể chúng ta đang ở một điểm cực tiểu (hoặc cực đại/điểm yên ngựa) xét theo tham số đó.

Bằng cách tính đạo hàm theo tất cả các tham số, chúng ta có thể xác định “hướng tăng dốc nhất” của hàm chi phí. Để tối thiểu hóa chi phí, chúng ta chỉ cần thực hiện một bước di chuyển nhỏ theo hướng *ngược lại*. Đây là ý tưởng cốt lõi đằng sau **Gradient Descent**, một trong những thuật toán tối ưu hóa phổ biến nhất trong **Machine Learning**, mà chúng ta sẽ tìm hiểu chi tiết hơn ở phần sau.

Ở thời điểm này, điều quan trọng cần ghi nhớ là đạo hàm cung cấp một thước đo định lượng về cách đầu ra của hàm số thay đổi theo đầu vào. Thước đo này rất cần thiết để định hướng quá trình tìm hiểu hàm chi phí phức tạp của mô hình và điều chỉnh các tham số theo từng bước lặp nhằm tìm ra những giá trị tối ưu. Trong các chương tiếp theo, chúng ta sẽ xây dựng định nghĩa chính thức về đạo hàm, học cách tính đạo hàm của nhiều loại hàm số khác nhau và mở rộng những ý tưởng này sang các hàm số có nhiều đầu vào (tham số), vốn là trường hợp phổ biến trong **Machine Learning**.
