# **Hàm số và mô hình trong học máy (Functions and Models in ML)**

Một **mô hình (Model)** trong học máy là một cấu trúc toán học được thiết kế để học các quy luật từ dữ liệu. Cách trực tiếp nhất để hình dung nhiều mô hình **học có giám sát (supervised learning)** là xem chúng như những hàm số.

Cũng giống như hàm số toán học $f(x)=y$ nhận đầu vào $x$ và tạo ra đầu ra $y$, một mô hình học máy nhận dữ liệu đầu vào (các đặc trưng) và tạo ra kết quả đầu ra (một dự đoán hoặc quyết định).



## **1. Đầu vào, đầu ra và tham số (Inputs, Outputs, and Parameters)**

Hãy cùng phân tích các thành phần:

1. **Đầu vào – Đặc trưng (Inputs – Features):** Đây là những đặc điểm có thể đo lường được của dữ liệu mà bạn đưa vào mô hình. Nếu bạn muốn dự đoán giá nhà, các đầu vào có thể bao gồm diện tích nhà, số phòng ngủ và tuổi của ngôi nhà.

   Trong học máy, chúng ta thường biểu diễn các đầu vào này dưới dạng một vector, thường được ký hiệu là $x$. Mỗi phần tử trong vector tương ứng với một đặc trưng cụ thể. Ví dụ:

   $$
   x=[\text{diện tích nhà},\text{số phòng ngủ},\text{tuổi ngôi nhà}]
   $$

2. **Đầu ra – Dự đoán (Outputs – Predictions):** Đây là kết quả mà mô hình tạo ra dựa trên các đầu vào. Đối với bài toán dự đoán giá nhà, đầu ra sẽ là mức giá dự đoán (một giá trị liên tục).

   Đối với bài toán phân loại hình ảnh, đầu ra có thể là một nhãn như 'mèo' hoặc 'chó' (một nhóm rời rạc), hoặc là xác suất của từng nhóm. Chúng ta thường ký hiệu kết quả dự đoán của mô hình là $\hat{y}$ (đọc là "y-hat") để phân biệt với giá trị mục tiêu thực tế $y$.

3. **Tham số mô hình (Model Parameters):** Đây là những biến nội bộ mà mô hình học được từ dữ liệu huấn luyện. Chúng xác định phép biến đổi cụ thể từ đầu vào thành đầu ra.

   Hãy hình dung chúng như những "núm điều chỉnh" được tinh chỉnh trong quá trình huấn luyện để kết quả dự đoán $\hat{y}$ của mô hình càng gần với giá trị mục tiêu thực tế $y$ càng tốt.

   Các tham số thường được biểu diễn bằng những chữ cái Hy Lạp như $\theta$ (theta), hoặc bằng các ký hiệu cụ thể như $w$ cho trọng số và $b$ cho độ chệch.



## **2. Mô hình dưới dạng hàm số: $f(x;\theta)$ (The Model as a Function: $f(x;\theta)$)**

Chúng ta có thể biểu diễn mối quan hệ giữa đầu vào, tham số và đầu ra bằng ký hiệu hàm số:

$$
\hat{y}=f(x;\theta)
$$

Phương trình này cho biết dự đoán $\hat{y}$ được tạo ra bởi hàm số $f$. Hàm số này nhận các đặc trưng đầu vào $x$ và được xác định bởi các tham số $\theta$.

Dạng cụ thể của hàm số $f$ sẽ quyết định loại mô hình.




### **Ví dụ: Hồi quy tuyến tính đơn giản (Simple Linear Regression)**

Hãy xét một trong những mô hình đơn giản nhất: **hồi quy tuyến tính (linear regression)** với một đặc trưng đầu vào duy nhất.

Mục tiêu là tìm một đường thẳng mô tả tốt nhất mối quan hệ giữa đầu vào $x$ (ví dụ: số năm kinh nghiệm) và đầu ra $y$ (ví dụ: mức lương).

Hàm số của mô hình này chính là phương trình đường thẳng:

$$
\hat{y}=wx+b
$$

Trong đó:

- $x$ là đặc trưng đầu vào duy nhất (số năm kinh nghiệm).
- $\hat{y}$ là giá trị đầu ra được dự đoán (mức lương).
- $w$ (trọng số) và $b$ (độ chệch) là các tham số của mô hình, $\theta=\{w,b\}$. Chúng lần lượt biểu thị độ dốc và tung độ gốc của đường thẳng.

Quá trình học bao gồm việc tìm các giá trị tối ưu của $w$ và $b$, sao cho mức lương dự đoán $\hat{y}$ càng khớp với mức lương thực tế $y$ trong dữ liệu huấn luyện càng tốt.

> Một mô hình hồi quy tuyến tính đơn giản $f(x;w,b)=wx+b$ được biểu diễn bằng một đường thẳng khớp với các điểm dữ liệu. Các tham số $w$ và $b$ xác định độ dốc và tung độ gốc của đường thẳng.



### **Các mô hình tuyến tính (Linear Models)**

Mặc dù hồi quy tuyến tính là một ví dụ dễ hình dung, cách biểu diễn mô hình dưới dạng hàm số này còn áp dụng được cho nhiều mô hình phức tạp hơn:

- **Hồi quy logistic (Logistic Regression):** Được sử dụng cho các bài toán phân loại. Mô hình này áp dụng hàm sigmoid lên tổ hợp tuyến tính của các đầu vào để tạo ra một giá trị xác suất. Dù vậy, nó vẫn là một hàm số $f(x;\theta)$.

- **Mạng nơ-ron (Neural Networks):** Về bản chất, đây là những tổ hợp hàm số phức tạp (các lớp). Mỗi lớp thực hiện các phép biến đổi tuyến tính và áp dụng các hàm kích hoạt phi tuyến. Toàn bộ mạng có thể được xem như một hàm số phức tạp duy nhất $f(x;\theta)$, trong đó $\theta$ bao gồm tất cả trọng số và độ chệch của các lớp.

Việc hiểu mô hình dưới dạng hàm số $f(x;\theta)$ là nền tảng rất quan trọng. Cách nhìn này giúp chúng ta xem nhiệm vụ của học máy là tìm ra bộ tham số $\theta$ tốt nhất cho một hàm số $f$ đã chọn, sao cho đầu ra dự đoán $\hat{y}$ phản ánh chính xác kết quả thực tế $y$.

Cách nhìn này dẫn chúng ta đến khái niệm **tối ưu hóa (optimization)**, trong đó chúng ta cần những công cụ để đo lường mức độ ảnh hưởng của những thay đổi trong $\theta$ đến hiệu suất của mô hình.

Đây chính là lúc giải tích, đặc biệt là đạo hàm và vector gradient, trở nên không thể thiếu, như chúng ta sẽ tìm hiểu trong các phần tiếp theo.
