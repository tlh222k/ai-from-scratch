# **Định nghĩa đạo hàm (Defining the Derivative)**

**Giới hạn (Limit)** cung cấp khuôn khổ toán học để định nghĩa chính xác **tốc độ thay đổi tức thời (Instantaneous Rate of Change)**. Khuôn khổ này sau đó được áp dụng để đưa ra định nghĩa chính thức về **đạo hàm (derivative)**.

Hãy tưởng tượng bạn có một hàm số, chẳng hạn $f(x)$, biểu diễn một đại lượng mà bạn muốn tìm hiểu hoặc tối ưu hóa, có thể là chi phí của một mô hình Machine Learning phụ thuộc vào một tham số duy nhất $x$. Bạn muốn biết đầu ra $f(x)$ thay đổi như thế nào khi bạn điều chỉnh nhẹ đầu vào $x$.

## **1. Từ tốc độ thay đổi trung bình đến tốc độ thay đổi tức thời (From Average Rate of Change to Instantaneous Rate of Change)**

Xét hai điểm trên đồ thị $y = f(x)$: một điểm tại $x$ và một điểm lân cận tại $x + h$. Các giá trị $y$ tương ứng là $f(x)$ và $f(x + h)$. Độ thay đổi của $x$ là:

$$
(x + h) - x = h
$$

và độ thay đổi của $y$ là:

$$
f(x + h) - f(x)
$$

**Tốc độ thay đổi trung bình (Average Rate of Change)** giữa hai điểm này chính là độ dốc của đường thẳng nối chúng (được gọi là **Đường cát tuyến (Secant Line)**):

$$
\begin{aligned}
\text{Tốc độ thay đổi trung bình}
&= \frac{\text{Độ thay đổi của } y}{\text{Độ thay đổi của } x} \\
&= \frac{f(x + h) - f(x)}{h}
\end{aligned}
$$

Đại lượng này cho biết trung bình, hàm số $f$ thay đổi bao nhiêu trên mỗi đơn vị thay đổi của $x$ trong khoảng từ $x$ đến $x + h$.

Nếu muốn biết tốc độ thay đổi chính xác tại điểm $x$, chúng ta có thể tìm Instantaneous Rate of Change bằng cách làm cho khoảng $h$ ngày càng nhỏ, về cơ bản là đưa điểm thứ hai ngày càng gần điểm thứ nhất. Đây chính là lúc Limit phát huy vai trò. Chúng ta lấy giới hạn của tốc độ thay đổi trung bình khi $h$ tiến đến $0$.

## **2. Định nghĩa toán học của Đạo hàm (The Derivative Defined)**

**Đạo hàm** của hàm số $f(x)$ theo $x$, được ký hiệu là $f'(x)$ (đọc là “$f$ phẩy của $x$”) hoặc $\frac{dy}{dx}$ (ký hiệu Leibniz, đọc là “$dy$ trên $dx$”), được định nghĩa là Instantaneous Rate of Change của hàm số tại điểm $x$. Về mặt hình thức, đạo hàm được xác định bằng giới hạn:

$$
f'(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}
$$

Nếu giới hạn này tồn tại, chúng ta nói hàm số $f$ **khả vi (differentiable)** tại $x$. Giá trị $f'(a)$ cho biết chính xác tốc độ thay đổi của hàm số $f(x)$ tại điểm cụ thể $x = a$.

## **3. Diễn giải hình học: Độ dốc của đường tiếp tuyến (Geometric Interpretation: Slope of the Tangent Line)**

Về mặt hình học, đạo hàm $f'(a)$ biểu thị độ dốc của đường thẳng tiếp xúc với đồ thị $y = f(x)$ tại điểm $(a, f(a))$. **Đường tiếp tuyến (Tangent Line)** là đường thẳng “chỉ vừa chạm” vào đường cong tại điểm đó và có cùng hướng với đường cong tại điểm ấy.

Khi $h \to 0$, đường Secant Line nối hai điểm $(x, f(x))$ và $(x + h, f(x + h))$ sẽ xoay và tiến gần đến Tangent Line tại $(x, f(x))$. Độ dốc của đường thẳng giới hạn này chính là đạo hàm.

Khi điểm thứ hai (được xác định bởi $h$) tiến gần đến điểm thứ nhất $(1, 1)$ trên đường cong $f(x) = x^2$, độ dốc của Secant Line tiến gần đến độ dốc của Tangent Line tại $x = 1$. Độ dốc của Tangent Line chính là đạo hàm $f'(1)$. Với $f(x) = x^2$, ta có $f'(x) = 2x$, do đó:

$$
f'(1) = 2
$$

## **4. Vai trò của Đạo hàm trong Học máy (Why is this Definition Important for Machine Learning?)**

Trong Machine Learning, chúng ta thường làm việc với các hàm chi phí (hoặc **Hàm mất mát (Loss Function)**) dùng để đo lường mức độ hoạt động kém hiệu quả của mô hình. Giả sử $J(w)$ là Cost Function, trong đó $w$ biểu thị một tham số (chẳng hạn như một trọng số) trong mô hình. Chúng ta muốn điều chỉnh $w$ để cực tiểu hóa $J(w)$.

Đạo hàm $\frac{dJ}{dw}$ cho biết mức độ nhạy của Cost Function $J$ đối với những thay đổi nhỏ của tham số $w$.

- Nếu $\frac{dJ}{dw} > 0$, việc tăng nhẹ $w$ sẽ làm tăng chi phí. Để giảm chi phí, chúng ta nên giảm $w$.
- Nếu $\frac{dJ}{dw} < 0$, việc tăng nhẹ $w$ sẽ làm giảm chi phí. Để giảm chi phí, chúng ta nên tăng $w$.
- Độ lớn của $\frac{dJ}{dw}$ cho biết Cost Function dốc đến mức nào tại điểm đó. Độ lớn càng lớn cho thấy chi phí thay đổi càng nhiều khi $w$ thay đổi một lượng nhỏ.

Thông tin về hướng và độ lớn của sự thay đổi này chính là điều mà các thuật toán tối ưu hóa như Gradient Descent sử dụng để cập nhật lặp đi lặp lại các tham số của mô hình và tìm ra chi phí cực tiểu. Vì vậy, hiểu về đạo hàm là nền tảng để hiểu cách các mô hình Machine Learning được huấn luyện.

Trong các phần tiếp theo, chúng ta sẽ tìm hiểu những quy tắc thực hành để tính đạo hàm mà không cần quay lại định nghĩa bằng giới hạn mỗi lần thực hiện. Đồng thời, chúng ta sẽ khám phá cách sử dụng đạo hàm để tìm các điểm cực tiểu và cực đại của hàm số.
