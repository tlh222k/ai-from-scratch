# **Tìm cực tiểu và cực đại bằng đạo hàm (Finding Minima and Maxima using Derivatives)**

**Derivative** $f'(x)$ cho biết **Slope** của hàm số $f(x)$ tại một điểm bất kỳ $x$. Vậy nếu độ dốc bằng $0$ thì điều đó có ý nghĩa gì? Trên đồ thị, độ dốc bằng $0$ tương ứng với một **Tangent Line** nằm ngang. Những điểm này thường xuất hiện ở đâu? Chúng thường nằm tại đáy của một thung lũng (cực tiểu) hoặc đỉnh của một ngọn đồi (cực đại) trên đồ thị hàm số. Đây là kiến thức nền tảng của **Optimization**.

## **1. Điểm tới hạn: Nơi diễn ra những điều quan trọng (Critical Points: Where the Action Happens)**

Những điểm mà **Derivative** $f'(x)$ bằng $0$ hoặc tại đó $f'(x)$ không xác định được gọi là **Critical Points**. Đây là những ứng viên chính có thể là **Local Minimum** hoặc **Local Maximum**. Một **Local Minimum** là điểm thấp hơn tất cả các điểm lân cận ngay sát nó, trong khi một **Local Maximum** là điểm cao hơn tất cả các điểm lân cận ngay sát nó.

Tại sao chúng ta tập trung vào $f'(x)=0$? Bởi vì nếu một hàm số trơn và khả vi, cách duy nhất để nó chuyển từ tăng (độ dốc dương) sang giảm (độ dốc âm), tạo thành một đỉnh, hoặc từ giảm sang tăng, tạo thành một đáy, là đi qua một điểm mà tại đó độ dốc tạm thời bằng $0$.

Ví dụ, hàm số $f(x)=x^2$ có **Derivative** $f'(x)=2x$. Cho $f'(x)=0$, ta có:

$$
2x=0
$$

Do đó, $x=0$ là **Critical Point** duy nhất. Theo trực giác, chúng ta biết rằng $x=0$ là điểm cực tiểu của parabol $y=x^2$.

> Hàm số $f(x)=x^2$ đạt cực tiểu tại $x=0$. Ở điểm này, **Tangent Line** nằm ngang, cho thấy **Slope** $f'(0)$ bằng $0$.

## **2. Sử dụng đạo hàm cấp một để phân loại các điểm tới hạn (Using the First Derivative to Classify Critical Points)**

Biết một điểm là điểm tới hạn vẫn chưa đủ. Đó là cực tiểu, cực đại hay một dạng điểm khác (chẳng hạn như một điểm phẳng trên một hàm số vốn đang tăng, ví dụ $f(x)=x^3$ tại $x=0$)? **Kiểm tra đạo hàm cấp một (First Derivative Test)** giúp chúng ta phân loại những điểm này bằng cách xem xét sự thay đổi của độ dốc xung quanh điểm tới hạn.

Cách thực hiện khá đơn giản:

1. Tìm các điểm tới hạn bằng cách giải phương trình $f'(x)=0$ (đồng thời ghi nhận những vị trí mà $f'(x)$ không xác định, mặc dù trường hợp này ít phổ biến hơn trong các **Cost Function** thường gặp của **Machine Learning**).
2. Chọn các giá trị kiểm tra nằm hơi lệch về bên trái và bên phải của mỗi điểm tới hạn $c$.
3. Xác định dấu của **Derivative** $f'(x)$ tại các điểm kiểm tra này:
   - Nếu $f'(x)$ đổi dấu từ âm (hàm số đang giảm) sang dương (hàm số bắt đầu tăng) khi đi qua $c$, thì hàm số $f$ có một **Local Minimum** tại $c$. Hãy hình dung việc trượt tuyết xuống một thung lũng ($f'<0$), rồi đi lên ở sườn bên kia ($f'>0$).
   - Nếu $f'(x)$ đổi dấu từ dương (hàm số đang tăng) sang âm (hàm số bắt đầu giảm) tại $c$, thì hàm số $f$ có một **Local Maximum** tại $c$. Hãy hình dung việc leo lên một ngọn đồi ($f'>0$), rồi đi xuống ở sườn bên kia ($f'<0$).
   - Nếu $f'(x)$ không đổi dấu tại $c$ (ví dụ, dương ở cả hai phía hoặc âm ở cả hai phía), thì $c$ không phải là cực tiểu hay cực đại địa phương. Điểm này có thể là một **Inflection Point** với **Tangent Line** nằm ngang.

**Ví dụ:** Hãy phân tích hàm số:

$$
f(x)=x^3-3x
$$

**Derivative** của hàm số là:

$$
f'(x)=3x^2-3
$$

Cho $f'(x)=0$:

$$
3x^2-3=0
$$

$$
3(x^2-1)=0
$$

$$
3(x-1)(x+1)=0
$$

Các điểm tới hạn là $x=1$ và $x=-1$.

Bây giờ, hãy kiểm tra dấu của $f'(x)=3(x-1)(x+1)$ trên các khoảng được xác định bởi những điểm này: $(-\infty,-1)$, $(-1,1)$ và $(1,\infty)$.

- Chọn $x=-2$ (thuộc khoảng $(-\infty,-1)$): $
f'(-2)=3(-2-1)(-2+1)=3(-3)(-1)=9$

  - Vì $f'(-2)>0$, hàm số tăng trên khoảng $(-\infty,-1)$.

- Chọn $x=0$ (thuộc khoảng $(-1,1)$): $ f'(0)=3(0-1)(0+1)=3(-1)(1)=-3 $
  - Vì $f'(0)<0$, hàm số giảm trên khoảng $(-1,1)$.

- Chọn $x=2$ (thuộc khoảng $(1,\infty)$): $ f'(2)=3(2-1)(2+1)=3(1)(3)=9$

  - Vì $f'(2)>0$, hàm số tăng trên khoảng $(1,\infty)$.

**Phân tích dựa trên sự thay đổi dấu tại các điểm tới hạn:**

- Tại $x=-1$, **Derivative** đổi dấu từ dương sang âm ($+\rightarrow-$). Điều này cho thấy có một **Local Maximum**. Giá trị của hàm số tại điểm này là:

$$
f(-1)=(-1)^3-3(-1)=-1+3=2
$$

- Tại $x=1$, **Derivative** đổi dấu từ âm sang dương ($-\rightarrow+$). Điều này cho thấy có một **Local Minimum**. Giá trị của hàm số tại điểm này là:

$$
f(1)=(1)^3-3(1)=1-3=-2
$$


> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/second_derivative_test.html)


Dấu của **First Derivative** $f'(x)$ cho chúng ta biết hàm số đang tăng hay giảm. Sự thay đổi dấu tại các điểm tới hạn ($x=-1$, $x=1$) giúp xác định các cực đại và cực tiểu địa phương.

## **3. Một phương pháp trực tiếp hơn: Kiểm tra đạo hàm cấp hai (A More Direct Approach: The Second Derivative Test)**

Đôi khi, việc tính đạo hàm có thể dễ dàng hơn so với kiểm tra các khoảng xung quanh điểm tới hạn. **Kiểm tra đạo hàm cấp hai (Second Derivative Test)** cung cấp một phương pháp khác để phân loại các điểm tới hạn, bằng cách tận dụng tính lõm của hàm số tại những điểm đó.

Hãy nhớ rằng **Second Derivative** $f''(x)$ là đạo hàm của **First Derivative** $f'(x)$. Nó cho biết tốc độ thay đổi của độ dốc.

- Nếu $f''(x)>0$ tại một điểm, độ dốc $f'(x)$ đang tăng. Điều này có nghĩa là đồ thị hàm số cong lên trên, giống như một chiếc bát có thể chứa nước. Chúng ta nói rằng hàm số **Concave Up** tại điểm đó.
- Nếu $f''(x)<0$ tại một điểm, độ dốc $f'(x)$ đang giảm. Đồ thị cong xuống dưới, giống như một chiếc bát úp ngược. Chúng ta nói rằng hàm số **Concave Down** tại điểm đó.

Điều này liên quan như thế nào đến cực tiểu và cực đại? Tại một điểm tới hạn $c$, nơi **Tangent Line** nằm ngang, nghĩa là $f'(c)=0$, chúng ta có thể xem xét những gì xảy ra tiếp theo.

1. Tính **Second Derivative** $f''(x)$.
2. Tính giá trị $f''(c)$, tức xác định tính lõm tại điểm tới hạn $c$.
3. Phân tích kết quả:
   - Nếu $f''(c)>0$: Hàm số có **Tangent Line** nằm ngang và **Concave Up**. Hình dạng này cho thấy một **Local Minimum**.
   - Nếu $f''(c)<0$: Hàm số có **Tangent Line** nằm ngang và **Concave Down**. Hình dạng này cho thấy một **Local Maximum**.
   - Nếu $f''(c)=0$: Phép kiểm tra không cung cấp thông tin nào. Điểm đó có thể là **Local Minimum**, **Local Maximum** hoặc **Inflection Point** (điểm mà tính lõm thay đổi). Trong trường hợp này, bạn cần sử dụng **First Derivative Test**.

**Ví dụ:** Hãy xét lại hàm số $f(x)=x^3-3x$. Trước đó, chúng ta đã tìm được:

$$
f'(x)=3x^2-3
$$

và các điểm tới hạn:

$$
x=1,\quad x=-1
$$

Bây giờ, hãy tìm **Second Derivative** bằng cách lấy đạo hàm của $f'(x)$:

$$
f''(x)=\frac{d}{dx}(3x^2-3)=6x
$$

Hãy tính giá trị $f''(x)$ tại các điểm tới hạn:

- Tại $x=-1$: $ f''(-1)=6(-1)=-6 $.
  
  - Vì $f''(-1)$ âm nên hàm số **Concave Down** tại $x=-1$. Do đó, $x=-1$ tương ứng với một **Local Maximum**.

- Tại $x=1$: $ f''(1)=6(1)=6$

  - Vì $f''(1)$ dương nên hàm số **Concave Up** tại $x=1$. Do đó, $x=1$ tương ứng với một **Local Minimum**.

Những kết luận này khớp với kết quả từ **First Derivative Test**. Trong ví dụ này, **Second Derivative** rất đơn giản để tính toán và xác định giá trị.

> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/derivative_sign_monotonicity.html)


> Dấu của **Second Derivative** $f''(x)$ cho biết tính lõm của hàm số. Tại các điểm tới hạn mà $f'(x)=0$, tính lõm dương ($f''>0$) cho thấy có một **Local Minimum**, trong khi tính lõm âm ($f''<0$) cho thấy có một **Local Maximum**. Tại nơi $f''(x)=0$ (trong ví dụ này là $x=0$), tính lõm thay đổi, xác định một **Inflection Point**.

## **4. Mối liên hệ với tối ưu hóa (Optimization Connection)**

Tại sao chúng ta dành thời gian tìm các cực tiểu và cực đại? Bởi vì đây là ý tưởng toán học cốt lõi đằng sau quá trình huấn luyện của nhiều **Model** trong **Machine Learning**. Các **Model** học bằng cách cố gắng tối thiểu hóa một **Cost Function** (còn được gọi là **Loss Function** hoặc **Objective Function**). Hàm này định lượng mức độ khớp giữa các dự đoán của **Model** và các giá trị mục tiêu thực tế trong dữ liệu huấn luyện. **Cost Function** có giá trị cao đồng nghĩa với hiệu suất kém. **Cost Function** có giá trị thấp đồng nghĩa với việc **Model** đang hoạt động tốt.

Quá trình huấn luyện bao gồm việc điều chỉnh các **Parameters** bên trong **Model** (chẳng hạn như **Slope** và hệ số chặn trong hồi quy tuyến tính, hoặc **weights** và **bias** trong **Neural Networks**) để tìm ra tập hợp giá trị **Parameters** giúp **Cost Function** đạt giá trị nhỏ nhất có thể.

Trong những trường hợp đơn giản, chẳng hạn như một số dạng hồi quy tuyến tính, chúng ta có thể tìm cực tiểu này bằng phương pháp giải tích: cho **Derivative** (hoặc **Gradient** trong không gian nhiều chiều) của **Cost Function** bằng 0 rồi giải phương trình, giống như cách chúng ta đã tìm các điểm tới hạn ở trên. Tuy nhiên, đối với hầu hết các **Model** phức tạp, đặc biệt là trong **Deep Learning**, **Cost Function** vô cùng phức tạp. Chúng ta không thể đơn giản giải phương trình $f'(x)=0$. Thay vào đó, chúng ta sử dụng các **iterative algorithms** như **Gradient Descent** (sẽ được trình bày chi tiết trong Chương 4). Những thuật toán này sử dụng **Derivative** (**Gradient**) tại vị trí hiện tại để xác định hướng đi xuống về phía cực tiểu, sau đó thực hiện một bước nhỏ theo hướng đó và lặp lại quá trình này nhiều lần.

Hiểu cách **First Derivative** và **Second Derivative** xác định cực tiểu và cực đại của các **single-variable functions** là nền tảng thiết yếu để hiểu những kỹ thuật **Optimization** mạnh mẽ được sử dụng trong toàn bộ lĩnh vực **Machine Learning**. Giờ đây, bạn đã có các công cụ **Calculus** để phân tích hình dạng của những hàm số đơn giản và xác định chính xác các điểm thấp nhất của chúng, trực tiếp phản ánh mục tiêu tối thiểu hóa một **Cost Function**. Tiếp theo, chúng sẽ áp dụng kiến thức này bằng cách tối ưu hóa một **Cost Function** cơ bản và sử dụng các công cụ Python để hỗ trợ tính toán.
