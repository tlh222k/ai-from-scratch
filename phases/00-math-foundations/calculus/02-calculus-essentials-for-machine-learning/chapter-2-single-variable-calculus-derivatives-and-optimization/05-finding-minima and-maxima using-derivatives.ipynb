# **Tìm cực tiểu và cực đại bằng đạo hàm (Finding Minima and Maxima using Derivatives)**

**Đạo hàm** $f'(x)$ cho biết độ dốc của hàm số $f(x)$ tại một điểm bất kỳ $x$. Vậy nếu độ dốc bằng $0$ thì có ý nghĩa gì? Trên đồ thị, độ dốc bằng $0$ tương ứng với một **Tangent Line** nằm ngang. Những điểm này thường xuất hiện ở đáy của một thung lũng hay đỉnh của một ngọn đồi trên đồ thị hàm số. Đây là kiến thức nền tảng của **Optimization**.

## **1. Điểm tới hạn (Critical Points)**

Những điểm mà đạo hàm $f'(x)$ bằng $0$ hoặc tại đó $f'(x)$ không xác định được gọi là **điểm tới hạn (Critical Points)**. Đây là những ứng viên chính có thể là cực tiểu địa phương hoặc cực đại địa phương. Một cực tiểu địa phương là điểm thấp hơn tất cả các điểm lân cận ngay sát nó, trong khi một cực đại địa phương là điểm cao hơn tất cả các điểm lân cận ngay sát nó.

Tại sao chúng ta tập trung vào $f'(x) = 0$? Bởi vì nếu một hàm số trơn và khả vi, cách duy nhất để nó chuyển từ tăng (độ dốc dương) sang giảm (độ dốc âm) để tạo thành một ngọn đồi, hoặc ngược lại để tạo thành một thung lũng, là đi qua một điểm mà tại đó độ dốc tạm thời bằng $0$.

Ví dụ, hàm số $f(x) = x^2$ có đạo hàm $f'(x) = 2x$. Cho $f'(x) = 0$, ta có:

$$
2x = 0
$$

Suy ra $x = 0$ là **Critical Point** duy nhất. Theo trực giác, chúng ta biết rằng $x = 0$ là điểm cực tiểu của parabol $y = x^2$, nơi **Tangent Line** nằm ngang và có độ dốc $f'(0) = 0$.

## **2. Kiểm tra đạo hàm cấp một (First Derivative Test)**

Biết một điểm là **Critical Point** vẫn chưa đủ. Đó là cực tiểu, cực đại hay một dạng điểm nào khác? **Phương pháp kiểm tra đạo hàm cấp một (First Derivative Test)** giúp chúng ta phân loại những điểm này bằng cách xem xét sự thay đổi của độ dốc xung quanh **Critical Point**.

Cách thực hiện khá đơn giản:

1. Tìm các **Critical Points** bằng cách giải phương trình $f'(x) = 0$.
2. Chọn các giá trị kiểm tra nằm hơi lệch về bên trái và bên phải của mỗi **Critical Point** $c$.
3. Xác định dấu của đạo hàm $f'(x)$ tại các điểm kiểm tra này:
   - Nếu $f'(x)$ đổi dấu từ âm (hàm số đang giảm) sang dương (hàm số bắt đầu tăng) khi đi qua $c$, thì hàm số $f$ đạt **cực tiểu địa phương (Local Minimum)** tại $c$.
   - Nếu $f'(x)$ đổi dấu từ dương (hàm số đang tăng) sang âm (hàm số bắt đầu giảm) tại $c$, thì hàm số $f$ đạt **cực đại địa phương (Local Maximum)** tại $c$.
   - Nếu $f'(x)$ không đổi dấu tại $c$, thì $c$ không phải là **Local Minimum** hay **Local Maximum** mà có thể là một **điểm uốn (Inflection Point)** với **Tangent Line** nằm ngang.

## **3. Ví dụ minh họa First Derivative Test**

Hãy phân tích hàm số:

$$
f(x) = x^3 - 3x
$$

Đạo hàm của hàm số là:

$$
f'(x) = 3 x^2 - 3
$$

Cho $f'(x) = 0$:

$$
3 x^2 - 3 = 0 \implies 3(x - 1)(x + 1) = 0
$$

Các **Critical Points** là $x = 1$ và $x = -1$.

Bây giờ, hãy kiểm tra dấu của $f'(x)$ trên các khoảng được xác định bởi những điểm này: $(-\infty, -1)$, $(-1, 1)$ và $(1, \infty)$:

- Chọn $x = -2$ (thuộc khoảng $(-\infty, -1)$):

  $$
  f'(-2) = 3(-2)^2 - 3 = 9 > 0 \quad (\text{Hàm số tăng})
  $$

- Chọn $x = 0$ (thuộc khoảng $(-1, 1)$):

  $$
  f'(0) = 3(0)^2 - 3 = -3 < 0 \quad (\text{Hàm số giảm})
  $$

- Chọn $x = 2$ (thuộc khoảng $(1, \infty)$):

  $$
  f'(2) = 3(2)^2 - 3 = 9 > 0 \quad (\text{Hàm số tăng})
  $$

**Phân tích sự thay đổi dấu:**
- Tại $x = -1$, đạo hàm đổi dấu từ dương sang âm ($+ \to -$), cho thấy có một **Local Maximum**. Giá trị cực đại là $f(-1) = (-1)^3 - 3(-1) = 2$.
- Tại $x = 1$, đạo hàm đổi dấu từ âm sang dương ($- \to +$), cho thấy có một **Local Minimum**. Giá trị cực tiểu là $f(1) = (1)^3 - 3(1) = -2$.

## **4. Kiểm tra đạo hàm cấp hai (Second Derivative Test)**

**Phương pháp kiểm tra đạo hàm cấp hai (Second Derivative Test)** cung cấp một phương pháp khác nhanh hơn để phân loại các **Critical Points** bằng cách tận dụng tính lõm của hàm số tại điểm đó:

1. Tính đạo hàm cấp hai $f''(x)$.
2. Tính giá trị $f''(c)$ tại **Critical Point** $c$ (nơi có $f'(c) = 0$):
   - Nếu $f''(c) > 0$: Đồ thị ở trạng thái **lõm lên (Concave Up)**. Điểm đó là một **Local Minimum**.
   - Nếu $f''(c) < 0$: Đồ thị ở trạng thái **lõm xuống (Concave Down)**. Điểm đó là một **Local Maximum**.
   - Nếu $f''(c) = 0$: Phép kiểm tra không đưa ra kết luận. Ta cần quay lại dùng **First Derivative Test**.

**Ví dụ:** Xét lại hàm số $f(x) = x^3 - 3x$ với các **Critical Points** $x = 1$ và $x = -1$:

$$
    f''(x) = \frac{d}{dx}(3 x^2 - 3) = 6x
$$

- Tại $x = -1$: $f''(-1) = 6(-1) = -6 < 0$ (đồ thị ở trạng thái **Concave Down**) $\implies$ **Local Maximum**.
- Tại $x = 1$: $f''(1) = 6(1) = 6 > 0$ (đồ thị ở trạng thái **Concave Up**) $\implies$ **Local Minimum**.

Tại nơi $f''(x) = 0$ (trong ví dụ này là $x = 0$), tính lõm thay đổi, xác định một **Inflection Point**.

## **5. Mối liên hệ với tối ưu hóa (Optimization Connection)**

Việc tìm các cực tiểu và cực đại là ý tưởng toán học cốt lõi đằng sau quá trình huấn luyện của nhiều mô hình **Machine Learning**. Các mô hình học bằng cách cố gắng tối thiểu hóa một hàm chi phí (còn được gọi là hàm mất mát hoặc hàm mục tiêu). Hàm này định lượng mức độ khớp giữa các dự đoán của mô hình và các giá trị mục tiêu thực tế trong dữ liệu huấn luyện. Chi phí cao có nghĩa là hiệu suất kém. Chi phí thấp có nghĩa là mô hình đang hoạt động tốt.

Quá trình huấn luyện bao gồm việc điều chỉnh các tham số bên trong mô hình để tìm ra tập hợp giá trị tham số giúp hàm chi phí đạt giá trị nhỏ nhất có thể.

Trong những trường hợp đơn giản, chẳng hạn như hồi quy tuyến tính, chúng ta có thể tìm cực tiểu này bằng phương pháp giải tích bằng cách cho đạo hàm (hoặc **Gradient** trong không gian nhiều chiều) của hàm chi phí bằng $0$ rồi giải phương trình. Tuy nhiên, đối với hầu hết các mô hình phức tạp như học sâu, chúng ta sử dụng các thuật toán lặp như **Gradient Descent**. Những thuật toán này sử dụng đạo hàm (**Gradient**) tại vị trí hiện tại để xác định hướng đi xuống về phía cực tiểu, sau đó thực hiện một bước nhỏ theo hướng đó và lặp lại quá trình này nhiều lần.
