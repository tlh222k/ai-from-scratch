# **Ứng dụng: Tối ưu hóa hàm chi phí đơn giản (Application: Simple Cost Function Optimization)**

**Derivative** giúp xác định độ dốc của một hàm số và tìm ra những điểm mà tại đó độ dốc bằng $0$, thường tương ứng với các giá trị cực tiểu hoặc cực đại. Khái niệm này được áp dụng trực tiếp vào một nhiệm vụ quan trọng trong **Machine Learning**: tối thiểu hóa một **Cost Function**.

## **1. Hàm chi phí là gì? (What is a Cost Function?)**

Trong **Machine Learning**, **Cost Function** (còn được gọi là **Loss Function** hoặc **Objective Function**) đo lường mức độ hoạt động của **Model**. Cụ thể, nó định lượng sự khác biệt giữa các giá trị **Predictions** của **Model** và các giá trị mục tiêu thực tế. **Cost** càng thấp thì hiệu suất càng tốt; mục tiêu của quá trình huấn luyện **Model** thường là tìm ra các **Parameters** giúp **Cost** đạt giá trị nhỏ nhất có thể.

Trước mắt, hãy xem xét một tình huống được đơn giản hóa rất nhiều, trong đó **Model** của chúng ta chỉ có một **Parameter**, gọi là $w$. Hãy tưởng tượng **Parameter** này điều khiển đầu ra của **Model**. **Cost Function** của chúng ta, $J(w)$, sẽ cho biết mức **Cost** tương ứng với một giá trị cụ thể của $w$. Mục tiêu của chúng ta là tìm giá trị $w$ giúp tối thiểu hóa $J(w)$.

## **2. Ví dụ đơn giản: Hàm chi phí bậc hai (A Simple Example: Quadratic Cost)**

Một dạng **Cost Function** phổ biến và thuận tiện về mặt toán học là hàm bậc hai. Hãy định nghĩa một **Cost Function** đơn giản như sau:

$$
J(w)=(w-3)^2+2
$$

Ở đây, $J(w)$ biểu thị **Cost** ứng với một giá trị **Parameter** $w$ cho trước. Chúng ta muốn tìm giá trị $w$ tại đó **Cost** $J(w)$ nhỏ nhất. Về mặt hình ảnh, hàm số này là một parabol mở lên trên.

**Cost Function** $J(w)=(w-3)^2+2$. **Cost** đạt giá trị nhỏ nhất tại $w=3$.

Làm thế nào để tìm giá trị cực tiểu này bằng phương pháp toán học sử dụng **Calculus**?

### **2.1. Bước 1: Tìm đạo hàm (Step 1: Find the Derivative)**

Chúng ta tính **First Derivative** của $J(w)$ theo $w$. Sử dụng **Power Rule** và **chain rule** (hoặc đơn giản là khai triển bình phương trước), ta có:

$$
J'(w)=\frac{dJ}{dw}
=\frac{d}{dw}\left((w-3)^2+2\right)
$$

$$
J'(w)=2(w-3)^1\cdot\frac{d}{dw}(w-3)+0
$$

$$
J'(w)=2(w-3)\cdot 1
$$

$$
J'(w)=2w-6
$$

**Derivative** này, $J'(w)=2w-6$, cho chúng ta biết độ dốc của **Cost Function** tại bất kỳ giá trị $w$ nào.

### **2.2. Bước 2: Tìm điểm tới hạn (Cho đạo hàm bằng 0) (Step 2: Find Critical Points (Set Derivative to Zero))**

Để tìm các điểm có khả năng là cực tiểu hoặc cực đại, chúng ta tìm những điểm mà độ dốc bằng 0. Ta cho **First Derivative** bằng 0 rồi giải phương trình theo $w$:

$$
J'(w)=0
$$

$$
2w-6=0
$$

$$
2w=6
$$

$$
w=3
$$

Điều này cho biết $w=3$ là một **Critical Point**, tại đó **Tangent Line** của hàm số nằm ngang.

### **2.3. Bước 3: Sử dụng kiểm tra đạo hàm cấp hai (Step 3: Use the Second Derivative Test)**

Liệu $w=3$ là cực tiểu, cực đại hay một dạng điểm khác? Chúng ta có thể sử dụng **Second Derivative Test**. Hãy tìm **Second Derivative**, $J''(w)$:

$$
J''(w)=\frac{d^2J}{dw^2}
=\frac{d}{dw}(J'(w))
=\frac{d}{dw}(2w-6)
$$

$$
J''(w)=2
$$

Bây giờ, hãy tính giá trị **Second Derivative** tại **Critical Point** $w=3$:

$$
J''(3)=2
$$

Vì $J''(3)=2>0$, hàm số **Concave Up** tại $w=3$. Điều này xác nhận rằng $w=3$ tương ứng với một **Local Minimum**. Vì **Cost Function** của chúng ta là một parabol đơn giản mở lên trên nên cực tiểu địa phương này cũng chính là cực tiểu toàn cục.

## **3. Mối liên hệ với học máy (Connection to Machine Learning)**

Ví dụ đơn giản này minh họa ý tưởng cốt lõi đằng sau việc tối ưu hóa nhiều **Model** trong **Machine Learning**:

- **Xác định hàm chi phí (Define a Cost Function):** Tạo một hàm $J$ đo lường sai số của **Model** dựa trên các **Parameters** của nó (ở đây chỉ có $w$).
- **Tính đạo hàm (hoặc Gradient) (Calculate the Derivative (or Gradient)):** Tìm **Derivative** của **Cost Function** theo từng **Parameter**. Điều này cho biết **Cost** thay đổi như thế nào khi bạn điều chỉnh các **Parameters**.
- **Tìm cực tiểu (Find the Minimum):** Sử dụng các kỹ thuật **Calculus** để tìm giá trị của **Parameters** tại đó **Derivative** bằng 0 (đồng thời thỏa mãn các điều kiện của cực tiểu, chẳng hạn như **Second Derivative** dương). Điều này xác định các **Parameters** giúp tối thiểu hóa **Cost**, từ đó tạo ra **Model** có hiệu suất “tốt nhất” theo **Cost Function** đó.

Trong các bài toán **Machine Learning** thực tế, **Cost Function** thường phụ thuộc vào hàng triệu **Parameters**, chứ không chỉ một **Parameter**. Hơn nữa, việc tìm cực tiểu bằng cách trực tiếp cho **Derivative** (hoặc **Gradient** trong trường hợp nhiều biến) bằng 0 có thể không khả thi về mặt tính toán hoặc không thể thực hiện được nếu hàm số quá phức tạp.

Đây là lý do chúng ta thường dựa vào các **iterative algorithms** như **Gradient Descent** (sẽ được tìm hiểu chi tiết trong Chương 4). **Gradient Descent** sử dụng **Derivative** không phải để trực tiếp giải tìm cực tiểu, mà để xác định hướng điều chỉnh các **Parameters** theo từng bước, qua đó dần tiến về phía **Cost** cực tiểu. Tuy nhiên, khái niệm nền tảng vẫn không thay đổi: **Derivative** định hướng quá trình **Optimization**.
