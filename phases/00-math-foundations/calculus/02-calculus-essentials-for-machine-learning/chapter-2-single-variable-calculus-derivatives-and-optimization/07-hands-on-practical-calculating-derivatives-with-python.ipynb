## **1. Tính đạo hàm ký hiệu bằng SymPy (Symbolic Differentiation with SymPy)**

**SymPy** là một thư viện Python dành cho toán học ký hiệu. Thư viện này cho phép chúng ta định nghĩa các biểu thức toán học bằng các ký hiệu và thực hiện những phép toán như tính đạo hàm, tích phân và giải phương trình, đồng thời duy trì các biểu diễn ký hiệu chính xác. Điều này đặc biệt hữu ích khi cần tìm biểu thức chính xác của một **Derivative**.

Trước tiên, hãy bảo đảm rằng bạn đã cài đặt SymPy (**`pip install sympy`**). Hãy bắt đầu bằng cách định nghĩa một biến ký hiệu và một hàm số.



```python
!pip install sympy
```

```python
import sympy

# Define 'x' as a symbolic variable
x = sympy.symbols('x')

# Define a function, for example, f(x) = x^3 + 2*x^2 - 5*x + 1
f = x**3 + 2*x**2 - 5*x + 1

print(f"Original function: f(x) = {f}")
```

Bây giờ, chúng ta có thể sử dụng hàm **`diff`** của SymPy để tính **Derivative** của $f(x)$ theo $x$.




```python
# Calculate the derivative of f with respect to x
f_prime = sympy.diff(f, x)

print(f"Derivative: f'(x) = {f_prime}")
```

SymPy áp dụng chính xác các quy tắc tính đạo hàm (**Power Rule**, **Sum Rule**) để đưa ra **Derivative** giải tích chính xác:

$$
f'(x)=3x^2+4x-5
$$

Chúng ta cũng có thể sử dụng SymPy để tính giá trị của **Derivative** tại một điểm cụ thể. Hãy tìm **Slope** của **Tangent Line** tại $x=2$.



```python
# Evaluate the derivative at x = 2
value_at_2 = f_prime.subs(x, 2)

print(f"Value of the derivative at x = 2: f'(2) = {value_at_2}")
```

Kết quả: $
f'(2)=3(2)^2+4(2)-5=12+8-5=15
$ cho chúng ta biết **Instantaneous Rate of Change** của hàm số $f(x)$ tại điểm $x=2$.

SymPy cũng có thể xử lý các hàm phức tạp hơn, bao gồm những thành phần lượng giác, hàm mũ và hàm logarit.



```python
import sympy
import numpy as np # Often used alongside SymPy

# Define x as a symbol
x = sympy.symbols('x')

# Define a more complex function: g(x) = sin(x) * exp(-x^2 / 2)
g = sympy.sin(x) * sympy.exp(-x**2 / 2)
print(f"\nAnother function: g(x) = {g}")

# Calculate its derivative
g_prime = sympy.diff(g, x)
print(f"Derivative: g'(x) = {g_prime}")

# Evaluate the derivative at x = 0
# Use evalf() for numerical evaluation if needed
value_at_0 = g_prime.subs(x, 0).evalf()
print(f"Value of the derivative at x = 0: g'(0) = {value_at_0}")
```

**Symbolic Differentiation** là lựa chọn hoàn hảo khi bạn cần biểu thức toán học chính xác của **Derivative** hoặc cần độ chính xác cao.



## **2. Liên hệ với tối ưu hóa (Connecting to Optimization)**

Tại sao việc tính **Derivative** lại quan trọng đến vậy trong **Machine Learning**? Như đã thảo luận trước đó trong chương này và trong [Chương 1](https://github.com/tlh222k/ai-from-scratch/tree/main/phases/00-math-foundations/calculus/02-calculus-essentials-for-machine-learning/chapter-1-introduction-why-calculus-matters-in-machine-learning), **Optimization** bao gồm việc tìm các **Parameters** (chẳng hạn như **weights** và **bias** trong một **Model**) giúp tối thiểu hóa **Cost Function** hoặc **Loss Function**. **Cost Function** này thường đo lường mức độ hoạt động kém hiệu quả của **Model**.

**Derivative** (hoặc khái niệm tương ứng trong trường hợp nhiều biến là **Gradient**, sẽ được tìm hiểu trong chương tiếp theo) cho chúng ta biết **Slope** của **Cost Function**. Bằng cách tính **Derivative**, các thuật toán **Optimization** như **Gradient Descent** biết cần điều chỉnh các **Parameters** theo hướng nào để giảm **Cost**, tức là thực hiện quá trình “đi xuống” theo độ dốc về phía một điểm cực tiểu. Vì vậy, khả năng tính toán hiệu quả các **Derivative**, dù bằng phương pháp ký hiệu hay phương pháp số, là điều thiết yếu để huấn luyện nhiều **Model** trong **Machine Learning**.

Khả năng thực hành tính **Derivative** bằng các thư viện Python như SymPy giúp chúng ta không còn phải thực hiện những phép tính thủ công tẻ nhạt, đồng thời cho phép tập trung vào việc xây dựng và tối ưu hóa các **Model** trong **Machine Learning**.
