# **Các quy tắc tính đạo hàm phổ biến (Common Differentiation Rules)**

Định nghĩa đạo hàm bằng giới hạn,

$$
\lim_{h \to 0} \frac{f(x + h) - f(x)}{h}
$$

cung cấp một phương pháp cơ bản để tính đạo hàm. Tuy nhiên, việc áp dụng trực tiếp định nghĩa này có thể trở nên tẻ nhạt đối với các hàm số phức tạp. Hãy tưởng tượng phải áp dụng định nghĩa này lặp đi lặp lại trong một vòng lặp tối ưu hóa của Machine Learning. Điều đó sẽ kém hiệu quả về mặt tính toán.

May mắn thay, giải tích cung cấp một tập hợp các quy tắc cho phép chúng ta tìm đạo hàm dễ dàng hơn nhiều bằng cách phân tách các hàm phức tạp thành những thành phần đơn giản hơn. Những quy tắc này là nền tảng để tính đạo hàm của các loại hàm số mà chúng ta thường gặp khi xác định chi phí của mô hình hoặc các hàm kích hoạt (activation functions) trong Machine Learning.

Hãy cùng tìm hiểu những quy tắc tính đạo hàm phổ biến và hữu ích nhất. Giả sử $f(x)$ và $g(x)$ là các hàm số khả vi, còn $c$ là một hằng số.

## **1. Quy tắc hằng số (Constant Rule)**

Quy tắc đơn giản nhất liên quan đến các hàm số luôn cho ra cùng một giá trị, chẳng hạn $f(x) = 5$. Vì giá trị của hàm số không thay đổi nên tốc độ thay đổi của nó (đạo hàm) luôn bằng $0$.

- **Quy tắc:** Nếu $f(x) = c$, trong đó $c$ là một hằng số, thì:

  $$
  f'(x) = 0
  $$

- **Theo ký hiệu Leibniz:**

  $$
  \frac{d}{dx}(c) = 0
  $$

- **Ví dụ:** Nếu $f(x) = 10$, thì $f'(x) = 0$.

## **2. Quy tắc lũy thừa (Power Rule)**

Đây là một trong những quy tắc được sử dụng thường xuyên nhất, đặc biệt đối với các hàm đa thức (polynomial functions) phổ biến trong mô hình hóa. Quy tắc này áp dụng cho các hàm có dạng $x$ được nâng lên một lũy thừa.

- **Quy tắc:** Nếu $f(x) = x^n$, trong đó $n$ là một số thực bất kỳ, thì:

  $$
  f'(x) = n x^{n - 1}
  $$

- **Theo ký hiệu Leibniz:**

  $$
  \frac{d}{dx}(x^n) = n x^{n - 1}
  $$

- **Giải thích:** Để tính đạo hàm của $x^n$, chúng ta đưa số mũ $n$ xuống làm hệ số nhân, sau đó giảm số mũ ban đầu đi $1$.

- **Ví dụ:**
  - Nếu $f(x) = x^3$, thì:

    $$
    f'(x) = 3 x^{3 - 1} = 3 x^2
    $$

  - Nếu $g(x) = x$ (tức là $x^1$), thì:

    $$
    g'(x) = 1 x^{1 - 1} = 1 x^0 = 1
    $$

    Điều này hợp lý vì độ dốc của đường thẳng $y = x$ bằng $1$.

  - Nếu $h(x) = \sqrt{x} = x^{1/2}$, thì:

    $$
    h'(x) = \frac{1}{2} x^{1/2 - 1} = \frac{1}{2} x^{-1/2} = \frac{1}{2\sqrt{x}}
    $$

  - Nếu $k(x) = \frac{1}{x^2} = x^{-2}$, thì:

    $$
    k'(x) = -2 x^{-2 - 1} = -2 x^{-3} = -\frac{2}{x^3}
    $$

## **3. Quy tắc nhân với hằng số (Constant Multiple Rule)**

Điều gì xảy ra nếu hàm số được nhân với một hằng số, chẳng hạn $f(x) = 5 x^3$? Quy tắc này cho biết về cơ bản, chúng ta có thể đưa hằng số ra ngoài, tính đạo hàm của phần còn lại rồi nhân lại với hằng số đó.

- **Quy tắc:** Nếu $h(x) = c \cdot f(x)$, thì:

  $$
  h'(x) = c \cdot f'(x)
  $$

- **Theo ký hiệu Leibniz:**

  $$
  \frac{d}{dx}(c \cdot f(x)) = c \cdot \frac{d}{dx}(f(x))
  $$

- **Ví dụ:** Nếu $f(x) = 5 x^3$. Chúng ta biết rằng:

  $$
  \frac{d}{dx}(x^3) = 3 x^2
  $$

  Áp dụng quy tắc:

  $$
  f'(x) = 5 \cdot \frac{d}{dx}(x^3) = 5 \cdot (3 x^2) = 15 x^2
  $$

## **4. Quy tắc tổng và hiệu (Sum and Difference Rule)**

Các Cost Function trong Machine Learning thường là tổng của nhiều hạng tử (chẳng hạn tổng bình phương sai số). Quy tắc này cho phép chúng ta tính đạo hàm của những hàm số như vậy theo từng hạng tử.

- **Quy tắc:** Nếu $h(x) = f(x) + g(x)$, thì:

  $$
  h'(x) = f'(x) + g'(x)
  $$

  Tương tự, nếu $h(x) = f(x) - g(x)$, thì:

  $$
  h'(x) = f'(x) - g'(x)
  $$

- **Theo ký hiệu Leibniz:**

  $$
  \frac{d}{dx}(f(x) \pm g(x)) = \frac{d}{dx}(f(x)) \pm \frac{d}{dx}(g(x))
  $$

- **Ví dụ:** Cho $h(x) = x^4 + 6 x^2 - 7$. Chúng ta có thể tính đạo hàm theo từng hạng tử bằng cách sử dụng Power Rule, Constant Multiple Rule và Constant Rule:

  $$
  \begin{aligned}
  h'(x)
  &= \frac{d}{dx}(x^4) + \frac{d}{dx}(6 x^2) - \frac{d}{dx}(7) \\
  &= (4 x^3) + (6 \cdot 2 x) - 0 \\
  &= 4 x^3 + 12 x
  \end{aligned}
  $$

## **5. Quy tắc tích (Product Rule)**

Quy tắc này cần thiết khi tính đạo hàm của một hàm số là tích của hai hàm số khác, chẳng hạn $h(x) = f(x)g(x)$. Điều quan trọng là không được đơn giản lấy đạo hàm riêng của từng hàm rồi nhân chúng với nhau.

- **Quy tắc:** Nếu $h(x) = f(x)g(x)$, thì:

  $$
  h'(x) = f'(x)g(x) + f(x)g'(x)
  $$

- **Theo ký hiệu Leibniz:**

  $$
  \frac{d}{dx}(f(x)g(x)) = \frac{df}{dx}g(x) + f(x)\frac{dg}{dx}
  $$

- **Ví dụ:** Cho $h(x) = (x^2 + 1)(x^3 - x)$.

  Đặt $f(x) = x^2 + 1$, khi đó $f'(x) = 2x$.
  
  Đặt tiếp $g(x) = x^3 - x$, khi đó $g'(x) = 3 x^2 - 1$.

  Áp dụng Product Rule:

  $$
  h'(x) = f'(x)g(x) + f(x)g'(x)
  $$

  $$
  \begin{aligned}
  h'(x)
  &= (2x)(x^3 - x) + (x^2 + 1)(3 x^2 - 1) \\
  &= (2 x^4 - 2 x^2) + (3 x^4 - x^2 + 3 x^2 - 1) \\
  &= 2 x^4 - 2 x^2 + 3 x^4 + 2 x^2 - 1 \\
  &= 5 x^4 - 1
  \end{aligned}
  $$

  **Tự kiểm tra:** Trước tiên, khai triển $h(x)$:

  $$
  h(x) = x^5 - x^3 + x^3 - x = x^5 - x
  $$

  Sau đó tính đạo hàm trực tiếp:

  $$
  h'(x) = 5 x^4 - 1
  $$

  Quy tắc này hoạt động hoàn toàn chính xác!

## **6. Quy tắc thương (Quotient Rule)**

Quy tắc này xử lý các hàm số được tạo thành bằng cách chia một hàm số cho một hàm số khác:

$$
h(x) = \frac{f(x)}{g(x)} \quad (g(x) \ne 0)
$$

- **Quy tắc:**

  $$
  h'(x) = \frac{f'(x)g(x) - f(x)g'(x)}{[g(x)]^2}
  $$

- **Theo ký hiệu Leibniz:**

  $$
  \frac{d}{dx}\left(\frac{f(x)}{g(x)}\right) = \frac{\frac{df}{dx}g(x) - f(x)\frac{dg}{dx}}{[g(x)]^2}
  $$

- **Ví dụ:** Cho $h(x) = \frac{x^2}{x - 1}$.

  Đặt $f(x) = x^2$ (phần tử số), khi đó $f'(x) = 2x$.
  
  Đặt $g(x) = x - 1$ (phần mẫu số), khi đó $g'(x) = 1$.

  Áp dụng Quotient Rule:

  $$
  \begin{aligned}
  h'(x)
  &= \frac{(2x)(x - 1) - (x^2)(1)}{(x - 1)^2} \\
  &= \frac{2 x^2 - 2x - x^2}{(x - 1)^2} \\
  &= \frac{x^2 - 2x}{(x - 1)^2}
  \end{aligned}
  $$

---

Những quy tắc này, đặc biệt là Power Rule, Constant Multiple Rule và Sum/Difference Rule, đóng vai trò nền tảng trong việc tính đạo hàm của các hạng tử có dạng đa thức thường xuất hiện trong Cost Function như Loss Function bình phương sai số trung bình. Product Rule và Quotient Rule trở nên thiết yếu khi xử lý các cấu trúc mô hình hoặc hàm kích hoạt phức tạp hơn.

Việc thành thạo những quy tắc này cho phép chúng ta tính toán đạo hàm cần thiết cho quá trình tối ưu hóa bằng phương pháp giải tích mà không phải quay lại định nghĩa bằng giới hạn mỗi lần. Trong phần sau, chúng ta sẽ thấy cách **Quy tắc dây chuyền (Chain Rule)** kết hợp với những quy tắc này để xử lý các hàm lồng nhau phức tạp hơn, vốn là đặc trưng của mạng nơ-ron.
