# **Tìm hiểu về giới hạn: Nền tảng (Understanding Limits: The Foundation)**

Trước khi có thể bàn về **tốc độ thay đổi tức thời (Instantaneous Rate of Change)**, là đại lượng mà đạo hàm biểu diễn, chúng ta cần một cách để mô tả điều gì xảy ra với giá trị của một hàm số khi đầu vào tiến đến rất gần một điểm nhất định, nhưng không nhất thiết phải đạt đến điểm đó. Khái niệm này được gọi là **Giới hạn (Limit)**. Hãy hình dung đó là việc phóng to vô hạn đến gần một điểm trên đồ thị và quan sát xem hàm số đang tiến đến giá trị nào.

## **1. Định nghĩa chính thức về Giới hạn (Formalizing the Idea of Limit)**

Ví dụ, xét một hàm số $f(x)$. Chúng ta muốn tìm hiểu hành vi của $f(x)$ khi $x$ tiến đến một giá trị nào đó là $a$. Điều chúng ta quan tâm là giá trị mà $f(x)$ ngày càng tiến gần đến khi $x$ ngày càng tiến gần đến $a$, từ cả hai phía (các giá trị nhỏ hơn $a$ và các giá trị lớn hơn $a$), nhưng không nhất thiết phải bằng chính $a$.

Về mặt toán học, chúng ta nói rằng giới hạn của $f(x)$ khi $x$ tiến đến $a$ là $L$, và ký hiệu như sau:

$$
\lim_{x \to a} f(x) = L
$$

Phát biểu này có nghĩa là chúng ta có thể làm cho giá trị của $f(x)$ gần với $L$ tùy ý bằng cách chọn $x$ đủ gần với $a$, nhưng không bằng $a$.

Hàm số thậm chí không cần được xác định tại $x = a$ thì giới hạn vẫn có thể tồn tại. Đây là một điểm rất quan trọng. Giới hạn mô tả xu hướng của hàm số xung quanh điểm đó, chứ không phải giá trị của hàm số ngay tại điểm đó.

## **2. Ví dụ minh họa tính Giới hạn (A Concrete Limit Example)**

Hãy xem một ví dụ cụ thể. Xét hàm số:

$$
f(x) = \frac{x^2 - 1}{x - 1}
$$

Điều gì xảy ra khi $x$ tiến đến $1$? Nếu thử thay trực tiếp $x = 1$ vào hàm số, chúng ta nhận được:

$$
\frac{1^2 - 1}{1 - 1} = \frac{0}{0}
$$

Biểu thức này không xác định. Tuy nhiên, chúng ta có thể rút gọn hàm số khi $x \ne 1$ bằng cách phân tích thành nhân tử:

$$
f(x) = \frac{(x - 1)(x + 1)}{x - 1} = x + 1 \quad (x \ne 1)
$$

Bây giờ, khi $x$ tiến rất gần đến $1$ (chẳng hạn như $0.9$, $0.99$, $0.999$ hoặc $1.1$, $1.01$, $1.001$), giá trị của $f(x)$ cũng tiến rất gần đến $1 + 1 = 2$. Mặc dù $f(1)$ không xác định, giới hạn tại điểm này vẫn tồn tại:

$$
\lim_{x \to 1} \frac{x^2 - 1}{x - 1} = 2
$$

> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/limit_removable_discontinuity.html)

Hàm số $f(x) = \frac{x^2 - 1}{x - 1}$ có hành vi hoàn toàn giống với hàm số $f(x) = x + 1$, ngoại trừ ngay tại $x = 1$, nơi hàm số không xác định (được biểu thị bằng một vòng tròn rỗng trên đồ thị). Giới hạn khi $x$ tiến đến $1$ là $2$, biểu thị giá trị $y$ mà đồ thị hàm số hướng tới.

## **3. Giới hạn một phía (One-Sided Limits)**

Đôi khi, chúng ta có thể chỉ quan tâm đến điều xảy ra khi $x$ tiến đến $a$ từ một hướng cụ thể:

- **Giới hạn bên trái (Limit from the left):** $x$ tiến đến $a$ thông qua các giá trị nhỏ hơn $a$. Ký hiệu:

  $$
  \lim_{x \to a^-} f(x)
  $$

- **Giới hạn bên phải (Limit from the right):** $x$ tiến đến $a$ thông qua các giá trị lớn hơn $a$. Ký hiệu:

  $$
  \lim_{x \to a^+} f(x)
  $$

Để giới hạn tổng thể $\lim_{x \to a} f(x)$ tồn tại và bằng $L$, cả giới hạn bên trái và giới hạn bên phải đều phải tồn tại đồng thời và cùng bằng $L$.

## **4. Mối liên hệ giữa Giới hạn và Đạo hàm (Why Limits Matter for Derivatives)**

Khái niệm giới hạn là nền tảng cốt lõi để xây dựng định nghĩa đạo hàm. Đạo hàm đo lường **Instantaneous Rate of Change**, về cơ bản chính là độ dốc của một hàm số tại một điểm duy nhất.

Làm thế nào để tìm độ dốc tại một điểm duy nhất?

1. Trước tiên, chúng ta tìm độ dốc của đường thẳng nối hai điểm trên đường cong (gọi là **Đường cát tuyến (Secant Line)**).
2. Sau đó, chúng ta sử dụng khái niệm giới hạn để đưa một trong hai điểm này tiến gần vô hạn đến điểm còn lại.

Giới hạn của độ dốc các đường cát tuyến này, khi khoảng cách giữa hai điểm tiến dần về $0$, cho chúng ta độ dốc của **Đường tiếp tuyến (Tangent Line)** tại điểm duy nhất đó. Quá trình lấy giới hạn này chính là định nghĩa toán học chính thống của đạo hàm.

Nhờ có giới hạn, chúng ta thực hiện thành công bước chuyển từ **tốc độ thay đổi trung bình (Average Rate of Change)** trên một khoảng sang **Instantaneous Rate of Change** tại một điểm cố định.
