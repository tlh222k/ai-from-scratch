# **Đạo hàm cấp cao (Higher-Order Derivatives)**

**Đạo hàm cấp một** $f'(x)$ mô tả **tốc độ thay đổi tức thời**, hay **độ dốc**, của hàm số $f(x)$ tại một điểm bất kỳ $x$. Độ dốc này cho biết hàm số đang tăng hay giảm. Nhưng nếu chúng ta cần tìm hiểu xem chính độ dốc đó thay đổi như thế nào thì sao? Để làm được điều này, chúng ta sử dụng **Higher-Order Derivatives**.


## **1. Đạo hàm cấp hai: Đo độ cong (The Second Derivative: Measuring Curvature)**

Tương tự như cách chúng ta lấy đạo hàm của hàm số $f(x)$ để thu được đạo hàm $f'(x)$, chúng ta cũng có thể tiếp tục lấy đạo hàm của $f'(x)$. Kết quả thu được được gọi là **đạo hàm cấp hai (Second Derivative)** của $f(x)$.

Các ký hiệu phổ biến của **Second Derivative** bao gồm:

- $f''(x)$ (đọc là "f phẩy hai của x").
- $\dfrac{d^2y}{dx^2}$ (đọc là "đạo hàm cấp hai của y theo x").
- $y''$.

**Second Derivative** đo tốc độ thay đổi của đạo hàm cấp một (độ dốc). Hãy hình dung việc lái một chiếc ô tô:

- Vị trí của bạn tương ứng với $f(x)$.
- Vận tốc của bạn (tốc độ và hướng chuyển động) tương ứng với đạo hàm cấp một, $f'(x)$.
- Gia tốc của bạn (mức độ thay đổi vận tốc nhanh hay chậm) tương ứng với đạo hàm cấp hai, $f''(x)$.

Trong ngữ cảnh đồ thị của một hàm số, **Second Derivative** cho chúng ta biết về tính lõm của đồ thị:

- **Lõm lên (Concave Up):** Nếu $f''(x) > 0$ trên một khoảng, độ dốc $f'(x)$ đang tăng. Đồ thị cong lên trên, giống như một chiếc cốc có thể chứa nước.
- **Lõm xuống (Concave Down):** Nếu $f''(x) < 0$ trên một khoảng, độ dốc $f'(x)$ đang giảm. Đồ thị cong xuống dưới, giống như một chiếc cốc úp ngược.
- **Điểm uốn (Inflection Point):** Một điểm mà tại đó tính lõm thay đổi (từ lõm lên sang lõm xuống hoặc ngược lại) được gọi là **Inflection Point**. Tại điểm như vậy, chúng ta thường thấy $f''(x) = 0$.

## **2. Ví dụ minh họa tính lõm và Inflection Point**

Xét hàm số $f(x) = x^3$. Đạo hàm cấp một của hàm số là $f'(x) = 3 x^2$. Đạo hàm cấp hai là $f''(x) = 6x$.

- Với $x > 0$, $f''(x) > 0$, do đó đồ thị của hàm số $f(x) = x^3$ ở trạng thái **Concave Up**.
- Với $x < 0$, $f''(x) < 0$, do đó đồ thị của hàm số $f(x) = x^3$ ở trạng thái **Concave Down**.
- Tại $x = 0$, $f''(x) = 0$. Điểm $(0, 0)$ là một **Inflection Point**, tại đó tính lõm thay đổi.

Hàm số $f(x) = x^3$ chuyển từ **Concave Down** ($f''(x) < 0$) sang **Concave Up** ($f''(x) > 0$) tại $x = 0$. Hãy chú ý rằng độ dốc $f'(x)$ giảm cho đến khi đạt $x = 0$, sau đó bắt đầu tăng.

## **3. Các đạo hàm sau đạo hàm cấp hai (Derivatives After the Second)**

Chúng ta có thể tiếp tục quá trình này:
 - Lấy đạo hàm của đạo hàm cấp hai $f''(x)$ sẽ thu được **đạo hàm cấp ba (Third Derivative)**, được ký hiệu là $f'''(x)$ hoặc $\dfrac{d^3y}{dx^3}$.

 - Tiếp tục lấy đạo hàm một lần nữa sẽ thu được **đạo hàm cấp bốn (Fourth Derivative)**, ký hiệu là $f^{(4)}(x)$ hoặc $\dfrac{d^4y}{dx^4}$.

 - Đạo hàm của $f(x)$ sau khi được lấy đạo hàm $n$ lần được gọi là **đạo hàm cấp $n$ (nth derivative)**, ký hiệu là $f^{(n)}(x)$ hoặc $\dfrac{d^ny}{dx^n}$.

Hãy tìm một vài đạo hàm đầu tiên của hàm đa thức sau:

$$
f(x) = 2 x^4 - 5 x^3 + x^2 - 7 x + 3
$$

- **Đạo hàm cấp một (Độ dốc):**

  $$
  f'(x) = \frac{d}{dx}(2 x^4 - 5 x^3 + x^2 - 7 x + 3) = 8 x^3 - 15 x^2 + 2x - 7
  $$

- **Đạo hàm cấp hai (Tính lõm):**

  $$
  f''(x) = \frac{d}{dx}(8 x^3 - 15 x^2 + 2x - 7) = 24 x^2 - 30x + 2
  $$

- **Đạo hàm cấp ba (Tốc độ thay đổi của tính lõm):**

  $$
  f'''(x) = \frac{d}{dx}(24 x^2 - 30x + 2) = 48x - 30
  $$

- **Đạo hàm cấp bốn:**

  $$
  f^{(4)}(x) = \frac{d}{dx}(48x - 30) = 48
  $$

- **Đạo hàm cấp năm (và các cấp cao hơn):**

  $$
  f^{(5)}(x) = \frac{d}{dx}(48) = 0
  $$

Tất cả các đạo hàm tiếp theo cũng sẽ bằng $0$.

## **4. Ứng dụng của Higher-Order Derivatives trong Optimization**

Trong khi đạo hàm cấp một cho chúng ta biết hướng thay đổi, đạo hàm cấp hai cung cấp thông tin quan trọng về hình dạng đồ thị của hàm số. Điều này đặc biệt hữu ích trong **Optimization**.

Chúng ta có thể kết hợp đạo hàm cấp một (để tìm các vị trí có thể có độ dốc bằng $0$, tại đó $f'(x) = 0$) với đạo hàm cấp hai (để kiểm tra độ cong tại những vị trí đó) nhằm xác định một cách đáng tin cậy các cực tiểu và cực đại địa phương. Phương pháp này được gọi là **kiểm tra đạo hàm cấp hai (Second Derivative Test)**:

- Nếu $f'(c) = 0$ và $f''(c) > 0$, hàm số ở trạng thái **Concave Up** tại $x = c$, cho thấy tại đó có một **cực tiểu địa phương (Local Minimum)**.
- Nếu $f'(c) = 0$ and $f''(c) < 0$, hàm số ở trạng thái **Concave Down** tại $x = c$, cho thấy tại đó có một **cực đại địa phương (Local Maximum)**.

Hiểu về độ cong thông qua đạo hàm cấp hai là nền tảng để phân tích các bài toán tối ưu hóa trong **Machine Learning**. Mặc dù chúng ta chủ yếu sử dụng đạo hàm cấp một (Gradient) trong các thuật toán như **Gradient Descent**, khái niệm độ cong (được biểu diễn thông qua đạo hàm cấp hai và phiên bản tương ứng trong nhiều biến là **ma trận Hessian (Hessian Matrix)**) giúp giải thích hành vi của những thuật toán này và tạo cơ sở cho các kỹ thuật tối ưu hóa nâng cao hơn.
