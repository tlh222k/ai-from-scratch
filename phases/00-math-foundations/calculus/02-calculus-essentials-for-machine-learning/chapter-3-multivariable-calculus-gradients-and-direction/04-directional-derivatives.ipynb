# **Đạo hàm theo hướng (Directional Derivatives)**

Các đạo hàm riêng đo lường tốc độ thay đổi của một hàm theo các trục tọa độ (chẳng hạn như chỉ di chuyển về hướng đông hoặc chỉ di chuyển về hướng bắc trên bản đồ). Vector gradient, $\nabla f$, chỉ theo hướng tăng nhanh nhất (hướng leo lên đồi nhanh nhất).

Nhưng nếu bạn muốn biết địa hình dốc như thế nào theo một hướng cụ thể khác thì sao? Có thể bạn muốn di chuyển theo hướng đông bắc hoặc dọc theo một đường đi bất kỳ được biểu diễn bằng một vector. Đây chính là điều mà đạo hàm theo hướng cho phép chúng ta tính toán. Nó định lượng tốc độ thay đổi của một hàm nhiều biến $f$ tại một điểm cụ thể $a$ khi di chuyển theo hướng được xác định bởi một vector đơn vị $u$.

## **1. Định nghĩa đạo hàm theo hướng (Defining the Directional Derivative)**

Đạo hàm theo hướng của một hàm $f$ tại điểm $a$, theo hướng của vector đơn vị $u$, được ký hiệu là $D_u f(a)$. Nó được tính bằng tích vô hướng giữa gradient tại điểm đó và vector chỉ hướng:

$$
D_u f(a)=\nabla f(a)\cdot u
$$

Hãy phân tích công thức này:

- **Gradient $\nabla f(a)$:** Như chúng ta đã tìm hiểu, vector này chứa tất cả các đạo hàm riêng của $f$ được tính tại điểm $a$. Đối với một hàm $f(x,y)$:

  $$
  \nabla f(a)=
  \left\langle
  \frac{\partial f}{\partial x}(a),
  \frac{\partial f}{\partial y}(a)
  \right\rangle
  $$

  Nó tập hợp thông tin về tốc độ thay đổi của hàm theo tất cả các hướng trục tọa độ tại điểm $a$.

- **Vector đơn vị $u$:** Vector này xác định hướng mà chúng ta quan tâm. Điều quan trọng là $u$ phải là một vector đơn vị, nghĩa là độ dài hay độ lớn của nó bằng $1$ ($\|u\|=1$). Tại sao? Bởi vì chúng ta chỉ muốn ghi nhận sự thay đổi do chính hướng di chuyển, chứ không muốn kết quả bị nhân thêm với độ dài của vector chỉ hướng. Nếu hướng di chuyển được xác định bằng một vector $v$ không phải vector đơn vị, trước tiên bạn phải chuẩn hóa nó bằng cách chia cho độ lớn của nó:

  $$
  u=\frac{v}{\|v\|}
  $$

- **Tích vô hướng ($\cdot$):** Tích vô hướng thực chất đo lường mức độ một vector "hướng theo" một vector khác.

## **2. Trực quan giác hình học: Phép chiếu (Geometric Intuition: Projection)**

Hãy nhớ rằng tích vô hướng giữa hai vector $v_1$ và $v_2$ cũng có thể được biểu diễn như sau:

$$
v_1\cdot v_2=\|v_1\|\|v_2\|\cos\theta
$$

Trong đó, $\theta$ là góc giữa hai vector.

Áp dụng công thức này vào công thức đạo hàm theo hướng và biết rằng $\|u\|=1$, chúng ta có:

$$
D_u f(a)
=\nabla f(a)\cdot u
=\|\nabla f(a)\|\|u\|\cos\theta
=\|\nabla f(a)\|\cos\theta
$$

Ở đây, $\theta$ là góc giữa vector gradient $\nabla f(a)$ và vector chỉ hướng $u$. Công thức này cho chúng ta một góc nhìn sâu sắc: đạo hàm theo hướng chính là hình chiếu vô hướng của vector gradient lên vector chỉ hướng $u$. Có thể hình dung điều này như việc đặt câu hỏi: "Độ lớn của gradient có bao nhiêu phần hướng theo hướng $u$?"

> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/directional_derivative.html)


> Đạo hàm theo hướng $D_u f$ là hình chiếu vô hướng của vector gradient $\nabla f$ lên vector đơn vị chỉ hướng $u$. Nó đo lường thành phần của gradient tác động theo hướng $u$.



Góc nhìn về phép chiếu này giúp chúng ta hiểu mối quan hệ giữa gradient và đạo hàm theo hướng:

- **Thay đổi lớn nhất (Maximum Change):** Khi $u$ cùng hướng với $\nabla f(a)$, góc $\theta$ bằng $0$, $\cos\theta=1$, và:

  $$
  D_u f(a)=\|\nabla f(a)\|
  $$

  Đạo hàm theo hướng đạt giá trị lớn nhất và bằng độ lớn của gradient. Điều này xác nhận rằng gradient chỉ theo hướng tăng nhanh nhất.

- **Thay đổi nhỏ nhất (Hướng giảm nhanh nhất – Steepest Descent):** Khi $u$ hướng hoàn toàn ngược với $\nabla f(a)$, góc $\theta$ bằng $\pi$ (180 độ), $\cos\theta=-1$, và:

  $$
  D_u f(a)=-\|\nabla f(a)\|
  $$

  Đây là hướng giảm nhanh nhất.

- **Không thay đổi (Zero Change):** Khi $u$ vuông góc (trực giao) với $\nabla f(a)$, góc $\theta$ bằng $\frac{\pi}{2}$ (90 độ), $\cos\theta=0$, và:

  $$
  D_u f(a)=0
  $$

  Di chuyển theo hướng này không làm thay đổi tức thời giá trị của hàm. Về mặt hình học, bạn đang di chuyển dọc theo một đường mức hoặc đường đồng mức trên bề mặt của hàm.

## **3. Ví dụ tính toán (Example Calculation)**

Hãy xét hàm số:

$$
f(x,y)=x^2+y^2
$$

Hàm này mô tả một chiếc bát parabol có tâm tại gốc tọa độ. Chúng ta muốn tìm tốc độ thay đổi tại điểm $a=(1,1)$ theo hướng của vector $v=\langle 1,2\rangle$.

1. **Tính gradient:**

   $$
   \nabla f(x,y)
   =
   \left\langle
   \frac{\partial f}{\partial x},
   \frac{\partial f}{\partial y}
   \right\rangle
   =
   \langle 2x,2y\rangle
   $$

2. **Tính gradient tại $a=(1,1)$:**

   $$
   \nabla f(1,1)
   =
   \langle 2(1),2(1)\rangle
   =
   \langle 2,2\rangle
   $$

   Vector này hướng thẳng ra xa gốc tọa độ, tức là hướng tăng nhanh nhất đối với hình dạng chiếc bát này.

3. **Tìm vector đơn vị $u$ theo hướng $v=\langle 1,2\rangle$:**

   Độ lớn:

   $$
   \|v\|
   =
   \sqrt{1^2+2^2}
   =
   \sqrt{1+4}
   =
   \sqrt{5}
   $$

   Chuẩn hóa:

   $$
   u=\frac{v}{\|v\|}
   =
   \left\langle
   \frac{1}{\sqrt{5}},
   \frac{2}{\sqrt{5}}
   \right\rangle
   $$

4. **Tính đạo hàm theo hướng bằng tích vô hướng:**

   $$
   D_u f(1,1)
   =
   \nabla f(1,1)\cdot u
   =
   \langle 2,2\rangle
   \cdot
   \left\langle
   \frac{1}{\sqrt{5}},
   \frac{2}{\sqrt{5}}
   \right\rangle
   $$

   $$
   D_u f(1,1)
   =
   (2)\left(\frac{1}{\sqrt{5}}\right)
   +(2)\left(\frac{2}{\sqrt{5}}\right)
   =
   \frac{2}{\sqrt{5}}+\frac{4}{\sqrt{5}}
   =
   \frac{6}{\sqrt{5}}
   $$

Như vậy, tại điểm $(1,1)$, nếu di chuyển theo hướng $\langle 1,2\rangle$, hàm $f(x,y)$ tăng với tốc độ $\frac{6}{\sqrt{5}}\approx 2.68$ đơn vị trên mỗi đơn vị khoảng cách di chuyển. Hãy lưu ý rằng giá trị này nhỏ hơn độ lớn của gradient:

$$
\|\nabla f(1,1)\|
=
\|\langle 2,2\rangle\|
=
\sqrt{2^2+2^2}
=
\sqrt{8}
\approx 2.83
$$

Đây là tốc độ thay đổi theo hướng dốc nhất $\langle 2,2\rangle$.

## **4. Mức độ liên quan đến học máy (Relevance in Machine Learning)**

Trong tối ưu hóa học máy, đặc biệt là với Gradient Descent, chúng ta chủ yếu quan tâm đến hướng giảm nhanh nhất, tức là $-\nabla f$. Tuy nhiên, hiểu về đạo hàm theo hướng giúp chúng ta có thêm góc nhìn hữu ích về không gian tham số. Khái niệm này giúp chúng ta lý giải tại sao di chuyển theo hướng ngược với gradient lại là bước đi hiệu quả nhất (trong phạm vi lân cận của một điểm) để giảm thiểu hàm mất mát.

Mặc dù đạo hàm theo hướng thường không được tính toán một cách tường minh trong quá trình thực hiện Gradient Descent tiêu chuẩn, khái niệm này là nền tảng giúp chúng ta hiểu cách hàm số hoạt động trong không gian tham số nhiều chiều mà chúng ta di chuyển qua trong quá trình huấn luyện mô hình. Nó củng cố vai trò trung tâm của gradient trong việc định hướng quá trình tối ưu hóa.
