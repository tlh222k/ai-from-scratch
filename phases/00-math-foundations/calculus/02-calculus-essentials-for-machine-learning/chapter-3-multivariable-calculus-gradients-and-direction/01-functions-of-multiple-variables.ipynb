# **Hàm số nhiều biến (Functions of Multiple Variables)**

Trong khi **hàm số một biến (single-variable functions)**, chẳng hạn như $f(x)$, và các **Derivative** của chúng (mô tả tốc độ thay đổi của hàm số) là những kiến thức nền tảng, nhiều tình huống trong **Machine Learning** đòi hỏi phải sử dụng các hàm số phụ thuộc đồng thời vào nhiều đầu vào. Ví dụ, khi dự đoán giá nhà, giá nhà không chỉ phụ thuộc vào diện tích sàn mà còn có thể phụ thuộc vào số phòng ngủ, tuổi của ngôi nhà và vị trí của nó. Tương tự, **Cost Function** trong quá trình huấn luyện **Model** thường phụ thuộc vào tất cả **Parameters** của **Model** (**weights** và **bias**), có thể lên đến hàng nghìn hoặc thậm chí hàng triệu đối với những **Model** phức tạp.

Để xử lý những tình huống này, chúng ta cần mở rộng bộ công cụ **Calculus** để áp dụng cho các hàm số nhiều biến.



## **1. Hàm số nhiều biến là gì? (What are Functions of Multiple Variables?)**

Một **hàm số nhiều biến (functions of multiple variables)** nhận nhiều hơn một giá trị đầu vào và tạo ra một giá trị đầu ra (trong ngữ cảnh của chúng ta, thường là một giá trị vô hướng duy nhất). Nếu một hàm số $f$ phụ thuộc vào $n$ biến, chẳng hạn như $x_1,x_2,\ldots,x_n$, chúng ta biểu diễn hàm số đó như sau:

$$
f(x_1,x_2,\ldots,x_n)
$$

Ngoài ra, chúng ta có thể nhóm các biến đầu vào thành một vector:

$$
\mathbf{x}=[x_1,x_2,\ldots,x_n]^T
$$

Khi đó, hàm số có thể được viết gọn hơn như sau:

$$
f(\mathbf{x})
$$

Đầu ra thường là một số thực duy nhất, đặc biệt khi xử lý các **Cost Function** trong **Machine Learning**, trong đó đầu ra biểu thị mức độ sai số hoặc “Cost”.

**Ví dụ:** Đây là một hàm số đơn giản có hai biến $x$ và $y$:

$$
f(x,y)=x^2+2y^2
$$

Ở đây, đầu vào bao gồm một cặp giá trị $(x,y)$ và đầu ra là một số duy nhất được tính bằng công thức trên. Chẳng hạn:

$$
f(1,2)=1^2+2(2^2)=1+8=9
$$



## **2. Trực quan hóa các hàm số nhiều biến (Visualizing Multivariable Functions)**

Việc trực quan hóa các hàm số trở nên khó khăn hơn khi số lượng biến đầu vào tăng lên.

- **Một biến ($y=f(x)$):** Chúng ta có thể dễ dàng trực quan hóa hàm số này dưới dạng một đường cong trên mặt phẳng 2D (trục $x$ và trục $y$).

- **Hai biến ($z=f(x,y)$):** Chúng ta có thể trực quan hóa hàm số này dưới dạng một **surface** trong không gian 3D, trong đó độ cao $z$ phía trên mặt phẳng $xy$ biểu thị giá trị của hàm số tại điểm $(x,y)$.

Hãy trực quan hóa hàm số:

$$
z=x^2+y^2
$$


Phương trình này mô tả một **paraboloid** mở lên trên, có giá trị cực tiểu tại $(0,0)$.

> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/paraboloid_surface.html) ·


> Đồ thị bề mặt 3D biểu diễn hàm số $z=x^2+y^2$. Độ cao ($z$) tương ứng với giá trị của hàm số tại mỗi cặp tọa độ $(x,y)$. Giá trị cực tiểu (0) xuất hiện tại $(x,y)=(0,0)$.

- **Ba biến trở lên ($w=f(x,y,z,\ldots)$):** Việc trực quan hóa trực tiếp trong không gian 3D không còn khả thi. Đối với các hàm số có ba biến, đôi khi chúng ta có thể sử dụng **tập mức (level sets)** (các **surface** trong không gian 3D mà tại đó giá trị của hàm số không đổi). Đối với các hàm số có nhiều hơn ba biến, chúng ta chủ yếu dựa vào phân tích toán học thay vì trực giác hình học trực tiếp. Chúng ta thường phân tích hàm số bằng cách xem xét cách nó biến đổi theo những hướng cụ thể hoặc khảo sát các mặt cắt của nó.



## **3. Mức độ liên quan trong học máy (Relevance in Machine Learning)**

Các hàm số nhiều biến đóng vai trò nền tảng trong **Machine Learning**:

- **Cost Function:** Như đã đề cập, **Cost Function** $J$ đo lường mức độ hoạt động của **Model** dựa trên các **Parameters** của nó (**weights** $w$ và **bias** $b$). Đối với một **Model** có $n$ **weights**, **Cost Function** được biểu diễn như sau:

$$
J(w_1,w_2,\ldots,w_n,b)
$$

Quá trình huấn luyện **Model** bao gồm việc tìm các giá trị $w_1,\ldots,w_n,b$ giúp tối thiểu hóa hàm số nhiều biến này. Ví dụ, **Mean Squared Error** của **Cost Function** trong hồi quy tuyến tính với hai **Features** ($x_1,x_2$) và các **Parameters** $w_1,w_2,b$ là:

$$
J(w_1,w_2,b)=\frac{1}{2m}\sum_{i=1}^{m}
\left(
w_1x_1^{(i)}+w_2x_2^{(i)}+b-y^{(i)}
\right)^2
$$

Ở đây:

- $m$ là số lượng mẫu huấn luyện.
- $(x_1^{(i)},x_2^{(i)})$ là các **Features** của mẫu thứ $i$.
- $y^{(i)}$ là nhãn thực tế của mẫu đó.

Hàm số $J$ phụ thuộc vào ba **Parameters**: $w_1,w_2$ và $b$.

- **Dự đoán của Model (Model Predictions):** Bản thân hàm dự đoán $h(x)$ thường cũng là một hàm số nhiều biến. Đối với hồi quy tuyến tính:

$$
h_{w,b}(x)=w_1x_1+w_2x_2+\cdots+w_nx_n+b
$$

Giá trị dự đoán phụ thuộc vào các **Features** đầu vào $x_1,\ldots,x_n$.

Hiểu cách phân tích những hàm số nhiều biến này, đặc biệt là cách chúng thay đổi khi chúng ta điều chỉnh các đầu vào (các **Parameters**), là điều thiết yếu đối với quá trình **Optimization**. Điều này dẫn chúng ta đến khái niệm **đạo hàm riêng (partial derivatives)**, dùng để đo lường tốc độ thay đổi theo một biến trong khi giữ các biến còn lại không đổi. Chúng ta sẽ tìm hiểu khái niệm này trong phần tiếp theo.
