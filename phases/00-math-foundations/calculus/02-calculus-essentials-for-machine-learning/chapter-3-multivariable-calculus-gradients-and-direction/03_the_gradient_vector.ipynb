# **Vector gradient (The Gradient Vector)**

Các đạo hàm riêng đo lường tốc độ thay đổi của một hàm nhiều biến theo các hướng trục tọa độ tiêu chuẩn (chẳng hạn như hướng $x$ hoặc hướng $y$). Nhưng nếu chúng ta muốn biết hàm số tăng nhanh nhất theo hướng nào từ một điểm cho trước thì sao? Đây chính là điều mà **vector gradient** cho chúng ta biết.

Hãy tưởng tượng bạn đang đứng trên một địa hình đồi núi được biểu diễn bằng hàm số $f(x,y)$. Các đạo hàm riêng cho bạn biết độ dốc của ngọn đồi nếu bạn đi chính xác về hướng đông ($\frac{\partial f}{\partial x}$) hoặc chính xác về hướng bắc ($\frac{\partial f}{\partial y}$). **Gradient** kết hợp thông tin này thành một vector duy nhất, hướng thẳng lên dốc theo hướng có độ dốc lớn nhất.

## **1. Định nghĩa Gradient (Defining the Gradient)**

Đối với một hàm vô hướng $f(x_1,x_2,\ldots,x_n)$ phụ thuộc vào $n$ biến, gradient của hàm là một vector chứa tất cả các đạo hàm riêng cấp một của hàm đó. Gradient thường được ký hiệu là $\nabla f$ (đọc là "nabla f" hoặc "del f").

$$ \nabla f(x_1, x_2, \dots, x_n) = \left\langle \frac{\partial f}{\partial x_1}, \frac{\partial f}{\partial x_2}, \dots, \frac{\partial f}{\partial x_n} \right\rangle $$

Mỗi thành phần của vector gradient là đạo hàm riêng của hàm theo một biến tương ứng. Vì vậy, gradient là một **trường vector (vector field)**; nó gán một vector (biểu thị hướng và độ lớn của hướng tăng nhanh nhất) cho mỗi điểm trong miền xác định của hàm.

## **2. Ví dụ tính toán (Example Calculation)**

Hãy xét hàm số: $ f(x,y)=x^2+y^3 $ Hàm này nhận hai đầu vào là $x$ và $y$.

1. **Tìm các đạo hàm riêng:**
    - $ \frac{\partial f}{\partial x} = \frac{\partial}{\partial x}(x^2+y^3) = 2x $ (coi $y$ là hằng số)
    - $ \frac{\partial f}{\partial y} = \frac{\partial}{\partial y}(x^2+y^3) = 3y^2 $ (coi $x$ là hằng số)

2. **Tập hợp thành vector gradient:**

$$
\nabla f(x,y)
=
\left\langle
\frac{\partial f}{\partial x},
\frac{\partial f}{\partial y}
\right\rangle
=
\langle 2x,3y^2\rangle
$$

Tại một điểm cụ thể, chẳng hạn $(x,y)=(1,2)$, gradient là:

$$
\nabla f(1,2)
=
\langle 2(1),3(2^2)\rangle
=
\langle 2,12\rangle
$$

Vector $\langle 2,12\rangle$ chỉ ra hướng xuất phát từ điểm $(1,2)$ mà theo đó hàm $f(x,y)=x^2+y^3$ tăng nhanh nhất.

## **3. Diễn giải hình học: Hướng và độ lớn (Geometric Interpretation: Direction and Magnitude)**

Tại mỗi điểm cho trước, gradient $\nabla f$ có hai tính chất quan trọng:

- **Hướng:** Vector $\nabla f$ chỉ theo hướng tăng nhanh nhất của hàm tại điểm đó. Nếu bạn muốn làm cho giá trị của hàm tăng nhanh nhất có thể, bạn nên di chuyển theo hướng mà gradient chỉ ra.
- **Độ lớn:** Độ lớn (hay độ dài) của vector gradient, được ký hiệu là $\|\nabla f\|$, biểu thị tốc độ tăng theo hướng tăng nhanh nhất đó. Độ lớn càng lớn thì hàm số tăng càng nhanh.

$$
\|\nabla f\|
=
\sqrt{
\left(\frac{\partial f}{\partial x_1}\right)^2
+
\left(\frac{\partial f}{\partial x_2}\right)^2
+\cdots+
\left(\frac{\partial f}{\partial x_n}\right)^2
}
$$

Ngược lại, hướng đối diện với gradient, $-\nabla f$, chỉ theo hướng giảm nhanh nhất. Đây là nguyên lý nền tảng của tối ưu hóa trong học máy. Nếu hàm $f$ biểu diễn một chi phí hoặc sai số mà chúng ta muốn giảm thiểu, việc di chuyển theo hướng $-\nabla f$ cho phép chúng ta giảm chi phí một cách hiệu quả nhất ở mỗi bước.

## **4. Trực quan hóa Gradient (Visualizing the Gradient)**

Đồ thị đường đồng mức rất hữu ích để trực quan hóa các hàm nhiều biến. Một đường đồng mức nối các điểm mà tại đó hàm có cùng giá trị. Vector gradient tại bất kỳ điểm nào luôn vuông góc (trực giao) với đường đồng mức đi qua điểm đó.

Hãy cùng trực quan hóa gradient của một hàm đơn giản:

$$
f(x,y)=x^2+y^2
$$

Gradient của hàm là:

$$
\nabla f=\langle 2x,2y\rangle
$$

Các đường đồng mức là những đường tròn có tâm tại gốc tọa độ.

> Đồ thị đường đồng mức của $f(x,y)=x^2+y^2$. Các đường tròn màu xám là những đường đồng mức tại đó $f$ không đổi. Các mũi tên màu xanh dương biểu diễn vector gradient $\nabla f=\langle 2x,2y\rangle$ tại những điểm được chọn. Hãy chú ý rằng các vector gradient hướng ra ngoài, vuông góc với các đường đồng mức, cho biết hướng tăng nhanh nhất khi di chuyển ra xa điểm cực tiểu $(0,0)$.

## **5. Tại sao Gradient quan trọng trong học máy (Why Gradients Matter in Machine Learning)**

Trong học máy, chúng ta thường xác định một hàm chi phí (hoặc hàm mất mát) để đo lường mức độ hoạt động kém hiệu quả của mô hình. Hàm chi phí này thường phụ thuộc vào nhiều tham số (trọng số và độ chệch). Việc huấn luyện mô hình bao gồm tìm các giá trị tham số sao cho hàm chi phí đạt giá trị nhỏ nhất.

Vì gradient âm $-\nabla f$ chỉ theo hướng giảm nhanh nhất, nó cung cấp hướng điều chỉnh các tham số của mô hình hiệu quả nhất để giảm chi phí. Các thuật toán như **Gradient Descent** được xây dựng trực tiếp dựa trên nguyên lý này: cập nhật các tham số theo cách lặp đi lặp lại bằng cách thực hiện những bước nhỏ theo hướng ngược với gradient của hàm chi phí.

Do đó, hiểu về vector gradient là điều thiết yếu để nắm bắt cách nhiều mô hình học máy được huấn luyện và tối ưu hóa. Nó giống như chiếc la bàn dẫn đường cho chúng ta qua không gian tham số phức tạp, nhiều chiều, hướng đến một điểm có chi phí thấp hơn. Chúng ta sẽ tìm hiểu các thuật toán tối ưu hóa dựa trên gradient trong chương tiếp theo.
