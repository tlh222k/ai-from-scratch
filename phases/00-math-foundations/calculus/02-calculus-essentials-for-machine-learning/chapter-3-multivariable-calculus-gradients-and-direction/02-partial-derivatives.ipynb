# **Đạo hàm riêng (Partial Derivatives)**

Đạo hàm $f'(x)$ đo lường **tốc độ thay đổi tức thời (Instantaneous Rate of Change)** của một hàm số một biến $f(x)$. Tuy nhiên, nhiều mô hình học máy phụ thuộc vào nhiều đầu vào hoặc tham số. Ví dụ, một hàm chi phí đơn giản $J(w_1,w_2)$ phụ thuộc vào hai trọng số $w_1$ và $w_2$. Làm thế nào để đo lường sự thay đổi của $J$? Hàm này thay đổi nhanh hơn nếu chúng ta điều chỉnh $w_1$ hay $w_2$?

Đây chính là lúc **đạo hàm riêng (partial derivatives)** phát huy tác dụng. Khi một hàm có nhiều biến đầu vào, đạo hàm riêng đo lường sự thay đổi của hàm khi một biến đầu vào cụ thể thay đổi, trong khi tất cả các biến đầu vào còn lại được giữ cố định.

## **1. Định nghĩa đạo hàm riêng (Defining the Partial Derivative)**

Hãy tưởng tượng bạn có một hàm $f(x,y)$. Đạo hàm riêng của $f$ theo $x$ cho chúng ta biết $f$ thay đổi như thế nào khi ta điều chỉnh $x$ một lượng rất nhỏ, với giả định rằng $y$ hoàn toàn không thay đổi. Tương tự, đạo hàm riêng của $f$ theo $y$ cho chúng ta biết $f$ thay đổi như thế nào khi ta điều chỉnh $y$, đồng thời giữ cố định $x$.

### **1.1 Ký hiệu (Notation)**

Chúng ta sử dụng một ký hiệu đặc biệt, $\partial$ (thường được gọi là "del" hoặc đơn giản là "partial"), để biểu thị các đạo hàm riêng.

- Đạo hàm riêng của $f$ theo $x$ được viết là $\frac{\partial f}{\partial x}$ hoặc đôi khi là $f_x$.
- Đạo hàm riêng của $f$ theo $y$ được viết là $\frac{\partial f}{\partial y}$ hoặc đôi khi là $f_y$.
- Nếu hàm số có dạng, chẳng hạn, $z=f(x,y)$, bạn cũng có thể bắt gặp ký hiệu $\frac{\partial z}{\partial x}$ và $\frac{\partial z}{\partial y}$.

## **2. Cách tính đạo hàm riêng (How to Calculate Partial Derivatives)**

Quy trình tính toán khá trực tiếp:

1. Xác định biến mà bạn đang lấy đạo hàm theo nó (ví dụ: $x$).
 Coi tất cả các biến độc lập khác là hằng số (ví dụ: coi $y$ như một số cố định, chẳng hạn $5$ hoặc $c$).
3. Áp dụng các quy tắc lấy đạo hàm tiêu chuẩn mà bạn đã học đối với hàm số một biến.

Hãy cùng giải một ví dụ. Giả sử chúng ta có hàm số:

$$
f(x,y)=x^2+3xy+y^3
$$



### **Tìm $\frac{\partial f}{\partial x}$ (đạo hàm riêng theo $x$)**

Chúng ta coi $y$ là một hằng số.

- Đạo hàm của $x^2$ theo $x$ là $2x$.
- Đạo hàm của $3xy$ theo $x$ là $3y$ (hãy xem $3y$ như một hệ số không đổi nhân với $x$).
- Đạo hàm của $y^3$ theo $x$ là $0$ (vì $y$ được coi là hằng số, nên $y^3$ cũng là hằng số, và đạo hàm của một hằng số bằng không).

Gộp các kết quả lại:

$$
\frac{\partial f}{\partial x}=2x+3y+0=2x+3y
$$



### **Tìm $\frac{\partial f}{\partial y}$ (đạo hàm riêng theo $y$):**

Lúc này, chúng ta coi $x$ là một hằng số.

- Đạo hàm của $x^2$ theo $y$ là $0$ (vì nó được coi là hằng số).
- Đạo hàm của $3xy$ theo $y$ là $3x$ (hãy xem $3x$ như một hệ số không đổi nhân với $y$).
- Đạo hàm của $y^3$ theo $y$ là $3y^2$.

Gộp các kết quả lại:

$$
\frac{\partial f}{\partial y}=0+3x+3y^2=3x+3y^2
$$

Lưu ý rằng, nhìn chung, các đạo hàm riêng vẫn là những hàm của cả $x$ và $y$. Chúng cho bạn biết tốc độ thay đổi tại một điểm cụ thể $(x,y)$ theo một hướng nhất định (hướng $x$ hoặc hướng $y$).

## **3. Diễn giải hình học (Geometric Interpretation)**

Đối với một hàm $z=f(x,y)$, mô tả một bề mặt trong không gian 3D, các đạo hàm riêng có ý nghĩa hình học rất trực quan.

- Giá trị $\frac{\partial f}{\partial x}$ tại điểm $(x_0,y_0)$ là độ dốc của đường tiếp tuyến với đường cong được tạo thành khi mặt phẳng $y=y_0$ cắt bề mặt $z=f(x,y)$. Đây là độ dốc theo hướng $x$.
- Giá trị $\frac{\partial f}{\partial y}$ tại điểm $(x_0,y_0)$ là độ dốc của đường tiếp tuyến với đường cong được tạo thành khi mặt phẳng $x=x_0$ cắt bề mặt $z=f(x,y)$. Đây là độ dốc theo hướng $y$.

Hình minh họa sau đây cho thấy bề mặt $z=0.5x^2+y^2$ và làm nổi bật các đường cong được tạo thành khi cắt bề mặt tại $x=1$ và $y=1$; hai đường cong giao nhau tại điểm $(1,1,1.5)$. Đạo hàm riêng $\frac{\partial z}{\partial x}$ tại $(1,1)$ cho biết độ dốc theo đường cong màu đỏ ($y=1$) tại điểm đó, còn đạo hàm riêng $\frac{\partial z}{\partial y}$ cho biết độ dốc theo đường cong màu xanh dương ($x=1$) tại điểm đó.

> Đạo hàm riêng $\frac{\partial f}{\partial x}$ tại $(1,1)$ biểu thị độ dốc của bề mặt theo đường cong màu đỏ (nơi $y=1$) tại điểm màu cam. Đạo hàm riêng $\frac{\partial f}{\partial y}$ tại $(1,1)$ biểu thị độ dốc theo đường cong màu xanh dương (nơi $x=1$) tại điểm màu cam.

## **4. Mức độ liên quan đến học máy (Relevance to Machine Learning)**

Tại sao điều này lại quan trọng đối với học máy? Ví dụ, xét hàm chi phí sai số bình phương của một mô hình tuyến tính rất đơn giản trên một điểm dữ liệu $(x,y)$:

$$
J(w,b)=(wx+b-y)^2
$$

Ở đây, chi phí $J$ là một hàm của các tham số mô hình: trọng số $w$ và độ chệch $b$. Mục tiêu của chúng ta trong quá trình huấn luyện thường là tìm các giá trị của $w$ và $b$ sao cho chi phí này đạt giá trị nhỏ nhất (hoặc tổng các chi phí tương tự trên nhiều điểm dữ liệu đạt giá trị nhỏ nhất).

Để sử dụng các thuật toán tối ưu hóa như Gradient Descent, chúng ta cần biết chi phí $J$ thay đổi như thế nào khi điều chỉnh $w$ và $b$. Chúng ta cần các đạo hàm riêng:

- **Đạo hàm riêng theo $w$:** Coi $b$, $x$ và $y$ là các hằng số. Áp dụng chain rule:

$$
\frac{\partial J}{\partial w}
=2(wx+b-y)^1\cdot\frac{\partial}{\partial w}(wx+b-y)
$$

$$
\frac{\partial J}{\partial w}
=2(wx+b-y)\cdot(x)
$$

- **Đạo hàm riêng theo $b$:** Coi $w$, $x$ và $y$ là các hằng số. Áp dụng chain rule:

$$
\frac{\partial J}{\partial b}
=2(wx+b-y)^1\cdot\frac{\partial}{\partial b}(wx+b-y)
$$

$$
\frac{\partial J}{\partial b}
=2(wx+b-y)\cdot(1)
$$

Các đạo hàm riêng này cho chúng ta biết mức độ nhạy của sai số đối với những thay đổi nhỏ tương ứng của trọng số và độ chệch. Chẳng hạn, $\frac{\partial J}{\partial w}$ cho biết sai số bình phương sẽ tăng hay giảm bao nhiêu khi $w$ tăng một lượng rất nhỏ. Đây chính là thông tin mà các thuật toán tối ưu hóa dựa trên gradient sử dụng để cập nhật các tham số và giảm thiểu hàm chi phí.

Hiểu về đạo hàm riêng là bước đầu tiên để nắm bắt khái niệm **Gradient**, kết hợp tất cả các đạo hàm riêng thành một vector duy nhất, chỉ theo hướng mà hàm số tăng nhanh nhất. Chúng ta sẽ tìm hiểu về Gradient trong phần tiếp theo.
