# **Kết hợp tất cả: Quy trình tối ưu hóa (Putting It All Together: The Optimization Process)**

Toàn bộ quy trình tối ưu hóa mô hình bao gồm việc xác định một mô hình, chẳng hạn như **Simple Linear Regression** ($y=mx+b$), đo lường sai số của mô hình bằng một hàm chi phí (chẳng hạn **Mean Squared Error – MSE**) và tính gradient (các đạo hàm riêng của hàm chi phí theo các tham số $m$ và $b$). Phần này trình bày phương pháp đầy đủ để sử dụng những công cụ giải tích này nhằm tối ưu hóa mô hình – về bản chất chính là huấn luyện mô hình.

Ý tưởng cốt lõi là cải thiện mô hình theo từng bước lặp. Chúng ta bắt đầu với một số giá trị ước lượng ban đầu cho các tham số $m$ và $b$. Những giá trị ước lượng ban đầu này có thể chưa tạo ra một đường thẳng phù hợp với dữ liệu, nghĩa là hàm chi phí sẽ có giá trị tương đối cao. Mục tiêu của chúng ta là điều chỉnh một cách có hệ thống các tham số $m$ và $b$ để giảm chi phí này.

Đây chính là lúc **Gradient Descent** phát huy tác dụng. Hãy hình dung hàm chi phí xác định một bề mặt, có thể giống như một ngọn đồi, trong đó hai chiều nằm ngang biểu diễn các giá trị của $m$ và $b$, còn chiều thẳng đứng biểu diễn chi phí (sai số). Mục tiêu của chúng ta là tìm điểm thấp nhất trên bề mặt này.

Dưới đây là quy trình tối ưu hóa từng bước:





## **1. Khởi tạo tham số (Initialize Parameters)**

Bắt đầu với các giá trị ban đầu cho $m$ và $b$. Những giá trị này có thể là số $0$, các số nhỏ được chọn ngẫu nhiên hoặc bất kỳ giá trị ước lượng khởi đầu nào khác. Chúng ta ký hiệu các giá trị ban đầu này là $m_0$ và $b_0$.

## **2. Tính gradient (Calculate the Gradient)**

Tại các giá trị tham số hiện tại ($m_i$, $b_i$), tính gradient của hàm chi phí $J(m,b)$. Việc này bao gồm tính các đạo hàm riêng đã được đề cập:

- $\frac{\partial J}{\partial m}$ (chi phí thay đổi như thế nào theo $m$).
- $\frac{\partial J}{\partial b}$ (chi phí thay đổi như thế nào theo $b$).

Vector gradient, $ \nabla J= \left[ \frac{\partial J}{\partial m}, \frac{\partial J}{\partial b} \right] $ chỉ theo hướng mà chi phí tăng nhanh nhất từ điểm hiện tại ($m_i$, $b_i$).



## **3. Cập nhật tham số (Update Parameters)**

Chúng ta muốn di chuyển theo hướng ngược lại với gradient để giảm chi phí. Các tham số được cập nhật theo những quy tắc sau:

$$
m_{i+1}
=
m_i-\alpha\frac{\partial J}{\partial m}
$$

$$
b_{i+1}
=
b_i-\alpha\frac{\partial J}{\partial b}
$$

Ở đây, $\alpha$ là **Learning Rate**. Đây là một giá trị dương nhỏ (chẳng hạn $0.01$ hoặc $0.001$), kiểm soát độ lớn của bước di chuyển theo hướng xuống dốc. Việc lựa chọn **Learning Rate** phù hợp rất quan trọng: nếu giá trị quá lớn, chúng ta có thể vượt quá điểm cực tiểu; nếu quá nhỏ, quá trình có thể mất quá nhiều thời gian. Phép trừ bảo đảm chúng ta di chuyển theo hướng xuống dốc.



## **4. Lặp lại (Repeat)**

Quay lại Bước 2, sử dụng các tham số vừa được cập nhật ($m_{i+1}$, $b_{i+1}$) để tính gradient tiếp theo. Lặp lại quy trình này (tính gradient, cập nhật tham số) trong một số vòng lặp xác định, hoặc cho đến khi hàm chi phí không còn giảm đáng kể, hoặc mức thay đổi của $m$ và $b$ trở nên rất nhỏ. Trạng thái này thường được gọi là **sự hội tụ (convergence)**.





## **5. Chu trình tối ưu hóa bằng Gradient Descent (Gradient Descent Optimization Cycle)**

Vòng lặp này tạo thành cốt lõi của thuật toán tối ưu hóa **Gradient Descent**. Giải tích, cụ thể là việc tính các đạo hàm riêng để tìm gradient, cung cấp thông tin cần thiết về "độ dốc" của hàm chi phí, từ đó hướng dẫn chúng ta tiến về phía điểm có sai số nhỏ nhất.

Bằng cách liên tục áp dụng những bước này, chúng ta điều chỉnh dần các tham số $m$ và $b$, giúp đường hồi quy tuyến tính ngày càng phù hợp hơn với dữ liệu và cực tiểu hóa hàm chi phí. Điều này minh họa cách các khái niệm cơ bản về đạo hàm và gradient được áp dụng để huấn luyện ngay cả những mô hình học máy đơn giản. Mặc dù chúng ta sử dụng hồi quy tuyến tính làm ví dụ, quy trình nền tảng này – sử dụng gradient để cực tiểu hóa hàm chi phí – cũng là cơ sở để huấn luyện nhiều thuật toán học máy phức tạp.
