# **Định nghĩa hàm chi phí cho hồi quy tuyến tính (Defining a Cost Function for Linear Regression)**

Một mô hình tuyến tính đơn giản, thường được biểu diễn bằng phương trình

$$
y=mx+b
$$

là nền tảng trong nhiều phân tích. Khi làm việc với các điểm dữ liệu, chẳng hạn như các cặp (số giờ học, điểm thi), mục tiêu thường là tìm một đường thẳng thể hiện tốt nhất mối quan hệ giữa các cặp này. Việc xác định ý nghĩa của từ "tốt nhất" bằng các thuật ngữ toán học là điều thiết yếu đối với quá trình này.

Hãy tưởng tượng bạn có một vài điểm dữ liệu được biểu diễn trên đồ thị. Bạn có thể vẽ nhiều đường thẳng khác nhau đi qua các điểm này. Một số đường sẽ đi rất gần các điểm, trong khi những đường khác sẽ cách chúng khá xa. Chúng ta cần một cách để đo lường mức độ phù hợp của một đường thẳng cụ thể, được xác định bởi một giá trị **Slope** `m` và **tung độ gốc (y-intercept)** `b` nhất định, với dữ liệu của mình. Thước đo này được gọi là **Cost Function** (đôi khi còn được gọi là **Loss Function** hoặc **Objective Function**).


Ý tưởng cốt lõi là định lượng "sai số" hoặc "lỗi" mà đường thẳng của chúng ta tạo ra đối với từng điểm dữ liệu. Giả sử chúng ta có $n$ điểm dữ liệu:

$$
(x_1,y_1),(x_2,y_2),\ldots,(x_n,y_n)
$$

Với mỗi đầu vào $x_i$ bất kỳ, đường thẳng hiện tại của chúng ta

$$
y=mx+b
$$

sẽ dự đoán một giá trị đầu ra. Hãy gọi giá trị dự đoán này là $\hat{y}_i$ (đọc là "y-hat"):

$$
\hat{y}_i=mx_i+b
$$

Giá trị đầu ra thực tế của điểm dữ liệu này là $y_i$. Hiệu giữa giá trị thực tế và giá trị dự đoán,

$$
y_i-\hat{y}_i
$$

cho chúng ta biết dự đoán đã sai lệch bao xa đối với điểm cụ thể đó. Hiệu này thường được gọi là **phần dư (residual)**.

- Nếu giá trị thực tế $y_i$ lớn hơn giá trị dự đoán $\hat{y}_i$, phần dư sẽ dương.

- Nếu giá trị dự đoán lớn hơn giá trị thực tế, phần dư sẽ âm.

Nếu chỉ cộng các phần dư của tất cả điểm dữ liệu lại với nhau, sai số dương và âm có thể triệt tiêu lẫn nhau, khiến tổng sai số có vẻ nhỏ một cách sai lệch, ngay cả khi đường thẳng không khớp tốt với dữ liệu.

Để tránh sự triệt tiêu này và bảo đảm rằng các sai số lớn hơn đóng góp nhiều hơn vào thước đo tổng thể về "mức độ không phù hợp", chúng ta thường bình phương từng phần dư:

$$
(y_i-\hat{y}_i)^2
$$

Việc bình phương khiến tất cả sai số đều trở thành số dương và phạt các độ lệch lớn nghiêm khắc hơn so với các độ lệch nhỏ (ví dụ: sai số bằng 2 trở thành 4, trong khi sai số bằng 3 trở thành 9).

Bây giờ, khi đã có thước đo sai số cho từng điểm riêng lẻ, chúng ta cần một con số duy nhất để biểu thị sai số tổng thể của đường thẳng trên toàn bộ $n$ điểm dữ liệu. Một phương pháp rất phổ biến là tính giá trị trung bình của các sai số bình phương này. Đây chính là **hàm chi phí sai số bình phương trung bình (Mean Squared Error – MSE)**.

Về mặt toán học, chúng ta định nghĩa hàm chi phí **Mean Squared Error**, ký hiệu là $J$, như một hàm của các tham số mô hình $m$ và $b$:

$$
J(m,b)=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Hãy phân tích công thức này:

- **$J(m,b)$**: Điều này cho thấy chi phí $J$ phụ thuộc vào các giá trị được chọn cho **Slope** $m$ và tung độ gốc $b$. Mục tiêu của chúng ta là tìm các giá trị $m$ và $b$ khiến $J$ nhỏ nhất có thể.

- **$n$**: Tổng số điểm dữ liệu mà chúng ta có.

- **$\sum_{i=1}^{n}$**: Đây là ký hiệu phép tính tổng. Nó có nghĩa là "cộng biểu thức sau đây cho từng điểm dữ liệu, bắt đầu từ $i=1$ và tăng dần cho đến $i=n$".

- **$(x_i,y_i)$**: Đầu vào ($x_i$) và đầu ra thực tế ($y_i$) của điểm dữ liệu thứ $i$.

- **$\hat{y}_i$**: Giá trị đầu ra được mô hình của chúng ta dự đoán ($mx_i+b$) đối với đầu vào $x_i$.

- **$(y_i-\hat{y}_i)^2$**: Sai số bình phương của điểm dữ liệu thứ $i$.

- **$\frac{1}{n}$**: Chia tổng các sai số bình phương cho $n$ sẽ cho chúng ta sai số bình phương trung bình.

Chúng ta có thể thay trực tiếp định nghĩa của $\hat{y}_i$ vào công thức hàm chi phí:

$$
J(m,b)=\frac{1}{n}\sum_{i=1}^{n}\left(y_i-(mx_i+b)\right)^2
$$

Công thức này cho chúng ta một con số duy nhất, $J$, biểu thị mức độ không phù hợp trung bình của đường thẳng $y=mx+b$ với dữ liệu. Giá trị $J$ càng nhỏ có nghĩa là các dự đoán của đường thẳng càng gần với các điểm dữ liệu thực tế xét trên trung bình, cho thấy mức độ phù hợp càng tốt. Giá trị lớn hơn cho thấy mức độ phù hợp kém hơn.

> *Đồ thị này thể hiện các điểm dữ liệu mẫu (chấm màu xanh dương), một mô hình tuyến tính có thể có (đường màu đỏ) và các sai số theo phương thẳng đứng (phần dư, biểu diễn bằng các đường chấm màu xám) giữa các điểm dữ liệu thực tế và các giá trị dự đoán của mô hình. Hàm chi phí MSE tính giá trị trung bình của bình phương độ dài các đường chấm này.*



Mục tiêu tối ưu hóa của chúng ta giờ đây đã trở nên cụ thể: tìm các giá trị chính xác của $m$ và $b$ sao cho giá trị của $J(m,b)$ đạt mức thấp nhất có thể. Việc cực tiểu hóa hàm chi phí này có nghĩa là tìm đường thẳng giúp giảm thiểu khoảng cách theo phương thẳng đứng bình phương trung bình đến các điểm dữ liệu của chúng ta.

-----

**Tại sao chọn Mean Squared Error?** Ngoài việc phạt các sai số lớn nghiêm khắc hơn, MSE còn có những tính chất toán học thuận lợi. Cụ thể, đây là một hàm trơn và liên tục; quan trọng hơn, chúng ta có thể dễ dàng tính các đạo hàm của nó theo $m$ và $b$. Như đã tìm hiểu trong các chương trước, những đạo hàm này (gradient) chính là thứ chúng ta cần để định hướng thuật toán **Gradient Descent** tiến về phía điểm có chi phí cực tiểu.

Trong phần tiếp theo, chúng ta sẽ lấy hàm chi phí $J(m,b)$ này và áp dụng những kiến thức giải tích đã học để tính gradient của nó. Đây sẽ là động lực thúc đẩy toàn bộ quá trình tối ưu hóa của chúng ta.
