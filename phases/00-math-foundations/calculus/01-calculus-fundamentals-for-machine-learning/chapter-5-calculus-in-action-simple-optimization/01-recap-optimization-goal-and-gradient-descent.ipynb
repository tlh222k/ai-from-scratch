# **Ôn tập: Mục tiêu tối ưu hóa và Gradient Descent (Recap: Optimization Goal and Gradient Descent)**

Mục tiêu tổng quát của nhiều bài toán **Machine Learning** là xây dựng một mô hình có khả năng đưa ra dự đoán hoặc phân loại chính xác. Vậy “chính xác” được xác định như thế nào và làm thế nào để cải thiện mô hình một cách có hệ thống nhằm đạt được độ chính xác đó? Thách thức này được giải quyết thông qua **tối ưu hóa (Optimization)**, một quá trình được định hướng bởi **giải tích (Calculus)**.







## **1. Mục tiêu: Tìm các tham số tốt nhất cho mô hình (The Goal: Finding the Best Model Parameters)**

Hãy nhớ lại khái niệm về hàm số. Về cốt lõi, một mô hình **Machine Learning** thường có thể được biểu diễn dưới dạng một hàm số với các đầu vào (các **đặc trưng (features)** của dữ liệu) và đầu ra (các giá trị dự đoán). Hàm số này cũng có các *tham số* hoặc *trọng số* bên trong, quyết định hành vi cụ thể của mô hình. Đối với một mô hình tuyến tính đơn giản như $y=mx+b$, các tham số chính là độ dốc $m$ và tung độ gốc $b$. Những giá trị khác nhau của $m$ và $b$ sẽ tạo ra những đường thẳng khác nhau, từ đó dẫn đến những kết quả dự đoán khác nhau.

Mục tiêu của chúng ta là tìm ra những giá trị cụ thể cho các tham số này (trong ví dụ đơn giản của chúng ta là $m$ và $b$), sao cho mô hình hoạt động tốt nhất trên dữ liệu. “Tốt nhất” thường có nghĩa là giảm thiểu các sai số mà mô hình mắc phải.

## **2. Đo lường hiệu suất: Hàm chi phí (Measuring Performance: The Cost Function)**

Để tìm ra các tham số tốt nhất một cách có hệ thống, trước tiên chúng ta cần một phương pháp đo lường mức độ hoạt động hiệu quả (hoặc kém hiệu quả) hiện tại của mô hình. Đây chính là nhiệm vụ của **hàm chi phí (cost function)**, còn được gọi là hàm mất mát hoặc hàm mục tiêu.

Một **Cost Function** nhận các giá trị dự đoán của mô hình và các giá trị mục tiêu thực tế từ dữ liệu, sau đó tính toán một con số duy nhất biểu thị tổng sai số hoặc “chi phí”. Một ví dụ phổ biến là **Mean Squared Error (MSE)**, tính trung bình bình phương độ chênh lệch giữa các giá trị dự đoán và giá trị thực tế.

- **Chi phí cao:** Các giá trị dự đoán của mô hình cách xa các giá trị thực tế. Các tham số có khả năng chưa tốt.
- **Chi phí thấp:** Các giá trị dự đoán của mô hình gần với các giá trị thực tế. Các tham số tốt hơn.

Mục tiêu tối ưu hóa của chúng ta trở nên rõ ràng: **Tìm các tham số của mô hình sao cho giá trị của hàm chi phí đạt mức thấp nhất.**



## **3. Thuật toán: Gradient Descent (The Algorithm: Gradient Descent)**

Làm thế nào để tìm được các giá trị tham số giúp hàm chi phí đạt mức thấp nhất? Chúng ta có thể thử các giá trị ngẫu nhiên, nhưng cách này cực kỳ kém hiệu quả, đặc biệt khi mô hình có nhiều tham số. Một phương pháp có hệ thống hơn nhiều là **Gradient Descent**.

Hãy hình dung hàm chi phí như một địa hình có những ngọn đồi và thung lũng. Độ cao tại mỗi điểm biểu thị chi phí tương ứng với một tập hợp giá trị tham số cụ thể. Mục tiêu của chúng ta là tìm điểm thấp nhất trong một thung lũng (điểm cực tiểu).

**Gradient Descent** là một thuật toán lặp giúp chúng ta “di chuyển” xuống theo độ dốc của hàm chi phí:

1. **Bắt đầu:** Khởi tạo bằng một giá trị ước lượng ban đầu cho các tham số.

2. **Tính toán hướng di chuyển:** Xác định hướng *tăng nhanh nhất* tại vị trí hiện tại. **Calculus** cung cấp cho chúng ta hướng này thông qua **gradient** của hàm chi phí, ký hiệu là $\nabla J$. Hãy nhớ rằng gradient là một vector chứa các **đạo hàm riêng (partial derivatives)** của hàm chi phí theo từng tham số (ví dụ: $\frac{\partial J}{\partial m}$ và $\frac{\partial J}{\partial b}$). Gradient chỉ hướng đi lên.

3. **Thực hiện một bước:** Di chuyển một bước nhỏ theo hướng *ngược lại* với gradient (hướng đi xuống). Độ lớn của bước di chuyển này được kiểm soát bởi **Learning Rate**.

4. **Lặp lại:** Tính gradient mới tại vị trí vừa di chuyển đến và tiếp tục thực hiện một bước theo hướng đi xuống.

5. **Dừng lại:** Tiếp tục lặp cho đến khi đạt được một điểm mà tại đó chi phí đủ thấp, hoặc các bước di chuyển trở nên rất nhỏ, cho thấy chúng ta có khả năng đã đạt đến một điểm cực tiểu.

Về bản chất, gradient (được xây dựng từ các đạo hàm riêng) đóng vai trò như chiếc la bàn, luôn chỉ hướng đi lên. Bằng cách liên tục di chuyển theo hướng ngược lại, **Gradient Descent** dẫn dắt chúng ta đến một điểm cực tiểu của hàm chi phí, qua đó tìm ra các giá trị tham số giúp mô hình hoạt động tốt hơn.
