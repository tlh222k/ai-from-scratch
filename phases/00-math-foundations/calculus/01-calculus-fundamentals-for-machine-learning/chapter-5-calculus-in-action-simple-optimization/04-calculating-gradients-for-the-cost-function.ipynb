# **Tính gradient cho hàm chi phí (Calculating Gradients for the Cost Function)**

Tính gradient cho hàm chi phí là một bước quan trọng trong quá trình tối ưu hóa. Một mô hình tuyến tính đơn giản:

$$
y=mx+b
$$

được sử dụng để biểu diễn dữ liệu. Để đo lường mức độ phù hợp của đường thẳng này với dữ liệu, một hàm chi phí, thường là **Bình phương sai số trung bình (Mean Squared Error - MSE)**, sẽ định lượng sai số. Mục tiêu là tìm các giá trị của $m$ (**Hệ số góc - Slope**) và $b$ (**Tung độ gốc - y-intercept**) sao cho hàm chi phí đạt giá trị nhỏ nhất.

**Tối ưu hóa xuống dốc (Gradient Descent)** cần có hướng di chuyển. Nó cần biết: "Nếu tôi thay đổi $m$ một chút, chi phí sẽ thay đổi như thế nào?" và "Nếu tôi thay đổi $b$ một chút, chi phí sẽ thay đổi như thế nào?". Đây chính xác là điều mà các **Đạo hàm riêng (Partial Derivatives)** cho chúng ta biết. Chúng ta cần tính gradient của hàm chi phí, $\nabla C$, bao gồm các đạo hàm riêng theo từng tham số: $m$ và $b$.

## **1. Hàm chi phí: Mean Squared Error (MSE)**

Giả sử chúng ta có $N$ điểm dữ liệu:

$$
(x_1,y_1),(x_2,y_2),\ldots,(x_N,y_N)
$$

Với mỗi cặp giá trị $m$ và $b$ bất kỳ, mô hình của chúng ta dự đoán giá trị

$$
y_{\text{pred},i}=mx_i+b
$$

cho mỗi đầu vào $x_i$. Giá trị thực tế là $y_i$.

Hàm chi phí **MSE**, ký hiệu là $C(m,b)$, được định nghĩa là giá trị trung bình của bình phương hiệu giữa các giá trị dự đoán và giá trị thực tế:

$$
C(m,b)=\frac{1}{N}\sum_{i=1}^{N}
(y_{\text{pred},i}-y_{\text{actual},i})^2
$$

Thay giá trị dự đoán của mô hình vào:

$$
C(m,b)=\frac{1}{N}\sum_{i=1}^{N}
((mx_i+b)-y_i)^2
$$

Hàm số $C(m,b)$ này nhận $m$ và $b$ làm đầu vào, đồng thời trả về một con số duy nhất biểu thị sai số trung bình. Nhiệm vụ của chúng ta lúc này là tìm các đạo hàm riêng:

$$
\frac{\partial C}{\partial m}
\qquad \text{và} \qquad
\frac{\partial C}{\partial b}
$$

## **2. Tính đạo hàm riêng theo $m$ ($\frac{\partial C}{\partial m}$)**

Để tìm $\frac{\partial C}{\partial m}$, chúng ta lấy đạo hàm của hàm chi phí $C(m,b)$ theo $m$, đồng thời xem $b$ (cùng tất cả các giá trị $x_i$, $y_i$ và $N$) là các hằng số.

Hãy xét biểu thức theo từng thành phần:

- **Hằng số (Constant Factor):** $\frac{1}{N}$ là một hằng số nhân, vì vậy nó được giữ nguyên.

- **Quy tắc tổng (Sum Rule):** Đạo hàm của một tổng bằng tổng các đạo hàm. Chúng ta có thể đưa phép lấy đạo hàm vào bên trong ký hiệu tổng:

$$
\frac{\partial C}{\partial m}
=
\frac{1}{N}\sum_{i=1}^{N}
\frac{\partial}{\partial m}
(mx_i+b-y_i)^2
$$

- **Quy tắc hàm hợp (Chain Rule):** Chúng ta cần lấy đạo hàm của biểu thức $(mx_i+b-y_i)^2$ theo $m$. Đặt: $ u=mx_i+b-y_i $. Chúng ta đang lấy đạo hàm của $u^2$ theo $m$. Quy tắc **Chain Rule** cho biết:

$$
\frac{d}{dm}(u^2)
=
2u\cdot\frac{\partial u}{\partial m}
$$

- **Thành phần thứ nhất:**

$$
2u=2(mx_i+b-y_i)
$$

- **Thành phần thứ hai:** Chúng ta cần tính:

$$
\frac{\partial u}{\partial m}
=
\frac{\partial}{\partial m}(mx_i+b-y_i)
$$

- Vì $b$ và $y_i$ được xem là các hằng số nên đạo hàm của chúng bằng 0. Đạo hàm của $mx_i$ theo $m$ đơn giản là $x_i$ (vì $x_i$ được xem là hệ số của $m$). Do đó:

$$
\frac{\partial u}{\partial m}=x_i
$$

- **Kết hợp các thành phần:** Thay trở lại công thức **Chain Rule**:

$$
\frac{\partial}{\partial m}(mx_i+b-y_i)^2
=
2(mx_i+b-y_i)\cdot x_i
$$

- **Kết quả cuối cùng cho $\frac{\partial C}{\partial m}$:** Bây giờ, thay kết quả này trở lại biểu thức tổng:

$$
\frac{\partial C}{\partial m}
=
\frac{1}{N}\sum_{i=1}^{N}
2(mx_i+b-y_i)x_i
$$

- Chúng ta có thể đưa hằng số 2 ra ngoài dấu tổng:

$$
\boxed{
\frac{\partial C}{\partial m}
=
\frac{2}{N}\sum_{i=1}^{N}
(mx_i+b-y_i)x_i
}
$$

Biểu thức này cho chúng ta biết chi phí thay đổi như thế nào khi chúng ta thay đổi **Slope** $m$ một lượng nhỏ. Hãy chú ý rằng nó phụ thuộc vào thành phần sai số $(mx_i+b-y_i)$ và giá trị đầu vào $x_i$ của từng điểm dữ liệu.

## **3. Tính đạo hàm riêng theo $b$ ($\frac{\partial C}{\partial b}$)**

Bây giờ, chúng ta lặp lại quy trình, nhưng lần này lấy đạo hàm của $C(m,b)$ theo $b$, đồng thời xem $m$ là một hằng số.

- **Hằng số và quy tắc tổng (Constant Factor and Sum Rule):** Tương tự như trước, chúng ta có:

$$
\frac{\partial C}{\partial b}
=
\frac{1}{N}\sum_{i=1}^{N}
\frac{\partial}{\partial b}
(mx_i+b-y_i)^2
$$

- **Quy tắc hàm hợp (Chain Rule):** Một lần nữa, đặt:

$$
u=mx_i+b-y_i
$$

- Chúng ta cần tính:

$$
\frac{d}{db}(u^2)
=
2u\cdot\frac{\partial u}{\partial b}
$$

- **Thành phần thứ nhất:**

$$
2u=2(mx_i+b-y_i)
$$

- **Thành phần thứ hai:** Chúng ta cần tính:

$$
\frac{\partial u}{\partial b}
=
\frac{\partial}{\partial b}(mx_i+b-y_i)
$$

Lần này, $m$, $x_i$ và $y_i$ được xem là các hằng số. Đạo hàm của $mx_i$ theo $b$ bằng 0. Đạo hàm của $b$ theo $b$ bằng 1. Đạo hàm của $y_i$ theo $b$ bằng 0. Do đó:

$$
\frac{\partial u}{\partial b}=1
$$

- **Kết hợp các thành phần:**

$$
\frac{\partial}{\partial b}(mx_i+b-y_i)^2
=
2(mx_i+b-y_i)\cdot 1
$$

- **Kết quả cuối cùng cho $\frac{\partial C}{\partial b}$:** Thay trở lại biểu thức tổng:

$$
\frac{\partial C}{\partial b}
=
\frac{1}{N}\sum_{i=1}^{N}
2(mx_i+b-y_i)(1)
$$

- Đưa hằng số 2 ra ngoài:

$$
\boxed{
\frac{\partial C}{\partial b}
=
\frac{2}{N}\sum_{i=1}^{N}
(mx_i+b-y_i)
}
$$

Biểu thức này cho chúng ta biết chi phí thay đổi như thế nào khi chúng ta thay đổi **y-intercept** $b$ một lượng nhỏ. Nó chỉ phụ thuộc vào thành phần sai số $(mx_i+b-y_i)$ của từng điểm dữ liệu.



## **4. Vector Gradient ($\nabla C$) (The Gradient Vector)**

Chúng ta đã tính thành công các đạo hàm riêng! Bây giờ, chúng ta có thể kết hợp chúng thành **Vector Gradient** của hàm chi phí $C(m,b)$:

$$
\nabla C(m,b)
=
\left[
\frac{\partial C}{\partial m},
\frac{\partial C}{\partial b}
\right]
$$

Thay các biểu thức đã tìm được vào:

$$
\boxed{
\nabla C(m,b)
=
\left[
\frac{2}{N}\sum_{i=1}^{N}(mx_i+b-y_i)x_i,
\frac{2}{N}\sum_{i=1}^{N}(mx_i+b-y_i)
\right]
}
$$

Vector này, $\nabla C$, có vai trò nền tảng. Với bất kỳ giá trị $m$ và $b$ nào đã cho:

- Thành phần thứ nhất ($\frac{\partial C}{\partial m}$) cho chúng ta biết tốc độ thay đổi của hàm chi phí nếu chúng ta chỉ di chuyển theo hướng $m$. Dấu của thành phần này cho biết việc tăng $m$ sẽ làm tăng hay giảm chi phí.

- Thành phần thứ hai ($\frac{\partial C}{\partial b}$) cho chúng ta biết tốc độ thay đổi của hàm chi phí nếu chúng ta chỉ di chuyển theo hướng $b$. Dấu của thành phần này cho biết việc tăng $b$ sẽ làm tăng hay giảm chi phí.

Điều quan trọng là **Vector Gradient** $\nabla C$ chỉ theo hướng **tăng nhanh nhất (steepest increase)** của hàm chi phí $C(m,b)$ tại điểm hiện tại $(m,b)$. Vì chúng ta muốn cực tiểu hóa chi phí, **Gradient Descent** sẽ thực hiện các bước di chuyển theo hướng ngược lại với gradient.

Với các gradient đã tính được này, giờ đây chúng ta có thông tin chính xác cần thiết để cập nhật $m$ và $b$ theo cách lặp đi lặp lại, nhằm tìm ra đường thẳng phù hợp nhất với dữ liệu. Bước tiếp theo là tìm hiểu cách sử dụng những gradient này bên trong thuật toán **Gradient Descent**.
