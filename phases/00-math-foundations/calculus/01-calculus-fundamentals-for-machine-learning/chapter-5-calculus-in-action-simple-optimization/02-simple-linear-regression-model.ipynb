# **Ví dụ: Mô hình hồi quy tuyến tính đơn giản (Example: Simple Linear Regression Model)**

Giải tích cung cấp các công cụ để tìm giá trị cực tiểu và cực đại thông qua **đạo hàm (derivatives)** và khái niệm **Gradient Descent**. Những công cụ này được áp dụng trong các mô hình học máy cơ bản, chẳng hạn như **hồi quy tuyến tính đơn giản (Simple Linear Regression)**.

Mục tiêu của hồi quy tuyến tính đơn giản là tìm ra đường thẳng phù hợp nhất để mô tả mối quan hệ giữa một **đặc trưng đầu vào (input feature)** duy nhất và một **đầu ra mục tiêu (target output)**. Hãy tưởng tượng bạn có một số điểm dữ liệu được biểu diễn trên đồ thị. Hồi quy tuyến tính cố gắng vẽ một đường thẳng đi qua các điểm này sao cho đường thẳng khớp với chúng nhất có thể.





## **1. Phương trình mô hình (The Model Equation)**

Có lẽ bạn đã quen thuộc với phương trình đường thẳng:

$$
y=mx+b
$$

Hãy phân tích các thành phần trong bối cảnh học máy:

- **$x$**: Đây là đặc trưng đầu vào của chúng ta. Nó là thông tin mà chúng ta có, còn được gọi là **biến độc lập (independent variable)**. Ví dụ, $x$ có thể là diện tích của một ngôi nhà.
- **$y$**: Đây là biến mục tiêu mà chúng ta muốn dự đoán, còn được gọi là **biến phụ thuộc (dependent variable)**. Ví dụ, $y$ có thể là giá bán của ngôi nhà.
- **$m$**: Đây là **Slope** của đường thẳng. Nó xác định mức thay đổi của $y$ khi $x$ tăng thêm một đơn vị. Trong ví dụ về ngôi nhà, nó biểu thị mức giá tăng thêm ứng với mỗi đơn vị diện tích tăng thêm.
- **$b$**: Đây là **tung độ gốc (y-intercept)** của đường thẳng. Đó là giá trị của $y$ khi $x$ bằng không. Trong một số trường hợp, giá trị này có thể biểu thị một mức cơ sở (ví dụ: giá cơ bản của một ngôi nhà bất kể diện tích, mặc dù cách diễn giải này không phải lúc nào cũng có ý nghĩa, tùy thuộc vào bối cảnh).

Trong thuật ngữ học máy, $m$ và $b$ là các **Parameters** (đôi khi còn được gọi là **trọng số (weights)** hoặc **hệ số (coefficients)**) của mô hình. Mục tiêu của chúng ta trong quá trình huấn luyện là tìm các giá trị tối ưu cho $m$ và $b$, sao cho đường thẳng khớp với dữ liệu tốt nhất.



## **2. Trực quan hóa mục tiêu (Visualizing the Goal)**

Ví dụ, hãy xét một tập dữ liệu nhỏ thể hiện mối quan hệ giữa số giờ học ($x$) và điểm thi ($y$).



> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/linear_regression_goal.html)

> *Các chấm màu xanh dương biểu thị dữ liệu của chúng ta. Đường nét đứt màu xanh lá có vẻ nắm bắt xu hướng khá tốt, trong khi đường chấm màu đỏ rõ ràng không khớp tốt với dữ liệu. Hồi quy tuyến tính hướng đến việc tìm các tham số của đường thẳng ($m$ và $b$) để đạt được mức độ khớp tốt nhất có thể, chẳng hạn như đường màu xanh lá hoặc thậm chí tốt hơn.*



Quá trình tìm đường thẳng "khớp tốt nhất" này là một nhiệm vụ thuộc **supervised learning**. Chúng ta được cung cấp các cặp ví dụ $(x,y)$ (số giờ học, điểm thi) và muốn thuật toán học máy học được mối quan hệ giữa chúng, được biểu diễn thông qua các tham số $m$ và $b$. Sau khi học được mối quan hệ này, chúng ta có thể sử dụng mô hình $y=mx+b$ để dự đoán điểm thi tương ứng với một số giờ học mới.

Ý tưởng cốt lõi là chúng ta có thể đo lường mức độ *phù hợp* của một đường thẳng cụ thể (được xác định bởi các giá trị cụ thể của $m$ và $b$) với dữ liệu. Nếu có thể đo lường "mức độ khớp" này (hoặc ngược lại, "mức độ không khớp" hay sai số), chúng ta có thể sử dụng giải tích, cụ thể là **Gradient Descent**, để điều chỉnh $m$ và $b$ một cách có hệ thống, giúp đường thẳng ngày càng khớp với dữ liệu tốt hơn.

"Thước đo mức độ khớp" này được gọi là **hàm chi phí (cost function)**, sẽ được định nghĩa trong phần tiếp theo. Hãy nhớ rằng, trong các thuật toán tối ưu hóa của chúng ta, $m$ và $b$ là những biến sẽ được điều chỉnh để giảm thiểu hàm chi phí này.
