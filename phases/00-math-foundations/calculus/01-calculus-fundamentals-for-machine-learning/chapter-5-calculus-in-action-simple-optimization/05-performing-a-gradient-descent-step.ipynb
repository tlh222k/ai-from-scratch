# **Thực hiện một bước Gradient Descent (Performing a Gradient Descent Step)**

Một mô hình **hồi quy tuyến tính đơn giản (Simple Linear Regression)**,

$$
y=mx+b
$$

sử dụng hàm chi phí $J(m,b)$ để đo lường mức độ phù hợp của đường thẳng với dữ liệu. Gradient của hàm chi phí này được tính dưới dạng một vector chứa các đạo hàm riêng $\frac{\partial J}{\partial m}$ và $\frac{\partial J}{\partial b}$. Những đạo hàm này cho biết chi phí thay đổi như thế nào khi chúng ta điều chỉnh nhẹ $m$ hoặc $b$.

Vậy, làm thế nào để thực sự sử dụng những gradient này nhằm cải thiện các tham số mô hình và giảm thiểu chi phí? Đây chính là lúc thuật toán **Gradient Descent** phát huy tác dụng.

Hãy hình dung hàm chi phí $J(m,b)$ xác định một bề mặt, có thể giống như một địa hình với những ngọn đồi và thung lũng. Mục tiêu của chúng ta là tìm điểm thấp nhất trên địa hình này, tức là điểm có chi phí nhỏ nhất. Các tham số $m$ và $b$ xác định vị trí hiện tại của chúng ta trên bề mặt đó.

Gradient,

$$
\nabla J=
\left[
\frac{\partial J}{\partial m},
\frac{\partial J}{\partial b}
\right]
$$

tại vị trí hiện tại chỉ theo hướng **tăng nhanh nhất (steepest ascent)**, tức là con đường đi lên nhanh nhất. Vì chúng ta muốn giảm thiểu chi phí nên cần đi xuống dốc. Do đó, chúng ta thực hiện một bước theo hướng ngược lại với gradient.

Điều này dẫn chúng ta đến quy tắc cập nhật cốt lõi của **Gradient Descent**. Đối với mỗi tham số, chúng ta điều chỉnh giá trị hiện tại bằng cách trừ đi một lượng nhỏ tỷ lệ thuận với đạo hàm riêng tương ứng:

- **Cập nhật hệ số góc $m$:** $
m_{\text{new}}
=
m_{\text{old}}
-
\alpha\frac{\partial J}{\partial m}
$

- **Cập nhật tung độ gốc $b$:** $
b_{\text{new}}
=
b_{\text{old}}
-
\alpha\frac{\partial J}{\partial b}
$

Ở đây, $m_{\text{old}}$ và $b_{\text{old}}$ là các giá trị tham số trước bước cập nhật, còn $m_{\text{new}}$ và $b_{\text{new}}$ là các giá trị sau bước cập nhật. Các gradient $\frac{\partial J}{\partial m}$ và $\frac{\partial J}{\partial b}$ được tính bằng các giá trị hiện tại ($m_{\text{old}}$, $b_{\text{old}}$).

Hãy chú ý đến ký hiệu $\alpha$ (alpha) trong các quy tắc cập nhật. Đây là **Learning Rate**, sẽ được thảo luận trong phần tiếp theo. Trước mắt, hãy xem nó là đại lượng kiểm soát kích thước của bước di chuyển xuống dốc. Đây là một số dương nhỏ (ví dụ: 0.01, 0.1).

- Nếu thành phần gradient (chẳng hạn $\frac{\partial J}{\partial m}$) là số dương, điều đó có nghĩa là việc tăng $m$ sẽ làm tăng chi phí. Vì vậy, quy tắc cập nhật trừ đi một giá trị dương ($\alpha \times$ gradient dương), qua đó làm giảm $m$ một cách hiệu quả.
- Nếu thành phần gradient là số âm, việc tăng $m$ sẽ làm giảm chi phí. Quy tắc cập nhật trừ đi một giá trị âm ($\alpha \times$ gradient âm), nghĩa là chúng ta cộng thêm vào $m$, đưa nó di chuyển theo hướng làm giảm chi phí.

Phép tính và cập nhật duy nhất này cho cả $m$ và $b$ tạo thành một bước của **Gradient Descent**.

Sơ đồ cho thấy quy trình của một bước **Gradient Descent**. Bắt đầu với các tham số mô hình hiện tại, tính gradient của hàm chi phí bằng những tham số đó, sau đó cập nhật các tham số bằng cách di chuyển một khoảng nhỏ theo hướng ngược lại với gradient. Quy trình này thường được lặp lại nhiều lần.

Điều quan trọng cần hiểu là **Gradient Descent** là một **thuật toán lặp (iterative algorithm)**. Thông thường, chỉ một bước là chưa đủ để đạt đến điểm có chi phí nhỏ nhất. Chúng ta lặp lại quy trình tính gradient và cập nhật các tham số nhiều lần. Qua mỗi bước, chúng ta (hy vọng) sẽ tiến gần hơn đến các giá trị $m$ và $b$ giúp cực tiểu hóa hàm chi phí, từ đó tạo ra một mô hình hồi quy tuyến tính có mức độ phù hợp tốt hơn.

Bước này, được định hướng bởi các đạo hàm mà chúng ta đã tính, là cơ chế nền tảng giúp nhiều mô hình học máy học từ dữ liệu. Bằng cách liên tục điều chỉnh các tham số theo hướng làm giảm sai số (chi phí), mô hình dần cải thiện khả năng dự đoán của mình.
