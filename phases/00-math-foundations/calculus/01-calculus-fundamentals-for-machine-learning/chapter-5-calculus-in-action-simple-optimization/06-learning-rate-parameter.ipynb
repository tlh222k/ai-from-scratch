# **Tham số Tỷ lệ học (Learning Rate Parameter)**

Bước **Tối ưu hóa xuống dốc (Gradient Descent)** bao gồm việc tính gradient của **Hàm chi phí (Cost Function)**. Gradient này chỉ ra hướng **tăng** chi phí nhanh nhất. Để **cực tiểu hóa** chi phí, chúng ta cần di chuyển theo hướng **ngược lại** với gradient.

Nhưng trong mỗi bước, chúng ta sẽ di chuyển **bao xa** theo hướng ngược lại đó? Đây chính là lúc **Tỷ lệ học (Learning Rate)** phát huy tác dụng. Đây là một số dương nhỏ, thường được ký hiệu bằng chữ cái Hy Lạp alpha ($\alpha$), dùng để điều chỉnh độ lớn của bước di chuyển mà chúng ta thực hiện.

Hãy nhớ lại quy tắc cập nhật cho một tham số (chẳng hạn `m` hoặc `b` trong ví dụ hồi quy tuyến tính). Dạng tổng quát của quy tắc này như sau:

  **`parameter = parameter - learning_rate * gradient_of_cost_wrt_parameter`**

Đối với hai tham số cụ thể `m` và `b`, sử dụng ký hiệu đạo hàm riêng cho các thành phần gradient, ta có các công thức cập nhật:

$$
b=b-\alpha\frac{\partial J}{\partial b}
$$

$$
m=m-\alpha\frac{\partial J}{\partial m}
$$

Ở đây, $J$ biểu thị *Cost Function*, $\frac{\partial J}{\partial b}$ là đạo hàm riêng của chi phí theo `b`, $\frac{\partial J}{\partial m}$ là đạo hàm riêng của chi phí theo `m`, và $\alpha$ là *Learning Rate*.

## **1. Tại sao chúng ta cần Learning Rate?**

Gradient cho chúng ta biết hướng cần di chuyển, nhưng không cho biết khoảng cách cần di chuyển. Độ lớn của vector gradient cho biết độ dốc của bề mặt lớn đến mức nào. Nếu trong mỗi bước, chúng ta chỉ đơn giản trừ toàn bộ gradient khỏi các tham số, chúng ta có thể thực hiện những bước rất lớn khi độ dốc cao và những bước rất nhỏ khi bề mặt thoải. Việc thực hiện những bước quá lớn, đặc biệt khi chúng ta còn cách xa điểm cực tiểu, có thể gây ra vấn đề.

Hãy tưởng tượng bạn đang đi bộ xuống một ngọn núi trong điều kiện sương mù. Bạn có thể cảm nhận được độ dốc dưới chân mình (gradient), cho biết hướng đi xuống dốc nhất. *Learning Rate* giống như việc bạn quyết định độ dài mỗi bước chân.

## **2. Ảnh hưởng của độ lớn Learning Rate**

Việc lựa chọn *Learning Rate* phù hợp ($\alpha$) rất quan trọng để quá trình tối ưu hóa đạt hiệu quả. Hãy xem điều gì xảy ra với những lựa chọn khác nhau:

- **Nếu Learning Rate quá nhỏ:** Bạn thực hiện những bước rất ngắn xuống dốc. Cuối cùng, bạn vẫn sẽ đến đáy (điểm có chi phí nhỏ nhất), nhưng quá trình này có thể mất rất nhiều thời gian và cần nhiều vòng lặp. Tiến độ sẽ chậm.
- **Nếu Learning Rate quá lớn:** Bạn thực hiện những bước rất dài xuống dốc. Bạn có thể vượt quá đáy thung lũng và đi sang phía bên kia, thậm chí có khả năng đến một vị trí cao hơn cả điểm xuất phát! Chi phí có thể dao động thất thường và không giảm xuống, hoặc thậm chí tăng dần theo thời gian (phân kỳ).
- **Nếu Learning Rate ở mức "vừa đủ":** Bạn thực hiện những bước có độ dài hợp lý, tiến đều đặn về phía điểm cực tiểu mà không vượt quá điểm đó quá nhiều. Điều này thường giúp tìm được một nghiệm tốt một cách hiệu quả.

Hãy cùng hình dung điều này. Giả sử chúng ta có một *Cost Function* đơn giản chỉ phụ thuộc vào một tham số, được biểu diễn trên đồ thị theo giá trị của tham số đó. Chúng ta muốn tìm điểm thấp nhất. Biểu đồ sau mô phỏng cách chi phí có thể thay đổi qua các vòng lặp với những giá trị *Learning Rate* khác nhau.

> *Mô phỏng hàm chi phí $J(w)=w^2$, bắt đầu tại $w=5$, sử dụng Gradient Descent với quy tắc cập nhật $w=w-\alpha(2w)$. Learning Rate phù hợp (màu xanh lá) giúp chi phí giảm đều đặn. Learning Rate nhỏ (màu vàng) khiến chi phí giảm rất chậm. Learning Rate lớn (màu cam) có thể gây ra dao động mà không tạo ra sự cải thiện. Learning Rate còn lớn hơn (màu đỏ) có thể khiến chi phí tăng lên (phân kỳ).*

## **Lựa chọn một giá trị**

Vậy, làm thế nào để lựa chọn *Learning Rate*? Việc tìm ra giá trị tối ưu thường đòi hỏi một quá trình thử nghiệm. Một số giá trị khởi đầu phổ biến có thể là 0.1, 0.01, 0.001 hoặc 0.0001. Bạn có thể thử một vài giá trị khác nhau và quan sát xem giá trị nào khiến *Cost Function* giảm đều đặn với tốc độ hợp lý trong những vòng lặp huấn luyện ban đầu. Ngoài ra, còn có những kỹ thuật nâng cao hơn để tự động điều chỉnh *Learning Rate* trong quá trình huấn luyện. Tuy nhiên, ở thời điểm hiện tại, hãy hiểu rằng đây là một tham số mà thông thường bạn cần thiết lập trước khi bắt đầu quá trình tối ưu hóa.

Tóm lại, *Learning Rate* $\alpha$ là một con số nhỏ nhưng có ý nghĩa quan trọng, kiểm soát độ lớn của mỗi bước mà *Gradient Descent* thực hiện trong từng vòng lặp. Tham số này cần được lựa chọn cẩn thận: giá trị quá nhỏ dẫn đến hội tụ chậm, trong khi giá trị quá lớn có thể gây mất ổn định và phân kỳ. Lựa chọn đúng giá trị sẽ giúp thuật toán tìm ra các giá trị tham số cực tiểu hóa *Cost Function* một cách hiệu quả.
