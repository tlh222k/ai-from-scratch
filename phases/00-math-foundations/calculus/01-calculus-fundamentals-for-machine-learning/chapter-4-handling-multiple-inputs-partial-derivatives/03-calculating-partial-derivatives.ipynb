# **Cách tính đạo hàm riêng (Calculating Partial Derivatives)**

Để tính **đạo hàm riêng (partial derivatives)**, chúng ta xem các biến còn lại là hằng số. Đây là ý tưởng quan trọng nhất cần ghi nhớ. Nếu bạn đã biết cách tính đạo hàm của **hàm số một biến (single-variable functions)** như $f(x)$, thì bạn đã có gần như đầy đủ những kỹ năng cần thiết.





## **1. Quy trình: Giữ cố định và lấy đạo hàm (The Process: Freeze and Differentiate)**

Khi cần tính đạo hàm riêng của một hàm số theo một biến cụ thể, hãy thực hiện các bước sau:

1. **Xác định biến cần lấy đạo hàm:** Xác định biến mà bạn muốn tính đạo hàm theo nó. Ví dụ, nếu đang tính $\frac{\partial f}{\partial x}$, thì biến cần lấy đạo hàm là $x$.

2. **Xem các biến còn lại là hằng số:** Hãy hình dung (hoặc thay thế trực tiếp nếu điều đó giúp bạn dễ hiểu hơn) tất cả các biến còn lại trong hàm số bằng những hằng số cố định. Hãy xem chúng như những con số chẳng hạn $5$, $-2$ hoặc $\pi$.

3. **Áp dụng các quy tắc tính đạo hàm thông thường:** Bây giờ, hãy tính đạo hàm của hàm số bằng những quy tắc quen thuộc như **Quy tắc lũy thừa (Power Rule)**, **Quy tắc hằng số (Constant Rule)** và **Quy tắc tổng (Sum Rule)**, nhưng chỉ lấy đạo hàm theo biến đã chọn. Hãy nhớ rằng đạo hàm của mọi hạng tử là hằng số đều bằng 0.

Hãy cùng thực hiện một vài ví dụ để hiểu rõ hơn cách làm này.



### **Ví dụ 1: Đa thức đơn giản (A Simple Polynomial)**

Xét hàm số:

$$f(x,y)=x^2+y^3+4$$

Hãy tìm các đạo hàm riêng theo $x$ và $y$.

**Tính $\frac{\partial f}{\partial x}$ (đạo hàm riêng theo $x$):**

- **Biến cần lấy đạo hàm:** $x$.
- **Xem các biến còn lại là hằng số:** Xem $y$ là một hằng số. Điều này có nghĩa là $y^3$ cũng được xem là một hằng số. Số $4$ vốn đã là một hằng số.
- **Tính đạo hàm:**
  - Đạo hàm của $x^2$ theo $x$ là $2x$.
  - Đạo hàm của $y^3$ theo $x$ là $0$, vì $y^3$ được xem là hằng số.
  - Đạo hàm của $4$ theo $x$ là $0$, vì đây là một hằng số.

Kết hợp các kết quả bằng **Quy tắc tổng (Sum Rule)**:

$$\frac{\partial f}{\partial x}=2x+0+0=2x$$

**Tính $\frac{\partial f}{\partial y}$ (đạo hàm riêng theo $y$):**

- **Biến cần lấy đạo hàm:** $y$.
- **Xem các biến còn lại là hằng số:** Xem $x$ là một hằng số. Điều này có nghĩa là $x^2$ cũng được xem là một hằng số. Số $4$ cũng là một hằng số.
- **Tính đạo hàm:**
  - Đạo hàm của $x^2$ theo $y$ là $0$, vì $x^2$ được xem là hằng số.
  - Đạo hàm của $y^3$ theo $y$ là $3y^2$.
  - Đạo hàm của $4$ theo $y$ là $0$, vì đây là một hằng số.

Kết hợp các kết quả:

$$\frac{\partial f}{\partial y}=0+3y^2+0=3y^2$$

Hãy chú ý rằng quy trình này giúp chúng ta tách riêng ảnh hưởng của từng biến. Khi tính $\frac{\partial f}{\partial x}$, hạng tử $y^3$ biến mất vì xét theo góc độ của $x$, biến $y$ không thay đổi.



### **Ví dụ 2: Các biến được nhân với nhau (Variables Multiplied Together)**

Hãy xem một dạng hàm số thường gặp khi làm việc với các tham số của mô hình, chẳng hạn trọng số ($w$) và độ chệch ($b$):

$$g(w,b)=w^2b+5w-2b+7$$

**Tính $\frac{\partial g}{\partial w}$ (đạo hàm riêng theo $w$):**

- **Biến cần lấy đạo hàm:** $w$.
- **Xem các biến còn lại là hằng số:** Xem $b$ là một hằng số.
- **Tính đạo hàm:**
  - Xét hạng tử $w^2b$. Vì $b$ được xem là hệ số không đổi, đạo hàm của $w^2b$ theo $w$ là:

    $$(b)\times(2w)=2wb$$

    Hãy hình dung phép tính này giống như tìm đạo hàm của $5x^2$, bằng $5\times 2x=10x$. Ở đây, $b$ đóng vai trò giống như số $5$.

  - Đạo hàm của $5w$ theo $w$ là $5$.
  - Đạo hàm của $-2b$ theo $w$ là $0$, vì $-2b$ được xem là một hằng số.
  - Đạo hàm của $7$ theo $w$ là $0$, vì đây là một hằng số.

Kết hợp các kết quả:

$$\frac{\partial g}{\partial w}=2wb+5+0+0=2wb+5$$

**Tính $\frac{\partial g}{\partial b}$ (đạo hàm riêng theo $b$):**

- **Biến cần lấy đạo hàm:** $b$.
- **Xem các biến còn lại là hằng số:** Xem $w$ là một hằng số. Điều này có nghĩa là $w^2$ và $5w$ cũng được xem là các hằng số.
- **Tính đạo hàm:**
  - Xét hạng tử $w^2b$. Vì $w^2$ được xem là một hệ số không đổi, đạo hàm của $w^2b$ theo $b$ là:

    $$w^2\times 1=w^2$$

    Hãy hình dung phép tính này giống như tìm đạo hàm của $ax$ theo $x$, kết quả là $a$. Ở đây, $w^2$ đóng vai trò giống như $a$, còn $b$ đóng vai trò giống như $x$.
  - Đạo hàm của $5w$ theo $b$ là $0$, vì $5w$ được xem là một hằng số.
  - Đạo hàm của $-2b$ theo $b$ là $-2$.
  - Đạo hàm của $7$ theo $b$ là $0$, vì đây là một hằng số.

Kết hợp các kết quả:

$$\frac{\partial g}{\partial b}=w^2+0-2+0=w^2-2$$



## **2. Điều cần ghi nhớ (Takeaway)**

Cách tính **đạo hàm riêng (partial derivatives)** sử dụng lại các quy tắc tính đạo hàm mà bạn đã học đối với hàm số một biến. Kỹ thuật quan trọng nhất là tạm thời "giữ cố định" tất cả các biến, ngoại trừ biến mà bạn đang lấy đạo hàm, bằng cách xem chúng là những hằng số trong suốt quá trình tính toán.

Nhờ đó, bạn có thể xác định đầu ra của hàm số thay đổi như thế nào khi chỉ một biến được chọn thay đổi, còn mọi yếu tố khác đều giữ nguyên. Đây là kỹ thuật nền tảng để hiểu cách điều chỉnh các tham số trong mô hình học máy bằng những phương pháp như **Gradient Descent**.
