# **Hàm số nhiều biến (Functions of Multiple Variables)**

Những hàm số như $f(x)=x^2$ hoặc $f(x)=3x+5$ là **hàm số một biến (single-variable functions)**: chúng nhận một đầu vào ($x$) và tạo ra một đầu ra. Ta có thể dễ dàng biểu diễn chúng dưới dạng một đường cong trên đồ thị 2D, cho thấy đầu ra thay đổi như thế nào khi đầu vào duy nhất thay đổi.

Tuy nhiên, trong nhiều tình huống, đặc biệt là trong **Machine Learning**, kết quả không chỉ phụ thuộc vào một yếu tố. Hãy nghĩ đến việc dự đoán giá một căn nhà. Giá nhà có chỉ phụ thuộc vào diện tích tính bằng feet vuông không? Có lẽ là không. Ta có thể còn cần xét đến số phòng ngủ, tuổi của căn nhà hoặc điểm đánh giá chất lượng khu vực xung quanh. Mỗi yếu tố này là một biến đầu vào độc lập.

Tương tự, khi huấn luyện một mô hình học máy, chẳng hạn như **mô hình hồi quy tuyến tính (linear regression model)** đơn giản $y=mx+b$ mà chúng ta sẽ tìm hiểu sau, ta cần tìm giá trị tốt nhất cho cả $m$ (độ dốc) và $b$ (tung độ gốc). Hiệu quả hoạt động của mô hình, thường được đo bằng một **hàm chi phí (Cost Function)** hoặc hàm sai số, phụ thuộc vào giá trị cụ thể được chọn cho cả $m$ và $b$.

Đây chính là lúc ta cần đến **hàm số nhiều biến (functions of multiple variables)**. Thay vì chỉ làm việc với $f(x)$, giờ đây ta sẽ xét những hàm như:

- $f(x,y)$ nhận hai đầu vào là $x$ và $y$.
- $f(x_1,x_2,x_3)$ nhận ba đầu vào.
- Tổng quát hơn, $f(x_1,x_2,\ldots,x_n)$ nhận $n$ đầu vào.

Tất cả những hàm này vẫn tạo ra một giá trị đầu ra duy nhất. Giá trị đầu ra phụ thuộc vào tổ hợp cụ thể của tất cả các giá trị đầu vào được cung cấp.



## **1. Định nghĩa hàm số nhiều biến (Defining Multivariable Functions)**

Một hàm số nhiều biến ánh xạ một tập hợp các giá trị đầu vào thành một giá trị đầu ra duy nhất. Với hai biến, ta có thể viết:

$$z=f(x,y)$$

Với $n$ biến, ta có:

$$y_{\text{output}}=f(x_1,x_2,\ldots,x_n)$$

Ở đây, $x_1,x_2,\ldots,x_n$ được gọi là **các biến độc lập (independent variables)**, còn giá trị đầu ra thu được ($z$ hoặc $y_{\text{output}}$) là **biến phụ thuộc (dependent variable)**.

Hãy xem một ví dụ đơn giản:

$$f(x,y)=x^2+y^2$$

Hàm này nhận hai đầu vào là $x$ và $y$. Để tính đầu ra, ta bình phương từng đầu vào rồi cộng hai kết quả lại.

- Nếu $x=1$ và $y=2$, ta có: $f(1,2)=1^2+2^2=1+4=5$

- Nếu $x=3$ và $y=-1$, ta có: $f(3,-1)=3^2+(-1)^2=9+1=10$


Một ví dụ khác, có lẽ gần với những gì ta thường gặp khi nói về sai số của mô hình hơn:

$$g(w,b)=(w\cdot 5+b-12)^2$$

Ở đây, hàm $g$ phụ thuộc vào hai đầu vào là $w$ và $b$. Nếu $w=2$ và $b=3$, ta có:

$$g(2,3)=(2\cdot 5+3-12)^2=(10+3-12)^2=(1)^2=1$$

Trong **Machine Learning**, $w$ và $b$ có thể đại diện cho các trọng số hoặc tham số của một mô hình đơn giản. Giá trị $5$ có thể là một đặc trưng đầu vào từ dữ liệu, còn $12$ có thể là giá trị mục tiêu thực tế. Hàm $g(w,b)$ có thể biểu diễn sai số bình phương của một điểm dữ liệu duy nhất. Mục tiêu khi huấn luyện mô hình là tìm các giá trị $w$ và $b$ sao cho sai số này (và sai số trên toàn bộ các điểm dữ liệu) nhỏ nhất có thể.



## **2. Biểu diễn trực quan hàm số nhiều biến (Visualizing Functions of Multiple Variables)**

Ta đã thấy rằng có thể biểu diễn **hàm số một biến** $f(x)$ dưới dạng một đường cong 2D. Vậy còn những hàm có hai biến, chẳng hạn như $z=f(x,y)$, thì sao?

Vì ta có hai đầu vào ($x$, $y$) và một đầu ra ($z$), nên cần đến ba chiều để vẽ đồ thị hàm số này. Ta có thể hình dung mặt phẳng $xy$ biểu diễn các tổ hợp đầu vào có thể có, còn trục $z$ thẳng đứng biểu diễn đầu ra. Với mỗi điểm $(x,y)$ trên mặt phẳng đầu vào, hàm $f(x,y)$ cho ta một độ cao $z$. Khi vẽ các điểm $(x,y,z)$ này, ta thường thu được một **bề mặt (surface)** trong không gian 3D.

Ví dụ, xét hàm:

$$f(x,y)=x^2+y^2$$

Hàm này mô tả một bề mặt có hình dạng như chiếc bát, được gọi là **mặt paraboloid (paraboloid)**, mở lên trên. Điểm thấp nhất của bề mặt nằm tại $(x,y)=(0,0)$, nơi: $f(0,0)=0$



> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/paraboloid_surface.html) ·


> Đồ thị bề mặt của $z=x^2+y^2$. Hai đầu vào $x$ và $y$ tạo thành mặt phẳng đáy, còn độ cao $z$ biểu diễn giá trị đầu ra của hàm. Hãy chú ý rằng giá trị nhỏ nhất xuất hiện tại gốc tọa độ $(0,0)$.



Vậy còn những hàm có nhiều hơn hai đầu vào thì sao? Để biểu diễn trực quan hàm $f(x_1,x_2,x_3)$, ta cần 4 chiều (3 chiều cho đầu vào và 1 chiều cho đầu ra). Tương tự, hàm $f(x_1,\ldots,x_n)$ cần $n+1$ chiều. Bộ não con người không được trang bị để trực tiếp hình dung những không gian có nhiều hơn ba chiều!

Tuy nhiên, các khái niệm toán học mà chúng ta xây dựng vẫn hoạt động tốt trong không gian có nhiều chiều hơn. Ngay cả khi không thể vẽ hình một hàm chi phí phụ thuộc vào hàng nghìn tham số của mô hình, ta vẫn có thể phân tích xem chi phí thay đổi như thế nào khi điều chỉnh các tham số đó.



## **3. Tại sao điều này quan trọng? (Why This Matters)**

Hiểu về **hàm số nhiều biến** là điều cần thiết vì hầu hết các mô hình học máy đều có nhiều tham số có thể điều chỉnh (chẳng hạn như trọng số và độ lệch trong mạng nơ-ron) và thường xử lý dữ liệu đầu vào gồm nhiều đặc trưng. Vì vậy, các **hàm chi phí**, dùng để đo lường mức độ hoạt động của mô hình, tự nhiên sẽ là những hàm phụ thuộc vào nhiều biến (chính là các tham số).

Khi huấn luyện mô hình, mục tiêu của chúng ta thường là tìm cách giảm thiểu hàm chi phí này. Vì chi phí phụ thuộc vào nhiều tham số, ta cần một cách để xác định việc thay đổi từng tham số riêng lẻ ảnh hưởng đến chi phí như thế nào. Đây chính là lúc khái niệm **đạo hàm riêng (partial derivatives)** phát huy tác dụng. Chúng ta sẽ tìm hiểu khái niệm này trong phần tiếp theo. Đạo hàm riêng cho phép ta xem xét tốc độ thay đổi theo một biến tại mỗi lần, ngay cả khi hàm số còn phụ thuộc vào nhiều biến khác.
