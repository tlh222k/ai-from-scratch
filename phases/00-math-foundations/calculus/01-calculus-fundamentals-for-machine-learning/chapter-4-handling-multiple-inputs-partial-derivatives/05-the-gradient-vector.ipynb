# **The Gradient Vector**

**Đạo hàm riêng (partial derivatives)** giúp chúng ta hiểu cách các hàm số có nhiều đầu vào, chẳng hạn $f(x,y)$, thay đổi như thế nào.

Ví dụ, $\frac{\partial f}{\partial x}$ mô tả cách $f$ thay đổi khi $x$ thay đổi, trong khi $y$ được giữ cố định.

Tương tự, $\frac{\partial f}{\partial y}$ cho biết cách $f$ thay đổi khi $y$ thay đổi, còn $x$ được giữ nguyên.

Nhưng nếu chúng ta muốn có một đối tượng duy nhất thể hiện tốc độ thay đổi theo tất cả các biến đầu vào cùng một lúc thì sao? Đó chính là lúc **vector gradient (gradient vector)** phát huy tác dụng.


**Gradient** của một hàm số $f$ đơn giản là một vector mà mỗi thành phần của nó là một đạo hàm riêng của $f$. Nếu hàm số có hai đầu vào là $x$ và $y$, **gradient** của nó là một vector hai chiều. Nếu hàm số có $n$ đầu vào là $x_1,x_2,\ldots,x_n$, **gradient** của nó là một vector có $n$ chiều.

Thông thường, chúng ta biểu diễn **gradient** bằng ký hiệu nabla $\nabla$. Với hàm số $f(x,y)$, **gradient** được viết là $\nabla f$ hoặc $\nabla f(x,y)$, và được xác định như sau:

$$
\nabla f(x,y)=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
$$

Đôi khi, bạn cũng có thể thấy **gradient** được viết theo chiều ngang bằng dấu ngoặc nhọn:

$$
\nabla f(x,y)=
\left\langle
\frac{\partial f}{\partial x},
\frac{\partial f}{\partial y}
\right\rangle
$$

Đối với hàm số có $n$ biến, $f(x_1,x_2,\ldots,x_n)$, **gradient** được viết như sau:

$$
\nabla f=
\begin{bmatrix}
\frac{\partial f}{\partial x_1}\\
\frac{\partial f}{\partial x_2}\\
\vdots\\
\frac{\partial f}{\partial x_n}
\end{bmatrix}
$$

Về cơ bản, **gradient** tập hợp tất cả các đạo hàm riêng cấp một vào trong một vector duy nhất, thuận tiện cho việc sử dụng.





### **Ví dụ: Tính gradient (Example: Calculating a Gradient)**

Xét hàm số:

$$f(x,y)=x^2+5xy$$

Hãy tìm các đạo hàm riêng:

**Xem $y$ là hằng số để tính $\frac{\partial f}{\partial x}$:**

$$
\begin{aligned}
\frac{\partial f}{\partial x}
&=\frac{\partial}{\partial x}(x^2)
+\frac{\partial}{\partial x}(5xy)\\
&=2x+5y
\end{aligned}
$$

**Xem $x$ là hằng số để tính $\frac{\partial f}{\partial y}$:**

$$
\begin{aligned}
\frac{\partial f}{\partial y}
&=\frac{\partial}{\partial y}(x^2)
+\frac{\partial}{\partial y}(5xy)\\
&=0+5x\\
&=5x
\end{aligned}
$$

**Ghép các đạo hàm riêng thành vector gradient:**

$$
\begin{aligned}
\nabla f(x,y)
&=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}\\
&=
\begin{bmatrix}
2x+5y\\
5x
\end{bmatrix}
\end{aligned}
$$

**Gradient** này, $\nabla f(x,y)$, cho chúng ta một vector phụ thuộc vào điểm $(x,y)$ mà chúng ta đang xét. Chẳng hạn, tại điểm $(1,2)$:

$$
\begin{aligned}
\nabla f(1,2)
&=
\begin{bmatrix}
2(1)+5(2)\\
5(1)
\end{bmatrix}\\
&=
\begin{bmatrix}
2+10\\
5
\end{bmatrix}\\
&=
\begin{bmatrix}
12\\
5
\end{bmatrix}
\end{aligned}
$$

Tại điểm $(-1,0)$:

$$
\begin{aligned}
\nabla f(-1,0)
&=
\begin{bmatrix}
2(-1)+5(0)\\
5(-1)
\end{bmatrix}\\
&=
\begin{bmatrix}
-2\\
-5
\end{bmatrix}
\end{aligned}
$$

Như vậy, bản thân **gradient** cũng là một hàm số: nó nhận một điểm $(x,y)$ làm đầu vào và trả về một vector. Vector này chứa thông tin quan trọng về cách hàm số $f$ hoạt động xung quanh điểm cụ thể đó. Chúng ta sẽ tìm hiểu kỹ hơn về điều này trong phần tiếp theo.

Trong **Machine Learning**, **gradient** của một hàm chi phí cho chúng ta biết cần điều chỉnh các tham số của mô hình (chẳng hạn các đầu vào $x_1,x_2,\ldots,x_n$) như thế nào để làm thay đổi chi phí.
