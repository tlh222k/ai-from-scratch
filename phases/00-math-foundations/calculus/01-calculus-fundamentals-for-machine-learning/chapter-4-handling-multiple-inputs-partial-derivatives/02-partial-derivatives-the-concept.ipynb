# **Đạo hàm riêng: Khái niệm (Partial Derivatives: The Concept)**

Các hàm số có nhiều hơn một đầu vào đóng vai trò quan trọng trong việc mô hình hóa những hiện tượng phức tạp. Ví dụ, để dự đoán giá nhà, chúng ta cần xem xét nhiều yếu tố. Giá nhà không chỉ phụ thuộc vào diện tích căn nhà ($x_1$), mà còn phụ thuộc vào số phòng ngủ ($x_2$), tuổi của căn nhà ($x_3$) và khoảng cách từ căn nhà đến trung tâm thành phố ($x_4$). Hàm dự đoán trong trường hợp này có thể được viết như sau:

$$Price=f(x_1,x_2,x_3,x_4)$$

Khi có một hàm số như $f(x,y)$, chúng ta thường muốn biết: **đầu ra sẽ thay đổi như thế nào nếu chỉ điều chỉnh $x$, còn $y$ giữ nguyên?** Hoặc, **đầu ra sẽ thay đổi như thế nào nếu chỉ điều chỉnh $y$, còn $x$ giữ nguyên?** Đây chính là điều mà **đạo hàm riêng (partial derivatives)** giúp chúng ta tìm hiểu.

Hãy tưởng tượng bạn đang đứng trên một sườn đồi. Độ cao tại vị trí bạn đứng (giá trị đầu ra, ký hiệu là $z$) phụ thuộc vào vị trí của bạn theo hai hướng: bạn đi về phía Đông bao xa (ký hiệu là $x$) và đi về phía Bắc bao xa (ký hiệu là $y$). Như vậy:

$$z=f(x,y)$$

Nếu muốn biết sườn đồi dốc như thế nào theo hướng Đông, bạn chỉ cần xem độ cao thay đổi ra sao khi di chuyển về phía Đông hoặc phía Tây, mà không di chuyển về phía Bắc hoặc phía Nam. Nói cách khác, tại thời điểm đó, bạn xem vị trí theo hướng Bắc – Nam ($y$) là không đổi và chỉ đo độ dốc theo hướng Đông – Tây ($x$).

Đó chính là ý tưởng cốt lõi của **đạo hàm riêng (partial derivatives)**.

**Đạo hàm riêng** đo tốc độ thay đổi của một hàm số nhiều biến theo một biến cụ thể, trong khi giữ tất cả các biến còn lại không đổi. Nó cho biết độ dốc của hàm số theo một hướng song song với một trong các trục đầu vào.

Thay vì dùng ký hiệu $d$ như trong $\frac{df}{dx}$, vốn được sử dụng với **hàm số một biến (single-variable functions)**, chúng ta dùng một ký hiệu có dạng cong là $\partial$, được gọi là "del" hoặc đơn giản là ký hiệu đạo hàm riêng.

- Đạo hàm riêng của $f$ theo $x$ được viết là $\frac{\partial f}{\partial x}$, đôi khi cũng được viết là $f_x$. Ký hiệu này cho biết $f$ thay đổi như thế nào khi chỉ có $x$ thay đổi.
- Đạo hàm riêng của $f$ theo $y$ được viết là $\frac{\partial f}{\partial y}$, đôi khi cũng được viết là $f_y$. Ký hiệu này cho biết $f$ thay đổi như thế nào khi chỉ có $y$ thay đổi.

Ví dụ, hàm số:

$$f(x,y)=x^2+y^2$$

mô tả một bề mặt có hình dạng như chiếc bát trong không gian 3D.

> *Đồ thị của hàm số $f(x,y)=x^2+y^2$. Độ cao của bề mặt biểu thị giá trị đầu ra của hàm số.*



Nếu muốn tìm $\frac{\partial f}{\partial x}$ tại một điểm cụ thể, chẳng hạn $(1,1)$, chúng ta đang đặt câu hỏi: nếu đứng tại điểm trên bề mặt có $x=1$ và $y=1$, rồi chỉ di chuyển theo hướng $x$ (giữ $y=1$ không đổi), thì độ dốc của bề mặt tại đó là bao nhiêu?

Để hiểu cách tính $\frac{\partial f}{\partial x}$, chúng ta xem biến $y$ như một hằng số. Tương tự, khi tính $\frac{\partial f}{\partial y}$, chúng ta xem $x$ là một hằng số.

Các quy tắc tính đạo hàm thông thường, chẳng hạn **Quy tắc lũy thừa (Power Rule)** mà bạn đã học trước đó, vẫn được áp dụng. Tuy nhiên, chúng ta chỉ áp dụng các quy tắc này cho biến đang lấy đạo hàm; những biến còn lại được xem như các số bình thường. Trong phần tiếp theo, chúng ta sẽ tìm hiểu cụ thể cách thực hiện các phép tính này.

Điều quan trọng nhất ở đây là **tách riêng ảnh hưởng của từng biến**: đạo hàm riêng giúp xác định mức độ ảnh hưởng của việc thay đổi một đầu vào duy nhất đến đầu ra của hàm số. Đây là nền tảng để hiểu cách các mô hình phức tạp phản ứng khi những đặc trưng hoặc tham số khác nhau thay đổi. Đồng thời, đạo hàm riêng cũng là nền tảng để xây dựng **gradient**, một khái niệm mà chúng ta sẽ tìm hiểu trong phần tiếp theo.
