# **Ký hiệu đạo hàm riêng (Partial Derivative Notation)**

Đạo hàm của các hàm số một biến, chẳng hạn $f(x)$, thường được biểu diễn bằng những ký hiệu như $\frac{df}{dx}$ hoặc $f'(x)$. Đối với các hàm số có nhiều biến, chẳng hạn $f(x,y)$ hoặc $f(x_1,x_2,\ldots,x_n)$, chúng ta cần chỉ rõ chính xác biến nào đang được lấy đạo hàm, đồng thời giữ tất cả các biến còn lại không đổi.



## **1. Ký hiệu đạo hàm riêng: ∂ (The Partial Derivative Symbol: ∂)**

Để phân biệt **đạo hàm riêng (partial derivatives)** với đạo hàm thông thường mà chúng ta đã học trước đó, chúng ta sử dụng một ký hiệu khác: $\partial$.

Ký hiệu này thường được gọi là "chữ d cong", "del" hoặc đơn giản là ký hiệu đạo hàm riêng. Nó cho biết chúng ta đang thực hiện phép lấy đạo hàm riêng.




## **2. Ký hiệu theo Leibniz (Leibniz-Style Notation)**

Ký hiệu phổ biến nhất mà bạn sẽ gặp, tương tự như ký hiệu $\frac{dy}{dx}$ dùng cho đạo hàm thông thường, sử dụng ký hiệu $\partial$.

Với hàm số $f(x,y)$, đạo hàm riêng theo $x$ được viết như sau:

$$\frac{\partial f}{\partial x}$$

Ký hiệu này được đọc là "đạo hàm riêng của $f$ theo $x$". Nó cho chúng ta biết rõ những điều sau:

- $\partial f$: Chúng ta đang xem xét sự thay đổi của hàm số $f$.
- $\partial x$: Chúng ta xem xét sự thay đổi đó khi biến $x$ thay đổi.
- Ngầm hiểu rằng: Tất cả các biến còn lại (trong trường hợp này chỉ có $y$) được xem là hằng số trong quá trình lấy đạo hàm.

Tương tự, đạo hàm riêng của $f$ theo $y$ được viết như sau:

$$\frac{\partial f}{\partial y}$$

Nếu giá trị đầu ra của hàm số được gán cho một biến khác, chẳng hạn $z=f(x,y)$, bạn cũng có thể thấy các ký hiệu được viết là $\frac{\partial z}{\partial x}$ hoặc $\frac{\partial z}{\partial y}$.




## **3. Ký hiệu chỉ số dưới (Subscript Notation)**

Một cách ký hiệu phổ biến khác, thường ngắn gọn hơn, là sử dụng chỉ số dưới để cho biết biến đang được lấy đạo hàm. Với cùng hàm số $f(x,y)$:

- Đạo hàm riêng theo $x$ có thể được viết là $f_x$ hoặc $f_x(x,y)$.
- Đạo hàm riêng theo $y$ có thể được viết là $f_y$ hoặc $f_y(x,y)$.

Cách ký hiệu này đặc biệt thuận tiện khi viết **vector gradient (gradient vector)**, một khái niệm mà chúng ta sẽ sớm tìm hiểu.



### **Ví dụ**

Hãy xem xét một hàm số đơn giản:

$$g(a,b)=3a^2b+5b^3$$

- Đạo hàm riêng của $g$ theo $a$:
  - Sử dụng ký hiệu theo Leibniz, là $\frac{\partial g}{\partial a}$.
  - Sử dụng ký hiệu chỉ số dưới, ta viết là $g_a$.

- Đạo hàm riêng của $g$ theo $b$:
  - Sử dụng ký hiệu theo Leibniz, là $\frac{\partial g}{\partial b}$.
  - Sử dụng ký hiệu chỉ số dưới, ta viết là $g_b$.

(Chúng ta sẽ học cách tính các đạo hàm này trong phần tiếp theo. Còn bây giờ, hãy tập trung hiểu ý nghĩa của các ký hiệu.)



## **4. Tính giá trị tại một điểm (Evaluating at a Point)**

Cũng giống như với đạo hàm thông thường, chúng ta thường muốn tính giá trị của một đạo hàm riêng tại một điểm cụ thể.

Nếu muốn tìm đạo hàm riêng của $f(x,y)$ theo $x$ tại điểm $(x_0,y_0)$, chúng ta có thể viết theo một trong các cách sau:

$$ \left.\frac{\partial f}{\partial x}\right|_{(x_0,y_0)} \quad \text{hoặc} \quad \frac{\partial f}{\partial x}(x_0,y_0) \quad \text{hoặc} \quad f_x(x_0,y_0) $$


Tất cả các ký hiệu này đều có cùng ý nghĩa: tính biểu thức đạo hàm riêng theo $x$, sau đó thay $x=x_0$ và $y=y_0$ vào kết quả.

----

Hiểu được các ký hiệu này là điều rất quan trọng, vì đây chính là ngôn ngữ được sử dụng để mô tả cách các hàm số phức tạp, chẳng hạn như hàm chi phí trong các mô hình học máy, thay đổi khi từng đầu vào hoặc tham số riêng lẻ của chúng được điều chỉnh.
