# **Luyện tập: Tính đạo hàm riêng và vector gradient (Practice: Calculating Partial Derivatives and Gradients)**

**Đạo hàm riêng (partial derivatives)** mô tả tốc độ thay đổi của một hàm số nhiều biến theo một biến, trong khi giữ các biến còn lại không đổi. **Vector gradient (gradient vector)** kết hợp các đạo hàm riêng này và chỉ hướng mà hàm số tăng nhanh nhất. Phần này cung cấp các bài tập tính đạo hàm riêng và **gradient** cho nhiều hàm số khác nhau.

Hãy nhớ rằng, quá trình tính toán này chủ yếu dựa vào các quy tắc đạo hàm mà bạn đã học trước đó (chẳng hạn như **Quy tắc lũy thừa (Power Rule)** và **Quy tắc tổng (Sum Rule)**). Chỉ có thêm một bước: xem một số biến nhất định là hằng số trong khi tính đạo hàm.



## **Ví dụ 1: Hàm đa thức đơn giản (A Simple Polynomial Function)**

Trước tiên, hãy xét một hàm số có hai biến $x$ và $y$:

$$
f(x,y)=2x^3+4xy^2-y^5+7
$$

Mục tiêu của chúng ta là tìm các đạo hàm riêng $\frac{\partial f}{\partial x}$ và $\frac{\partial f}{\partial y}$, sau đó lập **vector gradient** $\nabla f(x,y)$.



### **Tính $\frac{\partial f}{\partial x}$ (Calculating Partial Derivative with Respect to $x$)**

Để tìm $\frac{\partial f}{\partial x}$, ta xem $y$ như một hằng số, giống như một số cố định (chẳng hạn 5, -2 hoặc $\pi$).

- **Hạng tử 1 ($2x^3$):** Đạo hàm của $2x^3$ theo $x$ là:

  $$
  2\cdot(3x^{3-1})=6x^2
  $$

  (Áp dụng **Quy tắc lũy thừa** thông thường).

- **Hạng tử 2 ($4xy^2$):** Ở đây, ta xem cả 4 và $y^2$ là các hằng số được nhân với $x$. Đạo hàm theo $x$ của (hằng số $\times x$) chính là hằng số đó. Vì vậy, đạo hàm của $4xy^2$ theo $x$ là:

  $$
  4y^2
  $$

- **Hạng tử 3 ($-y^5$):** Vì ta xem $y$ là hằng số nên $y^5$ cũng là hằng số. Đạo hàm theo $x$ của bất kỳ hằng số nào đều bằng 0.

- **Hạng tử 4 ($7$):** Đây là một hằng số nên đạo hàm theo $x$ bằng 0.

Gộp các kết quả lại:

$$
\begin{aligned}
\frac{\partial f}{\partial x}
&=6x^2+4y^2+0+0\\
&=6x^2+4y^2
\end{aligned}
$$



### **Tính $\frac{\partial f}{\partial y}$ (Calculating Partial Derivative with Respect to $y$)**

Bây giờ, ta đổi cách xét. Để tìm $\frac{\partial f}{\partial y}$, ta xem $x$ như một hằng số.

- **Hạng tử 1 ($2x^3$):** Vì xem $x$ là hằng số nên $2x^3$ cũng là hằng số. Đạo hàm của nó theo $y$ bằng 0.
- **Hạng tử 2 ($4xy^2$):** Xem $4x$ là hệ số hằng số của $y^2$. Đạo hàm theo $y$ của (hằng số $\times y^2$) là (hằng số $\times 2y$). Do đó, đạo hàm là:

  $$
  4x\cdot(2y)=8xy
  $$

- **Hạng tử 3 ($-y^5$):** Đạo hàm của $-y^5$ theo $y$ là:

  $$
  -5y^{5-1}=-5y^4
  $$

  (Áp dụng **Quy tắc lũy thừa** thông thường).

- **Hạng tử 4 ($7$):** Đây là một hằng số nên đạo hàm theo $y$ bằng 0.

Gộp các kết quả lại:

$$
\begin{aligned}
\frac{\partial f}{\partial y}
&=0+8xy-5y^4+0\\
&=8xy-5y^4
\end{aligned}
$$



### **Lập vector gradient $\nabla f(x,y)$ (Forming the Gradient Vector)**

**Vector gradient** đơn giản là một vector chứa các đạo hàm riêng. Theo quy ước, ta liệt kê chúng theo thứ tự của các biến ($x$ trước, rồi đến $y$):

$$
\nabla f(x,y)=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
=
\begin{bmatrix}
6x^2+4y^2\\
8xy-5y^4
\end{bmatrix}
$$



### **Tính giá trị gradient tại một điểm (Evaluating the Gradient at a Point)**

Bản thân **gradient** là một hàm số; nó cho ta biết hướng tăng nhanh nhất tại bất kỳ điểm $(x,y)$ nào. Hãy tính **gradient** tại điểm cụ thể $(x,y)=(1,2)$.

Ta thay $x=1$ và $y=2$ vào các biểu thức đạo hàm riêng:

- Giá trị của $\frac{\partial f}{\partial x}$ tại $(1,2)$:

  $$
  \begin{aligned}
  6(1)^2+4(2)^2
  &=6(1)+4(4)\\
  &=6+16\\
  &=22
  \end{aligned}
  $$

- Giá trị của $\frac{\partial f}{\partial y}$ tại $(1,2)$:

  $$
  \begin{aligned}
  8(1)(2)-5(2)^4
  &=16-5(16)\\
  &=16-80\\
  &=-64
  \end{aligned}
  $$

Vậy **vector gradient** tại điểm $(1,2)$ là:

$$
\nabla f(1,2)=
\begin{bmatrix}
22\\
-64
\end{bmatrix}
$$

Vector này cho biết rằng nếu bắt đầu từ điểm $(1,2)$ trên bề mặt được xác định bởi hàm số $f(x,y)$, hướng tăng nhanh nhất chủ yếu theo chiều dương của trục $x$ (giá trị 22) và theo chiều âm của trục $y$ (giá trị -64) với mức độ mạnh hơn.

Trong **tối ưu hóa (optimization)**, chẳng hạn như khi sử dụng **Gradient Descent**, chúng ta thường di chuyển theo hướng ngược lại, $-\nabla f(1,2)$, để làm giảm giá trị của hàm số.



## **Ví dụ 2: Hàm số có hạng tử tương tác (Function with Interaction Term)**

Hãy thử thêm một ví dụ nữa. Xét hàm số $g(w_1,w_2)$, có thể biểu diễn một hàm chi phí đơn giản phụ thuộc vào hai trọng số $w_1$ và $w_2$:

$$
g(w_1,w_2)=3w_1^2-5w_1w_2+2w_2^2
$$



### **Tính $\frac{\partial g}{\partial w_1}$ (Calculating $\frac{\partial g}{\partial w_1}$)**

Xem $w_2$ là một hằng số.

- **Hạng tử 1 ($3w_1^2$):** Đạo hàm theo $w_1$ là:

  $$
  3(2w_1)=6w_1
  $$

- **Hạng tử 2 ($-5w_1w_2$):** Xem $-5w_2$ là hệ số hằng số của $w_1$. Đạo hàm theo $w_1$ là:

  $$
  -5w_2
  $$

- **Hạng tử 3 ($2w_2^2$):** Xem $w_2$ là hằng số nên $2w_2^2$ cũng là hằng số. Đạo hàm theo $w_1$ bằng 0.

Kết quả:

$$
\frac{\partial g}{\partial w_1}=6w_1-5w_2
$$



### **Tính $\frac{\partial g}{\partial w_2}$ (Calculating $\frac{\partial g}{\partial w_2}$)**

Xem $w_1$ là một hằng số.

- **Hạng tử 1 ($3w_1^2$):** Xem $w_1$ là hằng số nên $3w_1^2$ cũng là hằng số. Đạo hàm theo $w_2$ bằng 0.
- **Hạng tử 2 ($-5w_1w_2$):** Xem $-5w_1$ là hệ số hằng số của $w_2$. Đạo hàm theo $w_2$ là:

  $$
  -5w_1
  $$

- **Hạng tử 3 ($2w_2^2$):** Đạo hàm theo $w_2$ là:

  $$
  2(2w_2)=4w_2
  $$

Kết quả:

$$
\frac{\partial g}{\partial w_2}=-5w_1+4w_2
$$



### **Lập vector gradient $\nabla g(w_1,w_2)$ (Forming the Gradient Vector)**

$$
\nabla g(w_1,w_2)=
\begin{bmatrix}
\frac{\partial g}{\partial w_1}\\
\frac{\partial g}{\partial w_2}
\end{bmatrix}
=
\begin{bmatrix}
6w_1-5w_2\\
-5w_1+4w_2
\end{bmatrix}
$$



### **Tính giá trị gradient tại một điểm (Evaluating the Gradient at a Point)**

Hãy tính **gradient** tại điểm $(w_1,w_2)=(2,-1)$.

- Giá trị của $\frac{\partial g}{\partial w_1}$ tại $(2,-1)$:

  $$
  6(2)-5(-1)=12+5=17
  $$

- Giá trị của $\frac{\partial g}{\partial w_2}$ tại $(2,-1)$:

  $$
  -5(2)+4(-1)=-10-4=-14
  $$

Vậy **vector gradient** tại điểm $(2,-1)$ là:

$$
\nabla g(2,-1)=
\begin{bmatrix}
17\\
-14
\end{bmatrix}
$$



## **Tại sao cần luyện tập? (Why Practice This?)**

Tính đạo hàm riêng và **gradient** là kỹ năng tính toán cơ bản, cần thiết để hiểu và triển khai các thuật toán tối ưu hóa trong học máy.

Khi huấn luyện mô hình, chúng ta thường có một hàm chi phí phụ thuộc vào nhiều tham số (trọng số và độ chệch). **Gradient Descent** sử dụng **gradient** của hàm chi phí để cập nhật các tham số theo từng bước, di chuyển theo hướng làm giảm giá trị hàm chi phí.
