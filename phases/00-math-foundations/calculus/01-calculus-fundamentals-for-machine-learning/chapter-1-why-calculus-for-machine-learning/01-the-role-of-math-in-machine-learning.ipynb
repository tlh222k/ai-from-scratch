# **Vai trò của Toán học trong Machine Learning (The Role of Math in Machine Learning)**

Bạn có thể đã nghe nói rằng **Machine Learning** thường liên quan đến những kiến thức toán học phức tạp, và điều đó hoàn toàn đúng. Nhưng tại sao lại như vậy? Tại sao chúng ta không thể đơn giản xem các **Model** trong **Machine Learning** như những **hộp đen (Black Boxes)**, chỉ cần đưa dữ liệu vào và nhận kết quả đầu ra? Mặc dù các thư viện và framework đã trừu tượng hóa nhiều chi tiết, việc hiểu những nguyên lý toán học nền tảng vẫn có ý nghĩa quan trọng vì một số lý do.

**Machine Learning** là quá trình tìm kiếm **các mẫu hình (patterns)**  trong dữ liệu và sử dụng những **patterns** đó để đưa ra dự đoán hoặc quyết định. Các phần **patterns** và “đưa ra dự đoán” được định nghĩa và thực hiện bằng toán học. Hãy hình dung một **Model** trong **Machine Learning** như một **hàm toán học (Mathematical Function)** tinh vi. Nó tiếp nhận đầu vào (dữ liệu của bạn, chẳng hạn như các **Features** của một ngôi nhà) và tạo ra đầu ra (một dự đoán, chẳng hạn như giá của ngôi nhà đó).

Toán học cung cấp ngôn ngữ để:

1. **Biểu diễn dữ liệu (Represent Data):** Dữ liệu cần được cấu trúc dưới dạng số (thường là các **Vectors** hoặc **Matrices**) để các thuật toán có thể xử lý. **Linear Algebra** đóng vai trò nền tảng ở đây, mặc dù chúng ta sẽ không tập trung quá nhiều vào lĩnh vực này trong khóa học này.


2. **Xác định mô hình (Define Models):** Cách thức hoạt động bên trong của một **Model**, chẳng hạn như đường hồi quy tuyến tính (`y=mx+b`) hoặc các tầng phức tạp của một mạng neuron, được mô tả bằng những phương trình và cấu trúc toán học.


3. **Đo lường hiệu suất (Measure Performance):** Chúng ta cần một phương pháp để định lượng mức độ hoạt động của một **Model**. Dự đoán của nó có gần với giá trị thực tế hay không? Các khái niệm như **Error** hoặc **Loss Function** là những công cụ toán học được sử dụng để đo lường điều này.


4. **Cải thiện mô hình – Tối ưu hóa (Improve Models – Optimization):** Đây là lúc **Calculus** giữ vai trò trung tâm. Việc tìm ra phiên bản “tốt nhất” của một **Model** thường liên quan đến việc giảm thiểu sai số. **Calculus** cung cấp cho chúng ta các công cụ (như **Derivatives** và **Gradients**) để điều chỉnh một cách có hệ thống các **parameters** của **Model** (chẳng hạn như $m$ và $b$ trong phương trình $y=mx+b$), nhằm đạt được mức sai số thấp nhất có thể. Quá trình điều chỉnh này được gọi là **Optimization** và là nền tảng của quá trình huấn luyện các **Model** trong **Machine Learning**.



Mặc dù bạn *có thể* sử dụng các thư viện **Machine Learning** được xây dựng sẵn mà không cần hiểu biết toán học chuyên sâu, nhưng việc nắm vững những kiến thức nền tảng sẽ giúp bạn:

- **Lựa chọn thuật toán phù hợp:** Những giả định toán học khác nhau là cơ sở của các **Model** khác nhau. Hiểu được những giả định này giúp bạn lựa chọn công cụ thích hợp cho nhiệm vụ của mình.

- **Điều chỉnh các parameters của Model:** Những khái niệm như **Learning Rate** trong quá trình **Optimization** có liên hệ trực tiếp với **Calculus**. Hiểu biết về toán học giúp bạn nắm được cách việc điều chỉnh các **parameters** này ảnh hưởng đến quá trình huấn luyện **Model**.

- **Gỡ lỗi và xử lý vấn đề:** Khi một **Model** không hoạt động như mong đợi, việc hiểu nền tảng toán học của nó có thể giúp bạn chẩn đoán những vấn đề liên quan đến **Data Scaling**, **Vanishing Gradients** hoặc **Poor Convergence**.


- **Hiểu các bài báo nghiên cứu:** Những tiến bộ trong lĩnh vực **Machine Learning** thường được trình bày bằng ký hiệu và khái niệm toán học.

Khóa học này tập trung cụ thể vào những kiến thức nền tảng về **Calculus** cần thiết cho phần **Optimization**. Đây được xem là một trong những khía cạnh toán học quan trọng nhất đối với quá trình *huấn luyện* nhiều **Model** phổ biến trong **Machine Learning**. Đừng lo lắng nếu những nội dung này nghe có vẻ khó tiếp cận; chúng ta sẽ lần lượt tìm hiểu các khái niệm cần thiết như **Functions**, **Limits** và **Derivatives**, bắt đầu từ những kiến thức cơ bản mà bạn cần nắm vững để hiểu cách máy móc học từ dữ liệu.
