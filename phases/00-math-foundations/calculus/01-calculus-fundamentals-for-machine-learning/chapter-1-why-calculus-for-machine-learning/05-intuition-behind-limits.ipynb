# **Trực giác về giới hạn (Intuition Behind Limits)**

**Giới hạn (Limit)** mô tả giá trị mà một **hàm số (Function)** tiến gần đến khi **đầu vào (Input)** của nó ngày càng tiến gần đến một điểm nhất định. Trên thực tế, điều này thực sự có nghĩa là gì? Chúng ta sẽ xây dựng một số trực giác về khái niệm này.

Hãy tưởng tượng bạn đang quan sát **đồ thị (Graph)** của một hàm số, chẳng hạn như hàm số tuyến tính đơn giản $f(x)=2x+1$ mà chúng ta đã thấy trước đó. Hãy hình dung **Limit** đang đặt ra câu hỏi:
> "Nếu tôi dùng ngón tay lần theo đồ thị và tiến thật gần đến một giá trị đầu vào cụ thể (chẳng hạn $x=3$), thì trên trục tung, ngón tay của tôi dường như đang hướng đến giá trị nào (giá trị $f(x)$)?"

Hãy thử áp dụng điều này với $f(x)=2x+1$ khi $x$ tiến đến 3. Chúng ta không hỏi $f(3)$ *bằng bao nhiêu* (mặc dù trong trường hợp đơn giản này, giá trị đó là $2(3)+1=7$). Thay vào đó, chúng ta đang khảo sát **vùng lân cận (Neighborhood)** quanh $x=3$.

Ví dụ, xét các giá trị $x$ nhỏ hơn 3 một chút:

- Nếu $x=2.9$, thì $f(x)=2(2.9)+1=5.8+1=6.8$
- Nếu $x=2.99$, thì $f(x)=2(2.99)+1=5.98+1=6.98$
- Nếu $x=2.999$, thì $f(x)=2(2.999)+1=5.998+1=6.998$

Bây giờ, hãy xét các giá trị $x$ lớn hơn 3 một chút:

- Nếu $x=3.1$, thì $f(x)=2(3.1)+1=6.2+1=7.2$
- Nếu $x=3.01$, thì $f(x)=2(3.01)+1=6.02+1=7.02$
- Nếu $x=3.001$, thì $f(x)=2(3.001)+1=6.002+1=7.002$

Bạn có nhận thấy quy luật nào không? Khi $x$ ngày càng tiến gần đến 3 từ cả hai phía, giá trị **$f(x)$** cũng ngày càng tiến gần đến 7. Giá trị 7 chính là **Limit** của hàm số $f(x)=2x+1$ khi $x$ tiến đến 3. Ta biểu diễn điều này bằng ký hiệu toán học như sau:

$$
\lim_{x\to 3}(2x+1)=7
$$

Đồ thị bên dưới giúp trực quan hóa điều này. Khi chúng ta chọn các điểm trên đường thẳng ngày càng gần đường thẳng đứng tại $x=3$, các điểm tương ứng trên hàm số cũng ngày càng tiến gần đến độ cao $y=7$.


> [Limit Approach](https://tlh222k.github.io/ai-from-scratch/visualizations/limit_approach_function.html) ·



> *Khi các giá trị **$x$** tiến gần đến 3 từ cả hai phía: bên trái (chẳng hạn 2.9, 2.99) và bên phải (chẳng hạn 3.1, 3.01), các giá trị **đầu ra (Output)** **$f(x)$** tiến gần đến 7. **Limit** mô tả giá trị đích này, tức là 7.*



Phần quan trọng trong định nghĩa là cụm từ **"tiến gần đến một giá trị *mà không nhất thiết phải đạt tới giá trị đó*"**.

Đối với nhiều hàm số đơn giản như $f(x)=2x+1$, **Limit** khi $x$ tiến đến một điểm $a$ chính là giá trị của hàm số *tại* điểm đó, $f(a)$. Tuy nhiên, khái niệm **Limit** có tính tổng quát hơn. Nó cho phép chúng ta phân tích hành vi của hàm số ngay cả tại những điểm mà hàm số có thể không xác định (chẳng hạn như có một **điểm khuyết (Hole)** trên đồ thị, mặc dù chúng ta sẽ không tập trung vào những trường hợp phức tạp đó ở đây). **Limit** quan tâm đến *xu hướng* của **Output** khi **Input** tiến gần tùy ý đến một giá trị cụ thể.

Tại sao chúng ta cần dành thời gian tìm hiểu ý tưởng "tiến gần" này? Bởi vì nó tạo thành nền tảng để hiểu về **đạo hàm (Derivative)**, đại lượng đo lường tốc độ thay đổi *tức thời*. Chúng ta cần ý tưởng tiến gần vô hạn đến một điểm để có thể nói về độ dốc hoặc sự thay đổi *ngay tại điểm đó*. Chúng ta sẽ thấy rõ mối liên hệ này trong chương tiếp theo.
