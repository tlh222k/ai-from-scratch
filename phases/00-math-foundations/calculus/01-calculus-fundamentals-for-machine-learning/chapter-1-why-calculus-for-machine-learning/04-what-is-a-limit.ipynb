# **Giới hạn là gì? (What is a Limit?)**

**Hàm số (Function)** mô tả mối quan hệ giữa **đầu vào (Input)** và **đầu ra (Output)**, chẳng hạn như $y=f(x)$. Một khái niệm nền tảng trong **Giải tích (Calculus)** là **Giới hạn (Limit)**. **Limit** giúp chúng ta hiểu cách một **Function** hoạt động khi một giá trị **Input** tiến gần đến một giá trị cụ thể.

Hãy tưởng tượng bạn đang đi bộ về phía một điểm xác định trên con đường. **Limit** giống như việc đặt câu hỏi:

> "Khi tôi tiến ngày càng gần đến điểm đó, dường như tôi sẽ đến vị trí nào, ngay cả khi tôi không thực sự bước lên chính xác vị trí đó?"



## **1. Tại sao lại là "Tiến gần đến"? (Why "Close To"?)**

Đôi khi, chúng ta có thể thay trực tiếp một giá trị **Input** vào một **Function** để xác định **Output**. Nếu có: $ f(x)=x+2 $ và muốn biết điều gì xảy ra tại $x=3$, chúng ta tính: $f(3)=3+2=5$ .Điều này khá đơn giản.

Nhưng điều gì sẽ xảy ra nếu **Function** không được xác định tại điểm cụ thể mà chúng ta đang quan tâm? Hoặc nếu chúng ta muốn hiểu xu hướng hay chiều hướng biến thiên của **Function** trong vùng lân cận điểm đó thì sao?

Đây chính là lúc **Limit** trở nên thiết yếu. Chúng cho phép chúng ta phân tích hành vi của **Function** khi **Input** tiến gần đến một giá trị, bất kể điều gì xảy ra chính xác tại giá trị đó.



## **2. Tiến gần đến một giá trị: Ví dụ (Approaching a Value: An Example)**

Hãy tiếp tục sử dụng **Function** đơn giản: $ f(x)=x+2 $. Chúng ta đã biết: $
f(3)=5 $.

Nhưng hãy giả sử chúng ta không biết điều đó. Thay vào đó, hãy xem $f(x)$ tiến gần đến giá trị nào khi $x$ ngày càng tiến gần đến 3.

Chúng ta có thể thử các giá trị của $x$ nhỏ hơn một chút so với 3:

- Nếu $x=2.9$, thì: $ f(x)=2.9+2=4.9 $
- Nếu $x=2.99$, thì: $ f(x)=2.99+2=4.99 $
- Nếu $x=2.999$, thì: $ f(x)=2.999+2=4.999 $

Và các giá trị lớn hơn 3 một chút:

- Nếu $x=3.1$, thì: $ f(x)=3.1+2=5.1 $
- Nếu $x=3.01$, thì: $ f(x)=3.01+2=5.01 $
- Nếu $x=3.001$, thì: $ f(x)=3.001+2=5.001 $

Bạn có nhận thấy quy luật nào không?

Khi $x$ ngày càng tiến gần đến 3 (từ cả hai phía), giá trị đầu ra $f(x)$ cũng ngày càng tiến gần đến 5. Chúng ta nói rằng: "**Limit** của **Function** $f(x)=x+2$ khi $x$ tiến đến 3 bằng 5."

Trong ký hiệu toán học, chúng ta viết:

$$
\lim_{x\to 3}(x+2)=5
$$

Hãy phân tích từng thành phần trong ký hiệu này:

- $\lim$: Cho biết chúng ta đang tìm **Limit**.
- $x\to 3$ (nằm bên dưới $\lim$): Có nghĩa là "khi giá trị **Input** $x$ tiến gần đến giá trị 3".
- $(x+2)$: Là **Function** mà chúng ta đang xem xét.
- $=5$: Biểu thị giá trị mà **Function** tiến gần đến.



## **3. Khi hàm số có một điểm khuyết (When the Function Has a Hole)**

Giá trị thực sự của **Limit** được thể hiện rõ khi một **Function** có khoảng gián đoạn hoặc không được xác định tại một điểm. Ví dụ, xét **Function** sau:

$$
g(x)=\frac{x^2-1}{x-1}
$$

Điều gì xảy ra nếu chúng ta thử thay $x=1$ vào **Function**? Chúng ta nhận được:
$ \frac{1^2-1}{1-1}=\frac{0}{0} $. Biểu thức này không xác định. Phép chia cho $0$ không được phép thực hiện. Vì vậy, $g(1)$ không tồn tại.

Tuy nhiên, chúng ta có thể rút gọn **Function** này bằng các phép biến đổi đại số đối với những giá trị $x$ khác 1. Vì: $ x^2-1=(x-1)(x+1) $, nên chúng ta có:


$$ g(x) = \frac{x-1}{(x-1)(x+1)} = x+1, \quad \text{provided } x \neq 1 $$

Do đó, **đồ thị (Graph)** của $g(x)$ trông hoàn toàn giống đồ thị của:$ f(x)=x+1 $, ngoại trừ một **điểm khuyết (Hole)** tại vị trí $x=1$.

Hãy đặt câu hỏi: **Limit** của $g(x)$ khi $x$ tiến đến 1 bằng bao nhiêu?

Chúng ta có thể sử dụng biểu thức đã rút gọn $x+1$, bởi vì **Limit** chỉ quan tâm đến các giá trị nằm gần $x=1$, chứ không quan tâm đến giá trị chính xác tại $x=1$.

Hãy kiểm tra các giá trị gần $x=1$:

- Nếu $x=0.9$, thì: $ g(x)=0.9+1=1.9 $
- Nếu $x=0.99$, thì: $ g(x)=0.99+1=1.99 $
- Nếu $x=1.1$, thì: $ g(x)=1.1+1=2.1 $
- Nếu $x=1.01$, thì: $ g(x)=1.01+1=2.01 $

Khi $x$ tiến gần đến 1 tùy ý, $g(x)$ cũng tiến gần đến 2 tùy ý. Mặc dù $g(1)$ không xác định, **Limit** vẫn tồn tại.

Chúng ta viết:

$$
\lim_{x\to 1}\frac{x^2-1}{x-1}=2
$$

Dưới đây là hình minh họa của **Function**: $ g(x)=\frac{x^2-1}{x-1} $.**Function** này có hành vi giống đường thẳng $y=x+1$, nhưng có một điểm khuyết tại $x=1$.



> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/limit_function.html)


> *Đồ thị của $g(x)=(x^2-1)/(x-1)$ là đường thẳng $y=x+1$ bị loại bỏ một điểm duy nhất tại $(1,2)$, được biểu thị bằng một vòng tròn rỗng. **Limit** khi $x$ tiến đến 1 là giá trị $y$ mà **Function** tiến gần đến, tức là 2.*

### **4. Tóm tắt ý tưởng cốt lõi (The Core Idea Summarized)**

Một **Limit** $L$ tồn tại đối với **Function** $f(x)$ khi $x$ tiến đến một giá trị nào đó $c$, nếu chúng ta có thể làm cho giá trị đầu ra $f(x)$ gần với $L$ tùy ý bằng cách chọn giá trị đầu vào $x$ đủ gần với $c$ (nhưng không thực sự bằng $c$).

**Function** thậm chí không cần phải được xác định *tại* $c$ thì **Limit** vẫn có thể tồn tại.

Ý tưởng về việc "tiến gần đến một giá trị tùy ý" chính là nền tảng để xây dựng khái niệm **Đạo hàm (Derivative)**.

Như chúng ta sẽ tìm hiểu trong chương tiếp theo, **Derivative** đo lường **tốc độ thay đổi tức thời (Instantaneous Rate of Change)**, và **Limit** cung cấp công cụ toán học cần thiết để định nghĩa hành vi "tức thời" này.

Hiểu rõ **Limit** cung cấp nền tảng để chúng ta hiểu cách các **Function** thay đổi, đây là yếu tố cơ bản trong việc tối ưu hóa các mô hình **Học máy (Machine Learning)**.
