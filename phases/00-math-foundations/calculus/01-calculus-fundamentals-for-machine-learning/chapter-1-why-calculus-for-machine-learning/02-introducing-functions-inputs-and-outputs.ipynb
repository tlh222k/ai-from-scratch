# **Giới thiệu về Hàm: Đầu vào và đầu ra (Introducing Functions: Inputs and Outputs)**

Trong **Máy học (Machine Learning)**, chúng ta thường tìm kiếm các mẫu và đưa ra dự đoán dựa trên dữ liệu. Để thực hiện điều này bằng toán học, chúng ta cần một cách để mô tả mối quan hệ giữa các đại lượng khác nhau. Đây chính là lúc **Hàm (Functions)** phát huy vai trò. Hãy hình dung một **Function** như một quy tắc hoặc cơ chế chính xác, nhận một đầu vào và tạo ra duy nhất một đầu ra.

Hãy tưởng tượng một máy bán hàng tự động đơn giản: bạn đưa tiền vào (đầu vào) và chọn mã sản phẩm (một đầu vào khác), sau đó máy sẽ đưa ra một món ăn nhẹ cụ thể (đầu ra). Một **Function** toán học hoạt động tương tự, nhưng thay vào đó sử dụng các con số hoặc những đối tượng toán học khác.

## **1. Inputs and Outputs**

Các giá trị mà chúng ta đưa vào một **Function** được gọi là **Inputs** hoặc **Arguments**. Giá trị mà **Function** tạo ra dựa trên đầu vào được gọi là **Output**. Chúng ta thường sử dụng các chữ cái như $x$ để biểu diễn đầu vào và $y$ hoặc $f(x)$ để biểu diễn đầu ra. Ký hiệu $f(x)$ được đọc là “f của x” và biểu thị đầu ra của **Function** $f$ khi đầu vào là $x$. Bản thân chữ cái $f$ đại diện cho **Function**, tức quy tắc biến đổi đầu vào thành đầu ra.

Ví dụ, một **Function** nhân đôi bất kỳ số nào bạn cung cấp rồi cộng thêm 1. Chúng ta có thể viết quy tắc này bằng ký hiệu hàm như sau:

$$
f(x) = 2x + 1
$$

Trong đó:

- $f$ là tên của **Function** (quy tắc: “nhân đôi rồi cộng thêm 1”).
- $x$ là biến đầu vào.
- $f(x)$ biểu thị đầu ra tương ứng với đầu vào $x$.

Nếu chúng ta cung cấp đầu vào $x = 3$, **Function** sẽ áp dụng quy tắc:

$$
\begin{aligned}
f(3) &= 2(3) + 1 \\
     &= 6 + 1 \\
     &= 7
\end{aligned}
$$

Như vậy, với đầu vào $3$, đầu ra là $7$.

Nếu đầu vào là $x = -5$, đầu ra sẽ là:

$$
\begin{aligned}
f(-5) &= 2(-5) + 1 \\
      &= -10 + 1 \\
      &= -9
\end{aligned}
$$



## **2. Các loại hàm phổ biến (Common Types of Functions)**

Có rất nhiều loại **Functions**, nhưng một số loại đặc biệt liên quan đến việc chúng ta bắt đầu tìm hiểu về **Calculus** trong **Machine Learning**:



### **2.1. Linear Functions**

Đây là những hàm tạo ra một đường thẳng khi được biểu diễn trên đồ thị. Chúng có dạng tổng quát:

  $$
  f(x) = mx + b
  $$

  - $m$ biểu thị hệ số góc (độ dốc của đường thẳng) và $b$ là tung độ gốc (vị trí đường thẳng cắt trục tung).
  
  - **Function** $f(x) = 2x + 1$ mà chúng ta vừa xét là một **Linear Function**, với $m = 2$ và $b = 1$.
  
  - **Linear Functions** là nền tảng trong **Machine Learning**, tạo cơ sở cho các **Models** như **Linear Regression**.



### **2.3. Other Functions** (các hàm khác)**

Khi tiếp tục học, bạn sẽ gặp những hàm như:
 - **Exponential Functions** (hàm mũ), chẳng hạn $e^x$
 - **Logarithmic Functions** (hàm logarit), chẳng hạn $log(x)$
 - **Trigonometric Functions** (hàm lượng giác), chẳng hạn $sin(x)$ và $cos(x)$.

 - Những hàm này xuất hiện trong nhiều ngữ cảnh khác nhau của **Machine Learning**, chẳng hạn như các **Activation Functions** trong mạng nơ-ron hoặc việc mô hình hóa các mẫu tăng trưởng.

Hiện tại, hãy tập trung vào việc hiểu mối quan hệ cơ bản giữa đầu vào và đầu ra. Xét **Linear Function**:
$f(x) = 0.5x - 1$
- Nếu $(x = 4)$: $ f(4) = 0.5 (4) - 1 = 2 - 1 = 1 $
- Nếu $(x = 0)$: $ f(0) = 0.5(0) - 1 = 0 - 1 = -1 $
- Nếu $(x = -2)$: $ f(-2) = 0.5(-2) - 1 = -1 - 1 = -2 $

Mỗi đầu vào $(4, 0, -2)$ tương ứng với duy nhất một đầu ra $(1, -1, -2)$.



## **3. Tại sao Functions quan trọng trong Machine Learning? (Why Functions Matter in Machine Learning)**

**Functions** là ngôn ngữ toán học được sử dụng để mô tả các **Models** mà chúng ta xây dựng.

- **Biểu diễn mô hình (Model Representation):** Một **Machine Learning Model** thường có thể được xem như một **Function**. Nó nhận **Input Data** (chẳng hạn như các **Features** của một ngôi nhà: diện tích và số lượng phòng ngủ), sau đó áp dụng một quy tắc đã học được (chính là **Function**) để tạo ra **Output** (chẳng hạn như giá nhà dự đoán). Huấn luyện một **Model** có nghĩa là điều chỉnh các **Parameters** của **Function** này (chẳng hạn như (m) và (b) trong biểu thức (mx+b)) để đưa ra những dự đoán chính xác.

- **Đo lường hiệu suất (Measuring Performance):** Chúng ta cũng sử dụng các **Functions**, được gọi là **Cost Functions** hoặc **Loss Functions**, để đo lường mức độ hoạt động của **Model**. Những hàm này nhận các dự đoán của **Model** và các giá trị mục tiêu thực tế làm đầu vào, sau đó trả về một con số biểu thị mức độ sai số. Giảm thiểu **Error Function** này là mục tiêu trọng tâm của quá trình huấn luyện **Model**, và **Calculus** cung cấp các công cụ để thực hiện điều đó một cách hiệu quả.
