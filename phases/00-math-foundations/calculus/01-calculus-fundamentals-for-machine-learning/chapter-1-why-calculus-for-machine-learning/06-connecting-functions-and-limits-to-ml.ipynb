# **Kết nối hàm số và giới hạn với học máy (Connecting Functions and Limits to ML)**

Hàm số ánh xạ **đầu vào (Input)** thành **đầu ra (Output)**. **Giới hạn (Limit)** liên quan đến việc tiến gần đến một giá trị cụ thể. Cả hàm số và giới hạn đều là những thành phần nền tảng trong học máy (**Machine Learning**) ứng dụng vào thực tế.



## **1. Hàm số với vai trò là mô hình (Functions as Models)**

Nhiều mô hình học máy có thể được xem như những hàm số. Ví dụ, hãy xét một nhiệm vụ đơn giản: dự đoán giá của một ngôi nhà dựa trên diện tích của nó (tính bằng feet vuông).

- **Đầu vào (Input):** Diện tích ngôi nhà (ví dụ: 1500 sq ft) – chúng ta gọi giá trị này là $x$.

- **Đầu ra (Output):** Giá nhà được dự đoán (ví dụ: 300,000) - chúng ta gọi giá trị này là $y$.

- **Mô hình (Model):** Cơ chế tiếp nhận **Input** $x$ và tạo ra **Output** $y$.

Về bản chất, mô hình này chính là một hàm số $f$. Chúng ta đưa **Input** $x$ vào và nhận được **Output** $y$, do đó: $ y=f(x) $

Ví dụ, một mô hình tuyến tính rất đơn giản có thể cố gắng mô tả mối quan hệ này bằng phương trình quen thuộc:

$$
y=mx+b
$$


Ở đây, $x$ là diện tích ngôi nhà, $y$ là giá nhà được dự đoán, còn $m$ và $b$ là các **tham số (Parameters)** mà mô hình cần học từ dữ liệu. Hàm số biểu diễn mô hình này là $f(x)=mx+b$. Những mô hình phức tạp hơn, chẳng hạn như mạng nơ-ron (**Neural Networks**), chỉ là những hàm số nhiều lớp và phức tạp hơn, nhưng nguyên lý vẫn không thay đổi: chúng ánh xạ **Input** thành **Output**.



## **2. Giới hạn là nền tảng của quá trình học (Limits as the Foundation for Learning)**

Vậy hàm số biểu diễn các mô hình của chúng ta. Còn giới hạn có vai trò gì? **Limit** là nền tảng lý thuyết để xây dựng **đạo hàm (Derivatives)**. Chúng ta vẫn chưa thảo luận chi tiết về **Derivatives** (nội dung này sẽ xuất hiện trong chương tiếp theo!), nhưng đây là mối liên hệ giữa chúng:

**Derivatives** đo lường **tốc độ thay đổi tức thời (Instantaneous Rate of Change)** của một hàm số. Hãy nghĩ về mô hình dự đoán giá nhà của chúng ta. Chúng ta muốn tìm các giá trị $m$ và $b$ *tốt nhất* để mô hình đưa ra dự đoán chính xác nhất. "Độ chính xác" thường được đo bằng mức độ *sai lệch* của mô hình, thông qua một đại lượng gọi là **hàm chi phí (Cost Function)** (chúng ta sẽ tìm hiểu thêm về nội dung này sau). Mục tiêu của chúng ta là giảm thiểu chi phí hoặc sai số này.

Làm thế nào để giảm thiểu sai số? Chúng ta cần biết việc thay đổi một chút giá trị $m$ hoặc $b$ sẽ ảnh hưởng đến sai số như thế nào. Tăng $m$ có làm sai số tăng lên hay giảm xuống không? Mức độ thay đổi là bao nhiêu? **Derivatives** trả lời chính xác những câu hỏi này. Chúng cho chúng ta biết độ dốc hoặc **gradient** của **Cost Function** theo các tham số $m$ và $b$.

Vậy **Derivatives** được tính như thế nào? Chúng được định nghĩa bằng **Limit**! Cụ thể, một **Derivative** được xác định bằng cách xét tỷ số giữa độ thay đổi của **Output** trong một hàm số và độ thay đổi của **Input**, khi độ thay đổi của **Input** tiến dần đến không. Chính phần "tiến dần đến không" này là nơi khái niệm **Limit** đóng vai trò thiết yếu.

>  *Một sơ đồ minh họa cách hàm số biểu diễn các mô hình, giới hạn tạo thành nền tảng cho đạo hàm, và đạo hàm định hướng quá trình tối ưu hóa được sử dụng để huấn luyện mô hình.*



Về bản chất:

1. **Functions** cung cấp cấu trúc cho các mô hình học máy của chúng ta (ánh xạ các đặc trưng thành các dự đoán).
2. **Limits** cung cấp nền tảng toán học cần thiết để định nghĩa chính xác cách một hàm số thay đổi.
3. Định nghĩa về sự thay đổi này (**Derivative**) cho phép chúng ta điều chỉnh có hệ thống các tham số của mô hình nhằm cải thiện hiệu suất (**Optimization**).

---
Nếu không hiểu về hàm số, chúng ta không thể biểu diễn các mô hình. Nếu không hiểu khái niệm giới hạn, chúng ta không thể nắm bắt được đạo hàm. Và nếu không có đạo hàm, chúng ta không thể huấn luyện hiệu quả nhiều mô hình học máy. Chương này đã đặt nền móng bằng cách giới thiệu hàm số và giới hạn; trong chương tiếp theo, chúng ta sẽ xây dựng trên nền tảng đó để tìm hiểu khái niệm chính về đạo hàm.
