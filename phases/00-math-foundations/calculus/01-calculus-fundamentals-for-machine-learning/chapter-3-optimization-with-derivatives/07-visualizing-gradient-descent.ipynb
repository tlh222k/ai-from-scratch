# **Trực quan hóa thuật toán Gradient Descent (Visualizing Gradient Descent)**

Trực quan hóa **Gradient Descent** giúp chúng ta hiểu rõ cách thuật toán tối ưu hóa này hoạt động. Bằng cách thực hiện các bước di chuyển dựa trên đạo hàm, chúng ta có thể tìm được điểm nhỏ nhất của một hàm số một cách hiệu quả.

Hãy tưởng tượng hàm chi phí của chúng ta có hình dạng giống như một thung lũng hoặc một cái bát. Mục tiêu là tìm đến điểm thấp nhất của thung lũng đó. Hãy hình dung bạn đang đứng ở đâu đó trên sườn thung lũng. Làm thế nào để xuống đến đáy? Bạn sẽ quan sát mặt đất ngay dưới chân để xác định hướng xuống dốc, rồi bước theo hướng đó.

Đạo hàm $f'(x)$ cung cấp cho chúng ta chính xác thông tin này: độ dốc của hàm số tại vị trí hiện tại $x$.

- Nếu độ dốc $f'(x)$ âm, hàm số đang giảm tại điểm đó. Để tiếp tục đi xuống dốc, chúng ta cần di chuyển theo chiều dương của trục $x$ (tăng $x$).
- Nếu độ dốc $f'(x)$ dương, hàm số đang tăng. Để đi xuống dốc, chúng ta cần di chuyển theo chiều âm của trục $x$ (giảm $x$).
- Nếu độ dốc $f'(x)$ bằng 0, chúng ta đang ở một điểm có độ dốc bằng 0. Điểm đó có thể là điểm cực tiểu (hoặc cực đại, hoặc **điểm yên ngựa (Saddle Point)**, nhưng trước mắt chúng ta tập trung vào điểm cực tiểu).

**Gradient Descent** sử dụng thông tin về độ dốc này để thực hiện các bước di chuyển lặp đi lặp lại. Ở mỗi bước, thuật toán tính đạo hàm tại vị trí hiện tại $x_{\text{old}}$, rồi cập nhật vị trí theo công thức:

$$x_{\text{new}}=x_{\text{old}}-\eta\times f'(x_{\text{old}})$$

Trong đó, $\eta$ (eta) là **Learning Rate**, một số dương nhỏ mà chúng ta đã tìm hiểu trước đó. Đại lượng này quyết định độ dài của mỗi bước di chuyển. Hãy chú ý đến dấu trừ:

- Nếu $f'(x_{\text{old}})$ dương (độ dốc hướng lên), chúng ta trừ đi một số dương. Vì vậy, $x$ di chuyển sang trái (giảm $x$).
- Nếu $f'(x_{\text{old}})$ âm (độ dốc hướng xuống), chúng ta trừ đi một số âm (tức là cộng thêm một số dương). Vì vậy, $x$ di chuyển sang phải (tăng $x$).

Trong cả hai trường hợp, công thức cập nhật đều đưa $x$ di chuyển theo hướng xuống dốc.

Hãy cùng hình dung quá trình này qua một **hàm bậc hai (Quadratic Function)** đơn giản:

$$f(x)=(x-3)^2+2$$

Chúng ta biết điểm cực tiểu của hàm số nằm tại $x=3$. Đạo hàm của hàm số là:

$$f'(x)=2(x-3)$$

Hãy bắt đầu tại $x=0$ và sử dụng tốc độ học $\eta=0.2$.




> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/gradient_descent_shifted.html) ·


> *Đường màu xanh biểu diễn hàm số $f(x)=(x-3)^2+2$. Các điểm màu cam và đường nét đứt biểu diễn đường đi của thuật toán Gradient Descent khi bắt đầu tại $x=0$. Mỗi bước di chuyển đưa chúng ta đến gần điểm cực tiểu tại $x=3$ hơn.*


Trong biểu đồ trên:

1. Chúng ta bắt đầu tại $x=0$. Giá trị hàm số là $f(0)=11$. Độ dốc tại đây là $f'(0)=2(0-3)=-6$ (độ dốc hướng xuống rất lớn).

2. Bước cập nhật đầu tiên là: $x_{\text{new}}=0-0.2\times(-6)=0+1.2=1.2$. Chúng ta di chuyển đến $x=1.2$.

3. Tại $x=1.2$, giá trị hàm số là $f(1.2)\approx5.24$. Độ dốc là $f'(1.2)=2(1.2-3)=-3.6$ (độ dốc đã thoải hơn nhưng vẫn hướng xuống).
4. Bước cập nhật tiếp theo là: $x_{\text{new}}=1.2-0.2\times(-3.6)=1.2+0.72=1.92$. Chúng ta di chuyển đến $x=1.92$.

5. Quá trình này tiếp tục. Hãy chú ý rằng các bước di chuyển ngày càng ngắn khi chúng ta tiến gần đến điểm cực tiểu. Điều này xảy ra vì độ dốc $f'(x)$ tiến gần về 0 khi chúng ta đến gần đáy thung lũng.

Hình ảnh trực quan này cho thấy ý tưởng cốt lõi: **Gradient Descent** liên tục sử dụng đạo hàm (độ dốc) để xác định hướng di chuyển tiếp theo, rồi từng bước tiến về một điểm cực tiểu của hàm số.

----

Trong học máy, hàm số mà chúng ta cần giảm giá trị chính là hàm chi phí, còn vị trí $x$ đại diện cho các tham số của mô hình. Khi tìm được những tham số giúp hàm chi phí đạt giá trị nhỏ nhất, chúng ta cũng tìm được mô hình phù hợp nhất với dữ liệu.
