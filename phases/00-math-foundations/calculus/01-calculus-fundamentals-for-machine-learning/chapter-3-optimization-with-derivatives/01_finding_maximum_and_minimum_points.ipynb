# **Tìm điểm cực đại và cực tiểu (Finding Maximum and Minimum Points)**

Đạo hàm của một hàm số, $f'(x)$, cho biết **tốc độ thay đổi tức thời (Instantaneous Rate of Change)** hoặc độ dốc của **đường tiếp tuyến (Tangent Line)** tại một điểm bất kỳ $x$. Khái niệm độ dốc này rất quan trọng khi xác định những điểm đặc biệt trên đồ thị hàm số, chẳng hạn như đỉnh và đáy của đường cong.

Hãy tưởng tượng bạn đang đi bộ trên một con đường có nhiều đoạn lên dốc và xuống dốc, được biểu diễn bằng hàm số $f(x)$. Khi bạn đến đỉnh của một ngọn đồi (điểm cực đại) hoặc đáy của một thung lũng (điểm cực tiểu), trong một khoảnh khắc ngắn, mặt đường ngay dưới chân bạn sẽ trở nên bằng phẳng, tức là nằm ngang. Một đường nằm ngang có độ dốc bằng $0$.

Trực giác này được áp dụng trực tiếp trong giải tích. Nếu một hàm số trơn $f(x)$ đạt **cực đại địa phương (Local Maximum)** hoặc **cực tiểu địa phương (Local Minimum)** tại điểm $x=c$, và đạo hàm của hàm số tồn tại tại điểm đó, thì độ dốc của đường tiếp tuyến tại điểm ấy phải bằng $0$.

Ta viết điều này dưới dạng toán học:

$$f'(c)=0$$

Những điểm mà đạo hàm bằng $0$ (hoặc không xác định, nhưng trước mắt chúng ta sẽ tập trung vào trường hợp đạo hàm bằng $0$) được gọi là **điểm tới hạn (Critical Points)**. Đây là những điểm *có khả năng* là nơi hàm số đạt cực đại hoặc cực tiểu địa phương.



## **1. Hình dung độ dốc bằng 0 (Visualizing the Flat Slope)**

Ví dụ, xét hàm số: $f(x)=x^2-4x+5$. Đây là một đường parabol. Hãy tìm đạo hàm của hàm số bằng **Power Rule**, **Sum Rule** và **Constant Rule** đã học ở chương trước:

$$f'(x)=\frac{d}{dx}(x^2-4x+5)=2x-4+0=2x-4$$

Bây giờ, ta tìm vị trí mà độ dốc bằng $0$ bằng cách cho đạo hàm bằng $0$:

$$ f'(x) = 0 \Rightarrow 2x - 4 = 0 \Rightarrow 2x = 4 \Rightarrow x = 2 $$


Điều này cho biết tại $x=2$, độ dốc của hàm số bằng $0$. Hãy quan sát đồ thị:

> *Đồ thị biểu diễn parabol $f(x)=x^2-4x+5$. Chấm đỏ đánh dấu điểm cực tiểu tại $(2,1)$. Đường nét đứt màu đỏ là tiếp tuyến tại điểm này. Vì đường tiếp tuyến nằm ngang nên độ dốc của nó bằng $f'(2)=0$.*

## **2. Quy trình tìm các điểm có khả năng là cực đại hoặc cực tiểu (The Process: Finding Candidates for Maxima and Minima)**

Như vậy, để tìm những vị trí có khả năng là nơi hàm số $f(x)$ đạt giá trị cực đại hoặc cực tiểu (tức các điểm tới hạn), ta thực hiện các bước sau:

1. **Tính đạo hàm (Calculate the Derivative):** Tìm hàm đạo hàm $f'(x)$.

2. **Cho đạo hàm bằng 0 (Set Derivative to Zero):** Đặt đạo hàm bằng $0$: $f'(x)=0$

3. **Giải tìm $x$ (Solve for x):** Giải phương trình vừa nhận được để tìm $x$. Những giá trị $x$ tìm được là các điểm tới hạn.

Điều quan trọng cần nhớ là việc giải phương trình $f'(x)=0$ **chỉ giúp chúng ta tìm ra những điểm *có khả năng* là cực đại hoặc cực tiểu**. Những điểm này thường là cực đại hoặc cực tiểu địa phương, nhưng đôi khi chúng cũng có thể là một dạng điểm khác. Chẳng hạn, đó có thể là điểm uốn, nơi đường cong tạm thời trở nên phẳng rồi tiếp tục đi theo cùng một hướng tổng thể như trước.

Trong phạm vi tối ưu hóa **Machine Learning**, chúng ta thường muốn tìm giá trị cực tiểu. Vì vậy, điều kiện $f'(x)=0$ là điểm khởi đầu quan trọng để tìm giá trị đó.



## **3. Tại sao điều này quan trọng đối với học máy? (Why is this Important for Machine Learning?)**

Việc tìm những điểm mà đạo hàm bằng $0$ không chỉ là một bài tập toán học. Đây còn là nền tảng cho cách nhiều thuật toán **Machine Learning** được "huấn luyện".

Hãy tưởng tượng bạn có một mô hình học máy và xác định một **hàm chi phí (Cost Function)**, còn được gọi là hàm mất mát (**Loss Function**). Hàm này đo lường mức độ sai lệch giữa dự đoán của mô hình và dữ liệu thực tế.

- Giá trị hàm chi phí cao có nghĩa là mô hình đang dự đoán chưa tốt.
- Giá trị hàm chi phí thấp có nghĩa là mô hình đang hoạt động tốt hơn.

----
Mục tiêu của quá trình huấn luyện là điều chỉnh các tham số bên trong mô hình để làm cho giá trị hàm chi phí nhỏ nhất có thể, lý tưởng nhất là đạt đến giá trị cực tiểu của hàm.

Nếu biểu diễn được hàm chi phí theo các tham số của mô hình, chúng ta có thể sử dụng giải tích để tìm những giá trị tham số giúp hàm chi phí đạt mức nhỏ nhất. Việc tìm những điểm mà đạo hàm của hàm chi phí bằng $0$ giúp chúng ta xác định vị trí có sai số nhỏ nhất. Từ đó, mô hình có thể đạt được hiệu suất tốt nhất theo thước đo chi phí mà chúng ta đã lựa chọn.


Trong các phần tiếp theo, chúng ta sẽ tìm hiểu về hàm chi phí và một thuật toán cụ thể có tên là **Gradient Descent** (thuật toán hạ gradient). Thuật toán này sử dụng đạo hàm lặp đi lặp lại để tìm các điểm cực tiểu, ngay cả trong những trường hợp việc giải trực tiếp phương trình $f'(x)=0$ rất khó hoặc không thể thực hiện được.
