# **Mục tiêu: Giảm hàm chi phí xuống mức thấp nhất (Goal: Minimizing the Cost Function)**

**Hàm chi phí (Cost Function)**, thường được gọi là **Loss Function**, dùng để đo lường mức độ hoạt động của một mô hình **Machine Learning**. Hàm này nhận các giá trị dự đoán của mô hình và các giá trị mục tiêu thực tế làm đầu vào, sau đó tính toán và trả về một con số duy nhất biểu thị tổng sai số hay "chi phí".

- Chi phí cao cho thấy dự đoán của mô hình sai lệch nhiều so với giá trị thực tế.
- Chi phí thấp cho thấy dự đoán gần với thực tế hơn.

Vậy mục tiêu của chúng ta khi huấn luyện một mô hình **Machine Learning** là gì?

Thông thường, chúng ta muốn mô hình đưa ra dự đoán chính xác nhất có thể. Xét theo khái niệm hàm chi phí, điều này có nghĩa là tìm các tham số của mô hình sao cho **Cost Function** đạt *giá trị nhỏ nhất có thể*.

Hãy tưởng tượng việc này giống như điều chỉnh một chiếc radio. Các tham số của mô hình (chẳng hạn độ dốc $m$ và hệ số chặn $b$ khi bạn tìm cách khớp một đường thẳng với dữ liệu) giống như những núm điều chỉnh mà bạn có thể xoay.

**Cost Function** cho biết mức độ nhiễu (tức sai số) tương ứng với mỗi cách thiết lập các núm điều chỉnh đó. Mục tiêu của bạn là điều chỉnh chúng cho đến khi tìm được thiết lập giúp tín hiệu rõ nhất. Điều đó có nghĩa là giảm nhiễu xuống mức thấp nhất, hay đạt được chi phí cực tiểu.



## **1. Tại sao phải giảm chi phí? (Why Minimization?)**

Việc giảm **Cost Function** xuống mức thấp nhất có vai trò nền tảng vì hai lý do:

1. **Định lượng thế nào là "khớp tốt nhất" (It Quantifies "Best Fit"):** **Cost Function** cung cấp một định nghĩa toán học về việc thế nào là "tốt nhất" đối với mô hình của bạn. Bằng cách tối thiểu hóa hàm này, bạn tìm được các tham số giúp mô hình khớp với dữ liệu huấn luyện tốt nhất có thể, theo tiêu chí mà hàm chi phí đã xác định.

   - Ví dụ, khi tối thiểu hóa **Mean Squared Error (MSE)**, chúng ta đang tìm các tham số giúp giảm trung bình bình phương chênh lệch giữa giá trị dự đoán và giá trị thực tế.

2. **Thúc đẩy quá trình học (It Drives Learning):** Trong nhiều trường hợp của **Machine Learning**, quá trình điều chỉnh các tham số của mô hình để giảm chi phí chính là bản chất của việc "học".

   - Mô hình liên tục thay đổi các tham số, kiểm tra giá trị chi phí rồi tiếp tục điều chỉnh. Quá trình này lặp đi lặp lại với mục tiêu luôn hướng đến giá trị chi phí thấp hơn.

Ví dụ, hãy xét một trường hợp rất đơn giản, trong đó chi phí chỉ phụ thuộc vào một tham số. Chúng ta gọi tham số đó là $w$. Khi biểu diễn hàm chi phí theo các giá trị khác nhau của $w$, đồ thị có thể có hình dạng giống một chiếc bát.


> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/quadratic_cost_function.html) ·



> *Một hàm chi phí dạng parabol đơn giản:* $\text{Cost}=w^2$
*Mục tiêu là tìm giá trị của tham số $w$ (trong trường hợp này là $w=0$) tương ứng với điểm thấp nhất trên đường cong, tức là điểm có chi phí nhỏ nhất.*

Mục tiêu của chúng ta là tìm giá trị $w$ nằm ở đáy chiếc bát đó. Quá trình tìm đầu vào (hoặc các đầu vào) của một hàm số sao cho đầu ra đạt giá trị nhỏ nhất được gọi là **Optimization**.

Trong **Machine Learning**, chúng ta tối ưu hóa các tham số của mô hình để giảm **Cost Function** xuống mức thấp nhất.

Với một hàm đơn giản như $\text{Cost}=w^2$, việc tìm giá trị nhỏ nhất khá dễ dàng (rõ ràng giá trị đó đạt được tại $w=0$). Tuy nhiên, các hàm chi phí trong những mô hình học máy thực tế thường phức tạp hơn nhiều và có thể phụ thuộc vào hàng triệu tham số.

Vậy làm thế nào để tìm điểm cực tiểu một cách hiệu quả, đặc biệt khi phải xử lý rất nhiều tham số?

Đây là lúc giải tích, cụ thể là **đạo hàm (Derivative)**, phát huy tác dụng. Như chúng ta đã tìm hiểu, đạo hàm cho biết độ dốc hoặc tốc độ thay đổi của một hàm số. Chúng ta có thể sử dụng thông tin về độ dốc này để định hướng quá trình tìm kiếm điểm cực tiểu của hàm chi phí.

Đây chính là ý tưởng cốt lõi đằng sau các thuật toán như **Gradient Descent** (thuật toán hạ gradient), mà chúng ta sẽ tìm hiểu trong phần tiếp theo.
