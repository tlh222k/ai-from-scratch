# **Hàm chi phí trong học máy (Cost Functions in Machine Learning)**

**Tối ưu hóa (Optimization)** là quá trình tìm giá trị nhỏ nhất hoặc lớn nhất của một hàm số. Đây là một khái niệm cốt lõi, không chỉ đơn thuần là một bài tập toán học trừu tượng. Trong **Machine Learning**, chúng ta áp dụng tối ưu hóa vào những hàm số cụ thể. Các mô hình có thể học hiệu quả từ dữ liệu bằng cách tối ưu hóa những hàm số này. Nhưng chính xác thì chúng ta đang cố gắng tìm giá trị nhỏ nhất hoặc lớn nhất của đại lượng nào?

Hãy nghĩ về quá trình huấn luyện một mô hình **Machine Learning**. Mục tiêu thường là đưa ra những dự đoán gần với kết quả thực tế nhất có thể.

Ví dụ:
- Nếu dự đoán giá nhà, chúng ta muốn giá mà mô hình dự đoán càng gần với giá bán thực tế của căn nhà càng tốt.
- Nếu phân loại hình ảnh, chúng ta muốn mô hình gán đúng nhãn cho hình ảnh, chẳng hạn như "mèo" hoặc "chó".

Chúng ta cần một cách để đo lường mức độ sai lệch giữa dự đoán của mô hình và giá trị thực tế. Đại lượng dùng để đo lường này được gọi là **hàm chi phí (Cost Function)**, thường cũng được gọi là **hàm mất mát (Loss Function)** hoặc đôi khi là **hàm mục tiêu (Objective Function)**.



## **1. Hàm chi phí là gì? (What is a Cost Function?)**

Một **Cost Function** nhận dự đoán của mô hình và các giá trị mục tiêu thực tế làm đầu vào, sau đó trả về một con số duy nhất. Con số này biểu thị **chi phí** hoặc **mức phạt** tương ứng với những sai số mà mô hình mắc phải trên tập dữ liệu đang được dùng để huấn luyện.

- **Chi phí cao (High Cost):** Có nghĩa là dự đoán của mô hình sai lệch nhiều so với giá trị thực tế. Mô hình đang hoạt động kém.
- **Chi phí thấp (Low Cost):** Có nghĩa là dự đoán gần với giá trị thực tế. Mô hình đang hoạt động tốt.

Hãy tưởng tượng bạn đang học chơi phi tiêu. **Cost Function** có thể được định nghĩa là khoảng cách trung bình từ vị trí phi tiêu cắm xuống đến hồng tâm. Khoảng cách trung bình càng nhỏ (chi phí càng thấp) thì có nghĩa là bạn chơi càng tốt. Tương tự, trong **Machine Learning**, chúng ta muốn điều chỉnh mô hình để đạt được chi phí thấp nhất có thể.



## **2. Ví dụ: Sai số bình phương trung bình (An Example: Mean Squared Error – MSE)**

Một trong những hàm chi phí phổ biến nhất, đặc biệt trong các bài toán **hồi quy (Regression)** như dự đoán giá cả, là **sai số bình phương trung bình (Mean Squared Error – MSE)**.

Hãy cùng tìm hiểu cách tính MSE trên một tập dữ liệu có $N$ mẫu:

1. **Tính sai số (Calculate the Error):** Với mỗi điểm dữ liệu $i$, tính hiệu giữa giá trị mô hình dự đoán ($y_{\text{pred},i}$) và giá trị thực tế ($y_{\text{actual},i}$). Hiệu này chính là sai số:

   $$e_i=y_{\text{pred},i}-y_{\text{actual},i}$$

2. **Bình phương sai số (Square the Error):** Bình phương từng sai số:

   $$e_i^2=(y_{\text{pred},i}-y_{\text{actual},i})^2$$

   Việc bình phương mang lại hai lợi ích:

   - Tất cả sai số đều trở thành số dương, nhờ đó sai số do dự đoán thấp hơn thực tế và sai số do dự đoán cao hơn thực tế không triệt tiêu lẫn nhau.
   - Sai số lớn bị phạt nặng hơn nhiều so với sai số nhỏ. Ví dụ, sai số bằng 4 đóng góp 16 vào tổng sai số, trong khi sai số bằng 2 chỉ đóng góp 4.

3. **Tính giá trị trung bình (Calculate the Mean):** Tính trung bình các sai số đã bình phương trên toàn bộ $N$ điểm dữ liệu trong tập huấn luyện.

Công thức tính MSE là:

$$\text{MSE}=\frac{1}{N}\sum_{i=1}^{N}(y_{\text{pred},i}-y_{\text{actual},i})^2$$

Trong công thức này:

- Ký hiệu $\sum$ (sigma) có nghĩa là "cộng tất cả lại".
- Các chỉ số từ $i=1$ đến $N$ cho biết chúng ta cần tính tổng trên tất cả các điểm dữ liệu, từ điểm đầu tiên ($i=1$) đến điểm cuối cùng ($i=N$).
- Hệ số $\frac{1}{N}$ dùng để tính giá trị trung bình.

Khi huấn luyện mô hình bằng MSE, mục tiêu của chúng ta là tìm các tham số của mô hình (chẳng hạn như độ dốc $m$ và hệ số chặn $b$ trong mô hình tuyến tính $y=mx+b$) sao cho giá trị MSE nhỏ nhất có thể.



## **3. Hình dung hàm chi phí (Visualizing the Cost)**

Một cách dễ hình dung là xem **Cost Function** như một bề mặt. "Vị trí" trên bề mặt này được xác định bởi các giá trị cụ thể mà chúng ta chọn cho các tham số của mô hình, chẳng hạn như $m$ và $b$. "Độ cao" của bề mặt tại mỗi vị trí biểu thị giá trị hàm chi phí tương ứng với các tham số đó.

Xét một mô hình tuyến tính đơn giản $y=mx+b$ đang cố gắng khớp với một tập dữ liệu. Chi phí **MSE** phụ thuộc vào các giá trị $m$ và $b$ mà chúng ta lựa chọn.

Nếu đơn giản hóa bài toán bằng cách cố định $b$ và chỉ thay đổi $m$, chúng ta có thể vẽ đồ thị biểu diễn chi phí (**MSE**) theo các giá trị khác nhau của $m$. Đồ thị này thường có hình dạng giống một chiếc bát.

> *Đây là hình dạng thường thấy của một hàm chi phí (chẳng hạn **MSE**) khi biểu diễn theo một tham số duy nhất của mô hình. Điểm thấp nhất trên đồ thị tương ứng với giá trị tham số giúp sai số đạt mức nhỏ nhất.*

Khi có nhiều tham số hơn (chẳng hạn cả $m$ và $b$), bề mặt này sẽ có nhiều chiều hơn. Ví dụ, với hai tham số, ta có thể hình dung nó giống như một chiếc bát trong không gian ba chiều (3D).

Tuy nhiên, ý tưởng cốt lõi vẫn không thay đổi: chúng ta đang tìm điểm thấp nhất trong không gian chi phí này. Tọa độ của điểm thấp nhất chính là các giá trị tham số tối ưu cho mô hình.

Điều này đưa chúng ta trở lại với **Optimization**. **Cost Function** xác định đại lượng cụ thể mà chúng ta muốn giảm xuống mức thấp nhất. Những kỹ thuật đang học trong chương này, sử dụng đạo hàm, chính là công cụ giúp chúng ta tìm điểm cực tiểu đó một cách hiệu quả.
