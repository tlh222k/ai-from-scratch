# **Giới thiệu về thuật toán hạ gradient (Introduction to Gradient Descent)**

Tìm điểm nhỏ nhất của một **hàm chi phí (Cost Function)** là một mục tiêu quan trọng khi huấn luyện các **mô hình học máy (Machine Learning Models)**. **Đạo hàm (Derivative)** của một hàm số cho biết độ dốc, hay tốc độ thay đổi của hàm tại một điểm bất kỳ. Đặt đạo hàm bằng $0$, tức là $f'(x)=0$, có thể giúp chúng ta tìm ra những điểm có khả năng là điểm cực tiểu bằng phương pháp giải tích, đặc biệt với các hàm đơn giản.

Nhưng điều gì sẽ xảy ra nếu hàm chi phí rất phức tạp, chẳng hạn có đến hàng triệu tham số (điều thường gặp trong học sâu)? Việc giải trực tiếp phương trình $f'(x)=0$ có thể đòi hỏi quá nhiều tài nguyên tính toán, thậm chí không thể thực hiện được. Chúng ta cần một cách tiếp cận lặp đi lặp lại, một phương pháp có thể tìm điểm nhỏ nhất theo từng bước mà không cần giải những phương trình phức tạp. Đó chính là lúc **Gradient Descent** phát huy tác dụng.

Hãy hình dung thế này: bạn đang đứng trên một sườn núi phủ đầy sương mù và muốn đi xuống điểm thấp nhất trong thung lũng. Bạn không thể nhìn thấy toàn bộ địa hình, nhưng có thể cảm nhận được độ dốc ngay dưới chân mình. Vậy cách đơn giản nhất để đi xuống là gì?

- Kiểm tra độ dốc và hướng dốc tại vị trí bạn đang đứng.
- Bước một bước nhỏ xuống dốc theo hướng dốc nhất.
- Lặp lại quá trình: kiểm tra độ dốc tại vị trí mới, rồi tiếp tục bước xuống.

Bằng cách lặp lại những bước này, bạn sẽ dần tiến xuống đáy thung lũng.

**Gradient Descent** chính là phiên bản toán học của quá trình này. Vị trí của chúng ta trên sườn núi tương ứng với giá trị hiện tại của biến đầu vào của hàm số (hoặc nhiều biến đầu vào, như chúng ta sẽ tìm hiểu sau). Độ dốc mà chúng ta cảm nhận được được biểu diễn bằng **Derivative** (hoặc đại lượng tương đương khi làm việc với hàm nhiều biến, gọi là **Gradient**). “Bước xuống dốc” có nghĩa là điều chỉnh biến đầu vào theo hướng ngược lại với hướng dốc.

Hãy xem xét một **hàm chi phí (Cost Function)** đơn giản $J(w)$, trong đó $w$ là một tham số mà chúng ta muốn tối ưu hóa (chẳng hạn hệ số góc $m$ trong mô hình tuyến tính $y=mx+b$).

- Đầu tiên, chúng ta chọn một giá trị ban đầu để ước lượng $w$.
- Tính đạo hàm của hàm chi phí tại giá trị đó, tức là $J'(w)$. Kết quả này cho biết độ dốc của hàm.
- Nếu $J'(w)$ dương, hàm số đang tăng tại điểm này. Để đi xuống dốc, chúng ta cần giảm giá trị $w$.
- Nếu $J'(w)$ âm, hàm số đang giảm tại điểm này. Để đi xuống dốc, chúng ta cần tăng giá trị $w$.

Hãy chú ý đến quy luật này: chúng ta luôn muốn điều chỉnh $w$ theo hướng ngược lại với dấu của đạo hàm. Điều này có thể thực hiện bằng một công thức cập nhật đơn giản:

$$w_{\text{new}}=w_{\text{old}}-\alpha\cdot J'(w_{\text{old}})$$

Trong đó:

- $w_{\text{new}}$ là giá trị mới của tham số sau khi cập nhật.
- $w_{\text{old}}$ là giá trị hiện tại của tham số.
- $J'(w_{\text{old}})$ là đạo hàm (độ dốc) được tính tại giá trị hiện tại $w_{\text{old}}$.
- $\alpha$ (alpha) là một số dương nhỏ, được gọi là **tốc độ học (Learning Rate)**. Nó quyết định mỗi lần chúng ta bước đi xa bao nhiêu. Chúng ta sẽ tìm hiểu kỹ hơn về đại lượng này sau. Trước mắt, hãy hiểu rằng nó giúp các bước di chuyển đủ nhỏ để không vượt quá điểm cực tiểu.

Nếu $J'(w_{\text{old}})$ dương (độ dốc hướng lên), việc trừ đi $\alpha\cdot J'(w_{\text{old}})$ sẽ khiến $w_{\text{new}}$ nhỏ hơn $w_{\text{old}}$. Điều đó có nghĩa là chúng ta di chuyển sang trái, theo hướng xuống dốc.

Ngược lại, nếu $J'(w_{\text{old}})$ âm (độ dốc hướng xuống), trừ đi một số âm đồng nghĩa với việc cộng thêm $\alpha\cdot |J'(w_{\text{old}})|$. Khi đó, $w_{\text{new}}$ sẽ lớn hơn $w_{\text{old}}$, nghĩa là chúng ta di chuyển sang phải, cũng theo hướng xuống dốc. Như vậy, công thức cập nhật tự động xác định hướng di chuyển!

Chúng ta lặp lại bước cập nhật này nhiều lần. Sau mỗi bước, chúng ta tính lại độ dốc tại vị trí mới rồi tiếp tục đi xuống. Khi tiến gần đến điểm cực tiểu, độ lớn của $J'(w)$ cũng tiến gần về 0, khiến các bước di chuyển ngày càng nhỏ. Thông thường, chúng ta dừng quá trình khi các bước cập nhật trở nên rất nhỏ. Đây là dấu hiệu cho thấy thuật toán có thể đã hội tụ đến gần điểm cực tiểu.


-----

**Gradient Descent** là một **thuật toán tối ưu hóa (Optimization Algorithm)** nền tảng, được sử dụng rộng rãi trong học máy để điều chỉnh các tham số của mô hình và giảm **Cost Function**, từ đó cải thiện hiệu suất của mô hình trên dữ liệu huấn luyện.

Thuật toán này không đảm bảo sẽ tìm được điểm cực tiểu toàn cục (điểm thấp nhất trong toàn bộ hàm số). Đôi khi, nó có thể dừng tại một điểm cực tiểu địa phương — một chỗ trũng nhỏ nhưng không phải điểm thấp nhất trên toàn bộ địa hình. Tuy vậy, trên thực tế, **Gradient Descent** vẫn hoạt động rất hiệu quả đối với nhiều bài toán học máy.
