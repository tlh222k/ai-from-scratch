# **Tối ưu hóa: Tại sao cần tìm giá trị nhỏ nhất hoặc lớn nhất? (Optimization: Why Minimize or Maximize?)**

Việc tìm các điểm có giá trị nhỏ nhất hoặc lớn nhất, thường được gọi là những “điểm phẳng” vì tại đó đạo hàm có thể bằng $0$, là một khái niệm rất quan trọng. Quá trình tìm giá trị tốt nhất có thể của một hàm số, tức là giá trị nhỏ nhất hoặc lớn nhất, được gọi là **tối ưu hóa (Optimization)**.

Hãy tưởng tượng bạn đang đi leo núi. Bạn có thể muốn tìm điểm thấp nhất trong một thung lũng (giá trị cực tiểu) để dựng lều, hoặc tìm đỉnh núi cao nhất (giá trị cực đại) để ngắm cảnh đẹp nhất.

Trong cuộc sống hằng ngày, bạn có thể muốn giảm thời gian đi làm xuống mức thấp nhất hoặc tăng số tiền tiết kiệm lên mức cao nhất. Các doanh nghiệp cũng muốn giảm chi phí sản xuất và tăng lợi nhuận. Tất cả những trường hợp này đều là bài toán tối ưu hóa: tìm ra kết quả tốt nhất có thể trong những điều kiện nhất định.



## **1. Tại sao cần tối ưu hóa trong học máy? (Why Optimize in Machine Learning?)**

Trong **Machine Learning**, mục tiêu của chúng ta thường là xây dựng một **Model** có khả năng đưa ra dự đoán.

Hãy xem xét một mô hình đơn giản có nhiệm vụ dự đoán giá nhà dựa trên diện tích. Về bản chất, mô hình này là một hàm số: bạn cung cấp một **Input** (diện tích căn nhà), và mô hình tạo ra một **Output** (giá nhà dự đoán).

Nhưng làm thế nào để biết mô hình dự đoán có tốt hay không? Chúng ta cần một cách để đo lường hiệu quả của mô hình, cụ thể là đo xem các dự đoán sai lệch bao nhiêu so với giá nhà thực tế.

Đại lượng dùng để đo mức độ sai lệch hay sai số này thường được gọi là **hàm chi phí (Cost Function)** hoặc **hàm mất mát (Loss Function)**. Chúng ta sẽ tìm hiểu kỹ hơn về hai hàm này trong phần tiếp theo.

Mối liên hệ giữa các khái niệm trên như sau:

1. **Chúng ta muốn có mô hình tốt nhất:** Mô hình tốt nhất là mô hình đưa ra những dự đoán chính xác nhất.

2. **Dự đoán chính xác nghĩa là sai số nhỏ:** Dự đoán càng chính xác thì chênh lệch giữa giá trị mô hình dự đoán và giá trị thực tế càng nhỏ.

3. **Sai số nhỏ nghĩa là chi phí thấp:** Hàm chi phí được thiết kế sao cho giá trị của nó nhỏ khi sai số nhỏ, và lớn khi sai số lớn.

Vì vậy, mục tiêu chính khi huấn luyện nhiều mô hình học máy là **tìm cách làm cho hàm chi phí đạt giá trị nhỏ nhất**.

Bằng cách tìm ra những giá trị **Parameters** của mô hình giúp hàm chi phí đạt mức thấp nhất có thể, chúng ta đang tìm cách xây dựng mô hình có sai số nhỏ nhất, tức là mô hình hoạt động tốt nhất trên dữ liệu đã được sử dụng để huấn luyện nó.

Đôi khi, chúng ta cũng muốn tìm giá trị lớn nhất của một đại lượng nào đó. Ví dụ, ta có thể muốn tối đa hóa xác suất xuất hiện của dữ liệu khi đã biết mô hình, đại lượng này được gọi là **likelihood** (hàm hợp lý). Tuy nhiên, trường hợp phổ biến nhất, đặc biệt khi mới bắt đầu học, vẫn là tìm cách giảm sai số xuống mức thấp nhất.

Việc tìm điểm cực tiểu của hàm chi phí chính là một bài toán tối ưu hóa.

Đây chính là lúc đạo hàm phát huy tác dụng. Như chúng ta đã học, đạo hàm giúp xác định độ dốc của một hàm số. Bằng cách xem xét đạo hàm (chính xác hơn là **Gradient** khi hàm số có nhiều đầu vào, nội dung này sẽ được trình bày ở phần sau), chúng ta có thể xác định nên điều chỉnh các tham số của mô hình theo hướng nào để làm giảm chi phí.

Việc tìm những điểm mà đạo hàm bằng $0$ cũng giúp chúng ta xác định các vị trí có khả năng là điểm cực tiểu của hàm chi phí.

Tóm lại, chúng ta thực hiện tối ưu hóa vì muốn tìm được mô hình hoạt động tốt nhất. Trong học máy, “tốt nhất” thường có nghĩa là “sai số nhỏ nhất” hoặc “chi phí thấp nhất”. Đạo hàm cung cấp công cụ toán học cần thiết để chúng ta tìm kiếm giá trị nhỏ nhất đó một cách có hệ thống.
