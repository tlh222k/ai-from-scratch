# **Cách đạo hàm định hướng thuật toán Gradient Descent (How Derivatives Guide Gradient Descent)**

Mục tiêu là tìm điểm nhỏ nhất của một hàm số, đặc biệt là **hàm chi phí (Cost Function)** trong học máy, vì điểm này tương ứng với mức sai số thấp nhất. Đạo hàm $f'(x)$ cho biết độ dốc, hay **tốc độ thay đổi tức thời (Instantaneous Rate of Change)** của hàm số $f(x)$ tại một điểm bất kỳ $x$. Đây là những kiến thức nền tảng giúp chúng ta hiểu cách đạo hàm định hướng cho thuật toán **Gradient Descent**.

Hãy tưởng tượng bạn đang đứng trên sườn một ngọn đồi được biểu diễn bằng đồ thị hàm chi phí và muốn đi xuống chân đồi (điểm cực tiểu). Xung quanh có sương mù nên bạn chỉ cảm nhận được mặt đất ngay tại vị trí mình đang đứng. Làm thế nào để quyết định nên bước theo hướng nào?

Đạo hàm cung cấp chính xác thông tin bạn cần.

- Nếu đạo hàm $f'(x)$ tại vị trí hiện tại $x$ **dương**, điều đó có nghĩa là độ dốc dương. Hàm số đang *tăng* tại điểm này. Để đi xuống dốc và tiến về điểm cực tiểu, bạn cần di chuyển theo hướng có giá trị $x$ nhỏ hơn, tức là đi sang *trái* (giảm $x$).

- Nếu đạo hàm $f'(x)$ tại vị trí hiện tại $x$ **âm**, điều đó có nghĩa là độ dốc âm. Hàm số đang *giảm* tại điểm này. Để tiếp tục đi xuống dốc và tiến về điểm cực tiểu, bạn cần di chuyển theo hướng có giá trị $x$ lớn hơn, tức là đi sang *phải* (tăng $x$).
- Nếu đạo hàm $f'(x)$ bằng **0**, độ dốc bằng 0, tức là tại đó đồ thị có tiếp tuyến nằm ngang. Điều này cho thấy bạn có thể đang ở đáy dốc (điểm cực tiểu), nhưng cũng có thể đang ở đỉnh dốc hoặc một đoạn bằng phẳng. Thuật toán **Gradient Descent** thường dừng tại đây.

Hãy chú ý đến quy luật: nếu đạo hàm dương, chúng ta giảm $x$; nếu đạo hàm âm, chúng ta tăng $x$. Trong cả hai trường hợp, về cơ bản, chúng ta đều di chuyển theo hướng *ngược lại* với dấu của đạo hàm.

Đây là ý tưởng cốt lõi về cách **Gradient Descent** sử dụng đạo hàm. Thuật toán tính đạo hàm tại điểm hiện tại, sau đó thực hiện một bước nhỏ theo hướng ngược lại. Độ dài bước đi thường cũng tỷ lệ với độ lớn của đạo hàm: độ dốc càng lớn thì bước đi càng dài, độ dốc càng thoải thì bước đi càng ngắn. Độ dài này còn được điều chỉnh bởi một đại lượng gọi là **tốc độ học (Learning Rate)**, chúng ta sẽ tìm hiểu về nó trong phần tiếp theo.

Về mặt toán học, có thể biểu diễn một bước của **Gradient Descent** như sau:

$$x_{\text{new}}=x_{\text{current}}-\alpha\cdot f'(x_{\text{current}})$$

Trong đó:

- $x_{\text{current}}$ là vị trí hiện tại của bạn.
- $f'(x_{\text{current}})$ là đạo hàm (độ dốc) tại vị trí hiện tại.
- $\alpha$ (alpha) là **Learning Rate**, một số dương nhỏ dùng để điều chỉnh độ dài bước đi. Chúng ta sẽ tìm hiểu kỹ hơn về tham số này trong phần “Tham số tốc độ học (The Learning Rate Parameter)”.
- $x_{\text{new}}$ là vị trí mới sau khi thực hiện bước di chuyển.

Điểm quan trọng nhất trong công thức là dấu trừ. Dấu này đảm bảo rằng bạn di chuyển *ngược hướng với gradient*:

- Nếu $f'(x_{\text{current}})$ dương (độ dốc hướng lên khi đi sang phải), việc trừ đi $\alpha\cdot f'(x_{\text{current}})$ sẽ khiến $x$ di chuyển sang trái (giảm $x$).
- Nếu $f'(x_{\text{current}})$ âm (độ dốc hướng xuống khi đi sang phải), việc trừ đi $\alpha\cdot f'(x_{\text{current}})$ thực chất là *cộng thêm* một giá trị nhỏ (vì trừ một số âm tương đương với cộng một số dương). Nhờ đó, $x$ di chuyển sang phải (tăng $x$).

Tóm lại, đạo hàm hoạt động giống như một chiếc la bàn chỉ lên dốc (hướng tăng nhanh nhất), còn **Gradient Descent** chỉ cần bước theo hướng ngược lại để đi xuống dốc. Bằng cách lặp lại quá trình này — tính đạo hàm rồi thực hiện một bước nhỏ theo hướng ngược lại — thuật toán sẽ dần đi xuống theo độ dốc của hàm chi phí và tiến về một điểm cực tiểu.

Hãy cùng hình dung quá trình này qua một ví dụ. Xét hàm số đơn giản $f(x)=x^2$. Rõ ràng, giá trị nhỏ nhất của hàm đạt được tại $x=0$. Đạo hàm của hàm số là $f'(x)=2x$.


> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/quadratic_cost_function.html) ·



> *Giả sử chúng ta đang ở vị trí $x=2$. Giá trị của hàm số tại đó là $f(2)=2^2=4$. Đạo hàm tại điểm này là $f'(2)=2\times 2=4$. Vì đạo hàm dương nên hàm số đang tăng tại đây (đường tiếp tuyến màu đỏ nét đứt thể hiện độ dốc dương). **Gradient Descent** sử dụng giá trị đạo hàm dương này để xác định rằng cần di chuyển $x$ sang trái (giảm $x$), như mũi tên màu xanh chỉ dẫn. Bước di chuyển này đưa chúng ta đến gần điểm cực tiểu tại $x=0$.*



Nếu thay vào đó, chúng ta đang ở vị trí $x=-1.5$, đạo hàm sẽ là:

$$f'(-1.5)=2\times(-1.5)=-3$$

Đạo hàm âm cho biết **Gradient Descent** cần *tăng* $x$ (di chuyển sang phải). Một lần nữa, bước đi này sẽ đưa chúng ta đến gần điểm cực tiểu.

Quá trình tiếp tục lặp lại. Đạo hàm tại mỗi vị trí mới sẽ định hướng cho bước tiếp theo, cho đến khi chúng ta lý tưởng nhất là đạt đến một điểm mà đạo hàm rất gần 0. Đây là dấu hiệu cho thấy chúng ta đã tìm được một điểm cực tiểu.
