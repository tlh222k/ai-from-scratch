# **Tốc độ thay đổi: Trung bình và tức thời (Rate of Change: Average vs Instantaneous)**

Hàm số xác định một mối quan hệ trong đó **đầu vào (Input)** được ánh xạ thành **đầu ra (Output)**. Việc hiểu cách các giá trị đầu ra thay đổi khi đầu vào biến thiên là một khía cạnh cốt lõi của **Giải tích (Calculus)**. Hãy tưởng tượng bạn đang lái một chiếc ô tô: vị trí của bạn thay đổi theo thời gian, với những thời điểm tăng tốc và giảm tốc. **Calculus** cung cấp các công cụ chính xác để mô tả những thay đổi động này.

Chúng ta thường nói về sự thay đổi thông qua tốc độ. Ví dụ, vận tốc là tốc độ thay đổi của quãng đường theo thời gian. Tuy nhiên, có nhiều cách khác nhau để đo lường tốc độ này.



## **1. Tốc độ thay đổi trung bình: Bức tranh tổng thể (Average Rate of Change: The Big Picture)**

Hãy tưởng tượng bạn đang thực hiện một chuyến đi đường dài. Bạn bắt đầu tại cột mốc dặm số 10 lúc 1:00 PM và đến cột mốc dặm số 130 lúc 3:00 PM. Vậy vận tốc trung bình của bạn là bao nhiêu?

- Quãng đường bạn đã đi là: $ 130-10=120 \text{ dặm}$

- Thời gian đã sử dụng là: $ 3{:}00\text{ PM}-1{:}00\text{ PM}=2\text{ giờ} $

- Vận tốc trung bình của bạn là:

$$\text{Vận tốc trung bình} = \frac{\text{Tổng quãng đường}}{\text{Tổng thời gian}} = \frac{120\text{ dặm}}{2\text{ giờ}} = 60\text{ dặm/giờ}$$


Đây chính là **Tốc độ thay đổi trung bình (Average Rate of Change)** của vị trí trong khoảng thời gian 2 giờ đó. Nó cho bạn biết khái quát vị trí đã thay đổi nhanh như thế nào trong suốt chuyến đi, nhưng không cho biết liệu bạn có lái xe chính xác ở vận tốc 60 dặm/giờ trong toàn bộ thời gian hay không. Có thể bạn đã dừng lại để đổ xăng hoặc tăng tốc khi đi trên đường cao tốc thông thoáng.

Về mặt toán học, nếu chúng ta có một hàm số $f(x)$, **Average Rate of Change** giữa hai điểm, chẳng hạn $x_1$ và $x_2$, là độ thay đổi của giá trị đầu ra của hàm số (giá trị $y$) chia cho độ thay đổi của đầu vào (giá trị $x$). Chúng ta thường sử dụng chữ cái Hy Lạp delta ($Δ$) để biểu thị "độ thay đổi". Vì vậy, $Δy$ có nghĩa là độ thay đổi của $y$, còn $Δx$ có nghĩa là độ thay đổi của $x$.

- Công thức là:

$$
\text{Average Rate of Change}
=\frac{\Delta y}{\Delta x}
=\frac{f(x_2)-f(x_1)}{x_2-x_1}
$$

Xét về mặt hình học, công thức này tính **độ dốc (Slope)** của đường thẳng nối hai điểm $(x_1, f(x_1))$ và $(x_2, f(x_2))$ trên **đồ thị (Graph)** của hàm số. Đường thẳng này được gọi là **đường cát tuyến (Secant Line)**.

Hãy xét một hàm số đơn giản: $ f(x)=x^2 $. Tốc độ thay đổi trung bình giữa $x_1=1$ và $x_2=3$ là bao nhiêu?

Đầu tiên, hãy tìm các giá trị $y$ tương ứng:

$$
f(1)=1^2=1
$$

$$
f(3)=3^2=9
$$

Bây giờ, áp dụng công thức:

$$\text{Average Rate of Change} = \frac{f(3)-f(1)}{3-1} = \frac{9-1}{2} = \frac{8}{2} = 4$$


Như vậy, trung bình, hàm số $f(x)=x^2$ tăng 4 đơn vị theo $y$ với mỗi 1 đơn vị tăng của $x$ trong khoảng từ $x=1$ đến $x=3$.

## **2. Tốc độ thay đổi tức thời: Góc nhìn tại từng thời điểm (Instantaneous Rate of Change: The Moment-by-Moment View)**

**Average Rate of Change** rất hữu ích, nhưng thường chúng ta cần biết một đại lượng đang thay đổi nhanh như thế nào ngay lúc này, tại một thời điểm cụ thể. Nếu bạn nhìn vào đồng hồ tốc độ của ô tô, nó cho biết vận tốc của bạn tại chính thời điểm đó, chứ không phải vận tốc trung bình trong một giờ vừa qua. Đây chính là **Tốc độ thay đổi tức thời (Instantaneous Rate of Change)**.

Làm thế nào để tìm tốc độ thay đổi tại một điểm duy nhất, chẳng hạn tại $x=x_1$?

Công thức **Average Rate of Change** của chúng ta: $$\frac{f(x_2)-f(x_1)}{x_2-x_1} $$

gặp phải một vấn đề. Nếu chúng ta cố gắng thu nhỏ khoảng này ngày càng nhiều bằng cách đưa $x_2$ đến gần $x_1$, thì cuối cùng $x_2$ sẽ bằng $x_1$. Khi đó, mẫu số $x_2-x_1$ bằng không và chúng ta không thể thực hiện phép chia cho không.

Đây chính là lúc khái niệm **giới hạn (Limit)** phát huy tác dụng. Thay vì đặt $x_2$ bằng $x_1$, chúng ta đặt câu hỏi:

> **Average Rate of Change** tiến đến giá trị nào khi $x_2$ tiến gần tùy ý đến $x_1$?

Hãy nghĩ về **Secant Line** trên Graph. Khi chúng ta di chuyển điểm $(x_2, f(x_2))$ ngày càng gần đến điểm $(x_1, f(x_1))$, **Secant Line** sẽ xoay quanh điểm đó. Đường thẳng mà nó tiến đến theo **Limit** được gọi là **đường tiếp tuyến (Tangent Line)** tại điểm $(x_1, f(x_1))$. **Tangent Line** tiếp xúc với đường cong tại điểm duy nhất đó và biểu thị hướng của đường cong ngay tại vị trí chính xác ấy.

Slope của **Tangent Line** chính là **Instantaneous Rate of Change** của hàm số tại $x=x_1$.



> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/secant_to_tangent_function.html)

*Khi điểm thứ hai (các điểm đánh dấu màu cam, vàng, xanh lá và xanh lơ) tiến gần đến điểm thứ nhất (điểm đánh dấu màu đỏ tại $x=1$), Slope của **Secant Line** tiến gần đến Slope của **Tangent Line** màu tím. Slope của **Tangent Line** này biểu thị **Instantaneous Rate of Change** tại $x=1$. (Nhấp vào Play để chạy hoạt ảnh).*



**Instantaneous Rate of Change** chính là điều chúng ta muốn nói đến khi đề cập đến **đạo hàm (Derivative)** của một hàm số tại một điểm. Nó cho chúng ta biết **Output** của hàm số nhạy cảm như thế nào với những thay đổi rất nhỏ của **Input** ngay tại vị trí đó.
