# **Giới thiệu về đạo hàm cấp cao (Introduction to Higher-Order Derivatives)**

**Đạo hàm (Derivative)** $f'(x)$ cho biết **tốc độ thay đổi tức thời (Instantaneous Rate of Change)** của hàm số $f(x)$. Nói cách khác, nó đo lường mức độ thay đổi nhanh hay chậm của giá trị đầu ra khi giá trị đầu vào thay đổi. Về mặt hình học, đạo hàm chính là độ dốc của đường tiếp tuyến với đồ thị hàm số tại một điểm.

Nhưng nếu chúng ta muốn biết tốc độ thay đổi đó đang thay đổi như thế nào thì sao?

Cũng như cách chúng ta tính tốc độ thay đổi của một hàm số, chúng ta có thể tiếp tục tính tốc độ thay đổi của chính đạo hàm của hàm số đó. Từ đây, chúng ta có khái niệm **đạo hàm cấp cao (Higher-Order Derivatives)**.




## **1. Đạo hàm cấp hai (The Second Derivative)**

Trong các loại đạo hàm cấp cao, loại thường gặp nhất là **đạo hàm cấp hai (Second Derivative)**. Đạo hàm này đặc biệt quan trọng khi nghiên cứu các bài toán tối ưu hóa trong học máy.

Nói đơn giản, đạo hàm cấp hai chính là đạo hàm của đạo hàm cấp một.

Với một hàm số $f(x)$, chúng ta thực hiện hai bước:

1. Tính đạo hàm cấp một, tức là $f'(x)$.
2. Tiếp tục tính đạo hàm của $f'(x)$ để thu được đạo hàm cấp hai.


### **1.1 Ký hiệu (Notation)**

Tương tự đạo hàm cấp một, đạo hàm cấp hai cũng có nhiều cách ký hiệu:

- **Ký hiệu Lagrange (Lagrange Notation):** Nếu đạo hàm cấp một được viết là $f'(x)$, thì đạo hàm cấp hai được viết là $f''(x)$, đọc là “f hai phẩy của x” (*f double prime of x*).

- **Ký hiệu Leibniz (Leibniz Notation):** Nếu đạo hàm cấp một được viết là $\frac{dy}{dx}$, thì đạo hàm cấp hai được viết là: $\frac{d^2y}{dx^2}$. Đọc là “đạo hàm bậc hai của y theo x” (*d squared y by dx squared*).

  Hãy chú ý vị trí của các số mũ 2. Chúng cho biết chúng ta thực hiện phép lấy đạo hàm hai lần theo biến $x$, tức là áp dụng toán tử đạo hàm $\frac{d}{dx}$ hai lần.



### **1.2. Cách tính**

Để tính đạo hàm cấp hai, chúng ta chỉ cần áp dụng các quy tắc tính đạo hàm đã học thêm một lần nữa.

Hãy xét ví dụ sau với hàm số:

$$
f(x)=x^3+2x^2-5x+1
$$



#### **Bước 1: Tính đạo hàm cấp một**

Sử dụng **Quy tắc lũy thừa (Power Rule)**, **Quy tắc nhân với hằng số (Constant Multiple Rule)** và **Quy tắc tổng (Sum Rule)**:

$$ f'(x) = \frac{d}{dx}(x^3) + \frac{d}{dx}(2x^2) - \frac{d}{dx}(5x) + \frac{d}{dx}(1) $$

Áp dụng các quy tắc tính đạo hàm cho từng hạng tử:

$$ f'(x) = 3x^{3-1} + 2(2x^{2-1}) - 5(1x^{1-1}) + 0 $$


Rút gọn, ta được:

$$ f'(x)=3x^2+4x-5 $$



#### **Bước 2: Tính đạo hàm cấp hai.**

Để tìm đạo hàm cấp hai $f''(x)$, chúng ta tiếp tục lấy đạo hàm của $f'(x)$ theo biến $x$:

$$ f''(x) = \frac{d}{dx}(f'(x)) = \frac{d}{dx}(3x^2+4x-5) $$


Tiếp tục lấy đạo hàm từng hạng tử:

$$
f''(x)
=\frac{d}{dx}(3x^2)
+\frac{d}{dx}(4x)
-\frac{d}{dx}(5)
$$

Áp dụng các quy tắc đã học:

$$
f''(x)=3(2x^{2-1})+4(1x^{1-1})-0
$$

Rút gọn, ta có:

$$
f''(x)=6x+4
$$

Như vậy, với hàm số $f(x)=x^3+2x^2-5x+1$:

- Đạo hàm cấp một là:

  $$
  f'(x)=3x^2+4x-5
  $$

- Đạo hàm cấp hai là:

  $$
  f''(x)=6x+4
  $$



## **2. Đạo hàm cấp hai cho chúng ta biết điều gì? (What Does the Second Derivative Tell Us?)**

Đạo hàm cấp một $f'(x)$ cho biết độ dốc của đồ thị hàm số tại mỗi điểm.

Còn đạo hàm cấp hai $f''(x)$ cho biết **độ dốc đó đang tăng hay giảm**. Nói cách khác, nó cho chúng ta biết đồ thị đang cong theo hướng nào.

Có ba trường hợp cần chú ý:

- **Nếu $f''(x)>0$ tại một điểm:** Độ dốc $f'(x)$ đang tăng khi $x$ tăng. Đồ thị có xu hướng cong lên trên, giống như hình dạng của một chiếc cốc đang ngửa để đựng nước. Ta gọi đây là **lõm lên (Concave Up)**.

- **Nếu $f''(x)<0$ tại một điểm:** Độ dốc $f'(x)$ đang giảm khi $x$ tăng. Đồ thị có xu hướng cong xuống dưới, giống như một chiếc cốc bị úp ngược. Ta gọi đây là **lõm xuống (Concave Down)**.

- **Nếu $f''(x)=0$ tại một điểm:** Độ cong của đồ thị có thể đang thay đổi tại điểm đó. Điểm này có khả năng là **điểm uốn (Inflection Point)**. Tuy nhiên, chỉ dựa vào điều kiện $f''(x)=0$ thì chưa thể kết luận chắc chắn đó là điểm uốn.

Hãy liên hệ với các khái niệm vận tốc và gia tốc trong vật lý:

Giả sử $f(x)$ biểu thị vị trí của một vật thể.

- Đạo hàm cấp một $f'(x)$ biểu thị **vận tốc**, tức tốc độ thay đổi vị trí theo thời gian.
- Đạo hàm cấp hai $f''(x)$ biểu thị **gia tốc**, tức tốc độ thay đổi của vận tốc theo thời gian.

Gia tốc dương có nghĩa là vận tốc đang tăng. Gia tốc âm có nghĩa là vận tốc đang giảm (trong trường hợp này có thể được hiểu là sự giảm tốc).

Đồ thị minh họa hai hàm số:

- $y=x^2$: Đồ thị luôn **lõm lên**, vì đạo hàm cấp hai luôn bằng $f''(x)=2$.
- $y=-x^2$: Đồ thị luôn **lõm xuống**, vì đạo hàm cấp hai luôn bằng $f''(x)=-2$.

## **3. Sau đạo hàm cấp hai (After the Second Derivative)**

Chúng ta có thể tiếp tục tính đạo hàm nhiều lần nữa không? Câu trả lời là có.

Chúng ta có thể lấy đạo hàm của đạo hàm cấp hai để thu được **đạo hàm cấp ba (Third Derivative)**, được ký hiệu là:

$$
f'''(x)
\quad \text{hoặc} \quad
\frac{d^3y}{dx^3}
$$

Chúng ta cũng có thể tiếp tục tính đạo hàm cấp bốn, cấp năm và các cấp cao hơn. Tất cả những đạo hàm này được gọi chung là **Higher-Order Derivatives**.

----
Mặc dù có thể tính đạo hàm ở nhiều cấp khác nhau, đạo hàm cấp một và cấp hai là những loại thường gặp nhất và có tính ứng dụng thực tế cao nhất trong các kiến thức nền tảng về học máy, đặc biệt là trong lĩnh vực tối ưu hóa.

Như chúng ta sẽ tìm hiểu ở phần sau, đạo hàm cấp hai còn giúp xác định một điểm được tìm thấy bằng đạo hàm cấp một có thực sự là điểm cực tiểu (chẳng hạn đáy của một vùng trũng trên đồ thị hàm chi phí) hay điểm cực đại hay không.

Ở thời điểm này, điều quan trọng cần nhớ là chúng ta có thể lấy đạo hàm của một hàm số nhiều lần. Trong đó, đạo hàm cấp hai cung cấp thông tin về độ cong của đồ thị, hay nói cách khác, cho biết đồ thị lõm lên hay lõm xuống
