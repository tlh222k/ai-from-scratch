## **1. Đạo hàm của một hằng số: Không thay đổi nghĩa là độ dốc bằng không (The Derivative of a Constant: No Change Means Zero Slope)**

## **2. Quy tắc hằng số (Constant Rule)**

Nếu $c$ là một số hằng, thì đạo hàm của hàm số $f(x)=c$ bằng $0$.

Sử dụng ký hiệu đạo hàm:

- Nếu $f(x)=c$, thì: $ f'(x)=0 $

- Ngoài ra, sử dụng ký hiệu **Leibniz**: $ \frac{d}{dx}(c)=0 $

Điều này phù hợp với trực giác: nếu một đại lượng không bao giờ thay đổi (nó là hằng số), thì tốc độ thay đổi của nó bằng $0$.

**Ví dụ:**

- Nếu $f(x)=100$, thì $f'(x)=0$.
- Nếu $y=-3.14$, thì $\frac{dy}{dx}=0$.
- $\frac{d}{dx}(2)=0$.


## **3. Kết hợp các hàm số: Quy tắc tổng và hiệu (Combining Functions: The Sum and Difference Rules)**

Bây giờ, điều gì sẽ xảy ra nếu chúng ta có một hàm số được tạo thành bằng cách cộng hoặc trừ các hàm số đơn giản hơn? Chẳng hạn:

$$
h(x)=x^2+x
$$

Chúng ta đã biết cách tìm đạo hàm của $x^2$ (kết quả là $2x$ theo **Power Rule**) và đạo hàm của $x$ (mà $x$ chính là $x^1$, nên đạo hàm của nó là $1x^0=1$, cũng theo **Power Rule**).

> Vậy phép cộng ảnh hưởng như thế nào đến đạo hàm?

**Calculus** cung cấp một quy tắc đơn giản: đạo hàm của tổng các hàm số chính là tổng các đạo hàm riêng của chúng. Điều tương tự cũng áp dụng cho phép trừ.



### **3.1. Quy tắc tổng (Sum Rule)**

Nếu $h(x)=f(x)+g(x)$, thì: $$ h'(x)=f'(x)+g'(x)$$

Sử dụng ký hiệu **Leibniz**:

$$
\frac{d}{dx}[f(x)+g(x)]
=
\frac{d}{dx}f(x)+\frac{d}{dx}g(x)
$$



### **3.2. Quy tắc hiệu (Difference Rule)**

Nếu $h(x)=f(x)-g(x)$, thì: $$ h'(x)=f'(x)-g'(x)$$ Sử dụng ký hiệu **Leibniz**:

$$
\frac{d}{dx}[f(x)-g(x)]
=
\frac{d}{dx}f(x)-\frac{d}{dx}g(x)
$$

Về cơ bản, bạn có thể tính đạo hàm của một hàm số bằng cách lấy đạo hàm từng hạng tử.

Hãy áp dụng điều này cho ví dụ $h(x)=x^2+x$:

- Tìm đạo hàm của hạng tử thứ nhất, $x^2$. Sử dụng **Power Rule**:

  $$
  \frac{d}{dx}(x^2)=2x
  $$

- Tìm đạo hàm của hạng tử thứ hai, $x$. Sử dụng **Power Rule**:

  $$
  \frac{d}{dx}(x)=1
  $$

- Cộng các kết quả lại:

  $$
  h'(x)=2x+1
  $$



## **4. Xử lý các thừa số hằng số (Handling Constant Multiples)**

Điều gì sẽ xảy ra với một hàm số như $f(x)=3x^2$?

Ở đây, $x^2$ được nhân với một hằng số là 3. Trường hợp này cũng có một quy tắc đơn giản. Các thừa số hằng số chỉ đơn giản là được giữ nguyên khi bạn tính đạo hàm.



### **4.1. Quy tắc nhân với hằng số (Constant Multiple Rule)**

Nếu $h(x)=c\cdot f(x)$, trong đó $c$ là một hằng số, thì:

$$
h'(x)=c\cdot f'(x)
$$

Sử dụng ký hiệu **Leibniz**:

$$
\frac{d}{dx}[c\cdot f(x)]
=
c\cdot\frac{d}{dx}f(x)
$$

Vì vậy, với $f(x)=3x^2$:

- Xác định thừa số hằng số: $c=3$.
- Xác định hàm số được nhân với hằng số: $f(x)=x^2$.
- Tìm đạo hàm của hàm số đó:

  $$
  \frac{d}{dx}(x^2)=2x
  $$

- Nhân kết quả với hằng số:

  $$
  f'(x)=3\cdot(2x)=6x
  $$



## **5. Kết hợp tất cả (Putting It All Together)**

Bây giờ, chúng ta có thể kết hợp **Constant Rule**, **Sum Rule**, **Difference Rule**, **Constant Multiple Rule** và **Power Rule** để tính đạo hàm của bất kỳ hàm đa thức nào. Đa thức đơn giản là tổng của các hạng tử, trong đó mỗi hạng tử là một hằng số nhân với $x$ được nâng lên một lũy thừa nguyên không âm (chẳng hạn như $5x^3-2x+7$).

Hãy tìm đạo hàm của:

$$
p(x)=4x^3-5x^2+x-2
$$

Chúng ta có thể tính đạo hàm của hàm số này theo từng hạng tử:

**Hạng tử 1: $4x^3$**

- Đạo hàm của $x^3$ là $3x^2$.
- Áp dụng **Constant Multiple Rule**:

  $$
  \frac{d}{dx}(4x^3)=4\cdot(3x^2)=12x^2
  $$

**Hạng tử 2: $-5x^2$**

- Đạo hàm của $x^2$ là $2x$.
- Áp dụng **Constant Multiple Rule**:

  $$
  \frac{d}{dx}(-5x^2)=-5\cdot(2x)=-10x
  $$

**Hạng tử 3: $+x$ (hoặc $+1x^1$)**

- Đạo hàm của $x^1$ là $1x^0=1$.
- Có thể áp dụng **Constant Multiple Rule** (không bắt buộc ở đây):

  $$
  \frac{d}{dx}(1x)=1\cdot(1)=1
  $$

**Hạng tử 4: $-2$**

- Đây là một hằng số.
- Áp dụng **Constant Rule**:

  $$
  \frac{d}{dx}(-2)=0
  $$

Bây giờ, hãy kết hợp đạo hàm của từng hạng tử bằng cách sử dụng **Sum Rule** và **Difference Rule**:

$$
p'(x)=12x^2-10x+1-0
$$

$$
p'(x)=12x^2-10x+1
$$
