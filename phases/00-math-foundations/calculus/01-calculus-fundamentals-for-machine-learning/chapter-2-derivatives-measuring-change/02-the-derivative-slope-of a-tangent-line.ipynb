# **Đạo hàm: Độ dốc của đường tiếp tuyến (The Derivative: Slope of a Tangent Line)**

Hàm số thể hiện sự thay đổi, có thể được đo lường bằng **tốc độ thay đổi trung bình (Average Rate of Change)** trên một khoảng hoặc **tốc độ thay đổi tức thời (Instantaneous Rate of Change)** tại một thời điểm cụ thể. **Average Rate of Change** cung cấp cái nhìn tổng quát, nhưng trong nhiều trường hợp, việc xác định chính xác một hàm số thay đổi nhanh như thế nào *tại một điểm cụ thể* là điều rất quan trọng.

Hãy nghĩ đến đồng hồ tốc độ trên ô tô của bạn. Nó không hiển thị vận tốc trung bình trong một giờ vừa qua mà hiển thị vận tốc của bạn *ngay lúc này*. Tốc độ thay đổi tại thời điểm "ngay lúc này" chính là điều chúng ta đang tìm hiểu, và nó dẫn chúng ta đến khái niệm **đạo hàm (Derivative)**.

Làm thế nào để trực quan hóa tốc độ thay đổi tức thời này? Hãy tưởng tượng bạn phóng to thật gần một điểm trên **đồ thị (Graph)** của một hàm số. Nếu hàm số trơn (không có góc nhọn hoặc điểm gián đoạn), đường cong sẽ ngày càng trông giống một đường thẳng. Đường thẳng này chỉ tiếp xúc với đường cong tại điểm duy nhất mà chúng ta quan tâm và có cùng hướng với đường cong tại điểm đó. Nó được gọi là **đường tiếp tuyến (Tangent Line)**.

Ví dụ, **đường cát tuyến (Secant Line)** là đường thẳng được vẽ đi qua *hai* điểm trên đường cong. **Độ dốc (Slope)** của **Secant Line** biểu thị **Average Rate of Change** giữa hai điểm đó.

Bây giờ, hãy tưởng tượng di chuyển một trong hai điểm dọc theo đường cong, ngày càng tiến gần đến điểm còn lại. Khi khoảng cách giữa hai điểm thu hẹp, **Secant Line** sẽ xoay và ngày càng tiến gần đến việc trở thành **Tangent Line** tại điểm cố định. Slope của **Secant Line** tiến gần đến Slope của **Tangent Line**.


> [Graph](https://tlh222k.github.io/ai-from-scratch/visualizations/secant_line_function.html)


> *Hàm số $f(x)=x²$ (màu xanh dương). Khi điểm thứ hai của **Secant Line** (màu cam, đỏ) tiến gần đến điểm tại $x=1$, Slope của **Secant Line** tiến gần đến Slope của **Tangent Line** (đường nét đứt màu xanh lá) tại $x=1$.*



Điều này dẫn chúng ta đến định nghĩa hình học của **Derivative**:

> **Derivative của một hàm số $f(x)$ tại điểm $x=a$ là Slope của Tangent Line với Graph của $f(x)$ tại điểm đó.**



Tại sao Slope lại quan trọng? Hãy nhớ rằng Slope đo lường độ nghiêng hoặc tốc độ thay đổi. Slope càng lớn thì sự thay đổi càng nhanh. Slope của **Tangent Line** cho chúng ta biết chính xác **đầu ra (Output)** $y$ của hàm số đang thay đổi nhanh như thế nào so với **đầu vào (Input)** $x$ *ngay tại thời điểm đó*.

- Nếu **Tangent Line** hướng lên trên (Slope dương), hàm số đang tăng tại điểm đó.
- Nếu **Tangent Line** hướng xuống dưới (Slope âm), hàm số đang giảm tại điểm đó.
- Nếu **Tangent Line** nằm ngang (Slope bằng không), hàm số tạm thời không thay đổi; nó không tăng cũng không giảm ngay tại điểm đó. Điều này thường xảy ra tại các điểm cực đại và cực tiểu của đường cong.

Về mặt toán học, chúng ta biểu diễn ý tưởng Slope của **Secant Line** tiến gần đến Slope của **Tangent Line** bằng cách sử dụng **giới hạn (Limit)**. Nếu hai điểm trên **Secant Line** là $(x,f(x))$ và $(x+h,f(x+h))$, Slope của **Secant Line** là:

$$ \text{Slope}_{\text{secant}} = \frac{\Delta y}{\Delta x} = \frac{f(x+h)-f(x)}{(x+h)-x} = \frac{f(x+h)-f(x)}{h} $$

Để tìm Slope của **Tangent Line**, chúng ta lấy **Limit** của biểu thức này khi khoảng cách $h$ giữa hai điểm tiến đến không:

$$
\text{Slope}_{\text{tangent}}
=f'(x)
=\lim_{h\to 0}\frac{f(x+h)-f(x)}{h}
$$


----
**Limit** này, nếu tồn tại, *chính là* **Derivative** của hàm số $f$ tại điểm $x$. Hiện tại, bạn chưa cần quá lo lắng về việc tính toán trực tiếp. Điều quan trọng cần ghi nhớ là mối liên hệ trực quan này: **Derivative** đo lường **Instantaneous Rate of Change**, và xét về mặt hình học, đại lượng này tương ứng với Slope của đường thẳng tiếp tuyến với Graph hàm số tại điểm cụ thể đó. Mối liên hệ này là nền tảng để hiểu cách chúng ta sử dụng **Derivative** nhằm tìm ra các nghiệm tối ưu trong **Machine Learning**.
