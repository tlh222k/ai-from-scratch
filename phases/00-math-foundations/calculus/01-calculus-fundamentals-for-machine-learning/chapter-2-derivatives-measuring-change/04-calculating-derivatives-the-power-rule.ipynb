# **Tính đạo hàm: Quy tắc lũy thừa (Calculating Derivatives: The Power Rule)**

**Đạo hàm (Derivative)** định lượng **tốc độ thay đổi tức thời (Instantaneous Rate of Change)**, biểu thị **độ dốc (Slope)** của **đường tiếp tuyến (Tangent Line)** với đồ thị của một hàm số tại bất kỳ điểm nào.

Việc tính trực tiếp **độ dốc (Slope)** này bằng định nghĩa đạo hàm thông qua **giới hạn (Limit)**, mặc dù có ý nghĩa nền tảng, nhưng có thể khá cồng kềnh đối với những hàm số phức tạp hơn các hàm đơn giản nhất. Hãy thử tưởng tượng việc phải tìm **giới hạn (Limit)** cho hàm số $f(x)=x^7$ hoặc một đa thức phức tạp hơn!

May mắn thay, các nhà toán học đã xây dựng những quy tắc tính nhanh để tìm đạo hàm của các dạng hàm số phổ biến. Một trong những quy tắc nền tảng và được sử dụng thường xuyên nhất là **Quy tắc lũy thừa (Power Rule)**. Quy tắc này cung cấp một cách đơn giản để tìm đạo hàm của các hàm số có biến được nâng lên một lũy thừa, chẳng hạn như $x^2$, $x^3$, hoặc thậm chí chỉ là $x$.

**Power Rule** áp dụng cho các hàm số có dạng $f(x)=x^n$, trong đó $n$ là số mũ thực bất kỳ. Trong phạm vi khóa học nhập môn này, chúng ta chủ yếu tập trung vào các trường hợp mà $n$ là số nguyên không âm (chẳng hạn như $0, 1, 2, 3$, ...).

## **1. Phát biểu quy tắc lũy thừa (The Power Rule Stated)**

Với hàm số $f(x)=x^n$, đạo hàm của nó, được ký hiệu là $f'(x)$ hoặc $\frac{d}{dx}(x^n)$, được xác định bởi:

$$
f'(x)=nx^{n-1}
$$

Nói một cách đơn giản:

- **Đưa số mũ xuống:** Nhân hạng tử với số mũ ban đầu $n$.
- **Giảm số mũ:** Trừ 1 khỏi số mũ ban đầu.

Hãy cùng xem quy tắc này được áp dụng như thế nào qua một vài ví dụ.



#### **Ví dụ 1: $f(x)=x^2$ (Example 1: $f(x)=x^2$)**

Hàm số này biểu diễn một đường parabol đơn giản. Hãy áp dụng **Power Rule**.

Ở đây, số mũ $n=2$.

- **Đưa số mũ xuống:** Ta nhân với 2.
- **Giảm số mũ:** Số mũ mới là $2-1=1$.

Vì vậy, đạo hàm là:

$$
f'(x)=2x^{2-1}=2x^1=2x
$$

Kết quả này, $f'(x)=2x$, cho chúng ta biết **độ dốc (Slope)** của **Tangent Line** với parabol $y=x^2$ tại bất kỳ điểm $x$ nào. Chẳng hạn:

 - Tại $x=1$, **độ dốc (Slope)** là $2(1)=2$.
 - Tại $x=0$, **độ dốc (Slope)** là $2(0)=0$ (đáy của parabol).
 - Tại $x=-3$, **độ dốc (Slope)** là $2(-3)=-6$.



#### **Ví dụ 2: $f(x)=x^5$ (Example 2: $f(x)=x^5$)**

Ở đây, $n=5$.

- **Đưa số mũ xuống:** Nhân với 5.
- **Giảm số mũ:** Số mũ mới là $5-1=4$.

Đạo hàm là:

$$
f'(x)=5x^{5-1}=5x^4
$$



#### **Ví dụ 3: $f(x)=x$ (Example 3: $f(x)=x$)**

Điều này có thể trông khác một chút, nhưng hãy nhớ rằng $x$ cũng chính là $x^1$. Vì vậy, ở đây $n=1$.

- **Đưa số mũ xuống:** Nhân với 1.
- **Giảm số mũ:** Số mũ mới là $1-1=0$.

Đạo hàm là:

$$
f'(x)=1x^{1-1}=1x^0
$$

Bây giờ, hãy nhớ rằng bất kỳ số khác $0$ nào được nâng lên lũy thừa $0$ đều bằng 1 (do đó $x^0=1$, với giả thiết $x\neq 0$).

$$
f'(x)=1\times 1=1
$$

Đạo hàm của $f(x)=x$ là $f'(x)=1$. Điều này hoàn toàn phù hợp với trực giác!

Đồ thị của $y=x$ là một đường thẳng có **độ dốc (Slope)** bằng 1 tại mọi điểm. Đạo hàm đã phản ánh chính xác **độ dốc (Slope)** không đổi này.



## **2. Còn các hằng số thì sao? (Xem trước) (What about constants? A Sneak Peek)**

Chẳng hạn, xét một hàm hằng như $f(x)=7$. Làm thế nào chúng ta có thể xem hàm này dưới góc độ **Power Rule**? Ta có thể viết số 7 thành $7x^0$.

Mặc dù **Power Rule** áp dụng trực tiếp cho $x^n$, chúng ta sẽ sớm tìm hiểu một quy tắc riêng dành cho hằng số. Tuy nhiên, $x^0$ là một trường hợp đáng để xem xét.

Nếu chỉ xét $g(x)=x^0$, **Power Rule** sẽ cho kết quả:

$$
g'(x)=0x^{0-1}=0x^{-1}=0
$$

Đạo hàm bằng 0. Điều này phù hợp với trực giác rằng một hàm hằng như $y=7$ là một đường thẳng nằm ngang, và các đường thẳng nằm ngang luôn có **độ dốc (Slope)** bằng 0. Chúng ta sẽ trình bày chính thức điều này bằng **Quy tắc hằng số (Constant Rule)** trong phần tiếp theo.

----

**Power Rule** giúp đơn giản hóa đáng kể việc tính toán. Thay vì phải xử lý các **giới hạn (Limit)** mỗi lần, giờ đây chúng ta có thể nhanh chóng tìm đạo hàm (tốc độ thay đổi) cho bất kỳ hạng tử nào có dạng $x^n$. Quy tắc này là một viên gạch nền tảng mà chúng ta sẽ sử dụng rộng rãi, đặc biệt khi bắt đầu tìm hiểu các hàm chi phí thường gặp trong **Học máy (Machine Learning)**, vốn thường chứa các hạng tử bình phương hoặc những lũy thừa khác của biến.
